# **2. Distribución de Bernoulli**
## **a. Definición, Probabilidad y Momentos**
+ Una variable aleatoria $X$ tiene distribución de probabilidad de Bernoulli si únicamente puede tomar dos valores  {0, 1} = {“fracaso”,  “éxito”}, y se define:
  + $P\{X=1\}=p$
  + $P\{X=0\}=1-p$

+ La distribución de bernoulli modela el número de éxitos en una secuencia de $n=1$ ensayo, donde hay dos posibles resultados: éxito o fracaso, y la probabilidad de éxito es $p$.

- **Función de probabilidad**: $$P(X = k) = p^k (1 - p)^{1 - k}, k=0,1$$

- **Parámetros**:$p$ es la probabilidad de éxito en el ensayo.


### **Distribución Bernoulli - Momentos**
- **Valor esperado (Esperanza)**: $$E(X) = p$$
- **Varianza**: $$\text{Var}(X) =p(1 - p)$$
- **Desviación estándar**: $$\sigma(X) = \sqrt{p \cdot (1 - p)}$$



```python
# @title **b. Experimento de Bernoulli - sea $X=$# de caras**: $$P(X = k) = p^k (1 - p)^{1 - k}, k = 0, 1$$
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, nbinom, poisson, hypergeom, randint

# @title **A. Para Graficar cualquier distribucion discreta binomial, binomial negativa, Poisson e Hipergeométrica**
def plot_distribution2(dist, values, pmf, title):
    plt.bar(values, pmf)
    plt.title(title)
    plt.xlabel('Valores')
    plt.ylabel('Probabilidad')
    plt.show()

n, p = 1, 0.25
x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)
plot_distribution2(binom, x_binom, pmf_binom, "Distribución Bernolulli")
```

```python
# ==========================================
#@title **c. Experimento de Bernoulli en R**
# Sea X = número de caras
# P(X = k) = p^k (1-p)^(1-k), k = 0,1
# ==========================================
%%R
# Parámetros
n <- 1
p <- 0.5

# Valores posibles de X
x <- 0:n

# Función de probabilidad (PMF)
pmf <- dbinom(x, size = n, prob = p)

# Mostrar probabilidades
datos <- data.frame(
  X = x,
  Probabilidad = pmf
)

print(datos)

# Gráfico
barplot(
  pmf,
  names.arg = x,
  col = c("#3498DB", "#E74C3C"),
  border = "black",
  ylim = c(0, max(pmf) + 0.1),
  main = "Distribución de Bernoulli",
  xlab = "Número de caras (X)",
  ylab = "P(X = k)"
)

# Etiquetas con probabilidades
text(
  x = c(0.7, 1.9),
  y = pmf,
  labels = round(pmf, 2),
  pos = 3,
  cex = 1.1
)

grid()
```

# **3. Distribución Binomial**

## **a. Definición y Momentos**
+ La distribución binomial modela el número de éxitos en una secuencia de $n$ ensayos independientes de Bernoulli, donde cada ensayo tiene dos posibles resultados: éxito o fracaso, y la probabilidad de éxito es $p$.

- **Función de probabilidad**: $$P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}$$
$$\binom{n}{k} = \frac{n!}{k!(n-k)!}$$
- **Parámetros**:$n$ es el número de ensayos y $p$ es la probabilidad de éxito en cada ensayo.


### **Distribución Binomial - Momentos**
- **Valor esperado (Esperanza)**: $$E(X) = n \cdot p$$
- **Varianza**: $$\text{Var}(X) = n \cdot p \cdot (1 - p)$$
- **Desviación estándar**: $$\sigma(X) = \sqrt{n \cdot p \cdot (1 - p)}$$

## **5. Ejemplo 1. Lanzamiento de una moneda 10 veces. Gráfica de la Distribución Binomial con Función de probabilidad:**

```python
# @title **A. Ejemplo 1. Lanzamiento de una moneda 10 veces. Gráfica de la Distribución Binomial con Función de probabilidad**: $$P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}$$
n, p = 100, 0.75
x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)
plot_distribution2(binom, x_binom, pmf_binom, "Distribución Binomial")
```

```python
# ==========================================
# @title **B. Ejemplo 1. Lanzamiento de una moneda 10 veces en R**
# Distribución Binomial
#
# P(X = k) = choose(n,k) * p^k * (1-p)^(n-k)
# ==========================================
%%R
# Parámetros
n <- 10      # Número de lanzamientos
p <- 0.5     # Probabilidad de obtener cara

# Valores posibles de X
x <- 0:n

# Función de probabilidad (PMF)
pmf <- dbinom(x, size = n, prob = p)

# Tabla de probabilidades
datos <- data.frame(
  X = x,
  Probabilidad = round(pmf, 5)
)

print(datos)

# Gráfico de la distribución
barplot(
  pmf,
  names.arg = x,
  col = "#3498DB",
  border = "black",
  ylim = c(0, max(pmf) + 0.03),
  main = "Distribución Binomial (n = 10, p = 0.5)",
  xlab = "Número de caras (X)",
  ylab = "P(X = k)"
)

# Agregar etiquetas sobre cada barra
text(
  x = seq(0.7, by = 1.2, length.out = length(x)),
  y = pmf,
  labels = round(pmf, 3),
  pos = 3,
  cex = 0.8
)

grid()
```

## **6. Ejemplo 2. Supongamos que lanzamos una moneda justa (probabilidad de éxito p=0.5 , n=10 veces. ¿Cuál es la probabilidad de obtener exactamente 6 caras?:**

```python
# @title **A. Ejemplo 2. Supongamos que lanzamos una moneda justa (probabilidad de éxito  p=0.5 ,  n=10  veces. ¿Cuál es la probabilidad de obtener exactamente 6 caras?**: $$P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}$$

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, nbinom, poisson, hypergeom, randint
from scipy.stats import binom

# Parámetros
n = 10  # Número de ensayos
p = 0.5  # Probabilidad de éxito
k = 6  # Número de éxitos

# Probabilidad de obtener exactamente 6 caras
prob_binom_1 = binom.pmf(k, n, p)
print(f"Probabilidad de obtener exactamente 6 caras: P(X = 7) = {prob_binom_1:.5f}")

# Gráfica de la Distribución Binomial
x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)

plt.bar(x_binom, pmf_binom)
plt.title("Distribución Binomial")
plt.xlabel('Valores')
plt.ylabel('Probabilidad')

# Resaltar la barra de P(X = 6) en rojo
plt.bar(k, prob_binom_1, color='red')

# Agregar texto dentro del rectangulo rojo con la probabilidad
plt.text(k+1, prob_binom_1, f"P(X = 6) = {prob_binom_1:.5f}", ha='center', va='bottom')

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 2. Lanzamiento de una moneda justa 10 veces**
# ¿Cuál es la probabilidad de obtener exactamente 6 caras?
#
# P(X = k) = choose(n,k) p^k (1-p)^(n-k)
# ==========================================================
%%R
# Parámetros
n <- 10      # Número de lanzamientos
p <- 0.5     # Probabilidad de cara
k <- 6       # Número de caras

# Probabilidad de obtener exactamente 6 caras
prob_binom <- dbinom(k, size = n, prob = p)

cat(sprintf(
  "Probabilidad de obtener exactamente %d caras: P(X = %d) = %.5f\n",
  k, k, prob_binom
))

# Distribución Binomial
x <- 0:n
pmf <- dbinom(x, size = n, prob = p)

# Colores de las barras
colores <- rep("steelblue", length(x))
colores[k + 1] <- "red"   # En R los índices empiezan en 1

# Gráfico
bp <- barplot(
  pmf,
  names.arg = x,
  col = colores,
  border = "black",
  ylim = c(0, max(pmf) + 0.06),
  main = "Distribución Binomial",
  xlab = "Número de caras (X)",
  ylab = "Probabilidad"
)

# Etiqueta sobre la barra resaltada
text(
  x = bp[k + 1],
  y = prob_binom + 0.02,
  labels = paste0("P(X = ", k, ") = ", round(prob_binom, 5)),
  col = "red",
  cex = 1,
  font = 2
)

# Cuadrícula
grid()
```

## **7. Ejemplo 3. Momentos Modelo Binomial**

```python
# @title **A. Ejemplo 3. Momentos Modelo Binomial**
n = 10
p = 0.5
esperanza_binom = n * p
varianza_binom = n * p * (1 - p)
desviacion_binom = np.sqrt(varianza_binom)
print(f"Momentos en el Modelo Binomial")
print(f"Esperanza:E(X) = {n}*{p}= {esperanza_binom}")
print(f"Varianza: V(X) = {n} * {p} * {(1 - p)} = {varianza_binom}")
print(f"Desviación estándar: {desviacion_binom}")
```

```python
# =====================================================
# @title **B. Ejemplo 3. Momentos del Modelo Binomial en R**
# X ~ Bin(n = 25, p = 0.10)
# =====================================================
%%R
# Parámetros
n <- 10
p <- 0.5

# Momentos
esperanza <- n * p
varianza <- n * p * (1 - p)
desviacion <- sqrt(varianza)

# Resultados
cat("Momentos del Modelo Binomial\n")
cat("-------------------------------------\n")
cat(sprintf("Esperanza: E(X) = %d × %.2f = %.2f\n",
            n, p, esperanza))

cat(sprintf("Varianza: V(X) = %d × %.2f × %.2f = %.2f\n",
            n, p, 1 - p, varianza))

cat(sprintf("Desviación estándar: σ(X) = √%.2f = %.4f\n",
            varianza, desviacion))
```

## **8. Ejemplo 4. La probabilidad de que una pieza fabricada por una empresa sea defectuosa es 0.1. Halla la probabilidad de que en una muestra de 25 piezas se encuentren a lo sumo cuatro piezas defectuosas.**

```python
# @title **A. Ejemplo 4. La probabilidad de que una pieza fabricada por una empresa sea defectuosa es 0.1. Halla la probabilidad de que en una muestra de 25 piezas se encuentren a lo sumo cuatro piezas defectuosas.**

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 15  # Número de piezas
p = 0.65  # Probabilidad de que una pieza sea defectuosa
k = np.arange(0, n - 1)  # Número de piezas defectuosas posibles

# Calcula la probabilidad de obtener exactamente k piezas defectuosas
probabilidades = binom.pmf(k, n, p)

# Calcula la probabilidad de que haya a lo sumo 4 piezas defectuosas (P(X <= 4))
probabilidad_acumulada = binom.cdf(10, n, p)

# Crea la gráfica de la distribución binomial
plt.bar(k, probabilidades)
plt.title('Distribución Binomial')
plt.xlabel('Número de piezas defectuosas (X)')
plt.ylabel('Probabilidad')
plt.grid(True)

# Sobrescribe P(X <= 10) con rojo
for i in range(11):
    plt.bar(i, probabilidades[i], color='red')


# Agrega el valor de P(X <= 4) al gráfico
plt.text(4, 0.2, f"P(X <= 10) = {probabilidad_acumulada:.5f}", fontsize=12, color='black')

plt.show()

print(f"La probabilidad de que en una muestra de 25 piezas se encuentren a lo sumo cuatro piezas defectuosas es: {probabilidad_acumulada:.5f}")
```

```python
# ==========================================================
# @title **B. Ejemplo 4. Distribución Binomial en R**
#
# La probabilidad de que una pieza sea defectuosa es 0.10.
# Hallar la probabilidad de que en una muestra de 25 piezas
# se encuentren a lo sumo 4 piezas defectuosas.
# ==========================================================
%%R
# Parámetros
n <- 15      # Número de piezas
p <- 0.65    # Probabilidad de pieza defectuosa

# Valores posibles de X
x <- 0:n

# Función de probabilidad
pmf <- dbinom(x, size = n, prob = p)

# Probabilidad acumulada P(X <= 4)
prob_acumulada <- pbinom(10, size = n, prob = p)

# Colores de las barras
colores <- rep("steelblue", length(x))
colores[1:11] <- "red"      # En R: X = 0,1,2,3,4

# Gráfico
bp <- barplot(
  pmf,
  names.arg = x,
  col = colores,
  border = "black",
  ylim = c(0, max(pmf) + 0.08),
  main = "Distribución Binomial",
  xlab = "Número de piezas defectuosas (X)",
  ylab = "Probabilidad"
)

# Agregar texto con la probabilidad acumulada
text(
  x = bp[5],
  y = max(pmf) + 0.04,
  labels = paste0("P(X ≤ 10) = ", round(prob_acumulada, 3)),
  col = "black",
  cex = 1,
  font = 2
)

grid()

# Resultado
cat("\n")
cat("=============================================\n")
cat("Distribución Binomial\n")
cat("=============================================\n")
cat(sprintf("P(X ≤ 10) = %.3f\n", prob_acumulada))
```

## **9. Ejemplo 5. El 60% de los Psicólogo de una facultad encuentran trabajo el primer año después de acabar la carrera. De los 100 Psicólogos en un curso, ¿Cuántos se espera que se coloque en el primer año? ¿Cuál es la varianza?**

```python
# @title **A. Ejemplo 5. El 60% de los Psicólogo de una facultad encuentran trabajo el primer año después de acabar la carrera. De los 100 Psicólogos en un curso, ¿Cuántos se espera que se coloque en el primer año? ¿Cuál es la varianza?**

# Datos del problema
n = 100  # Número total de psicólogos
p = 0.6  # Probabilidad de encontrar trabajo en el primer año

# Calcular el número esperado de psicólogos que encontrarán trabajo
esperanza_binom = n * p

# Calcular la varianza
varianza_binom = n * p * (1 - p)

x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)
plot_distribution2(binom, x_binom, pmf_binom, "Distribución Binomial")

print(f"Se espera que {esperanza_binom:.0f} psicólogos encuentren trabajo en el primer año.")
print(f"La varianza es: {varianza_binom:.2f}")
```

```python
# ==========================================================
# @title **B. Ejemplo 5 en R.**
# El 60% de los Psicólogos de una facultad encuentran trabajo
# el primer año después de acabar la carrera.
#
# De los 100 Psicólogos en un curso:
# ¿Cuántos se espera que se coloque en el primer año?
# ¿Cuál es la varianza?
# ==========================================================

# Instalar ggplot2 si es necesario
# install.packages("ggplot2")
%%R
library(ggplot2)

# Parámetros
n <- 100
p <- 0.60

# ===========================================
# Momentos de la distribución binomial
# ===========================================

esperanza_binom <- n * p
varianza_binom  <- n * p * (1 - p)
desviacion_binom <- sqrt(varianza_binom)

# Mostrar resultados
cat("=========================================\n")
cat("Modelo Binomial\n")
cat("=========================================\n")
cat("Esperanza E(X) =", n, "*", p, "=", esperanza_binom, "\n")
cat("Varianza V(X) =", n, "*", p, "*", (1-p), "=", varianza_binom, "\n")
cat("Desviación estándar =", round(desviacion_binom,4), "\n\n")

cat("Se espera que", esperanza_binom,
    "psicólogos encuentren trabajo en el primer año.\n")
cat("La varianza es:", varianza_binom, "\n")

# ===========================================
# Función de probabilidad
# ===========================================

x <- 0:n
probabilidades <- dbinom(x, size = n, prob = p)

datos <- data.frame(
  Psicologos = x,
  Probabilidad = probabilidades
)

# ===========================================
# Gráfica
# ===========================================

ggplot(datos, aes(x = Psicologos, y = Probabilidad)) +
  geom_col(fill = "steelblue", color = "black") +
  geom_vline(xintercept = esperanza_binom,
             color = "red",
             linewidth = 1.2,
             linetype = "dashed") +
  annotate("text",
           x = esperanza_binom + 8,
           y = max(probabilidades)*0.95,
           label = paste("E(X) =", esperanza_binom),
           color = "red",
           size = 5) +
  labs(
    title = "Distribución Binomial",
    subtitle = paste("n =", n, ", p =", p),
    x = "Número de psicólogos que encuentran trabajo",
    y = "Probabilidad"
  ) +
  theme_minimal(base_size = 14)
```

## **10. Ejemplo 6. Resuelve: Suponiendo que el 40% de los Comunicadores Sociales tengan Rh-, si tomamos una muestra de siete Comunicadores Sociales, (a) ¿Qué probabilidad hay de que todas sean Rh-? (b) ¿Y de que máximo cuatro sean Rh-?**

```python
# @title **A. Ejemplo 6. Resuelve: Suponiendo que el 40% de los Comunicadores Sociales tengan Rh-, si tomamos una muestra de siete Comunicadores Sociales,   (a)  ¿Qué probabilidad hay de que todas sean Rh-?  (b) ¿Y de que máximo cuatro sean Rh-?**

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 7  # Número de comunicadores sociales en la muestra
p = 0.4  # Probabilidad de que un comunicador social tenga Rh-

# (a) Probabilidad de que todas sean Rh- prob_minimo_diez
prob_todas_rh_neg = binom.pmf(n, n, p)
print(f"Probabilidad de que todas sean Rh-: P(X = 7) = {prob_todas_rh_neg:.5f}")

# Gráfico sombreando P(X = 7)
x_binom = np.arange(0, n + 3)
pmf_binom = binom.pmf(x_binom, n, p)
plt.bar(x_binom, pmf_binom)
plt.title('Distribución Binomial - P(X = 7)')
plt.xlabel('Número de comunicadores sociales con Rh-')
plt.ylabel('Probabilidad')

plt.bar(7, prob_todas_rh_neg, color='red')
plt.text(7, prob_todas_rh_neg + 0.01, f"P(X = 7) = {prob_todas_rh_neg:.5f}", fontsize=10, color='black')

plt.show()

# (b) Probabilidad de que máximo cuatro sean Rh-
prob_maximo_cuatro_rh_neg = binom.cdf(4, n, p)
print(f"Probabilidad de que máximo cuatro sean Rh-: P(X <= 4) = {prob_maximo_cuatro_rh_neg:.5f}")

# Gráfico sombreando P(X <= 4)
x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)
plt.bar(x_binom, pmf_binom)
plt.title('Distribución Binomial - P(X <= 4)')
plt.xlabel('Número de comunicadores sociales con Rh-')
plt.ylabel('Probabilidad')

for i in range(5):
    plt.bar(i, pmf_binom[i], color='red')

plt.text(4, 0.1, f"P(X <= 4) = {prob_maximo_cuatro_rh_neg:.4f}", fontsize=12, color='black')

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 6, usando R**
# Suponiendo que el 40% de los Comunicadores Sociales
# tengan Rh-, si tomamos una muestra de siete
# Comunicadores Sociales:
#
# (a) ¿Qué probabilidad hay de que todas sean Rh-?
# (b) ¿Qué probabilidad hay de que máximo cuatro sean Rh-?
# ==========================================================

# Instalar ggplot2 si es necesario
# install.packages("ggplot2")
%%R
library(ggplot2)

# Parámetros
n <- 7
p <- 0.40

#===========================================================
# (a) Probabilidad de que todas sean Rh-
#===========================================================

prob_todas <- dbinom(n, size = n, prob = p)

cat("=========================================\n")
cat("Apartado (a)\n")
cat("=========================================\n")
cat(sprintf("P(X = %d) = %.5f\n\n", n, prob_todas))

# Datos para la gráfica
x <- 0:n
probabilidades <- dbinom(x, size = n, prob = p)

datos1 <- data.frame(
  X = x,
  Probabilidad = probabilidades,
  Color = ifelse(x == n, "Resaltar", "Normal")
)

ggplot(datos1,
       aes(x = factor(X),
           y = Probabilidad,
           fill = Color)) +
  geom_col(color = "black") +
  scale_fill_manual(values = c("Normal" = "steelblue",
                               "Resaltar" = "red")) +
  geom_text(aes(label = ifelse(X == n,
                               sprintf("%.5f", Probabilidad),
                               "")),
            vjust = -0.5,
            size = 5) +
  labs(title = "Distribución Binomial",
       subtitle = "P(X = 7)",
       x = "Número de Comunicadores Sociales con Rh-",
       y = "Probabilidad") +
  theme_minimal(base_size = 14) +
  theme(legend.position = "none")


#===========================================================
# (b) Probabilidad de que máximo cuatro sean Rh-
#===========================================================

prob_max4 <- pbinom(4, size = n, prob = p)

cat("=========================================\n")
cat("Apartado (b)\n")
cat("=========================================\n")
cat(sprintf("P(X ≤ 4) = %.5f\n\n", prob_max4))

datos2 <- data.frame(
  X = x,
  Probabilidad = probabilidades,
  Color = ifelse(x <= 4, "Resaltar", "Normal")
)

ggplot(datos2,
       aes(x = factor(X),
           y = Probabilidad,
           fill = Color)) +
  geom_col(color = "black") +
  scale_fill_manual(values = c("Normal" = "steelblue",
                               "Resaltar" = "red")) +
  annotate("text",
           x = 5,
           y = max(probabilidades)*0.90,
           label = paste0("P(X ≤ 4) = ",
                          round(prob_max4,5)),
           size = 5) +
  labs(title = "Distribución Binomial",
       subtitle = "P(X ≤ 4)",
       x = "Número de Comunicadores Sociales con Rh-",
       y = "Probabilidad") +
  theme_minimal(base_size = 14) +
  theme(legend.position = "none")
```

## **11. Ejemplo 7. En 15 nacimientos, si la probabilidad de nacer varon es 0.58, halla la proabilidad de que hallan (a) 10 nacimientos varones, minimo 10 y maximo 7**

```python
# @title **A. Ejemplo 7. En 15 nacimientos, si la probabilidad de nacer varon es 0.58, halla la proabilidad de que hallan (a) 10 nacimientos varones, minimo 10 y maximo 7**

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 15  # Número de comunicadores sociales en la muestra
p = 0.58  # Probabilidad de que un comunicador social tenga Rh-

# (a) Probabilidad de que todas sean Rh- pnac_varones
pnac_varones = binom.pmf(10, 15, p)
print(f"Probabilidad de que todas sean Rh-: P(X = 10) = {pnac_varones:.5f}")

# Gráfico sombreando P(X = 7)
x_binom = np.arange(0, n + 3)
pmf_binom = binom.pmf(x_binom, n, p)
plt.bar(x_binom, pmf_binom)
plt.title('Distribución Binomial - P(X = 10)')
plt.xlabel('Número de comunicadores sociales con Rh-')
plt.ylabel('Probabilidad')

plt.bar(10, pnac_varones, color='red')
plt.text(10, pnac_varones + 0.01, f"P(X = 10) = {pnac_varones:.5f}", fontsize=10, color='black')

plt.show()

# (b) Probabilidad de que máximo 10 varones pmax_diez_varones
pmax_diez_varones = 1-binom.cdf(9, n, p)
print(f"Probabilidad de que mínimo 10 varones: P(X >= 10) = {pmax_diez_varones:.5f}")

# Gráfico sombreando P(X >= 10)
x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)
plt.bar(x_binom, pmf_binom)
plt.title('Distribución Binomial - P(X >= 10)')
plt.xlabel('Número de comunicadores sociales con Rh-')
plt.ylabel('Probabilidad')

for i in range(11,n+1):
    plt.bar(i, pmf_binom[i], color='red')

plt.text(10, 0.1, f"P(X >= 10) = {pmax_diez_varones:.4f}", fontsize=12, color='black')
plt.show()


# (c) Probabilidad de que máximo 7 varones pmax_diez_varones
pmax_siete_varones = binom.cdf(7, n, p)
print(f"Probabilidad de que máximo 7 varones: P(X <= 7) = {pmax_siete_varones:.5f}")


print(f"La probabilidad de que al menos tres enfermos sufran efectos secundarios es: {pmax_siete_varones:.5f}")

# Gráfico sombreando P(X <= 7)
x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)
plt.bar(x_binom, pmf_binom)
plt.title('Distribución Binomial - P(X <= 7)')
plt.xlabel('Número de enfermos con efectos secundarios')
plt.ylabel('Probabilidad')

for j in range(0, 8):
    plt.bar(j, pmf_binom[j], color='red')

plt.text(5, 0.1, f"P(X <= 7) = {pmax_siete_varones:.5f}", fontsize=12, color='black')

plt.show()


```

```python
%%R
# ==========================================================
# @title **B. Ejemplo 7, en R**
# En 15 nacimientos, si la probabilidad de nacer varón es 0.58,
# hallar la probabilidad de que haya:
#
# (a) Exactamente 10 varones.
# (b) Al menos 10 varones.
# (c) A lo sumo 7 varones.
# ==========================================================

# Instalar ggplot2 si es necesario
install.packages("gridExtra")
library(ggplot2)
library(gridExtra)  # Para organizar múltiples gráficos

# Parámetros
n <- 15
p <- 0.58

# Distribución Binomial
x <- 0:n
probabilidades <- dbinom(x, size = n, prob = p)

#===========================================================
# (a) P(X = 10)
#===========================================================

p10 <- dbinom(10, size = n, prob = p)

cat("=========================================\n")
cat("Apartado (a)\n")
cat("=========================================\n")
cat(sprintf("P(X = 10) = %.5f\n\n", p10))

datos1 <- data.frame(
  X = x,
  Probabilidad = probabilidades,
  Color = ifelse(x == 10, "Resaltar", "Normal")
)

plot1 <- ggplot(datos1,
       aes(x = factor(X),
           y = Probabilidad,
           fill = Color)) +
  geom_col(color="black") +
  scale_fill_manual(values = c("Normal"="steelblue",
                               "Resaltar"="red")) +
  geom_text(aes(label = ifelse(X==10,
                               sprintf("%.5f",Probabilidad),"")),
            vjust=-0.6,
            size=3) +
  labs(title="Distribución Binomial",
       subtitle="P(X = 10)",
       x="Número de nacimientos de varones",
       y="Probabilidad") +
  theme_minimal(base_size = 12) +
  theme(legend.position="none")


#===========================================================
# (b) P(X >= 10)
#===========================================================

p_menos10 <- 1 - pbinom(9, size=n, prob=p)

cat("=========================================\n")
cat("Apartado (b)\n")
cat("=========================================\n")
cat(sprintf("P(X ≥ 10) = %.5f\n\n", p_menos10))

datos2 <- data.frame(
  X = x,
  Probabilidad = probabilidades,
  Color = ifelse(x >= 10, "Resaltar", "Normal")
)

plot2 <- ggplot(datos2,
       aes(x=factor(X),
           y=Probabilidad,
           fill=Color))+
  geom_col(color="black")+
  scale_fill_manual(values=c("Normal"="steelblue",
                             "Resaltar"="red"))+
  annotate("text",
           x=11,
           y=max(probabilidades)*0.90,
           label=paste("P(X ≥ 10) =", round(p_menos10,5)),
           size=4)+
  labs(title="Distribución Binomial",
       subtitle="P(X ≥ 10)",
       x="Número de nacimientos de varones",
       y="Probabilidad")+
  theme_minimal(base_size=12)+
  theme(legend.position="none")


#===========================================================
# (c) P(X <= 7)
#===========================================================

p_max7 <- pbinom(7, size=n, prob=p)

cat("=========================================\n")
cat("Apartado (c)\n")
cat("=========================================\n")
cat(sprintf("P(X ≤ 7) = %.5f\n\n", p_max7))

datos3 <- data.frame(
  X = x,
  Probabilidad = probabilidades,
  Color = ifelse(x <= 7, "Resaltar", "Normal")
)

plot3 <- ggplot(datos3,
       aes(x=factor(X),
           y=Probabilidad,
           fill=Color))+
  geom_col(color="black")+
  scale_fill_manual(values=c("Normal"="steelblue",
                             "Resaltar"="red"))+
  annotate("text",
           x=5,
           y=max(probabilidades)*0.90,
           label=paste("P(X ≤ 7) =", round(p_max7,5)),
           size=4)+
  labs(title="Distribución Binomial",
       subtitle="P(X ≤ 7)",
       x="Número de nacimientos de varones",
       y="Probabilidad")+
  theme_minimal(base_size=12)+
  theme(legend.position="none")

# Mostrar los tres gráficos juntos
grid.arrange(plot1, plot2, plot3, ncol=3)

# Alternativamente, si prefieres verlos uno por uno:
#print(plot1)
#print(plot2)
#print(plot3)
```

## **12. Ejemplo 8. Un laboratorio ha comprobado que el 25% de los que toman un determinado antibiótico sufren efectos secundarios. De una muestra de ocho enfermos que toman dicho antibiótico, halla la probabilidad de que sufran efectos secundarios al menos tres.**

```python
# @title **A. Ejemplo 8. Un laboratorio ha comprobado que el 25% de los que toman un determinado antibiótico sufren efectos secundarios. De una muestra de ocho enfermos que toman dicho antibiótico, halla la probabilidad de que sufran efectos secundarios al menos tres.**

# Parámetros del problema
p = 0.25  # Probabilidad de sufrir efectos secundarios
n = 8  # Número de enfermos en la muestra

# Calcular la probabilidad de que al menos tres sufran efectos secundarios
# P(X >= 3) = 1 - P(X < 3) = 1 - [P(X = 0) + P(X = 1) + P(X = 2)]
prob_al_menos_tres = 1 - binom.cdf(2, n, p)

print(f"La probabilidad de que al menos tres enfermos sufran efectos secundarios es: {prob_al_menos_tres:.5f}")

# Gráfico sombreando P(X >= 3)
x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)
plt.bar(x_binom, pmf_binom)
plt.title('Distribución Binomial - P(X >= 3)')
plt.xlabel('Número de enfermos con efectos secundarios')
plt.ylabel('Probabilidad')

for i in range(3, n + 1):
    plt.bar(i, pmf_binom[i], color='red')

plt.text(5, 0.1, f"P(X >= 3) = {prob_al_menos_tres:.5f}", fontsize=12, color='black')

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 8, en R**
# Un laboratorio ha comprobado que el 25% de los que toman
# un determinado antibiótico sufren efectos secundarios.
#
# De una muestra de ocho enfermos, hallar la probabilidad
# de que sufran efectos secundarios al menos tres.
# ==========================================================

# Instalar ggplot2 si es necesario
# install.packages("ggplot2")
%%R
library(ggplot2)

# Parámetros
n <- 8
p <- 0.25

#===========================================================
# Probabilidad de que al menos tres sufran efectos secundarios
#===========================================================

prob_al_menos_tres <- 1 - pbinom(2, size = n, prob = p)

cat("=========================================\n")
cat("Ejemplo 8\n")
cat("=========================================\n")
cat(sprintf("P(X ≥ 3) = %.5f\n\n", prob_al_menos_tres))

#===========================================================
# Función de probabilidad
#===========================================================

x <- 0:n
probabilidades <- dbinom(x, size = n, prob = p)

datos <- data.frame(
  X = x,
  Probabilidad = probabilidades,
  Color = ifelse(x >= 3, "Resaltar", "Normal")
)

#===========================================================
# Gráfica
#===========================================================

grafico <- ggplot(datos,
                  aes(x = factor(X),
                      y = Probabilidad,
                      fill = Color)) +
  geom_col(color = "black") +
  scale_fill_manual(values = c("Normal" = "steelblue",
                               "Resaltar" = "red")) +
  annotate("text",
           x = 6,
           y = max(probabilidades) * 0.90,
           label = paste("P(X ≥ 3) =",
                         round(prob_al_menos_tres, 5)),
           size = 5) +
  labs(
    title = "Distribución Binomial",
    subtitle = "Probabilidad de que al menos tres enfermos sufran efectos secundarios",
    x = "Número de enfermos con efectos secundarios",
    y = "Probabilidad"
  ) +
  theme_minimal(base_size = 14) +
  theme(legend.position = "none")

print(grafico)
```

## **13. Ejemplo 9. Un laboratorio ha comprobado que el 25% de los que toman un determinado antibiótico sufren efectos secundarios. De una muestra de ocho enfermos que toman dicho antibiótico, halla la probabilidad de que sufran efectos secundarios menos de 3.**

```python
# @title **A. Ejemplo 9. Un laboratorio ha comprobado que el 25% de los que toman un determinado antibiótico sufren efectos secundarios. De una muestra de ocho enfermos que toman dicho antibiótico, halla la probabilidad de que sufran efectos secundarios menos de 3.**

# Parámetros del problema
p = 0.25  # Probabilidad de sufrir efectos secundarios
n = 8  # Número de enfermos en la muestra

# Calcular la probabilidad de que menos de 3 sufran efectos secundarios
# P(X < 3) = P(X = 0) + P(X = 1) + P(X = 2)
prob_menos_de_tres = binom.cdf(2, n, p)

print(f"La probabilidad de que menos de 3 enfermos sufran efectos secundarios es: {prob_menos_de_tres:.5f}")

# Gráfico sombreando P(X < 3)
x_binom = np.arange(0, n + 1)
pmf_binom = binom.pmf(x_binom, n, p)
plt.bar(x_binom, pmf_binom)
plt.title('Distribución Binomial - P(X < 3)')
plt.xlabel('Número de enfermos con efectos secundarios')
plt.ylabel('Probabilidad')

for i in range(3):
    plt.bar(i, pmf_binom[i], color='red')

plt.text(2, 0.1, f"P(X < 3) = {prob_menos_de_tres:.5f}", fontsize=12, color='black')

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 9 en R**
# Un laboratorio ha comprobado que el 25% de los que toman
# un determinado antibiótico sufren efectos secundarios.
#
# De una muestra de ocho enfermos, hallar la probabilidad
# de que sufran efectos secundarios menos de 3.
# ==========================================================
%%R
# Instalar ggplot2 si es necesario
# install.packages("ggplot2")

library(ggplot2)

# Parámetros
n <- 8
p <- 0.25

#===========================================================
# Probabilidad de que menos de tres sufran efectos secundarios
#===========================================================

# P(X < 3) = P(X ≤ 2)
prob_menos_tres <- pbinom(2, size = n, prob = p)

cat("=========================================\n")
cat("Ejemplo 9\n")
cat("=========================================\n")
cat(sprintf("P(X < 3) = %.5f\n\n", prob_menos_tres))

#===========================================================
# Función de probabilidad
#===========================================================

x <- 0:n
probabilidades <- dbinom(x, size = n, prob = p)

datos <- data.frame(
  X = x,
  Probabilidad = probabilidades,
  Color = ifelse(x < 3, "Resaltar", "Normal")
)

#===========================================================
# Gráfica
#===========================================================

grafico <- ggplot(datos,
                  aes(x = factor(X),
                      y = Probabilidad,
                      fill = Color)) +
  geom_col(color = "black") +
  scale_fill_manual(values = c("Normal" = "steelblue",
                               "Resaltar" = "red")) +
  annotate("text",
           x = 3,
           y = max(probabilidades) * 0.90,
           label = paste("P(X < 3) =",
                         round(prob_menos_tres, 5)),
           size = 5) +
  labs(
    title = "Distribución Binomial",
    subtitle = "Probabilidad de que menos de tres enfermos sufran efectos secundarios",
    x = "Número de enfermos con efectos secundarios",
    y = "Probabilidad"
  ) +
  theme_minimal(base_size = 14) +
  theme(legend.position = "none")

print(grafico)
```

## **14. CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL**
#
# Esta calculadora te permite:
# 1. Ingresar parámetros n, p y k
# 2. Calcular P(X=k), P(X≤k) y P(X≥k)
# 3. Visualizar la distribución con áreas sombreadas
# 4. Mostrar probabilidades en el gráfico
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
from matplotlib.patches import Rectangle
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_binomial():
    """
    Calculadora interactiva para la distribución binomial
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL")
    print("="*70)
    print("\nEsta calculadora te permite analizar la distribución binomial")
    print("ingresando los parámetros n, p y k.\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            n = int(input("  Número total de ensayos (n): "))
            if n <= 0:
                print("  ❌ n debe ser un número positivo. Intenta de nuevo.")
                continue

            p = float(input("  Probabilidad de éxito (p, entre 0 y 1): "))
            if p < 0 or p > 1:
                print("  ❌ p debe estar entre 0 y 1. Intenta de nuevo.")
                continue

            k = int(input("  Número de éxitos a evaluar (k): "))
            if k < 0 or k > n:
                print(f"  ❌ k debe estar entre 0 y {n}. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

    #==========================================================
    # 2. CÁLCULO DE PROBABILIDADES
    #==========================================================

    # Rango de valores posibles
    x = np.arange(0, n + 1)
    pmf = binom.pmf(x, n, p)

    # Probabilidades solicitadas
    prob_exacta = binom.pmf(k, n, p)
    prob_acumulada_menor = binom.cdf(k, n, p)  # P(X ≤ k)
    prob_acumulada_mayor = 1 - binom.cdf(k-1, n, p) if k > 0 else 1  # P(X ≥ k)

    # Estadísticas
    media = n * p
    varianza = n * p * (1 - p)
    desviacion = np.sqrt(varianza)

    #==========================================================
    # 3. MOSTRAR RESULTADOS NUMÉRICOS
    #==========================================================

    print("\n" + "="*70)
    print("📈 RESULTADOS DEL ANÁLISIS")
    print("="*70)
    print(f"\nParámetros de la distribución:")
    print(f"  • n = {n} (número de ensayos)")
    print(f"  • p = {p:.4f} (probabilidad de éxito)")
    print(f"  • k = {k} (número de éxitos evaluado)")

    print(f"\n📊 Probabilidades calculadas:")
    print("-"*70)
    print(f"  • P(X = {k}) = {prob_exacta:.6f} ({prob_exacta*100:.4f}%)")
    print(f"  • P(X ≤ {k}) = {prob_acumulada_menor:.6f} ({prob_acumulada_menor*100:.4f}%)")
    print(f"  • P(X ≥ {k}) = {prob_acumulada_mayor:.6f} ({prob_acumulada_mayor*100:.4f}%)")

    print(f"\n📐 Estadísticas descriptivas:")
    print("-"*70)
    print(f"  • Media (μ) = {media:.4f}")
    print(f"  • Varianza (σ²) = {varianza:.4f}")
    print(f"  • Desviación estándar (σ) = {desviacion:.4f}")
    print(f"  • Coeficiente de variación = {desviacion/media:.4f}")

    #==========================================================
    # 4. CREAR GRÁFICO PRINCIPAL CON ÁREAS SOMBREADAS
    #==========================================================

    # Crear figura con subplots
    fig = plt.figure(figsize=(14, 10))

    # Gráfico 1: Distribución completa
    ax1 = plt.subplot(2, 2, 1)

    # Barras de la distribución
    colors = ['#2E86AB' if i != k else '#E63946' for i in range(n+1)]
    bars = ax1.bar(x, pmf, color=colors, alpha=0.8, edgecolor='black', linewidth=1)

    # Resaltar barra de k
    ax1.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black', linewidth=2)

    # Línea vertical en la media
    ax1.axvline(x=media, color='#2C3E50', linestyle='--', linewidth=2, alpha=0.7)
    ax1.axvline(x=media + desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)
    ax1.axvline(x=media - desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)

    # Etiquetas
    ax1.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax1.set_ylabel('Probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Binomial\nn={n}, p={p:.3f}', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.set_xticks(np.arange(0, n+1, max(1, n//20)))

    # Agregar etiquetas de valores
    for i, bar in enumerate(bars):
        if i % max(1, n//15) == 0:  # Mostrar solo algunas etiquetas para no saturar
            height = bar.get_height()
            if height > 0.001:
                ax1.text(bar.get_x() + bar.get_width()/2., height + 0.001,
                        f'{height:.3f}', ha='center', va='bottom', fontsize=8)

    # Agregar leyenda
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor='#2E86AB', alpha=0.8, label='Distribución'),
        Patch(facecolor='#E63946', alpha=0.9, label=f'P(X={k})'),
        plt.Line2D([0], [0], color='#2C3E50', linestyle='--', linewidth=2, label=f'μ = {media:.2f}'),
        plt.Line2D([0], [0], color='#2C3E50', linestyle=':', linewidth=1.5, label=f'μ ± σ')
    ]
    ax1.legend(handles=legend_elements, loc='best', fontsize=10)

    # Gráfico 2: P(X ≤ k) - Área izquierda
    ax2 = plt.subplot(2, 2, 2)

    # Barras de la distribución
    colors2 = ['#FF6B6B' if i <= k else '#2E86AB' for i in range(n+1)]
    bars2 = ax2.bar(x, pmf, color=colors2, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax2.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax2.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax2.set_ylabel('Probabilidad', fontsize=12)
    ax2.set_title(f'P(X ≤ {k}) = {prob_acumulada_menor:.4f}\n({prob_acumulada_menor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.set_xticks(np.arange(0, n+1, max(1, n//20)))

    # Texto adicional
    ax2.text(0.02, 0.98, f'Área sombreada:\nP(X ≤ {k}) = {prob_acumulada_menor:.6f}',
             transform=ax2.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 3: P(X ≥ k) - Área derecha
    ax3 = plt.subplot(2, 2, 3)

    # Barras de la distribución
    colors3 = ['#FF6B6B' if i >= k else '#2E86AB' for i in range(n+1)]
    bars3 = ax3.bar(x, pmf, color=colors3, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax3.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax3.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax3.set_ylabel('Probabilidad', fontsize=12)
    ax3.set_title(f'P(X ≥ {k}) = {prob_acumulada_mayor:.4f}\n({prob_acumulada_mayor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.set_xticks(np.arange(0, n+1, max(1, n//20)))

    # Texto adicional
    ax3.text(0.02, 0.98, f'Área sombreada:\nP(X ≥ {k}) = {prob_acumulada_mayor:.6f}',
             transform=ax3.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 4: Resumen de probabilidades
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Crear tabla de resultados
    resultados = [
        ['Concepto', 'Valor', 'Porcentaje'],
        ['P(X = k)', f'{prob_exacta:.6f}', f'{prob_exacta*100:.4f}%'],
        ['P(X ≤ k)', f'{prob_acumulada_menor:.6f}', f'{prob_acumulada_menor*100:.4f}%'],
        ['P(X ≥ k)', f'{prob_acumulada_mayor:.6f}', f'{prob_acumulada_mayor*100:.4f}%'],
        ['Media (μ)', f'{media:.4f}', ''],
        ['Varianza (σ²)', f'{varianza:.4f}', ''],
        ['Desv. Estándar (σ)', f'{desviacion:.4f}', '']
    ]

    # Crear tabla
    table = ax4.table(cellText=resultados, loc='center', cellLoc='center',
                      colWidths=[0.3, 0.35, 0.35])

    # Estilo de la tabla
    table.auto_set_font_size(False)
    table.set_fontsize(11)
    table.scale(1, 1.5)

    # Colores de la tabla
    for i in range(len(resultados)):
        for j in range(len(resultados[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            elif i in [1, 2, 3]:
                cell.set_facecolor('#E8F4F8')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('#F5F5F5')

    # Título de la tabla
    ax4.set_title('📊 RESUMEN DE RESULTADOS', fontsize=13, fontweight='bold', pad=20)

    # Ajustar layout
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # 5. GRÁFICO ADICIONAL: FUNCIÓN DE DISTRIBUCIÓN ACUMULADA
    #==========================================================

    # Crear una figura separada para la CDF
    fig2, ax5 = plt.subplots(figsize=(10, 6))

    # Calcular CDF
    cdf = binom.cdf(x, n, p)

    # Graficar CDF
    ax5.step(x, cdf, where='mid', linewidth=2.5, color='#2E86AB', label='CDF')
    ax5.fill_between(x, 0, cdf, step='mid', alpha=0.3, color='#2E86AB')

    # Punto en k
    ax5.scatter(k, cdf[k], color='#E63946', s=150, zorder=5,
               label=f'F({k}) = {cdf[k]:.4f}')

    # Líneas de referencia
    ax5.axhline(y=cdf[k], color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.axvline(x=k, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax5.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax5.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax5.set_title(f'Función de Distribución Acumulada\nn={n}, p={p:.3f}',
                  fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.set_xticks(np.arange(0, n+1, max(1, n//20)))
    ax5.set_ylim(-0.05, 1.05)
    ax5.legend(loc='best')

    # Texto con la probabilidad
    ax5.text(0.02, 0.98, f'P(X ≤ {k}) = {cdf[k]:.6f}\nP(X > {k}) = {1-cdf[k]:.6f}',
             transform=ax5.transAxes, fontsize=11, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()

    #==========================================================
    # 6. MOSTRAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # 7. OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Binomial_n{n}_p{p:.2f}_k{k}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_cdf.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_cdf.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

    return {
        'n': n,
        'p': p,
        'k': k,
        'prob_exacta': prob_exacta,
        'prob_menor': prob_acumulada_menor,
        'prob_mayor': prob_acumulada_mayor,
        'media': media,
        'varianza': varianza,
        'desviacion': desviacion
    }

#==========================================================
# FUNCIÓN PARA MÚLTIPLES ANÁLISIS COMPARATIVOS
#==========================================================

def analisis_comparativo():
    """
    Permite comparar diferentes distribuciones binomiales
    """

    print("\n" + "="*70)
    print("📊 ANÁLISIS COMPARATIVO DE DISTRIBUCIONES BINOMIALES")
    print("="*70)

    # Obtener parámetros para múltiples distribuciones
    num_dist = int(input("\n¿Cuántas distribuciones quieres comparar? (2-4): "))
    num_dist = max(2, min(4, num_dist))

    distribuciones = []

    for i in range(num_dist):
        print(f"\n📌 Distribución {i+1}:")
        n = int(input(f"  n{i+1}: "))
        p = float(input(f"  p{i+1} (0-1): "))
        distribuciones.append((n, p))

    # Crear gráfico comparativo
    fig, axes = plt.subplots(1, num_dist, figsize=(5*num_dist, 5))
    if num_dist == 1:
        axes = [axes]

    for i, (n, p) in enumerate(distribuciones):
        x = np.arange(0, n+1)
        pmf = binom.pmf(x, n, p)

        axes[i].bar(x, pmf, color='#2E86AB', alpha=0.8, edgecolor='black')
        axes[i].set_title(f'n={n}, p={p:.2f}', fontsize=12, fontweight='bold')
        axes[i].set_xlabel('Éxitos (X)')
        axes[i].set_ylabel('Probabilidad')
        axes[i].grid(True, alpha=0.3)
        axes[i].set_xticks(np.arange(0, n+1, max(1, n//15)))

        # Mostrar media
        media = n * p
        axes[i].axvline(x=media, color='#E63946', linestyle='--', linewidth=2)
        axes[i].text(0.95, 0.95, f'μ={media:.2f}', transform=axes[i].transAxes,
                    fontsize=10, verticalalignment='top', horizontalalignment='right',
                    bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.suptitle('Comparación de Distribuciones Binomiales', fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis simple de una distribución")
        print("  2. Análisis comparativo de múltiples distribuciones")
        print("  3. Salir")

        opcion = input("\nSelecciona una opción (1-3): ")

        if opcion == '1':
            resultados = calculadora_binomial()
        elif opcion == '2':
            analisis_comparativo()
        elif opcion == '3':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2 o 3.")
```

#**15. Distribución Binomial Negativa**
+ *La distribución binomial negativa modela el número de ensayos requeridos para obtener un número fijo de éxitos $r$ en ensayos independientes de Bernoulli, donde la probabilidad de éxito es $p$.*

- **Función de probabilidad**: $$P(X = k) = \binom{k + r - 1}{r - 1} p^r (1 - p)^k$$
- **Parámetros**:$r$ es el número de éxitos deseados, $p$ es la probabilidad de éxito y $k$ es el número de fracasos antes de obtener $r$ éxitos.

### **Distribución Binomial Negativa - Momentos**
- **Valor esperado (Esperanza)**: $$E(X) = \frac{r(1 - p)}{p}$$
- **Varianza**: $$\text{Var}(X) = \frac{r(1 - p)}{p^2}$$
- **Desviación estándar**: $$\sigma(X) = \sqrt{\frac{r(1 - p)}{p^2}}$$



```python
# @title **A. Gráfica de la Distribución Binomial Negativa en Pythton** $$P(X = k) = \binom{k + r - 1}{r - 1} p^r (1 - p)^k$$

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, nbinom, poisson, hypergeom, randint

# @title **A. Para Graficar cualquier distribucion discreta binomial, binomial negativa, Poisson e Hipergeométrica**
def plot_distribution2(dist, values, pmf, title):
    plt.bar(values, pmf)
    plt.title(title)
    plt.xlabel('Valores')
    plt.ylabel('Probabilidad')
    plt.show()
r, p = 5, 0.7
x_nbinom = np.arange(0, 20)
pmf_nbinom = nbinom.pmf(x_nbinom, r, p)
plot_distribution2(nbinom, x_nbinom, pmf_nbinom, "Distribución Binomial Negativa")
```

```python
# ==========================================================
# @title **B. Gráfica de la Distribución Binomial Negativa en R** $$P(X = k) = \binom{k + r - 1}{r - 1} p^r (1 - p)^k$$
# P(X = k) = choose(k+r-1,r-1) * p^r * (1-p)^k
# ==========================================================
%%R
# Instalar paquetes si es necesario
# install.packages("ggplot2")

library(ggplot2)

#----------------------------------------------------------
# Función para graficar distribuciones discretas
#----------------------------------------------------------
plot_distribution2 <- function(values, pmf, title){

  datos <- data.frame(
    x = values,
    Probabilidad = pmf
  )

  ggplot(datos, aes(x = factor(x), y = Probabilidad)) +
    geom_col(fill = "steelblue", color = "black") +
    labs(
      title = title,
      x = "Número de fracasos antes del éxito número r",
      y = "Probabilidad"
    ) +
    theme_minimal(base_size = 14)
}

#----------------------------------------------------------
# Parámetros
#----------------------------------------------------------
r <- 5      # Número de éxitos
p <- 0.7    # Probabilidad de éxito

# Valores posibles de X
x_nbinom <- 0:19

# Función de probabilidad
pmf_nbinom <- dnbinom(x_nbinom,
                      size = r,
                      prob = p)

# Mostrar probabilidades
tabla <- data.frame(
  x = x_nbinom,
  Probabilidad = round(pmf_nbinom, 5)
)

print(tabla)

#----------------------------------------------------------
# Gráfico
#----------------------------------------------------------
plot_distribution2(
  x_nbinom,
  pmf_nbinom,
  "Distribución Binomial Negativa"
)
```

# **16. Ejemplo 1. Binomial Negativa: ¿Cuál es la probabilidad de obtener 5 éxitos antes de 3 fracasos si la probabilidad de éxito es p=0.7?**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import nbinom

# @title **A. Ejemplo 4: (Distribución Geométrica $r=1$) Probabilidad de obtener el primer antes del 3 fracasos (p=0.7) en Pyhton** $$P(X = k) =  p (1 - p)^k$$

# Parámetros del problema
r = 1  # Número de éxitos deseados
p = 0.7  # Probabilidad de éxito
k_failures = 3  # Número de fracasos antes de obtener los éxitos

# Calcular la probabilidad P(X = k_failures)
prob_nbinom = nbinom.pmf(k_failures, r, p)
print(f"La probabilidad de obtener 5 éxitos antes de {k_failures} fracasos es: P(X = {k_failures}) = {prob_nbinom:.4f}")

# Generar valores de x (número de fracasos) para la gráfica
x_nbinom = np.arange(0, 15)  # Abarcar un rango razonable de fracasos
pmf_nbinom = nbinom.pmf(x_nbinom, r, p)

# Crear la gráfica
plt.figure(figsize=(10, 6))
plt.bar(x_nbinom, pmf_nbinom, color='skyblue', label='PMF')

# Resaltar la barra para el número específico de fracasos (k_failures)
plt.bar(k_failures, prob_nbinom, color='red', label=f'P(X = {k_failures}) = {prob_nbinom:.4f}')

plt.title(f'Distribución Binomial Negativa (r={r}, p={p})')
plt.xlabel('Número de Fracasos (k)')
plt.ylabel('Probabilidad')
plt.xticks(np.arange(min(x_nbinom), max(x_nbinom) + 1, 1))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 4: (Distribución Geométrica $r=1$) Probabilidad de obtener el primer antes de 3 fracasos (p=0.7) en r** $$P(X = k) =  p (1 - p)^k$$
# después de exactamente 3 fracasos (p = 0.7)
#
# P(X = k) = p(1-p)^k
# ==========================================================
%%R
# Instalar ggplot2 si es necesario
# install.packages("ggplot2")

library(ggplot2)

#----------------------------------------------------------
# Parámetros del problema
#----------------------------------------------------------
r <- 1               # Primer éxito
p <- 0.7             # Probabilidad de éxito
k_failures <- 3      # Número de fracasos

#----------------------------------------------------------
# Probabilidad de exactamente 3 fracasos antes del primer éxito
#----------------------------------------------------------
prob_nbinom <- dnbinom(k_failures,
                       size = r,
                       prob = p)

cat("La probabilidad de obtener el primer éxito",
    "después de", k_failures,
    "fracasos es:\n")
cat(sprintf("P(X = %d) = %.5f\n\n",
            k_failures, prob_nbinom))

#----------------------------------------------------------
# Valores para la gráfica
#----------------------------------------------------------
x_nbinom <- 0:14
pmf_nbinom <- dnbinom(x_nbinom,
                      size = r,
                      prob = p)

datos <- data.frame(
  Fracasos = x_nbinom,
  Probabilidad = pmf_nbinom,
  Color = ifelse(x_nbinom == k_failures,
                 "Resaltado",
                 "Normal")
)

#----------------------------------------------------------
# Gráfica
#----------------------------------------------------------
ggplot(datos,
       aes(x = factor(Fracasos),
           y = Probabilidad,
           fill = Color)) +

  geom_col(color = "black") +

  scale_fill_manual(values = c(
    "Normal" = "skyblue",
    "Resaltado" = "red"
  )) +

  geom_text(
    data = subset(datos,
                  Fracasos == k_failures),
    aes(label = paste0(
      "P(X=", k_failures, ") = ",
      round(prob_nbinom, 5))),
    vjust = -0.6,
    size = 5
  ) +

  labs(
    title = paste("Distribución Binomial Negativa (r =", r,
                  ", p =", p, ")"),
    x = "Número de fracasos (k)",
    y = "Probabilidad"
  ) +

  theme_minimal(base_size = 14) +
  theme(legend.position = "none")
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import nbinom

# ==================================================
# Ejemplo:
# @title **C. (Distribución Geométrica $r=1$) Probabilidad de obtener el primer éxito antes de acumular 3 fracasos en Python** $$P(X = k) = \binom{k + r - 1}{r - 1} p^r (1 - p)^k$$
# ==================================================

r = 1
p = 0.7

# P(X < 3) = P(X<=2)
probabilidad = nbinom.cdf(2, r, p)

print(f"P(X < 3) = {probabilidad:.4f}")

# Valores para la gráfica
x = np.arange(0, 10)
pmf = nbinom.pmf(x, r, p)

plt.figure(figsize=(10,6))

# Barras normales
plt.bar(
    x,
    pmf,
    color='skyblue',
    edgecolor='skyblue',
    label='PMF'
)

# Resaltar los casos favorables
x_favorable = np.arange(0,3)
pmf_favorable = nbinom.pmf(x_favorable, r, p)

plt.bar(
    x_favorable,
    pmf_favorable,
    color='red',
    edgecolor='skyblue',
    label=f'P(X < 3) = {probabilidad:.4f}'
)

plt.title(
    'Primer éxito antes de 3 fracasos\nDistribución Binomial Negativa'
)

plt.xlabel('Número de fracasos antes del primer éxito')
plt.ylabel('Probabilidad')

plt.xticks(np.arange(0,10))
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.legend()
plt.show()
```

```python

# ==================================================
# Ejemplo:
# @title **D. (Distribución Geométrica $r=1$) Probabilidad de obtener el primer éxito antes de acumular 3 fracasos en R** $$P(X = k) = \binom{k + r - 1}{r - 1} p^r (1 - p)^k$$
# ==================================================

%%R
# Instalar ggplot2 si es necesario
# install.packages("ggplot2")

library(ggplot2)

#----------------------------------------------------------
# Parámetros
#----------------------------------------------------------
r <- 1
p <- 0.7

#----------------------------------------------------------
# Probabilidad P(X < 3) = P(X <= 2)
#----------------------------------------------------------
probabilidad <- pnbinom(
  q = 2,
  size = r,
  prob = p
)

cat(sprintf("P(X < 3) = %.5f\n", probabilidad))

#----------------------------------------------------------
# Valores para la gráfica
#----------------------------------------------------------
x <- 0:9
pmf <- dnbinom(
  x,
  size = r,
  prob = p
)

datos <- data.frame(
  Fracasos = x,
  Probabilidad = pmf,
  Tipo = ifelse(x <= 2, "Favorables", "No favorables")
)

#----------------------------------------------------------
# Gráfico
#----------------------------------------------------------
ggplot(datos,
       aes(x = factor(Fracasos),
           y = Probabilidad,
           fill = Tipo)) +

  geom_col(color = "black") +

  scale_fill_manual(values = c(
    "Favorables" = "red",
    "No favorables" = "skyblue"
  )) +

  annotate(
    "text",
    x = 4,
    y = max(pmf)*0.95,
    label = paste0(
      "P(X < 3) = ",
      round(probabilidad,5)
    ),
    size = 5
  ) +

  labs(
    title = "Primer éxito antes de 3 fracasos\nDistribución Binomial Negativa",
    x = "Número de fracasos antes del primer éxito",
    y = "Probabilidad"
  ) +

  theme_minimal(base_size = 14) +
  theme(legend.position = "bottom")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import nbinom


# ==================================================
# Ejemplo:
# @title **E. (Distribución Geométrica $r=1$) (Distribución Biinomial Negativa $r> 1$) Ejemplo 1 ¿Cuál es la probabilidad de obtener 5 éxitos antes de 3 fracasos si la probabilidad de éxito es  p=0.7 ?** $$P(X = k) = \binom{k + r - 1}{r - 1} p^r (1 - p)^k$$
# ==================================================



# Parameters from the previous example
r = 5  # Number of successes
p = 0.7  # Probability of success
k = 3  # Number of failures to highlight

# Generate x values for the plot (number of failures)
x_nbinom = np.arange(0, 15)
pmf_nbinom_values = nbinom.pmf(x_nbinom, r, p)

# Calculate the specific probability to highlight
pmf_at_k = nbinom.pmf(k, r, p)

# Plot the distribution using plot_distribution2
# (Assuming plot_distribution2 is defined in a previous cell and accessible)
plt.bar(x_nbinom, pmf_nbinom_values)
plt.title(f"Distribución Binomial Negativa (r={r}, p={p})")
plt.xlabel('Número de Fracasos (k)')
plt.ylabel('Probabilidad')

# Highlight the bar for k
plt.bar(k, pmf_at_k, color='red')

# Add text annotation for the highlighted probability
plt.text(k + 0.5, pmf_at_k + 0.01, f"P(X={k}) = {pmf_at_k:.4f}", ha='center', va='bottom')

plt.show()
```

```python
%%R

# @title **F. (Distribución Biinomial Negativa $r> 1$ en R) Ejemplo 1 ¿Cuál es la probabilidad de obtener 5 éxitos antes de 3 fracasos si la probabilidad de éxito es  p=0.7 ?** $$P(X = k) = \binom{k + r - 1}{r - 1} p^r (1 - p)^k$$

library(ggplot2)

# Parámetros
r <- 5
p <- 0.7
k <- 3

# Distribución
x <- 0:14
pmf <- dnbinom(x, size = r, prob = p)

datos <- data.frame(
  Fracasos = x,
  Probabilidad = pmf,
  Color = ifelse(x == k, "Resaltado", "Normal")
)

probabilidad <- dnbinom(k, size = r, prob = p)

ggplot(datos,
       aes(x = factor(Fracasos),
           y = Probabilidad,
           fill = Color)) +

  geom_col(color = "black") +

  scale_fill_manual(values = c(
    "Normal" = "skyblue",
    "Resaltado" = "red"
  )) +

  geom_text(
    data = subset(datos, Fracasos == k),
    aes(label = paste0("P(X=", k, ") = ",
                       round(probabilidad,5))),
    vjust = -0.6,
    size = 5
  ) +

  labs(
    title = paste("Distribución Binomial Negativa (r =", r,
                  ", p =", p, ")"),
    x = "Número de fracasos (k)",
    y = "Probabilidad"
  ) +

  theme_minimal(base_size = 14) +
  theme(legend.position = "none")
```

# **17. Ejemplo 2. Binomial Negativa: Momentos Modelo Binomial Negativo**

```python
# @title **Momentos Modelo Binomial Negativo**
r = 5
p = 0.7
esperanza_nbinom = r * (1 - p) / p
varianza_nbinom = r * (1 - p) / (p ** 2)
desviacion_nbinom = np.sqrt(varianza_nbinom)
print(f"Momentos en el Modelo Binomial Negativo r= 5, p = 0.7")
print(f"Esperanza:E(X) =  {esperanza_nbinom}")
print(f"Varianza: V(X) =  {varianza_nbinom}")
print(f"Desviación estándar: {desviacion_nbinom}")
```

# **18. Ejemplo 3. Binomial Negativa: La probabilidad de un niño expuesto al COVID-19 se contagie es de 0.6. ¿Cuál es la probabilidad de que el 10° niño expuesto sea el tercero en contraerla?.**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import nbinom



# ==================================================
# Ejemplo:
# @title **A.  (Ejemplo 3. Binomial Negativa) La probabilidad de un niño expuesto al COVID-19 se contagie es de 0.6. ¿Cuál es la probabilidad de que el 10° niño expuesto sea el tercero en contraerla?**
# ==================================================



# Parámetros del ejemplo anterior
r = 3  # Número de éxitos (3er niño contagiado)
p = 0.6  # Probabilidad de éxito (contagio)
n_trials = 10 # Total de niños expuestos (10° niño)
k_failures = n_trials - r # Número de fracasos (10 - 3 = 7)

# Generar valores de x (número de fracasos) para la gráfica
x_nbinom_plot = np.arange(0, 15) # Abarcar un rango razonable de fracasos
pmf_nbinom_plot = nbinom.pmf(x_nbinom_plot, r, p)

# Calcular la probabilidad específica a resaltar
prob_at_k_failures = nbinom.pmf(k_failures, r, p)

plt.figure(figsize=(10, 6))
plt.bar(x_nbinom_plot, pmf_nbinom_plot, color='skyblue', label='PMF')

# Resaltar la barra para k_failures
plt.bar(k_failures, prob_at_k_failures, color='red', label=f'P(X={k_failures}) = {prob_at_k_failures:.4f}')

plt.title(f'Distribución Binomial Negativa (r={r}, p={p})')
plt.xlabel('Número de Fracasos (k)')
plt.ylabel('Probabilidad')
plt.xticks(np.arange(min(x_nbinom_plot), max(x_nbinom_plot)+1, 1))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

```python

# ==================================================
# Ejemplo:
# @title **B.  (Ejemplo 3. Binomial Negativa) La probabilidad de un niño expuesto al COVID-19 se contagie es de 0.6. ¿Cuál es la probabilidad de que el 10° niño expuesto sea el tercero en contraerla?**
# ==================================================


%%R
library(ggplot2)

# Parámetros
r <- 3
p <- 0.6
n_trials <- 10
k_failures <- n_trials - r

# Distribución
x <- 0:14
pmf <- dnbinom(x, size = r, prob = p)

datos <- data.frame(
  Fracasos = x,
  Probabilidad = pmf,
  Tipo = ifelse(x == k_failures,
                "Resaltado",
                "Normal")
)

probabilidad <- dnbinom(
  k_failures,
  size = r,
  prob = p
)

ggplot(datos,
       aes(x = factor(Fracasos),
           y = Probabilidad,
           fill = Tipo)) +

  geom_col(color = "black") +

  scale_fill_manual(values = c(
    "Normal" = "skyblue",
    "Resaltado" = "red"
  )) +

  geom_text(
    data = subset(datos, Fracasos == k_failures),
    aes(label = paste0(
      "P(X=", k_failures, ") = ",
      round(probabilidad, 5)
    )),
    vjust = -0.6,
    size = 5
  ) +

  labs(
    title = paste("Distribución Binomial Negativa (r =", r,
                  ", p =", p, ")"),
    x = "Número de niños no contagiados (fracasos)",
    y = "Probabilidad"
  ) +

  theme_minimal(base_size = 14) +
  theme(legend.position = "none")
```

# **19. CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA**



```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA**
#
# La distribución binomial negativa modela el número de fracasos
# antes de obtener un número específico de éxitos.
#
# Fórmula: P(X = k) = C(k+r-1, r-1) * p^r * (1-p)^k
# donde:
#   r = número de éxitos deseados
#   p = probabilidad de éxito
#   k = número de fracasos
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import nbinom
from scipy.special import comb
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_binomial_negativa():
    """
    Calculadora interactiva para la distribución binomial negativa
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)
    print("\nLa binomial negativa modela el número de fracasos (k)")
    print("antes de obtener r éxitos, donde cada ensayo tiene")
    print("probabilidad p de éxito.\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            r = int(input("  Número de éxitos deseados (r, > 0): "))
            if r <= 0:
                print("  ❌ r debe ser un número positivo. Intenta de nuevo.")
                continue

            p = float(input("  Probabilidad de éxito (p, entre 0 y 1): "))
            if p <= 0 or p >= 1:
                print("  ❌ p debe estar entre 0 y 1. Intenta de nuevo.")
                continue

            k = int(input("  Número de fracasos a evaluar (k, ≥ 0): "))
            if k < 0:
                print("  ❌ k debe ser mayor o igual a 0. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

    #==========================================================
    # 2. CÁLCULO DE PROBABILIDADES
    #==========================================================

    # Rango de valores posibles para k (número de fracasos)
    max_k = max(20, k * 3)  # Asegurar que se muestre suficiente rango
    x = np.arange(0, max_k + 1)
    pmf = nbinom.pmf(x, r, p)

    # Probabilidades solicitadas
    prob_exacta = nbinom.pmf(k, r, p)
    prob_acumulada_menor = nbinom.cdf(k, r, p)  # P(X ≤ k)
    prob_acumulada_mayor = 1 - nbinom.cdf(k-1, r, p) if k > 0 else 1  # P(X ≥ k)

    # Estadísticas
    media = r * (1 - p) / p
    varianza = r * (1 - p) / (p**2)
    desviacion = np.sqrt(varianza)
    moda = int((r - 1) * (1 - p) / p) if r > 1 else 0

    #==========================================================
    # 3. MOSTRAR RESULTADOS NUMÉRICOS
    #==========================================================

    print("\n" + "="*70)
    print("📈 RESULTADOS DEL ANÁLISIS")
    print("="*70)
    print(f"\nParámetros de la distribución binomial negativa:")
    print(f"  • r = {r} (número de éxitos deseados)")
    print(f"  • p = {p:.4f} (probabilidad de éxito)")
    print(f"  • k = {k} (número de fracasos evaluado)")

    print(f"\n📊 Probabilidades calculadas:")
    print("-"*70)
    print(f"  • P(X = {k}) = {prob_exacta:.6f} ({prob_exacta*100:.4f}%)")
    print(f"  • P(X ≤ {k}) = {prob_acumulada_menor:.6f} ({prob_acumulada_menor*100:.4f}%)")
    print(f"  • P(X ≥ {k}) = {prob_acumulada_mayor:.6f} ({prob_acumulada_mayor*100:.4f}%)")

    print(f"\n📐 Estadísticas descriptivas:")
    print("-"*70)
    print(f"  • Media (μ) = {media:.4f} fracasos")
    print(f"  • Varianza (σ²) = {varianza:.4f}")
    print(f"  • Desviación estándar (σ) = {desviacion:.4f}")
    print(f"  • Moda = {moda} fracasos")
    print(f"  • Coeficiente de variación = {desviacion/media:.4f}")

    #==========================================================
    # 4. CREAR GRÁFICO PRINCIPAL CON ÁREAS SOMBREADAS
    #==========================================================

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    # Gráfico 1: Distribución completa
    ax1 = plt.subplot(2, 2, 1)

    # Barras de la distribución
    colors = ['#2E86AB' if i != k else '#E63946' for i in range(max_k + 1)]
    bars = ax1.bar(x, pmf, color=colors, alpha=0.8, edgecolor='black', linewidth=1)

    # Resaltar barra de k
    ax1.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black', linewidth=2)

    # Líneas de referencia
    ax1.axvline(x=media, color='#2C3E50', linestyle='--', linewidth=2, alpha=0.7, label=f'μ = {media:.2f}')
    ax1.axvline(x=media + desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)
    ax1.axvline(x=media - desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)

    # Línea de la moda
    if moda >= 0 and moda <= max_k:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=2, alpha=0.6, label=f'Moda = {moda}')

    ax1.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax1.set_ylabel('Probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Binomial Negativa\nr={r}, p={p:.3f}', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))
    ax1.legend(loc='best', fontsize=10)

    # Gráfico 2: P(X ≤ k) - Área izquierda
    ax2 = plt.subplot(2, 2, 2)

    # Barras de la distribución
    colors2 = ['#FF6B6B' if i <= k else '#2E86AB' for i in range(max_k + 1)]
    bars2 = ax2.bar(x, pmf, color=colors2, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax2.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax2.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax2.set_ylabel('Probabilidad', fontsize=12)
    ax2.set_title(f'P(X ≤ {k}) = {prob_acumulada_menor:.4f}\n({prob_acumulada_menor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Texto adicional
    ax2.text(0.02, 0.98, f'Área sombreada:\nP(X ≤ {k}) = {prob_acumulada_menor:.6f}',
             transform=ax2.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 3: P(X ≥ k) - Área derecha
    ax3 = plt.subplot(2, 2, 3)

    # Barras de la distribución
    colors3 = ['#FF6B6B' if i >= k else '#2E86AB' for i in range(max_k + 1)]
    bars3 = ax3.bar(x, pmf, color=colors3, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax3.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax3.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax3.set_ylabel('Probabilidad', fontsize=12)
    ax3.set_title(f'P(X ≥ {k}) = {prob_acumulada_mayor:.4f}\n({prob_acumulada_mayor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Texto adicional
    ax3.text(0.02, 0.98, f'Área sombreada:\nP(X ≥ {k}) = {prob_acumulada_mayor:.6f}',
             transform=ax3.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 4: Resumen de resultados y fórmula
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Mostrar la fórmula
    formula_text = (
        f"Fórmula de la Binomial Negativa:\n\n"
        f"P(X = k) = C({k+r-1}, {r-1}) · {p:.2f}^{r} · ({1-p:.2f})^{k}\n\n"
        f"C({k+r-1}, {r-1}) = {comb(k+r-1, r-1):.0f}\n\n"
        f"P(X = {k}) = {comb(k+r-1, r-1):.0f} · {p**r:.6f} · {(1-p)**k:.6f}\n"
        f"P(X = {k}) = {prob_exacta:.6f}"
    )

    ax4.text(0.5, 0.7, formula_text, transform=ax4.transAxes,
             fontsize=11, verticalalignment='center', horizontalalignment='center',
             bbox=dict(boxstyle='round,pad=0.5', facecolor='#F0F8FF', edgecolor='#2C3E50'),
             family='monospace')

    # Tabla de resultados
    resultados = [
        ['Concepto', 'Valor', 'Porcentaje'],
        ['P(X = k)', f'{prob_exacta:.6f}', f'{prob_exacta*100:.4f}%'],
        ['P(X ≤ k)', f'{prob_acumulada_menor:.6f}', f'{prob_acumulada_menor*100:.4f}%'],
        ['P(X ≥ k)', f'{prob_acumulada_mayor:.6f}', f'{prob_acumulada_mayor*100:.4f}%'],
        ['Media (μ)', f'{media:.4f}', ''],
        ['Varianza (σ²)', f'{varianza:.4f}', ''],
        ['Desv. Estándar (σ)', f'{desviacion:.4f}', ''],
        ['Moda', f'{moda}', '']
    ]

    # Crear tabla en la parte inferior
    table = ax4.table(cellText=resultados, loc='bottom', cellLoc='center',
                      colWidths=[0.3, 0.35, 0.35], bbox=[0, 0, 1, 0.4])

    # Estilo de la tabla
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.3)

    # Colores de la tabla
    for i in range(len(resultados)):
        for j in range(len(resultados[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            elif i in [1, 2, 3]:
                cell.set_facecolor('#E8F4F8')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('#F5F5F5')

    # Ajustar layout
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # 5. GRÁFICO ADICIONAL: FUNCIÓN DE DISTRIBUCIÓN ACUMULADA
    #==========================================================

    # Crear una figura separada para la CDF
    fig2, ax5 = plt.subplots(figsize=(10, 6))

    # Calcular CDF
    cdf = nbinom.cdf(x, r, p)

    # Graficar CDF
    ax5.step(x, cdf, where='mid', linewidth=2.5, color='#2E86AB', label='CDF')
    ax5.fill_between(x, 0, cdf, step='mid', alpha=0.3, color='#2E86AB')

    # Punto en k
    ax5.scatter(k, cdf[k], color='#E63946', s=150, zorder=5,
               label=f'F({k}) = {cdf[k]:.4f}')

    # Líneas de referencia
    ax5.axhline(y=cdf[k], color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.axvline(x=k, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles
    cuantiles = [0.25, 0.5, 0.75]
    colores_cuantiles = ['#F39C12', '#2ECC71', '#3498DB']
    for i, q in enumerate(cuantiles):
        try:
            q_val = nbinom.ppf(q, r, p)
            ax5.axvline(x=q_val, color=colores_cuantiles[i], linestyle=':',
                       linewidth=1.5, alpha=0.5)
            ax5.text(q_val, -0.03, f'Q{int(q*100)}={q_val}',
                    ha='center', fontsize=9, color=colores_cuantiles[i])
        except:
            pass

    ax5.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax5.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax5.set_title(f'Función de Distribución Acumulada\nr={r}, p={p:.3f}',
                  fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))
    ax5.set_ylim(-0.05, 1.05)
    ax5.legend(loc='best')

    # Texto con la probabilidad
    ax5.text(0.02, 0.98,
             f'P(X ≤ {k}) = {cdf[k]:.6f}\nP(X > {k}) = {1-cdf[k]:.6f}\n'
             f'Media = {media:.2f} (límite teórico)',
             transform=ax5.transAxes, fontsize=11, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()

    #==========================================================
    # 6. GRÁFICO DE PROBABILIDADES ACUMULADAS INVERSAS
    #==========================================================

    # Gráfico que muestra la probabilidad de necesitar exactamente k fracasos
    fig3, ax6 = plt.subplots(figsize=(10, 6))

    # Crear un rango extendido para mostrar el comportamiento asintótico
    x_ext = np.arange(0, max_k + 20)
    pmf_ext = nbinom.pmf(x_ext, r, p)
    cdf_ext = nbinom.cdf(x_ext, r, p)

    # Probabilidad de necesitar exactamente k fracasos
    ax6.bar(x, pmf, color='#2E86AB', alpha=0.6, edgecolor='black', label='PMF')
    ax6.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black',
           linewidth=2, label=f'P(X={k})')

    # Curva de probabilidad acumulada (eje derecho)
    ax6_twin = ax6.twinx()
    ax6_twin.plot(x, cdf, 'r-', linewidth=2, label='CDF')
    ax6_twin.fill_between(x, 0, cdf, alpha=0.1, color='red')

    ax6.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax6.set_ylabel('Probabilidad P(X=k)', fontsize=12)
    ax6_twin.set_ylabel('Probabilidad acumulada F(X)', fontsize=12, color='red')
    ax6_twin.tick_params(axis='y', labelcolor='red')

    ax6.set_title(f'Distribución Binomial Negativa: PMF y CDF\nr={r}, p={p:.3f}',
                  fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3, linestyle='--')
    ax6.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Línea vertical en la media
    ax6.axvline(x=media, color='green', linestyle='--', linewidth=2, alpha=0.7, label=f'μ={media:.2f}')

    # Leyendas combinadas
    lines1, labels1 = ax6.get_legend_handles_labels()
    lines2, labels2 = ax6_twin.get_legend_handles_labels()
    ax6.legend(lines1 + lines2, labels1 + labels2, loc='best', fontsize=10)

    plt.tight_layout()

    #==========================================================
    # 7. MOSTRAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # 8. OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"BinomialNegativa_r{r}_p{p:.2f}_k{k}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_cdf.png", dpi=300, bbox_inches='tight')
        fig3.savefig(f"{nombre_base}_pmf_cdf.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_cdf.png")
        print(f"   • {nombre_base}_pmf_cdf.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

    return {
        'r': r,
        'p': p,
        'k': k,
        'prob_exacta': prob_exacta,
        'prob_menor': prob_acumulada_menor,
        'prob_mayor': prob_acumulada_mayor,
        'media': media,
        'varianza': varianza,
        'desviacion': desviacion,
        'moda': moda
    }

#==========================================================
# FUNCIÓN PARA PROBLEMAS DE EJEMPLO
#==========================================================

def ejemplo_binomial_negativa():
    """
    Ejecuta ejemplos clásicos de la binomial negativa
    """

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Lanzamiento de moneda',
            'descripcion': '¿Probabilidad de obtener 3 caras antes de 2 cruces?',
            'r': 3,
            'p': 0.5,
            'k': 2
        },
        {
            'titulo': 'Ejemplo 2: Control de calidad',
            'descripcion': 'Probabilidad de encontrar 2 piezas defectuosas antes de 5 buenas (p=0.9)',
            'r': 5,
            'p': 0.9,
            'k': 2
        },
        {
            'titulo': 'Ejemplo 3: Paciencia en juegos',
            'descripcion': '¿Probabilidad de ganar 4 partidas antes de perder 3?',
            'r': 4,
            'p': 0.6,
            'k': 3
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")
        print(f"   r={ej['r']}, p={ej['p']}, k={ej['k']}")

        prob = nbinom.pmf(ej['k'], ej['r'], ej['p'])
        prob_acum = nbinom.cdf(ej['k'], ej['r'], ej['p'])
        media = ej['r'] * (1 - ej['p']) / ej['p']

        print(f"   P(X = {ej['k']}) = {prob:.6f} ({prob*100:.2f}%)")
        print(f"   P(X ≤ {ej['k']}) = {prob_acum:.6f} ({prob_acum*100:.2f}%)")
        print(f"   Media esperada = {media:.2f} fracasos")

    print("\n" + "="*70)
    print("¿Deseas analizar alguno de estos ejemplos en detalle?")
    print("Selecciona 1, 2, 3 o ejecuta la calculadora principal.")

#==========================================================
# FUNCIÓN PARA COMPARAR CON DISTRIBUCIÓN DE POISSON
#==========================================================

def comparar_con_poisson():
    """
    Compara la distribución binomial negativa con la distribución de Poisson
    """

    print("\n" + "="*70)
    print("📊 COMPARACIÓN: BINOMIAL NEGATIVA vs POISSON")
    print("="*70)
    print("\nLa binomial negativa se aproxima a Poisson cuando:")
    print("  • r es grande")
    print("  • p es cercano a 1")
    print("  • λ = r(1-p)/p se mantiene constante\n")

    # Parámetros
    r = 10
    p = 0.9
    lam = r * (1 - p) / p

    print(f"Parámetros: r={r}, p={p:.2f}, λ={lam:.2f}")

    # Crear gráfico de comparación
    fig, ax = plt.subplots(figsize=(12, 6))

    # Binomial negativa
    x = np.arange(0, 30)
    pmf_nb = nbinom.pmf(x, r, p)
    ax.bar(x - 0.2, pmf_nb, width=0.4, color='#2E86AB', alpha=0.7,
           label=f'Binomial Negativa (r={r}, p={p:.2f})')

    # Poisson
    from scipy.stats import poisson
    pmf_poisson = poisson.pmf(x, lam)
    ax.bar(x + 0.2, pmf_poisson, width=0.4, color='#E63946', alpha=0.7,
           label=f'Poisson (λ={lam:.2f})')

    ax.set_xlabel('Número de fracasos/eventos', fontsize=12)
    ax.set_ylabel('Probabilidad', fontsize=12)
    ax.set_title('Comparación: Binomial Negativa vs Poisson', fontsize=13, fontweight='bold')
    ax.grid(True, alpha=0.3, linestyle='--')
    ax.legend(loc='best')
    ax.set_xticks(np.arange(0, 31, 2))

    plt.tight_layout()
    plt.show()

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado de una distribución")
        print("  2. Ver ejemplos clásicos")
        print("  3. Comparar con distribución de Poisson")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_binomial_negativa()
        elif opcion == '2':
            ejemplo_binomial_negativa()
        elif opcion == '3':
            comparar_con_poisson()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA**
#
# La distribución binomial negativa modela el número de fracasos
# antes de obtener un número específico de éxitos.
#
# Fórmula: P(X = k) = C(k+r-1, r-1) * p^r * (1-p)^k
# donde:
#   r = número de éxitos deseados
#   p = probabilidad de éxito
#   k = número de fracasos
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import nbinom
from scipy.special import comb
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_binomial_negativa():
    """
    Calculadora interactiva para la distribución binomial negativa
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)
    print("\nLa binomial negativa modela el número de fracasos (k)")
    print("antes de obtener r éxitos, donde cada ensayo tiene")
    print("probabilidad p de éxito.\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            r = int(input("  Número de éxitos deseados (r, > 0): "))
            if r <= 0:
                print("  ❌ r debe ser un número positivo. Intenta de nuevo.")
                continue

            p = float(input("  Probabilidad de éxito (p, entre 0 y 1): "))
            if p <= 0 or p >= 1:
                print("  ❌ p debe estar entre 0 y 1. Intenta de nuevo.")
                continue

            k = int(input("  Número de fracasos a evaluar (k, ≥ 0): "))
            if k < 0:
                print("  ❌ k debe ser mayor o igual a 0. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

    #==========================================================
    # 2. CÁLCULO DE PROBABILIDADES
    #==========================================================

    # Rango de valores posibles para k (número de fracasos)
    max_k = max(20, k * 3)  # Asegurar que se muestre suficiente rango
    x = np.arange(0, max_k + 1)
    pmf = nbinom.pmf(x, r, p)

    # Probabilidades solicitadas
    prob_exacta = nbinom.pmf(k, r, p)
    prob_acumulada_menor = nbinom.cdf(k, r, p)  # P(X ≤ k)
    prob_acumulada_mayor = 1 - nbinom.cdf(k-1, r, p) if k > 0 else 1  # P(X ≥ k)

    # Estadísticas
    media = r * (1 - p) / p
    varianza = r * (1 - p) / (p**2)
    desviacion = np.sqrt(varianza)
    moda = int((r - 1) * (1 - p) / p) if r > 1 else 0

    #==========================================================
    # 3. MOSTRAR RESULTADOS NUMÉRICOS
    #==========================================================

    print("\n" + "="*70)
    print("📈 RESULTADOS DEL ANÁLISIS")
    print("="*70)
    print(f"\nParámetros de la distribución binomial negativa:")
    print(f"  • r = {r} (número de éxitos deseados)")
    print(f"  • p = {p:.4f} (probabilidad de éxito)")
    print(f"  • k = {k} (número de fracasos evaluado)")

    print(f"\n📊 Probabilidades calculadas:")
    print("-"*70)
    print(f"  • P(X = {k}) = {prob_exacta:.6f} ({prob_exacta*100:.4f}%)")
    print(f"  • P(X ≤ {k}) = {prob_acumulada_menor:.6f} ({prob_acumulada_menor*100:.4f}%)")
    print(f"  • P(X ≥ {k}) = {prob_acumulada_mayor:.6f} ({prob_acumulada_mayor*100:.4f}%)")

    print(f"\n📐 Estadísticas descriptivas:")
    print("-"*70)
    print(f"  • Media (μ) = {media:.4f} fracasos")
    print(f"  • Varianza (σ²) = {varianza:.4f}")
    print(f"  • Desviación estándar (σ) = {desviacion:.4f}")
    print(f"  • Moda = {moda} fracasos")
    print(f"  • Coeficiente de variación = {desviacion/media:.4f}")

    #==========================================================
    # 4. CREAR GRÁFICO PRINCIPAL CON ÁREAS SOMBREADAS
    #==========================================================

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    # Gráfico 1: Distribución completa
    ax1 = plt.subplot(2, 2, 1)

    # Barras de la distribución
    colors = ['#2E86AB' if i != k else '#E63946' for i in range(max_k + 1)]
    bars = ax1.bar(x, pmf, color=colors, alpha=0.8, edgecolor='black', linewidth=1)

    # Resaltar barra de k
    ax1.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black', linewidth=2)

    # Líneas de referencia
    ax1.axvline(x=media, color='#2C3E50', linestyle='--', linewidth=2, alpha=0.7, label=f'μ = {media:.2f}')
    ax1.axvline(x=media + desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)
    ax1.axvline(x=media - desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)

    # Línea de la moda
    if moda >= 0 and moda <= max_k:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=2, alpha=0.6, label=f'Moda = {moda}')

    ax1.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax1.set_ylabel('Probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Binomial Negativa\nr={r}, p={p:.3f}', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))
    ax1.legend(loc='best', fontsize=10)

    # Gráfico 2: P(X ≤ k) - Área izquierda
    ax2 = plt.subplot(2, 2, 2)

    # Barras de la distribución
    colors2 = ['#FF6B6B' if i <= k else '#2E86AB' for i in range(max_k + 1)]
    bars2 = ax2.bar(x, pmf, color=colors2, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax2.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax2.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax2.set_ylabel('Probabilidad', fontsize=12)
    ax2.set_title(f'P(X ≤ {k}) = {prob_acumulada_menor:.4f}\n({prob_acumulada_menor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Texto adicional
    ax2.text(0.02, 0.98, f'Área sombreada:\nP(X ≤ {k}) = {prob_acumulada_menor:.6f}',
             transform=ax2.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 3: P(X ≥ k) - Área derecha
    ax3 = plt.subplot(2, 2, 3)

    # Barras de la distribución
    colors3 = ['#FF6B6B' if i >= k else '#2E86AB' for i in range(max_k + 1)]
    bars3 = ax3.bar(x, pmf, color=colors3, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax3.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax3.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax3.set_ylabel('Probabilidad', fontsize=12)
    ax3.set_title(f'P(X ≥ {k}) = {prob_acumulada_mayor:.4f}\n({prob_acumulada_mayor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Texto adicional
    ax3.text(0.02, 0.98, f'Área sombreada:\nP(X ≥ {k}) = {prob_acumulada_mayor:.6f}',
             transform=ax3.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 4: Resumen de resultados y fórmula
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Mostrar la fórmula
    formula_text = (
        f"Fórmula de la Binomial Negativa:\n\n"
        f"P(X = k) = C({k+r-1}, {r-1}) · {p:.2f}^{r} · ({1-p:.2f})^{k}\n\n"
        f"C({k+r-1}, {r-1}) = {comb(k+r-1, r-1):.0f}\n\n"
        f"P(X = {k}) = {comb(k+r-1, r-1):.0f} · {p**r:.6f} · {(1-p)**k:.6f}\n"
        f"P(X = {k}) = {prob_exacta:.6f}"
    )

    ax4.text(0.5, 0.7, formula_text, transform=ax4.transAxes,
             fontsize=11, verticalalignment='center', horizontalalignment='center',
             bbox=dict(boxstyle='round,pad=0.5', facecolor='#F0F8FF', edgecolor='#2C3E50'),
             family='monospace')

    # Tabla de resultados
    resultados = [
        ['Concepto', 'Valor', 'Porcentaje'],
        ['P(X = k)', f'{prob_exacta:.6f}', f'{prob_exacta*100:.4f}%'],
        ['P(X ≤ k)', f'{prob_acumulada_menor:.6f}', f'{prob_acumulada_menor*100:.4f}%'],
        ['P(X ≥ k)', f'{prob_acumulada_mayor:.6f}', f'{prob_acumulada_mayor*100:.4f}%'],
        ['Media (μ)', f'{media:.4f}', ''],
        ['Varianza (σ²)', f'{varianza:.4f}', ''],
        ['Desv. Estándar (σ)', f'{desviacion:.4f}', ''],
        ['Moda', f'{moda}', '']
    ]

    # Crear tabla en la parte inferior
    table = ax4.table(cellText=resultados, loc='bottom', cellLoc='center',
                      colWidths=[0.3, 0.35, 0.35], bbox=[0, 0, 1, 0.4])

    # Estilo de la tabla
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.3)

    # Colores de la tabla
    for i in range(len(resultados)):
        for j in range(len(resultados[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            elif i in [1, 2, 3]:
                cell.set_facecolor('#E8F4F8')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('#F5F5F5')

    # Ajustar layout
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # 5. GRÁFICO ADICIONAL: FUNCIÓN DE DISTRIBUCIÓN ACUMULADA
    #==========================================================

    # Crear una figura separada para la CDF
    fig2, ax5 = plt.subplots(figsize=(10, 6))

    # Calcular CDF
    cdf = nbinom.cdf(x, r, p)

    # Graficar CDF
    ax5.step(x, cdf, where='mid', linewidth=2.5, color='#2E86AB', label='CDF')
    ax5.fill_between(x, 0, cdf, step='mid', alpha=0.3, color='#2E86AB')

    # Punto en k
    ax5.scatter(k, cdf[k], color='#E63946', s=150, zorder=5,
               label=f'F({k}) = {cdf[k]:.4f}')

    # Líneas de referencia
    ax5.axhline(y=cdf[k], color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.axvline(x=k, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles
    cuantiles = [0.25, 0.5, 0.75]
    colores_cuantiles = ['#F39C12', '#2ECC71', '#3498DB']
    for i, q in enumerate(cuantiles):
        try:
            q_val = nbinom.ppf(q, r, p)
            ax5.axvline(x=q_val, color=colores_cuantiles[i], linestyle=':',
                       linewidth=1.5, alpha=0.5)
            ax5.text(q_val, -0.03, f'Q{int(q*100)}={q_val}',
                    ha='center', fontsize=9, color=colores_cuantiles[i])
        except:
            pass

    ax5.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax5.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax5.set_title(f'Función de Distribución Acumulada\nr={r}, p={p:.3f}',
                  fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))
    ax5.set_ylim(-0.05, 1.05)
    ax5.legend(loc='best')

    # Texto con la probabilidad
    ax5.text(0.02, 0.98,
             f'P(X ≤ {k}) = {cdf[k]:.6f}\nP(X > {k}) = {1-cdf[k]:.6f}\n'
             f'Media = {media:.2f} (límite teórico)',
             transform=ax5.transAxes, fontsize=11, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()

    #==========================================================
    # 6. GRÁFICO DE PROBABILIDADES ACUMULADAS INVERSAS
    #==========================================================

    # Gráfico que muestra la probabilidad de necesitar exactamente k fracasos
    fig3, ax6 = plt.subplots(figsize=(10, 6))

    # Crear un rango extendido para mostrar el comportamiento asintótico
    x_ext = np.arange(0, max_k + 20)
    pmf_ext = nbinom.pmf(x_ext, r, p)
    cdf_ext = nbinom.cdf(x_ext, r, p)

    # Probabilidad de necesitar exactamente k fracasos
    ax6.bar(x, pmf, color='#2E86AB', alpha=0.6, edgecolor='black', label='PMF')
    ax6.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black',
           linewidth=2, label=f'P(X={k})')

    # Curva de probabilidad acumulada (eje derecho)
    ax6_twin = ax6.twinx()
    ax6_twin.plot(x, cdf, 'r-', linewidth=2, label='CDF')
    ax6_twin.fill_between(x, 0, cdf, alpha=0.1, color='red')

    ax6.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax6.set_ylabel('Probabilidad P(X=k)', fontsize=12)
    ax6_twin.set_ylabel('Probabilidad acumulada F(X)', fontsize=12, color='red')
    ax6_twin.tick_params(axis='y', labelcolor='red')

    ax6.set_title(f'Distribución Binomial Negativa: PMF y CDF\nr={r}, p={p:.3f}',
                  fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3, linestyle='--')
    ax6.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Línea vertical en la media
    ax6.axvline(x=media, color='green', linestyle='--', linewidth=2, alpha=0.7, label=f'μ={media:.2f}')

    # Leyendas combinadas
    lines1, labels1 = ax6.get_legend_handles_labels()
    lines2, labels2 = ax6_twin.get_legend_handles_labels()
    ax6.legend(lines1 + lines2, labels1 + labels2, loc='best', fontsize=10)

    plt.tight_layout()

    #==========================================================
    # 7. MOSTRAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # 8. OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"BinomialNegativa_r{r}_p{p:.2f}_k{k}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_cdf.png", dpi=300, bbox_inches='tight')
        fig3.savefig(f"{nombre_base}_pmf_cdf.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_cdf.png")
        print(f"   • {nombre_base}_pmf_cdf.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

    return {
        'r': r,
        'p': p,
        'k': k,
        'prob_exacta': prob_exacta,
        'prob_menor': prob_acumulada_menor,
        'prob_mayor': prob_acumulada_mayor,
        'media': media,
        'varianza': varianza,
        'desviacion': desviacion,
        'moda': moda
    }

#==========================================================
# FUNCIÓN PARA PROBLEMAS DE EJEMPLO
#==========================================================

def ejemplo_binomial_negativa():
    """
    Ejecuta ejemplos clásicos de la binomial negativa
    """

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Lanzamiento de moneda',
            'descripcion': '¿Probabilidad de obtener 3 caras antes de 2 cruces?',
            'r': 3,
            'p': 0.5,
            'k': 2
        },
        {
            'titulo': 'Ejemplo 2: Control de calidad',
            'descripcion': 'Probabilidad de encontrar 2 piezas defectuosas antes de 5 buenas (p=0.9)',
            'r': 5,
            'p': 0.9,
            'k': 2
        },
        {
            'titulo': 'Ejemplo 3: Paciencia en juegos',
            'descripcion': '¿Probabilidad de ganar 4 partidas antes de perder 3?',
            'r': 4,
            'p': 0.6,
            'k': 3
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")
        print(f"   r={ej['r']}, p={ej['p']}, k={ej['k']}")

        prob = nbinom.pmf(ej['k'], ej['r'], ej['p'])
        prob_acum = nbinom.cdf(ej['k'], ej['r'], ej['p'])
        media = ej['r'] * (1 - ej['p']) / ej['p']

        print(f"   P(X = {ej['k']}) = {prob:.6f} ({prob*100:.2f}%)")
        print(f"   P(X ≤ {ej['k']}) = {prob_acum:.6f} ({prob_acum*100:.2f}%)")
        print(f"   Media esperada = {media:.2f} fracasos")

    print("\n" + "="*70)
    print("¿Deseas analizar alguno de estos ejemplos en detalle?")
    print("Selecciona 1, 2, 3 o ejecuta la calculadora principal.")

#==========================================================
# FUNCIÓN PARA COMPARAR CON DISTRIBUCIÓN DE POISSON
#==========================================================

def comparar_con_poisson():
    """
    Compara la distribución binomial negativa con la distribución de Poisson
    """

    print("\n" + "="*70)
    print("📊 COMPARACIÓN: BINOMIAL NEGATIVA vs POISSON")
    print("="*70)
    print("\nLa binomial negativa se aproxima a Poisson cuando:")
    print("  • r es grande")
    print("  • p es cercano a 1")
    print("  • λ = r(1-p)/p se mantiene constante\n")

    # Parámetros
    r = 10
    p = 0.9
    lam = r * (1 - p) / p

    print(f"Parámetros: r={r}, p={p:.2f}, λ={lam:.2f}")

    # Crear gráfico de comparación
    fig, ax = plt.subplots(figsize=(12, 6))

    # Binomial negativa
    x = np.arange(0, 30)
    pmf_nb = nbinom.pmf(x, r, p)
    ax.bar(x - 0.2, pmf_nb, width=0.4, color='#2E86AB', alpha=0.7,
           label=f'Binomial Negativa (r={r}, p={p:.2f})')

    # Poisson
    from scipy.stats import poisson
    pmf_poisson = poisson.pmf(x, lam)
    ax.bar(x + 0.2, pmf_poisson, width=0.4, color='#E63946', alpha=0.7,
           label=f'Poisson (λ={lam:.2f})')

    ax.set_xlabel('Número de fracasos/eventos', fontsize=12)
    ax.set_ylabel('Probabilidad', fontsize=12)
    ax.set_title('Comparación: Binomial Negativa vs Poisson', fontsize=13, fontweight='bold')
    ax.grid(True, alpha=0.3, linestyle='--')
    ax.legend(loc='best')
    ax.set_xticks(np.arange(0, 31, 2))

    plt.tight_layout()
    plt.show()

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado de una distribución")
        print("  2. Ver ejemplos clásicos")
        print("  3. Comparar con distribución de Poisson")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_binomial_negativa()
        elif opcion == '2':
            ejemplo_binomial_negativa()
        elif opcion == '3':
            comparar_con_poisson()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA**
#
# La distribución binomial negativa modela el número de fracasos
# antes de obtener un número específico de éxitos.
#
# Fórmula: P(X = k) = C(k+r-1, r-1) * p^r * (1-p)^k
# donde:
#   r = número de éxitos deseados
#   p = probabilidad de éxito
#   k = número de fracasos
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import nbinom
from scipy.special import comb
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_binomial_negativa():
    """
    Calculadora interactiva para la distribución binomial negativa
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)
    print("\nLa binomial negativa modela el número de fracasos (k)")
    print("antes de obtener r éxitos, donde cada ensayo tiene")
    print("probabilidad p de éxito.\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            r = int(input("  Número de éxitos deseados (r, > 0): "))
            if r <= 0:
                print("  ❌ r debe ser un número positivo. Intenta de nuevo.")
                continue

            p = float(input("  Probabilidad de éxito (p, entre 0 y 1): "))
            if p <= 0 or p >= 1:
                print("  ❌ p debe estar entre 0 y 1. Intenta de nuevo.")
                continue

            k = int(input("  Número de fracasos a evaluar (k, ≥ 0): "))
            if k < 0:
                print("  ❌ k debe ser mayor o igual a 0. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

    #==========================================================
    # 2. CÁLCULO DE PROBABILIDADES
    #==========================================================

    # Rango de valores posibles para k (número de fracasos)
    max_k = max(20, k * 3)  # Asegurar que se muestre suficiente rango
    x = np.arange(0, max_k + 1)
    pmf = nbinom.pmf(x, r, p)

    # Probabilidades solicitadas
    prob_exacta = nbinom.pmf(k, r, p)
    prob_acumulada_menor = nbinom.cdf(k, r, p)  # P(X ≤ k)
    prob_acumulada_mayor = 1 - nbinom.cdf(k-1, r, p) if k > 0 else 1  # P(X ≥ k)

    # Estadísticas
    media = r * (1 - p) / p
    varianza = r * (1 - p) / (p**2)
    desviacion = np.sqrt(varianza)
    moda = int((r - 1) * (1 - p) / p) if r > 1 else 0

    #==========================================================
    # 3. MOSTRAR RESULTADOS NUMÉRICOS
    #==========================================================

    print("\n" + "="*70)
    print("📈 RESULTADOS DEL ANÁLISIS")
    print("="*70)
    print(f"\nParámetros de la distribución binomial negativa:")
    print(f"  • r = {r} (número de éxitos deseados)")
    print(f"  • p = {p:.4f} (probabilidad de éxito)")
    print(f"  • k = {k} (número de fracasos evaluado)")

    print(f"\n📊 Probabilidades calculadas:")
    print("-"*70)
    print(f"  • P(X = {k}) = {prob_exacta:.6f} ({prob_exacta*100:.4f}%)")
    print(f"  • P(X ≤ {k}) = {prob_acumulada_menor:.6f} ({prob_acumulada_menor*100:.4f}%)")
    print(f"  • P(X ≥ {k}) = {prob_acumulada_mayor:.6f} ({prob_acumulada_mayor*100:.4f}%)")

    print(f"\n📐 Estadísticas descriptivas:")
    print("-"*70)
    print(f"  • Media (μ) = {media:.4f} fracasos")
    print(f"  • Varianza (σ²) = {varianza:.4f}")
    print(f"  • Desviación estándar (σ) = {desviacion:.4f}")
    print(f"  • Moda = {moda} fracasos")
    print(f"  • Coeficiente de variación = {desviacion/media:.4f}")

    #==========================================================
    # 4. CREAR GRÁFICO PRINCIPAL CON ÁREAS SOMBREADAS
    #==========================================================

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    # Gráfico 1: Distribución completa
    ax1 = plt.subplot(2, 2, 1)

    # Barras de la distribución
    colors = ['#2E86AB' if i != k else '#E63946' for i in range(max_k + 1)]
    bars = ax1.bar(x, pmf, color=colors, alpha=0.8, edgecolor='black', linewidth=1)

    # Resaltar barra de k
    ax1.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black', linewidth=2)

    # Líneas de referencia
    ax1.axvline(x=media, color='#2C3E50', linestyle='--', linewidth=2, alpha=0.7, label=f'μ = {media:.2f}')
    ax1.axvline(x=media + desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)
    ax1.axvline(x=media - desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)

    # Línea de la moda
    if moda >= 0 and moda <= max_k:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=2, alpha=0.6, label=f'Moda = {moda}')

    ax1.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax1.set_ylabel('Probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Binomial Negativa\nr={r}, p={p:.3f}', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))
    ax1.legend(loc='best', fontsize=10)

    # Gráfico 2: P(X ≤ k) - Área izquierda
    ax2 = plt.subplot(2, 2, 2)

    # Barras de la distribución
    colors2 = ['#FF6B6B' if i <= k else '#2E86AB' for i in range(max_k + 1)]
    bars2 = ax2.bar(x, pmf, color=colors2, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax2.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax2.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax2.set_ylabel('Probabilidad', fontsize=12)
    ax2.set_title(f'P(X ≤ {k}) = {prob_acumulada_menor:.4f}\n({prob_acumulada_menor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Texto adicional
    ax2.text(0.02, 0.98, f'Área sombreada:\nP(X ≤ {k}) = {prob_acumulada_menor:.6f}',
             transform=ax2.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 3: P(X ≥ k) - Área derecha
    ax3 = plt.subplot(2, 2, 3)

    # Barras de la distribución
    colors3 = ['#FF6B6B' if i >= k else '#2E86AB' for i in range(max_k + 1)]
    bars3 = ax3.bar(x, pmf, color=colors3, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax3.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax3.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax3.set_ylabel('Probabilidad', fontsize=12)
    ax3.set_title(f'P(X ≥ {k}) = {prob_acumulada_mayor:.4f}\n({prob_acumulada_mayor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Texto adicional
    ax3.text(0.02, 0.98, f'Área sombreada:\nP(X ≥ {k}) = {prob_acumulada_mayor:.6f}',
             transform=ax3.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 4: Resumen de resultados y fórmula
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Mostrar la fórmula
    formula_text = (
        f"Fórmula de la Binomial Negativa:\n\n"
        f"P(X = k) = C({k+r-1}, {r-1}) · {p:.2f}^{r} · ({1-p:.2f})^{k}\n\n"
        f"C({k+r-1}, {r-1}) = {comb(k+r-1, r-1):.0f}\n\n"
        f"P(X = {k}) = {comb(k+r-1, r-1):.0f} · {p**r:.6f} · {(1-p)**k:.6f}\n"
        f"P(X = {k}) = {prob_exacta:.6f}"
    )

    ax4.text(0.5, 0.7, formula_text, transform=ax4.transAxes,
             fontsize=11, verticalalignment='center', horizontalalignment='center',
             bbox=dict(boxstyle='round,pad=0.5', facecolor='#F0F8FF', edgecolor='#2C3E50'),
             family='monospace')

    # Tabla de resultados
    resultados = [
        ['Concepto', 'Valor', 'Porcentaje'],
        ['P(X = k)', f'{prob_exacta:.6f}', f'{prob_exacta*100:.4f}%'],
        ['P(X ≤ k)', f'{prob_acumulada_menor:.6f}', f'{prob_acumulada_menor*100:.4f}%'],
        ['P(X ≥ k)', f'{prob_acumulada_mayor:.6f}', f'{prob_acumulada_mayor*100:.4f}%'],
        ['Media (μ)', f'{media:.4f}', ''],
        ['Varianza (σ²)', f'{varianza:.4f}', ''],
        ['Desv. Estándar (σ)', f'{desviacion:.4f}', ''],
        ['Moda', f'{moda}', '']
    ]

    # Crear tabla en la parte inferior
    table = ax4.table(cellText=resultados, loc='bottom', cellLoc='center',
                      colWidths=[0.3, 0.35, 0.35], bbox=[0, 0, 1, 0.4])

    # Estilo de la tabla
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.3)

    # Colores de la tabla
    for i in range(len(resultados)):
        for j in range(len(resultados[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            elif i in [1, 2, 3]:
                cell.set_facecolor('#E8F4F8')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('#F5F5F5')

    # Ajustar layout
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # 5. GRÁFICO ADICIONAL: FUNCIÓN DE DISTRIBUCIÓN ACUMULADA
    #==========================================================

    # Crear una figura separada para la CDF
    fig2, ax5 = plt.subplots(figsize=(10, 6))

    # Calcular CDF
    cdf = nbinom.cdf(x, r, p)

    # Graficar CDF
    ax5.step(x, cdf, where='mid', linewidth=2.5, color='#2E86AB', label='CDF')
    ax5.fill_between(x, 0, cdf, step='mid', alpha=0.3, color='#2E86AB')

    # Punto en k
    ax5.scatter(k, cdf[k], color='#E63946', s=150, zorder=5,
               label=f'F({k}) = {cdf[k]:.4f}')

    # Líneas de referencia
    ax5.axhline(y=cdf[k], color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.axvline(x=k, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles
    cuantiles = [0.25, 0.5, 0.75]
    colores_cuantiles = ['#F39C12', '#2ECC71', '#3498DB']
    for i, q in enumerate(cuantiles):
        try:
            q_val = nbinom.ppf(q, r, p)
            ax5.axvline(x=q_val, color=colores_cuantiles[i], linestyle=':',
                       linewidth=1.5, alpha=0.5)
            ax5.text(q_val, -0.03, f'Q{int(q*100)}={q_val}',
                    ha='center', fontsize=9, color=colores_cuantiles[i])
        except:
            pass

    ax5.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax5.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax5.set_title(f'Función de Distribución Acumulada\nr={r}, p={p:.3f}',
                  fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))
    ax5.set_ylim(-0.05, 1.05)
    ax5.legend(loc='best')

    # Texto con la probabilidad
    ax5.text(0.02, 0.98,
             f'P(X ≤ {k}) = {cdf[k]:.6f}\nP(X > {k}) = {1-cdf[k]:.6f}\n'
             f'Media = {media:.2f} (límite teórico)',
             transform=ax5.transAxes, fontsize=11, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()

    #==========================================================
    # 6. GRÁFICO DE PROBABILIDADES ACUMULADAS INVERSAS
    #==========================================================

    # Gráfico que muestra la probabilidad de necesitar exactamente k fracasos
    fig3, ax6 = plt.subplots(figsize=(10, 6))

    # Crear un rango extendido para mostrar el comportamiento asintótico
    x_ext = np.arange(0, max_k + 20)
    pmf_ext = nbinom.pmf(x_ext, r, p)
    cdf_ext = nbinom.cdf(x_ext, r, p)

    # Probabilidad de necesitar exactamente k fracasos
    ax6.bar(x, pmf, color='#2E86AB', alpha=0.6, edgecolor='black', label='PMF')
    ax6.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black',
           linewidth=2, label=f'P(X={k})')

    # Curva de probabilidad acumulada (eje derecho)
    ax6_twin = ax6.twinx()
    ax6_twin.plot(x, cdf, 'r-', linewidth=2, label='CDF')
    ax6_twin.fill_between(x, 0, cdf, alpha=0.1, color='red')

    ax6.set_xlabel('Número de fracasos (X)', fontsize=12)
    ax6.set_ylabel('Probabilidad P(X=k)', fontsize=12)
    ax6_twin.set_ylabel('Probabilidad acumulada F(X)', fontsize=12, color='red')
    ax6_twin.tick_params(axis='y', labelcolor='red')

    ax6.set_title(f'Distribución Binomial Negativa: PMF y CDF\nr={r}, p={p:.3f}',
                  fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3, linestyle='--')
    ax6.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Línea vertical en la media
    ax6.axvline(x=media, color='green', linestyle='--', linewidth=2, alpha=0.7, label=f'μ={media:.2f}')

    # Leyendas combinadas
    lines1, labels1 = ax6.get_legend_handles_labels()
    lines2, labels2 = ax6_twin.get_legend_handles_labels()
    ax6.legend(lines1 + lines2, labels1 + labels2, loc='best', fontsize=10)

    plt.tight_layout()

    #==========================================================
    # 7. MOSTRAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # 8. OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"BinomialNegativa_r{r}_p{p:.2f}_k{k}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_cdf.png", dpi=300, bbox_inches='tight')
        fig3.savefig(f"{nombre_base}_pmf_cdf.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_cdf.png")
        print(f"   • {nombre_base}_pmf_cdf.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

    return {
        'r': r,
        'p': p,
        'k': k,
        'prob_exacta': prob_exacta,
        'prob_menor': prob_acumulada_menor,
        'prob_mayor': prob_acumulada_mayor,
        'media': media,
        'varianza': varianza,
        'desviacion': desviacion,
        'moda': moda
    }

#==========================================================
# FUNCIÓN PARA PROBLEMAS DE EJEMPLO
#==========================================================

def ejemplo_binomial_negativa():
    """
    Ejecuta ejemplos clásicos de la binomial negativa
    """

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Lanzamiento de moneda',
            'descripcion': '¿Probabilidad de obtener 3 caras antes de 2 cruces?',
            'r': 3,
            'p': 0.5,
            'k': 2
        },
        {
            'titulo': 'Ejemplo 2: Control de calidad',
            'descripcion': 'Probabilidad de encontrar 2 piezas defectuosas antes de 5 buenas (p=0.9)',
            'r': 5,
            'p': 0.9,
            'k': 2
        },
        {
            'titulo': 'Ejemplo 3: Paciencia en juegos',
            'descripcion': '¿Probabilidad de ganar 4 partidas antes de perder 3?',
            'r': 4,
            'p': 0.6,
            'k': 3
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")
        print(f"   r={ej['r']}, p={ej['p']}, k={ej['k']}")

        prob = nbinom.pmf(ej['k'], ej['r'], ej['p'])
        prob_acum = nbinom.cdf(ej['k'], ej['r'], ej['p'])
        media = ej['r'] * (1 - ej['p']) / ej['p']

        print(f"   P(X = {ej['k']}) = {prob:.6f} ({prob*100:.2f}%)")
        print(f"   P(X ≤ {ej['k']}) = {prob_acum:.6f} ({prob_acum*100:.2f}%)")
        print(f"   Media esperada = {media:.2f} fracasos")

    print("\n" + "="*70)
    print("¿Deseas analizar alguno de estos ejemplos en detalle?")
    print("Selecciona 1, 2, 3 o ejecuta la calculadora principal.")

#==========================================================
# FUNCIÓN PARA COMPARAR CON DISTRIBUCIÓN DE POISSON
#==========================================================

def comparar_con_poisson():
    """
    Compara la distribución binomial negativa con la distribución de Poisson
    """

    print("\n" + "="*70)
    print("📊 COMPARACIÓN: BINOMIAL NEGATIVA vs POISSON")
    print("="*70)
    print("\nLa binomial negativa se aproxima a Poisson cuando:")
    print("  • r es grande")
    print("  • p es cercano a 1")
    print("  • λ = r(1-p)/p se mantiene constante\n")

    # Parámetros
    r = 10
    p = 0.9
    lam = r * (1 - p) / p

    print(f"Parámetros: r={r}, p={p:.2f}, λ={lam:.2f}")

    # Crear gráfico de comparación
    fig, ax = plt.subplots(figsize=(12, 6))

    # Binomial negativa
    x = np.arange(0, 30)
    pmf_nb = nbinom.pmf(x, r, p)
    ax.bar(x - 0.2, pmf_nb, width=0.4, color='#2E86AB', alpha=0.7,
           label=f'Binomial Negativa (r={r}, p={p:.2f})')

    # Poisson
    from scipy.stats import poisson
    pmf_poisson = poisson.pmf(x, lam)
    ax.bar(x + 0.2, pmf_poisson, width=0.4, color='#E63946', alpha=0.7,
           label=f'Poisson (λ={lam:.2f})')

    ax.set_xlabel('Número de fracasos/eventos', fontsize=12)
    ax.set_ylabel('Probabilidad', fontsize=12)
    ax.set_title('Comparación: Binomial Negativa vs Poisson', fontsize=13, fontweight='bold')
    ax.grid(True, alpha=0.3, linestyle='--')
    ax.legend(loc='best')
    ax.set_xticks(np.arange(0, 31, 2))

    plt.tight_layout()
    plt.show()

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BINOMIAL NEGATIVA")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado de una distribución")
        print("  2. Ver ejemplos clásicos")
        print("  3. Comparar con distribución de Poisson")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_binomial_negativa()
        elif opcion == '2':
            ejemplo_binomial_negativa()
        elif opcion == '3':
            comparar_con_poisson()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```

# **20. Distribución de Poisson**
+ *La distribución de Poisson modela el número de eventos que ocurren en un intervalo de tiempo o espacio, bajo la suposición de que los eventos ocurren con una tasa constante y de forma independiente entre sí.*

- **Función de probabilidad**: $$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$
- **Parámetros**: $\lambda$ es la tasa promedio de ocurrencia de eventos en el intervalo de tiempo o espacio.


### **Distribución de Poisson - Momentos**
- **Valor esperado (Esperanza)**: $$E(X) = \lambda$$
- **Varianza**: $$\text{Var}(X) = \lambda$$
- **Desviación estándar**: $$\sigma(X) = \sqrt{\lambda}$$




```python
# @title **A. Gráfica de la Distribución de Poisson** $$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$

lambda_poisson = 5
x_poisson = np.arange(0, 15)
pmf_poisson = poisson.pmf(x_poisson, lambda_poisson)
plot_distribution2(poisson, x_poisson, pmf_poisson, "Distribución de Poisson")
```

```python
# @title **B. Gráfica de la Distribución de Poisson** $$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$
%%R
# install.packages("ggplot2")
library(ggplot2)

# Parámetro
lambda_poisson <- 5

# Datos
x_poisson <- 0:14
pmf_poisson <- dpois(x_poisson, lambda = lambda_poisson)

datos <- data.frame(
  X = x_poisson,
  Probabilidad = pmf_poisson
)

# Gráfica
ggplot(datos, aes(x = factor(X), y = Probabilidad)) +
  geom_col(fill = "steelblue", color = "black") +
  labs(
    title = "Distribución de Poisson",
    x = "Valores (X)",
    y = "Probabilidad"
  ) +
  theme_minimal(base_size = 14)
```

# **21. Ejemplo 1. Distribución de Poisson: Una central telefónica recibe en promedio 2 llamadas por minuto. ¿Cuál es la probabilidad de que en un minuto se reciban exactamente 3 llamadas?**

```python
# @title  **A. Ejemplo 1:- Una central telefónica recibe en promedio 2 llamadas por minuto. ¿Cuál es la probabilidad de que en un minuto se reciban exactamente 3 llamadas? -Distribución de Poisson** $$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parámetros
lambda_poisson = 2  # Tasa promedio de ocurrencia
k = 3  # Número de eventos (llamadas)

# Probabilidad
prob_poisson_1 = poisson.pmf(k, lambda_poisson)
print(f"Probabilidad de recibir exactamente 3 llamadas: P(X=3) = {prob_poisson_1:.5f}")

# Gráfico sombreando P(X=3)
x_poisson = np.arange(0, 15)
pmf_poisson = poisson.pmf(x_poisson, lambda_poisson)
plt.bar(x_poisson, pmf_poisson)
plt.title('Distribución de Poisson - P(X=3)')
plt.xlabel('Número de eventos')
plt.ylabel('Probabilidad')

plt.bar(3, prob_poisson_1, color='red')
plt.text(3, prob_poisson_1 + 0.01, f"P(X=3) = {prob_poisson_1:.5f}", fontsize=10, color='black')

plt.show()
```

# **22. Ejemplo 2. Distribución de Poisson: En la zona que rodea un punto kilométrico de una autovía hay una población de erizos. Se ha realizado un estudio por el que se ha deducido que los erizos cruzan la autovía a razón de 4 cada 30 minutos**

```python
# @title  **A. Ejemplo 2: En la zona que rodea un punto kilométrico de una autovía hay una población de erizos. Se ha realizado un estudio por el que se ha deducido que los erizos cruzan la autovía a razón de 4 cada 30 minutos. La probabilidad de que 6 o más erizos crucen la autovía en una hora es: Dibuja la distribución de poisson y sombrea el rectangulo con P(X>=6)**

# Parámetros del problema
lambda_poisson = 8  # Tasa promedio de erizos que cruzan en una hora (4 erizos cada 30 minutos * 2 = 8 erizos por hora)

# Calcular la probabilidad de que 6 o más erizos crucen la autovía en una hora (P(X >= 6))
prob_6_o_mas = 1 - poisson.cdf(5, lambda_poisson)  # P(X >= 6) = 1 - P(X <= 5)

print(f"La probabilidad de que 6 o más erizos crucen la autovía en una hora es: {prob_6_o_mas:.5f}")

# Gráfico sombreando P(X >= 6)
x_poisson = np.arange(0, 20)
pmf_poisson = poisson.pmf(x_poisson, lambda_poisson)
plt.bar(x_poisson, pmf_poisson)
plt.title('Distribución de Poisson - P(X ≥ 6)')
plt.xlabel('Número de erizos que cruzan')
plt.ylabel('Probabilidad')

for i in range(6, 20):
    plt.bar(i, pmf_poisson[i], color='red')

plt.text(10, 0.05, f"P(X ≥ 6) = {prob_6_o_mas:.5f}", fontsize=12, color='black')

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 2. En la zona que rodea un punto kilométrico de una autovía hay una población de erizos. Se ha realizado un estudio por el que se ha deducido que los erizos cruzan la autovía a razón de 4 cada 30 minutos.**
#
# ¿Cuál es la probabilidad de que 6 o más erizos crucen la autovía
# en una hora?
#
# Distribución de Poisson
# $$P(X=k)=\frac{\lambda^k e^{-\lambda}}{k!}$$
# ==========================================================
%%R
library(ggplot2)

# Parámetro
lambda_poisson <- 8

# Datos
datos <- data.frame(
  X = 0:19
)

datos$Probabilidad <- dpois(datos$X, lambda_poisson)
datos$Grupo <- ifelse(datos$X >= 6, "P(X≥6)", "Resto")

# Probabilidad acumulada
prob_6_o_mas <- 1 - ppois(5, lambda_poisson)

ggplot(datos,
       aes(x = factor(X),
           y = Probabilidad,
           fill = Grupo)) +
  geom_col(color = "black") +
  scale_fill_manual(values = c("Resto" = "steelblue",
                               "P(X≥6)" = "red")) +
  annotate(
    "text",
    x = 13,
    y = max(datos$Probabilidad) * 0.92,
    label = paste0("P(X ≥ 6) = ", round(prob_6_o_mas, 5)),
    size = 5,
    fontface = "bold"
  ) +
  labs(
    title = "Distribución de Poisson",
    subtitle = expression(P(X >= 6)),
    x = "Número de erizos que cruzan",
    y = "Probabilidad"
  ) +
  theme_minimal(base_size = 14) +
  theme(legend.position = "bottom")
```

# **23. Ejemplo 3. Distribución de Poisson: En una carretera se producen un promedio de 2 accidentes anuales. Calcula la probabilidad de que estos cinco años se produzcan más de 3 accidentes.**

```python
# @title **Ejemplo 3. En una carretera se producen un promedio de 2 accidentes anuales. Calcula la probabilidad de que estos cinco años se produzcan más de 3 accidentes.**
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from scipy.stats import poisson
from IPython.display import HTML

# Parámetros del problema
lambda_poisson = 2  # Promedio de accidentes por año
lambda_5_years = lambda_poisson * 5  # Tasa promedio de accidentes en 5 años

# Calcular la probabilidad de que en 5 años se produzcan más de 3 accidentes
prob_mas_de_3 = 1 - poisson.cdf(3, lambda_5_years)  # P(X > 3) = 1 - P(X <= 3)

# Configuración inicial del gráfico
fig, ax = plt.subplots(figsize=(8, 6))
x_poisson = np.arange(0, 25)
pmf_poisson = poisson.pmf(x_poisson, lambda_5_years)
bars = ax.bar(x_poisson, pmf_poisson, color='gray')

# Texto para mostrar la probabilidad calculada
text = ax.text(0.6, 0.8, '', transform=ax.transAxes, fontsize=12, color='green')

# Función para actualizar la animación
def actualizar(frame):
    for i, bar in enumerate(bars):
        # Colorea de rojo los valores donde X > 3
        if i > 3:
            bar.set_color('red')
        else:
            bar.set_color('gray')
    text.set_text(f"P(X > 3) = {prob_mas_de_3:.5f}")

# Configuración de la animación
ani = animation.FuncAnimation(fig, actualizar, frames=20, interval=1000, repeat=False)

# Muestra la animación en el notebook
plt.title('Distribución de Poisson - Probabilidad de más de 3 accidentes en 5 años')
plt.xlabel('Número de accidentes en 5 años')
plt.ylabel('Probabilidad')
HTML(ani.to_html5_video())
```

```python
# ==========================================================
# @title **B. Ejemplo 3. En una carretera se producen un promedio de 2 accidentes anuales. Calcula la probabilidad de que en cinco años se produzcan más de 3 accidentes.**
#
# Distribución de Poisson
# ==========================================================
%%R
# Parámetros
lambda_anual <- 2
lambda_5_anios <- lambda_anual * 5

# Probabilidad P(X > 3)
prob_mas_de_3 <- 1 - ppois(3, lambda = lambda_5_anios)

cat("=============================================\n")
cat("Distribución de Poisson\n")
cat("=============================================\n")
cat(sprintf("Promedio anual de accidentes = %.0f\n", lambda_anual))
cat(sprintf("Promedio en 5 años = %.0f\n\n", lambda_5_anios))

cat(sprintf("P(X > 3) = %.5f\n", prob_mas_de_3))
cat(sprintf("P(X > 3) = %.2f%%\n", prob_mas_de_3*100))
cat("=============================================\n")

# Valores de la distribución
x <- 0:24
pmf <- dpois(x, lambda_5_anios)

# Colores
colores <- ifelse(x > 3, "red", "gray70")

# Gráfico
barplot(
  pmf,
  names.arg = x,
  col = colores,
  border = "black",
  main = "Distribución de Poisson\nP(X > 3)",
  xlab = "Número de accidentes en 5 años",
  ylab = "Probabilidad"
)

text(
  x = 18,
  y = max(pmf)*0.92,
  labels = paste0("P(X > 3) = ", round(prob_mas_de_3,5)),
  cex = 1,
  font = 2,
  col = "darkgreen"
)

grid()
```

```python
# ==========================================================
# @title **C. (version 2 en R) Ejemplo 3. En una carretera se producen un promedio de 2 accidentes anuales. Calcula la probabilidad de que en cinco años se produzcan más de 3 accidentes.**
#
# Distribución de Poisson
# ==========================================================
%%R

library(ggplot2)

# Parámetros
lambda_anual <- 2
lambda_5_anios <- lambda_anual*5

# Datos
datos <- data.frame(
  Accidentes = 0:24
)

datos$Probabilidad <- dpois(datos$Accidentes, lambda_5_anios)
datos$Zona <- ifelse(datos$Accidentes > 3,
                     "P(X > 3)",
                     "Resto")

# Probabilidad
prob_mas_de_3 <- 1 - ppois(3, lambda_5_anios)

ggplot(datos,
       aes(factor(Accidentes),
           Probabilidad,
           fill=Zona))+

geom_col(color="black")+

scale_fill_manual(values=c(
  "Resto"="gray70",
  "P(X > 3)"="red"
))+

annotate("text",
         x=18,
         y=max(datos$Probabilidad)*0.92,
         label=paste0("P(X > 3) = ",
                      round(prob_mas_de_3,5)),
         colour="darkgreen",
         size=5,
         fontface="bold")+

labs(
title="Distribución de Poisson",
subtitle="Probabilidad de más de 3 accidentes en cinco años",
x="Número de accidentes",
y="Probabilidad"
)+

theme_minimal(base_size=14)+

theme(
legend.position="bottom",
plot.title=element_text(face="bold"),
plot.subtitle=element_text(face="bold")
)
```

# **25. Ejemplo 5. Distribución de Poisson: Si un banco recibe en promedio 6 cheques sin fondo por día, ¿cuáles son las probabilidades de que reciba: a. cuatro cheques sin fondo en un día dado, b. cincuenta cheques sin fondo en una semana, c. diez cheques sin fondo en un 12 horas, d. 5 cheques sin fondos en 4 horas?**

```python
# @title **A. Ejemplo 5. Si un banco recibe en promedio 6 cheques sin fondo por día, ¿cuáles son las probabilidades de que reciba: a. cuatro cheques sin fondo en un día dado, b. cincuenta cheques sin fondo en una semana, c. diez cheques sin fondo en un 12 horas, d.  5 cheques sin fondos en 4 horas?**
# Parámetros
lambda_dia = 6  # Promedio de cheques sin fondo por día

# a. Cuatro cheques sin fondo en un día dado
k_a = 4
prob_a = poisson.pmf(k_a, lambda_dia)
print(f"a. Probabilidad de 4 cheques sin fondo en un día: {prob_a:.5f}")

# Gráfico para a.
x_poisson_a = np.arange(0, 15)
pmf_poisson_a = poisson.pmf(x_poisson_a, lambda_dia)
plt.bar(x_poisson_a, pmf_poisson_a)
plt.title('Distribución de Poisson - P(X=4)')
plt.xlabel('Número de cheques sin fondo')
plt.ylabel('Probabilidad')

plt.bar(k_a, prob_a, color='red')
plt.text(k_a, prob_a + 0.01, f"P(X=4) = {prob_a:.5f}", fontsize=10, color='black')

plt.show()

# b. Cincuenta cheques sin fondo en una semana
lambda_semana = lambda_dia * 7  # Promedio de cheques sin fondo en una semana
k_b = 50
prob_b = poisson.pmf(k_b, lambda_semana)
print(f"b. Probabilidad de 50 cheques sin fondo en una semana: {prob_b:.5f}")

# Gráfico para b.
x_poisson_b = np.arange(0, 70)
pmf_poisson_b = poisson.pmf(x_poisson_b, lambda_semana)
plt.bar(x_poisson_b, pmf_poisson_b)
plt.title('Distribución de Poisson - P(X=50)')
plt.xlabel('Número de cheques sin fondo')
plt.ylabel('Probabilidad')

plt.bar(k_b, prob_b, color='red')
plt.text(k_b, prob_b + 0.0001, f"P(X=50) = {prob_b:.5f}", fontsize=10, color='black')

plt.show()

# c. Diez cheques sin fondo en 12 horas
lambda_12horas = lambda_dia * 0.5  # Promedio de cheques sin fondo en 12 horas
k_c = 10
prob_c = poisson.pmf(k_c, lambda_12horas)
print(f"c. Probabilidad de 10 cheques sin fondo en 12 horas: {prob_c:.5f}")

# Gráfico para c.
x_poisson_c = np.arange(0, 20)
pmf_poisson_c = poisson.pmf(x_poisson_c, lambda_12horas)
plt.bar(x_poisson_c, pmf_poisson_c)
plt.title('Distribución de Poisson - P(X=10)')
plt.xlabel('Número de cheques sin fondo')
plt.ylabel('Probabilidad')

plt.bar(k_c, prob_c, color='red')
plt.text(k_c, prob_c + 0.01, f"P(X=10) = {prob_c:.5f}", fontsize=10, color='black')

plt.show()

# d. 5 cheques sin fondos en 4 horas
lambda_4horas = lambda_dia * (4/24)  # Promedio de cheques sin fondo en 4 horas
k_d = 5
prob_d = poisson.pmf(k_d, lambda_4horas)
print(f"d. Probabilidad de 5 cheques sin fondo en 4 horas: {prob_d:.5f}")

# Gráfico para d.
x_poisson_d = np.arange(0, 15)
pmf_poisson_d = poisson.pmf(x_poisson_d, lambda_4horas)
plt.bar(x_poisson_d, pmf_poisson_d)
plt.title('Distribución de Poisson - P(X=5)')
plt.xlabel('Número de cheques sin fondo')
plt.ylabel('Probabilidad')

plt.bar(k_d, prob_d, color='red')
plt.text(k_d, prob_d + 0.01, f"P(X=5) = {prob_d:.5f}", fontsize=10, color='black')

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 5. Distribución de Poisson en R**
#
# Si un banco recibe en promedio 6 cheques sin fondo por día,
# calcular las probabilidades de:
# a) 4 cheques en un día.
# b) 50 cheques en una semana.
# c) 10 cheques en 12 horas.
# d) 5 cheques en 4 horas.
# ==========================================================
%%R
library(ggplot2)

#==========================================================
# Función para graficar la distribución de Poisson
#==========================================================
graficar_poisson <- function(lambda, k, titulo, xmax){

  x <- 0:xmax
  prob <- dpois(x, lambda)

  datos <- data.frame(
    x = x,
    Probabilidad = prob,
    Color = ifelse(x == k, "Resaltar", "Normal")
  )

  p <- ggplot(datos,
              aes(x = factor(x),
                  y = Probabilidad,
                  fill = Color)) +

    geom_col(color="black") +

    scale_fill_manual(values=c("Normal"="steelblue",
                               "Resaltar"="red")) +

    geom_text(
      data=subset(datos, x==k),
      aes(label=sprintf("P(X=%d)\n%.5f",k,Probabilidad)),
      vjust=-0.5,
      size=4
    )+

    labs(title=titulo,
         x="Número de cheques sin fondo",
         y="Probabilidad")+

    theme_minimal(base_size=14)+
    theme(legend.position="none")

  print(p)
}

#==========================================================
# Parámetro base
#==========================================================
lambda_dia <- 6

cat("==============================================\n")
cat("DISTRIBUCIÓN DE POISSON\n")
cat("==============================================\n\n")

#==========================================================
# a. Cuatro cheques en un día
#==========================================================
k_a <- 4

prob_a <- dpois(k_a, lambda_dia)

cat("a) Probabilidad de recibir exactamente 4 cheques en un día\n")
cat(sprintf("P(X=4)= %.5f\n\n", prob_a))

graficar_poisson(
  lambda=lambda_dia,
  k=k_a,
  titulo="Poisson: P(X = 4)",
  xmax=15
)

#==========================================================
# b. 50 cheques en una semana
#==========================================================
lambda_semana <- lambda_dia*7

k_b <- 50

prob_b <- dpois(k_b, lambda_semana)

cat("b) Probabilidad de recibir exactamente 50 cheques en una semana\n")
cat(sprintf("P(X=50)= %.5f\n\n", prob_b))

graficar_poisson(
  lambda=lambda_semana,
  k=k_b,
  titulo="Poisson: P(X = 50)",
  xmax=70
)

#==========================================================
# c. 10 cheques en 12 horas
#==========================================================
lambda_12 <- lambda_dia*(12/24)

k_c <- 10

prob_c <- dpois(k_c, lambda_12)

cat("c) Probabilidad de recibir exactamente 10 cheques en 12 horas\n")
cat(sprintf("P(X=10)= %.5f\n\n", prob_c))

graficar_poisson(
  lambda=lambda_12,
  k=k_c,
  titulo="Poisson: P(X = 10)",
  xmax=20
)

#==========================================================
# d. Cinco cheques en 4 horas
#==========================================================
lambda_4 <- lambda_dia*(4/24)

k_d <- 5

prob_d <- dpois(k_d, lambda_4)

cat("d) Probabilidad de recibir exactamente 5 cheques en 4 horas\n")
cat(sprintf("P(X=5)= %.5f\n\n", prob_d))

graficar_poisson(
  lambda=lambda_4,
  k=k_d,
  titulo="Poisson: P(X = 5)",
  xmax=15
)
```

# **26. Ejemplo 6. Distribución de Poisson: En una empresa el término medio de accidentes es de 5 por mes. Calcular la probabilidad de que como mínimo ocurran 2 accidentes en un mes. Adaptalo a la solucion anterior**


```python
# @title **Ejemplo 6. En una empresa el término medio de accidentes es de 5 por mes. Calcular la probabilidad de que como mínimo ocurran 2 accidentes en un mes. Adaptalo a la solucion anterior**


lambda_poisson = 5  # Tasa promedio de accidentes por mes
k = 1  # Número de accidentes (queremos la probabilidad de al menos 2, es decir, P(X >= 2))

# La probabilidad de que ocurran como mínimo 2 accidentes es
# P(X >= 2) = 1 - P(X < 2) = 1 - [P(X = 0) + P(X = 1)]

prob_poisson_0 = poisson.pmf(0, lambda_poisson)  # P(X = 0)
prob_poisson_1 = poisson.pmf(1, lambda_poisson)  # P(X = 1)

prob_al_menos_2 = 1 - (prob_poisson_0 + prob_poisson_1)

print(f"La probabilidad de que ocurran como mínimo 2 accidentes en un mes es: {prob_al_menos_2:.5f}")


# Hacemos el gráfico
x_poisson = np.arange(0, 15)
pmf_poisson = poisson.pmf(x_poisson, lambda_poisson)

plt.bar(x_poisson, pmf_poisson, color='blue')
for i in range(2, len(x_poisson)):
    plt.bar(x_poisson[i], pmf_poisson[i], color='red')

plt.title("Distribución de Poisson: Número de accidentes en un mes")
plt.xlabel("Número de accidentes")
plt.ylabel("Probabilidad")
plt.text(8, 0.08, f"P(X≥2) = {prob_al_menos_2:.5f}", fontsize=12, color='green')  # Mostrar el cálculo dentro del gráfico
plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 6. Distribución de Poisson en R**
#
# En una empresa el término medio de accidentes es de
# 5 por mes. Calcular la probabilidad de que como
# mínimo ocurran 2 accidentes en un mes.
# ==========================================================
%%R
library(ggplot2)

#==========================================================
# Función para graficar P(X >= k)
#==========================================================
graficar_poisson_acumulada <- function(lambda, k, titulo, xmax){

  x <- 0:xmax
  prob <- dpois(x, lambda)

  datos <- data.frame(
    x = x,
    Probabilidad = prob,
    Color = ifelse(x >= k, "Resaltar", "Normal")
  )

  prob_acumulada <- 1 - ppois(k-1, lambda)

  p <- ggplot(datos,
              aes(x = factor(x),
                  y = Probabilidad,
                  fill = Color)) +

    geom_col(color="black") +

    scale_fill_manual(values=c("Normal"="steelblue",
                               "Resaltar"="red")) +

    annotate("text",
             x = xmax-3,
             y = max(prob)*0.90,
             label = paste0("P(X ≥ ",k,") = ",
                            sprintf("%.5f",prob_acumulada)),
             size = 5,
             color = "darkgreen",
             fontface="bold") +

    labs(title=titulo,
         x="Número de accidentes",
         y="Probabilidad") +

    theme_minimal(base_size = 14) +
    theme(legend.position="none")

  print(p)
}

#==========================================================
# Parámetros
#==========================================================
lambda_poisson <- 5
k <- 2

#==========================================================
# Probabilidad
#==========================================================
prob_al_menos_2 <- 1 - ppois(k-1, lambda_poisson)

cat("=============================================\n")
cat("MODELO DE POISSON\n")
cat("=============================================\n\n")

cat("Media de accidentes por mes =", lambda_poisson,"\n\n")

cat(sprintf("P(X ≥ %d) = %.5f\n",
            k,
            prob_al_menos_2))

#==========================================================
# Gráfica
#==========================================================
graficar_poisson_acumulada(
  lambda = lambda_poisson,
  k = k,
  titulo = "Distribución de Poisson: P(X ≥ 2)",
  xmax = 15
)
```

# **27.CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN DE POISSON**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN DE POISSON**
#
# La distribución de Poisson modela el número de eventos que
# ocurren en un intervalo de tiempo o espacio fijo.
#
# Fórmula: P(X = k) = (e^(-λ) * λ^k) / k!
# donde:
#   λ = tasa promedio de eventos por intervalo
#   k = número de eventos
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from scipy.special import factorial
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_poisson():
    """
    Calculadora interactiva para la distribución de Poisson
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN DE POISSON")
    print("="*70)
    print("\nLa distribución de Poisson modela el número de eventos")
    print("que ocurren en un intervalo fijo de tiempo o espacio.")
    print("\nEjemplos típicos:")
    print("  • Número de llamadas por hora en un call center")
    print("  • Número de accidentes por mes en una carretera")
    print("  • Número de clientes que llegan a una tienda por minuto")
    print("  • Número de defectos por metro cuadrado en una tela\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            lam = float(input("  Tasa promedio de eventos (λ > 0): "))
            if lam <= 0:
                print("  ❌ λ debe ser mayor que 0. Intenta de nuevo.")
                continue

            k = int(input("  Número de eventos a evaluar (k ≥ 0): "))
            if k < 0:
                print("  ❌ k debe ser mayor o igual a 0. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

    #==========================================================
    # 2. CÁLCULO DE PROBABILIDADES
    #==========================================================

    # Rango de valores posibles para k
    max_k = max(int(lam * 3) + 10, k + 10)
    x = np.arange(0, max_k + 1)
    pmf = poisson.pmf(x, lam)

    # Probabilidades solicitadas
    prob_exacta = poisson.pmf(k, lam)
    prob_acumulada_menor = poisson.cdf(k, lam)  # P(X ≤ k)
    prob_acumulada_mayor = 1 - poisson.cdf(k-1, lam) if k > 0 else 1  # P(X ≥ k)

    # Estadísticas
    media = lam
    varianza = lam
    desviacion = np.sqrt(lam)
    moda = int(np.floor(lam))
    asimetria = 1 / np.sqrt(lam)
    curtosis = 1 / lam

    #==========================================================
    # 3. MOSTRAR RESULTADOS NUMÉRICOS
    #==========================================================

    print("\n" + "="*70)
    print("📈 RESULTADOS DEL ANÁLISIS")
    print("="*70)
    print(f"\nParámetros de la distribución de Poisson:")
    print(f"  • λ = {lam:.4f} (tasa promedio de eventos)")
    print(f"  • k = {k} (número de eventos evaluado)")

    print(f"\n📊 Probabilidades calculadas:")
    print("-"*70)
    print(f"  • P(X = {k}) = {prob_exacta:.6f} ({prob_exacta*100:.4f}%)")
    print(f"  • P(X ≤ {k}) = {prob_acumulada_menor:.6f} ({prob_acumulada_menor*100:.4f}%)")
    print(f"  • P(X ≥ {k}) = {prob_acumulada_mayor:.6f} ({prob_acumulada_mayor*100:.4f}%)")
    print(f"  • P(X < {k}) = {poisson.cdf(k-1, lam):.6f} ({poisson.cdf(k-1, lam)*100:.4f}%)" if k > 0 else "  • P(X < 0) = 0")
    print(f"  • P(X > {k}) = {1 - poisson.cdf(k, lam):.6f} ({ (1 - poisson.cdf(k, lam))*100:.4f}%)")

    print(f"\n📐 Estadísticas descriptivas:")
    print("-"*70)
    print(f"  • Media (μ) = {media:.4f}")
    print(f"  • Varianza (σ²) = {varianza:.4f}")
    print(f"  • Desviación estándar (σ) = {desviacion:.4f}")
    print(f"  • Moda = {moda}")
    print(f"  • Coeficiente de asimetría = {asimetria:.4f}")
    print(f"  • Coeficiente de curtosis = {curtosis:.4f}")
    print(f"  • Coeficiente de variación = {desviacion/media:.4f}")

    #==========================================================
    # 4. CREAR GRÁFICO PRINCIPAL CON ÁREAS SOMBREADAS
    #==========================================================

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    # Gráfico 1: Distribución completa
    ax1 = plt.subplot(2, 2, 1)

    # Barras de la distribución
    colors = ['#2E86AB' if i != k else '#E63946' for i in range(max_k + 1)]
    bars = ax1.bar(x, pmf, color=colors, alpha=0.8, edgecolor='black', linewidth=1)

    # Resaltar barra de k
    ax1.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black', linewidth=2)

    # Líneas de referencia
    ax1.axvline(x=media, color='#2C3E50', linestyle='--', linewidth=2, alpha=0.7, label=f'μ = {media:.2f}')
    ax1.axvline(x=media + desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)
    ax1.axvline(x=media - desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)

    # Línea de la moda
    ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=2, alpha=0.6, label=f'Moda = {moda}')

    ax1.set_xlabel('Número de eventos (X)', fontsize=12)
    ax1.set_ylabel('Probabilidad', fontsize=12)
    ax1.set_title(f'Distribución de Poisson\nλ = {lam:.3f}', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))
    ax1.legend(loc='best', fontsize=10)

    # Agregar etiquetas de valores en barras seleccionadas
    for i, bar in enumerate(bars):
        if i % max(1, max_k//20) == 0 and pmf[i] > 0.001:
            height = bar.get_height()
            ax1.text(bar.get_x() + bar.get_width()/2., height + 0.002,
                    f'{pmf[i]:.3f}', ha='center', va='bottom', fontsize=8)

    # Gráfico 2: P(X ≤ k) - Área izquierda
    ax2 = plt.subplot(2, 2, 2)

    # Barras de la distribución
    colors2 = ['#FF6B6B' if i <= k else '#2E86AB' for i in range(max_k + 1)]
    bars2 = ax2.bar(x, pmf, color=colors2, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax2.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax2.set_xlabel('Número de eventos (X)', fontsize=12)
    ax2.set_ylabel('Probabilidad', fontsize=12)
    ax2.set_title(f'P(X ≤ {k}) = {prob_acumulada_menor:.4f}\n({prob_acumulada_menor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Texto adicional
    ax2.text(0.02, 0.98, f'Área sombreada:\nP(X ≤ {k}) = {prob_acumulada_menor:.6f}',
             transform=ax2.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 3: P(X ≥ k) - Área derecha
    ax3 = plt.subplot(2, 2, 3)

    # Barras de la distribución
    colors3 = ['#FF6B6B' if i >= k else '#2E86AB' for i in range(max_k + 1)]
    bars3 = ax3.bar(x, pmf, color=colors3, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax3.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax3.set_xlabel('Número de eventos (X)', fontsize=12)
    ax3.set_ylabel('Probabilidad', fontsize=12)
    ax3.set_title(f'P(X ≥ {k}) = {prob_acumulada_mayor:.4f}\n({prob_acumulada_mayor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Texto adicional
    ax3.text(0.02, 0.98, f'Área sombreada:\nP(X ≥ {k}) = {prob_acumulada_mayor:.6f}',
             transform=ax3.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 4: Resumen de resultados y fórmula
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Mostrar la fórmula
    formula_text = (
        f"Fórmula de Poisson:\n\n"
        f"P(X = k) = (e^(-λ) · λ^k) / k!\n\n"
        f"P(X = {k}) = (e^(-{lam:.2f}) · {lam:.2f}^{k}) / {k}!\n\n"
        f"P(X = {k}) = {prob_exacta:.6f}\n"
        f"({prob_exacta*100:.4f}%)\n\n"
        f"λ = {lam:.2f} eventos por intervalo"
    )

    ax4.text(0.5, 0.75, formula_text, transform=ax4.transAxes,
             fontsize=11, verticalalignment='center', horizontalalignment='center',
             bbox=dict(boxstyle='round,pad=0.5', facecolor='#F0F8FF', edgecolor='#2C3E50'),
             family='monospace')

    # Tabla de resultados
    resultados = [
        ['Concepto', 'Valor', 'Porcentaje'],
        ['P(X = k)', f'{prob_exacta:.6f}', f'{prob_exacta*100:.4f}%'],
        ['P(X ≤ k)', f'{prob_acumulada_menor:.6f}', f'{prob_acumulada_menor*100:.4f}%'],
        ['P(X ≥ k)', f'{prob_acumulada_mayor:.6f}', f'{prob_acumulada_mayor*100:.4f}%'],
        ['Media (μ)', f'{media:.4f}', ''],
        ['Varianza (σ²)', f'{varianza:.4f}', ''],
        ['Desv. Estándar (σ)', f'{desviacion:.4f}', ''],
        ['Moda', f'{moda}', '']
    ]

    # Crear tabla en la parte inferior
    table = ax4.table(cellText=resultados, loc='bottom', cellLoc='center',
                      colWidths=[0.3, 0.35, 0.35], bbox=[0, 0, 1, 0.4])

    # Estilo de la tabla
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.3)

    # Colores de la tabla
    for i in range(len(resultados)):
        for j in range(len(resultados[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            elif i in [1, 2, 3]:
                cell.set_facecolor('#E8F4F8')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('#F5F5F5')

    # Ajustar layout
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # 5. GRÁFICO ADICIONAL: FUNCIÓN DE DISTRIBUCIÓN ACUMULADA
    #==========================================================

    # Crear una figura separada para la CDF
    fig2, ax5 = plt.subplots(figsize=(10, 6))

    # Calcular CDF
    cdf = poisson.cdf(x, lam)

    # Graficar CDF
    ax5.step(x, cdf, where='mid', linewidth=2.5, color='#2E86AB', label='CDF')
    ax5.fill_between(x, 0, cdf, step='mid', alpha=0.3, color='#2E86AB')

    # Punto en k
    ax5.scatter(k, cdf[k], color='#E63946', s=150, zorder=5,
               label=f'F({k}) = {cdf[k]:.4f}')

    # Líneas de referencia
    ax5.axhline(y=cdf[k], color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.axvline(x=k, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles
    cuantiles = [0.25, 0.5, 0.75]
    colores_cuantiles = ['#F39C12', '#2ECC71', '#3498DB']
    for i, q in enumerate(cuantiles):
        try:
            q_val = poisson.ppf(q, lam)
            ax5.axvline(x=q_val, color=colores_cuantiles[i], linestyle=':',
                       linewidth=1.5, alpha=0.5)
            ax5.text(q_val, -0.03, f'Q{int(q*100)}={q_val}',
                    ha='center', fontsize=9, color=colores_cuantiles[i])
        except:
            pass

    ax5.set_xlabel('Número de eventos (X)', fontsize=12)
    ax5.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax5.set_title(f'Función de Distribución Acumulada\nλ = {lam:.3f}',
                  fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))
    ax5.set_ylim(-0.05, 1.05)
    ax5.legend(loc='best')

    # Texto con la probabilidad
    ax5.text(0.02, 0.98,
             f'P(X ≤ {k}) = {cdf[k]:.6f}\nP(X > {k}) = {1-cdf[k]:.6f}\n'
             f'Media = {lam:.2f} (λ)',
             transform=ax5.transAxes, fontsize=11, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()

    #==========================================================
    # 6. GRÁFICO DE PROBABILIDADES ACUMULADAS INVERSAS
    #==========================================================

    # Gráfico que muestra la probabilidad de exactamente k eventos
    fig3, ax6 = plt.subplots(figsize=(10, 6))

    # Crear un rango extendido para mostrar el comportamiento asintótico
    x_ext = np.arange(0, max_k + 20)
    pmf_ext = poisson.pmf(x_ext, lam)
    cdf_ext = poisson.cdf(x_ext, lam)

    # Probabilidad de exactamente k eventos
    ax6.bar(x, pmf, color='#2E86AB', alpha=0.6, edgecolor='black', label='PMF')
    ax6.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black',
           linewidth=2, label=f'P(X={k})')

    # Curva de probabilidad acumulada (eje derecho)
    ax6_twin = ax6.twinx()
    ax6_twin.plot(x, cdf, 'r-', linewidth=2, label='CDF')
    ax6_twin.fill_between(x, 0, cdf, alpha=0.1, color='red')

    ax6.set_xlabel('Número de eventos (X)', fontsize=12)
    ax6.set_ylabel('Probabilidad P(X=k)', fontsize=12)
    ax6_twin.set_ylabel('Probabilidad acumulada F(X)', fontsize=12, color='red')
    ax6_twin.tick_params(axis='y', labelcolor='red')

    ax6.set_title(f'Distribución de Poisson: PMF y CDF\nλ = {lam:.3f}',
                  fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3, linestyle='--')
    ax6.set_xticks(np.arange(0, max_k+1, max(1, max_k//15)))

    # Línea vertical en la media
    ax6.axvline(x=media, color='green', linestyle='--', linewidth=2, alpha=0.7, label=f'μ={media:.2f}')

    # Leyendas combinadas
    lines1, labels1 = ax6.get_legend_handles_labels()
    lines2, labels2 = ax6_twin.get_legend_handles_labels()
    ax6.legend(lines1 + lines2, labels1 + labels2, loc='best', fontsize=10)

    plt.tight_layout()

    #==========================================================
    # 7. GRÁFICO PARA DIFERENTES VALORES DE λ
    #==========================================================

    # Mostrar cómo cambia la distribución con diferentes λ
    fig4, ax7 = plt.subplots(figsize=(12, 6))

    lambdas = [1, 3, 5, 10, 15]
    colores = ['#2E86AB', '#E63946', '#F39C12', '#2ECC71', '#9B59B6']

    max_x = max([int(l * 3) + 5 for l in lambdas])
    x_comp = np.arange(0, max_x + 1)

    for i, lam_comp in enumerate(lambdas):
        pmf_comp = poisson.pmf(x_comp, lam_comp)
        ax7.plot(x_comp, pmf_comp, 'o-', color=colores[i],
                linewidth=2, markersize=6, label=f'λ = {lam_comp}')

    ax7.set_xlabel('Número de eventos (X)', fontsize=12)
    ax7.set_ylabel('Probabilidad', fontsize=12)
    ax7.set_title('Comparación de distribuciones de Poisson con diferentes λ',
                  fontsize=13, fontweight='bold')
    ax7.grid(True, alpha=0.3, linestyle='--')
    ax7.legend(loc='best')
    ax7.set_xticks(np.arange(0, max_x+1, 2))

    plt.tight_layout()

    #==========================================================
    # 8. MOSTRAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # 9. OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Poisson_lam{lam:.2f}_k{k}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_cdf.png", dpi=300, bbox_inches='tight')
        fig3.savefig(f"{nombre_base}_pmf_cdf.png", dpi=300, bbox_inches='tight')
        fig4.savefig(f"{nombre_base}_comparacion_lambdas.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_cdf.png")
        print(f"   • {nombre_base}_pmf_cdf.png")
        print(f"   • {nombre_base}_comparacion_lambdas.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

    return {
        'lam': lam,
        'k': k,
        'prob_exacta': prob_exacta,
        'prob_menor': prob_acumulada_menor,
        'prob_mayor': prob_acumulada_mayor,
        'media': media,
        'varianza': varianza,
        'desviacion': desviacion,
        'moda': moda
    }

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_poisson():
    """
    Ejecuta ejemplos clásicos de la distribución de Poisson
    """

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN DE POISSON")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Llamadas telefónicas',
            'descripcion': 'Un call center recibe en promedio 5 llamadas por hora. ¿Cuál es la probabilidad de recibir exactamente 3 llamadas en una hora?',
            'lam': 5,
            'k': 3
        },
        {
            'titulo': 'Ejemplo 2: Accidentes de tráfico',
            'descripcion': 'En una carretera ocurren en promedio 2 accidentes por día. ¿Cuál es la probabilidad de que ocurran al menos 4 accidentes en un día?',
            'lam': 2,
            'k': 4
        },
        {
            'titulo': 'Ejemplo 3: Clientes en una tienda',
            'descripcion': 'Una tienda recibe en promedio 10 clientes por hora. ¿Cuál es la probabilidad de recibir exactamente 8 clientes en una hora?',
            'lam': 10,
            'k': 8
        },
        {
            'titulo': 'Ejemplo 4: Defectos en producción',
            'descripcion': 'Una fábrica produce en promedio 3.5 defectos por cada 100 metros de tela. ¿Cuál es la probabilidad de encontrar exactamente 2 defectos en 100 metros?',
            'lam': 3.5,
            'k': 2
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")
        print(f"   λ = {ej['lam']}, k = {ej['k']}")

        prob = poisson.pmf(ej['k'], ej['lam'])
        prob_acum = poisson.cdf(ej['k'], ej['lam'])
        prob_mayor = 1 - poisson.cdf(ej['k'] - 1, ej['lam']) if ej['k'] > 0 else 1

        print(f"   P(X = {ej['k']}) = {prob:.6f} ({prob*100:.2f}%)")
        print(f"   P(X ≤ {ej['k']}) = {prob_acum:.6f} ({prob_acum*100:.2f}%)")
        print(f"   P(X ≥ {ej['k']}) = {prob_mayor:.6f} ({prob_mayor*100:.2f}%)")
        print(f"   Media = {ej['lam']} eventos por intervalo")
        print(f"   Desviación estándar = {np.sqrt(ej['lam']):.4f}")

    print("\n" + "="*70)
    print("¿Deseas analizar alguno de estos ejemplos en detalle?")
    print("Selecciona 1, 2, 3, 4 o ejecuta la calculadora principal.")

#==========================================================
# FUNCIÓN PARA VALIDAR SUPOSICIONES DE POISSON
#==========================================================

def validar_suposiciones():
    """
    Valida las suposiciones para usar la distribución de Poisson
    """

    print("\n" + "="*70)
    print("🔍 VALIDACIÓN DE SUPOSICIONES PARA POISSON")
    print("="*70)

    print("\nLa distribución de Poisson requiere que se cumplan las siguientes condiciones:")
    print("\n1. Los eventos ocurren de forma independiente")
    print("   ✓ La ocurrencia de un evento no afecta la probabilidad de otro")
    print("\n2. La tasa de ocurrencia es constante")
    print("   ✓ λ es constante en el intervalo de tiempo/espacio considerado")
    print("\n3. Los eventos ocurren de forma aleatoria")
    print("   ✓ No hay patrón predecible en las ocurrencias")
    print("\n4. La probabilidad de múltiples eventos en un intervalo pequeño es insignificante")
    print("   ✓ Los eventos son raros en intervalos muy pequeños")

    print("\n" + "="*70)
    print("📊 PRUEBA DE BONDAD DE AJUSTE")
    print("="*70)

    print("\nSi tienes datos observados, puedes comparar con la distribución")
    print("de Poisson usando una prueba de chi-cuadrado:")
    print("   χ² = Σ (O_i - E_i)² / E_i")
    print("\nDonde O_i son las frecuencias observadas y E_i las esperadas.")

#==========================================================
# FUNCIÓN PARA CALCULAR INTERVALOS DE CONFIANZA
#==========================================================

def intervalos_confianza_poisson():
    """
    Calcula intervalos de confianza para la tasa de Poisson λ
    """

    print("\n" + "="*70)
    print("📈 INTERVALOS DE CONFIANZA PARA λ")
    print("="*70)

    try:
        print("\nIngresa los datos observados:")
        n_obs = int(input("  Número de eventos observados: "))
        tiempo = float(input("  Tiempo de observación (en unidades del intervalo): "))

        lam_estimado = n_obs / tiempo

        print(f"\nλ estimado = {lam_estimado:.4f} eventos por unidad de tiempo")

        # Intervalos de confianza aproximados (basados en distribución normal)
        niveles = [0.90, 0.95, 0.99]
        z_values = [1.645, 1.96, 2.576]

        print("\nIntervalos de confianza para λ:")
        print("-"*70)
        for nivel, z in zip(niveles, z_values):
            error = z * np.sqrt(lam_estimado / tiempo)
            li = lam_estimado - error
            ls = lam_estimado + error
            print(f"  {nivel*100:.0f}%: ({li:.4f}, {ls:.4f})")

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN DE POISSON")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado de una distribución")
        print("  2. Ver ejemplos clásicos")
        print("  3. Validar suposiciones de Poisson")
        print("  4. Calcular intervalos de confianza para λ")
        print("  5. Salir")

        opcion = input("\nSelecciona una opción (1-5): ")

        if opcion == '1':
            resultados = calculadora_poisson()
        elif opcion == '2':
            ejemplos_poisson()
        elif opcion == '3':
            validar_suposiciones()
        elif opcion == '4':
            intervalos_confianza_poisson()
        elif opcion == '5':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3, 4 o 5.")
```

# **28. Distribución Hipergeométrica**

+ *La distribución hipergeométrica modela el número de éxitos en una muestra de tamaño \(n\) tomada sin reemplazo de una población de tamaño \(N\), que contiene \(K\) éxitos y \(N - K\) fracasos.*

- **Función de probabilidad**: $$P(X = k) = \frac{\binom{K}{k} \binom{N-K}{n-k}}{\binom{N}{n}}$$
- **Parámetros**: \(N\) es el tamaño de la población, \(K\) es el número de éxitos en la población, \(n\) es el tamaño de la muestra, y \(k\) es el número de éxitos en la muestra.

### **Distribución Hipergeométrica - Momentos**
- **Valor esperado (Esperanza)**: $$E(X) = n \cdot \frac{K}{N}$$
- **Varianza**: $$\text{Var}(X) = n \cdot \frac{K}{N} \cdot \frac{N - K}{N} \cdot \frac{N - n}{N - 1}$$
- **Desviación estándar**: $$\sigma(X) = \sqrt{\text{Var}(X)}$$




```python

# @title **A. Gráfica de la Distribución Hipergeométrica** $$P(X = k) = \frac{\binom{K}{k} \binom{N-K}{n-k}}{\binom{N}{n}}$$
N, K, n = 30, 10, 5
x_hyper = np.arange(0, n + 1)
pmf_hyper = hypergeom.pmf(x_hyper, N, K, n)
plot_distribution2(hypergeom, x_hyper, pmf_hyper, "Distribución Hipergeométrica")
```

```python
# ==========================================================
# @title **B. Gráfica de la Distribución Hipergeométrica**
# $$P(X = k)=\frac{\binom{K}{k}\binom{N-K}{n-k}}
# {\binom{N}{n}}$$
# ==========================================================
%%R
# Parámetros
N <- 30   # Tamaño de la población
K <- 10   # Número de elementos exitosos en la población
n <- 5    # Tamaño de la muestra

# Valores posibles de la variable aleatoria
x_hyper <- 0:n

# Función de probabilidad
pmf_hyper <- dhyper(
  x = x_hyper,
  m = K,
  n = N - K,
  k = n
)

# Gráfica
barplot(
  pmf_hyper,
  names.arg = x_hyper,
  col = "steelblue",
  border = "black",
  main = "Distribución Hipergeométrica",
  xlab = "Número de éxitos (X)",
  ylab = "Probabilidad"
)

# Agregar los valores sobre cada barra
text(
  x = seq_along(x_hyper),
  y = pmf_hyper,
  labels = round(pmf_hyper, 4),
  pos = 3,
  cex = 0.8,
  col = "blue"
)
```

# **29. Distribución Hipergeométrica: Ejemplo 1:-Una urna contiene 10 bolas rojas y 20 bolas azules. Se seleccionan 5 bolas sin reemplazo. ¿Cuál es la probabilidad de seleccionar exactamente 3 bolas rojas?**

```python
# @title **A. Ejemplo 1. Una urna contiene 10 bolas rojas y 20 bolas azules. Se seleccionan 5 bolas sin reemplazo. ¿Cuál es la probabilidad de seleccionar exactamente 3 bolas rojas? - Distribución Hipergeométrica**
# $$P(X = k) = \frac{\binom{K}{k}\binom{N-K}{n-k}}{\binom{N}{n}}$$

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import hypergeom

# Parámetros
N = 30   # Tamaño de la población
K = 10   # Número de bolas rojas
n = 5    # Tamaño de la muestra
k = 3    # Número de bolas rojas deseadas

# Probabilidad exacta
prob_hyper_1 = hypergeom.pmf(k, N, K, n)

print(f"Probabilidad de seleccionar exactamente 3 bolas rojas:")
print(f"P(X = {k}) = {prob_hyper_1:.5f}")

# Valores para la distribución
x_hyper = np.arange(0, n + 1)
pmf_hyper = hypergeom.pmf(x_hyper, N, K, n)

# Gráfico
plt.figure(figsize=(8,5))

# Todas las barras
plt.bar(
    x_hyper,
    pmf_hyper,
    color="skyblue",
    edgecolor="black"
)

# Resaltar P(X=3)
plt.bar(
    k,
    prob_hyper_1,
    color="red",
    edgecolor="black"
)

# Texto
plt.text(
    k + 0.1,
    prob_hyper_1 + 0.01,
    f"P(X={k}) = {prob_hyper_1:.5f}",
    fontsize=11,
    color="black"
)

plt.title("Distribución Hipergeométrica")
plt.xlabel("Número de bolas rojas seleccionadas")
plt.ylabel("Probabilidad")
plt.xticks(x_hyper)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 1. Una urna contiene 10 bolas rojas y 20 bolas azules. Se seleccionan 5 bolas sin reemplazo. ¿Cuál es la probabilidad de seleccionar exactamente 3 bolas rojas?**
# Distribución Hipergeométrica
#
# $$P(X=k)=\frac{\binom{K}{k}\binom{N-K}{n-k}}{\binom{N}{n}}$$
# ==========================================================
%%R
library(ggplot2)

# Parámetros
N <- 30   # Total de bolas
K <- 10   # Bolas rojas
n <- 5    # Tamaño de la muestra
k <- 3    # Número de bolas rojas deseadas

# Probabilidad exacta
prob_hiper <- dhyper(k,
                     m = K,
                     n = N - K,
                     k = n)

cat("===========================================\n")
cat("DISTRIBUCIÓN HIPERGEOMÉTRICA\n")
cat("===========================================\n")
cat(sprintf("P(X = %d) = %.5f\n", k, prob_hiper))
cat("===========================================\n")

# Distribución completa
x <- 0:n
pmf <- dhyper(x,
              m = K,
              n = N - K,
              k = n)

datos <- data.frame(
  x = x,
  Probabilidad = pmf,
  Resaltar = ifelse(x == k, "Sí", "No")
)

# Gráfico
ggplot(datos,
       aes(x = factor(x),
           y = Probabilidad,
           fill = Resaltar)) +

  geom_col(color = "black", width = 0.75) +

  scale_fill_manual(values = c("No" = "skyblue",
                               "Sí" = "red")) +

  geom_text(
    data = subset(datos, x == k),
    aes(label = sprintf("P(X=%d)\n%.5f", k, Probabilidad)),
    vjust = -0.5,
    size = 4
  ) +

  labs(
    title = "Distribución Hipergeométrica",
    subtitle = "Probabilidad de seleccionar exactamente 3 bolas rojas",
    x = "Número de bolas rojas seleccionadas",
    y = "Probabilidad"
  ) +

  theme_minimal(base_size = 13) +

  theme(
    legend.position = "none",
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5)
  )
```

# **30. Distribución Hipergeométrica: Ejemplo 2. Momentos en el modelo Hipergeométrico**

```python
# @title  **Momentos en el modelo Hipergeométrico**
N = 30  # Tamaño de la población
K = 10  # Éxitos en la población
n = 5   # Tamaño de la muestra

esperanza_hyper = n * (K / N)
varianza_hyper = n * (K / N) * ((N - K) / N) * ((N - n) / (N - 1))
desviacion_hyper = np.sqrt(varianza_hyper)

print(f"Momentos en el Modelo Hipergeométrico")
print(f"Esperanza:E(X) =  {esperanza_hyper}")
print(f"Varianza: V(X) =  {varianza_hyper}")
print(f"Desviación estándar: {desviacion_hyper}")
```

# **31. CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN HIPERGEOMÉTRICA**


```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN HIPERGEOMÉTRICA**
#
# La distribución hipergeométrica modela el número de éxitos
# en una muestra sin reemplazo de una población finita.
#
# Fórmula: P(X = k) = C(K, k) * C(N-K, n-k) / C(N, n)
# donde:
#   N = tamaño de la población total
#   K = número de éxitos en la población
#   n = tamaño de la muestra
#   k = número de éxitos en la muestra
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import hypergeom
from scipy.special import comb
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_hipergeometrica():
    """
    Calculadora interactiva para la distribución hipergeométrica
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN HIPERGEOMÉTRICA")
    print("="*70)
    print("\nLa distribución hipergeométrica modela el número de éxitos")
    print("en una muestra sin reemplazo de una población finita.")
    print("\nEjemplos típicos:")
    print("  • Número de piezas defectuosas en una muestra de producción")
    print("  • Número de personas con cierta característica en una encuesta")
    print("  • Número de bolas rojas al extraer sin reemplazo de una urna")
    print("  • Número de clientes satisfechos en una muestra de ventas\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            N = int(input("  Tamaño de la población (N > 0): "))
            if N <= 0:
                print("  ❌ N debe ser mayor que 0. Intenta de nuevo.")
                continue

            K = int(input(f"  Número de éxitos en la población (0 ≤ K ≤ {N}): "))
            if K < 0 or K > N:
                print(f"  ❌ K debe estar entre 0 y {N}. Intenta de nuevo.")
                continue

            n = int(input(f"  Tamaño de la muestra (0 ≤ n ≤ {N}): "))
            if n < 0 or n > N:
                print(f"  ❌ n debe estar entre 0 y {N}. Intenta de nuevo.")
                continue

            k = int(input(f"  Número de éxitos en la muestra (0 ≤ k ≤ min({K}, {n})): "))
            if k < 0 or k > min(K, n):
                print(f"  ❌ k debe estar entre 0 y {min(K, n)}. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

    #==========================================================
    # 2. CÁLCULO DE PROBABILIDADES
    #==========================================================

    # Rango de valores posibles para k
    min_k = max(0, n - (N - K))
    max_k = min(n, K)
    x = np.arange(min_k, max_k + 1)
    pmf = hypergeom.pmf(x, N, K, n)

    # Probabilidades solicitadas
    prob_exacta = hypergeom.pmf(k, N, K, n)
    prob_acumulada_menor = hypergeom.cdf(k, N, K, n)  # P(X ≤ k)
    prob_acumulada_mayor = 1 - hypergeom.cdf(k-1, N, K, n) if k > min_k else 1  # P(X ≥ k)

    # Estadísticas
    media = n * K / N
    varianza = n * (K/N) * (1 - K/N) * (N - n) / (N - 1)
    desviacion = np.sqrt(varianza)

    # Calcular moda (aproximada)
    moda_valores = np.arange(min_k, max_k + 1)
    moda_idx = np.argmax(pmf)
    moda = moda_valores[moda_idx]

    # Factor de corrección por población finita
    fpc = np.sqrt((N - n) / (N - 1))

    #==========================================================
    # 3. MOSTRAR RESULTADOS NUMÉRICOS
    #==========================================================

    print("\n" + "="*70)
    print("📈 RESULTADOS DEL ANÁLISIS")
    print("="*70)
    print(f"\nParámetros de la distribución hipergeométrica:")
    print(f"  • N = {N} (tamaño de la población)")
    print(f"  • K = {K} (número de éxitos en la población)")
    print(f"  • n = {n} (tamaño de la muestra)")
    print(f"  • k = {k} (número de éxitos en la muestra)")
    print(f"  • Proporción de éxitos en población = {K/N:.4f}")

    print(f"\n📊 Probabilidades calculadas:")
    print("-"*70)
    print(f"  • P(X = {k}) = {prob_exacta:.6f} ({prob_exacta*100:.4f}%)")
    print(f"  • P(X ≤ {k}) = {prob_acumulada_menor:.6f} ({prob_acumulada_menor*100:.4f}%)")
    print(f"  • P(X ≥ {k}) = {prob_acumulada_mayor:.6f} ({prob_acumulada_mayor*100:.4f}%)")
    print(f"  • P(X < {k}) = {hypergeom.cdf(k-1, N, K, n):.6f} ({hypergeom.cdf(k-1, N, K, n)*100:.4f}%)" if k > min_k else f"  • P(X < {k}) = 0")
    print(f"  • P(X > {k}) = {1 - hypergeom.cdf(k, N, K, n):.6f} ({ (1 - hypergeom.cdf(k, N, K, n))*100:.4f}%)")

    print(f"\n📐 Estadísticas descriptivas:")
    print("-"*70)
    print(f"  • Media (μ) = {media:.4f}")
    print(f"  • Varianza (σ²) = {varianza:.4f}")
    print(f"  • Desviación estándar (σ) = {desviacion:.4f}")
    print(f"  • Moda = {moda}")
    print(f"  • Factor de corrección por población finita = {fpc:.4f}")
    print(f"  • Proporción de muestra = {n/N:.4f}")

    #==========================================================
    # 4. CREAR GRÁFICO PRINCIPAL CON ÁREAS SOMBREADAS
    #==========================================================

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    # Gráfico 1: Distribución completa
    ax1 = plt.subplot(2, 2, 1)

    # Barras de la distribución
    colors = ['#2E86AB' if i != k else '#E63946' for i in range(min_k, max_k + 1)]
    bars = ax1.bar(x, pmf, color=colors, alpha=0.8, edgecolor='black', linewidth=1)

    # Resaltar barra de k
    ax1.bar(k, prob_exacta, color='#E63946', alpha=0.9, edgecolor='black', linewidth=2)

    # Líneas de referencia
    ax1.axvline(x=media, color='#2C3E50', linestyle='--', linewidth=2, alpha=0.7, label=f'μ = {media:.2f}')
    ax1.axvline(x=media + desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)
    ax1.axvline(x=media - desviacion, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)

    # Línea de la moda
    ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=2, alpha=0.6, label=f'Moda = {moda}')

    ax1.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax1.set_ylabel('Probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Hipergeométrica\nN={N}, K={K}, n={n}', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.set_xticks(np.arange(min_k, max_k+1, max(1, (max_k-min_k)//15)))
    ax1.legend(loc='best', fontsize=10)

    # Agregar etiquetas de valores
    for i, bar in enumerate(bars):
        if pmf[i] > 0.01:
            height = bar.get_height()
            ax1.text(bar.get_x() + bar.get_width()/2., height + 0.002,
                    f'{pmf[i]:.3f}', ha='center', va='bottom', fontsize=8)

    # Gráfico 2: P(X ≤ k) - Área izquierda
    ax2 = plt.subplot(2, 2, 2)

    # Barras de la distribución
    colors2 = ['#FF6B6B' if i <= k else '#2E86AB' for i in range(min_k, max_k + 1)]
    bars2 = ax2.bar(x, pmf, color=colors2, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax2.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax2.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax2.set_ylabel('Probabilidad', fontsize=12)
    ax2.set_title(f'P(X ≤ {k}) = {prob_acumulada_menor:.4f}\n({prob_acumulada_menor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.set_xticks(np.arange(min_k, max_k+1, max(1, (max_k-min_k)//15)))

    # Texto adicional
    ax2.text(0.02, 0.98, f'Área sombreada:\nP(X ≤ {k}) = {prob_acumulada_menor:.6f}',
             transform=ax2.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 3: P(X ≥ k) - Área derecha
    ax3 = plt.subplot(2, 2, 3)

    # Barras de la distribución
    colors3 = ['#FF6B6B' if i >= k else '#2E86AB' for i in range(min_k, max_k + 1)]
    bars3 = ax3.bar(x, pmf, color=colors3, alpha=0.8, edgecolor='black', linewidth=1)

    # Línea vertical en k
    ax3.axvline(x=k, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    ax3.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax3.set_ylabel('Probabilidad', fontsize=12)
    ax3.set_title(f'P(X ≥ {k}) = {prob_acumulada_mayor:.4f}\n({prob_acumulada_mayor*100:.2f}%)',
                  fontsize=13, fontweight='bold', color='#E63946')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.set_xticks(np.arange(min_k, max_k+1, max(1, (max_k-min_k)//15)))

    # Texto adicional
    ax3.text(0.02, 0.98, f'Área sombreada:\nP(X ≥ {k}) = {prob_acumulada_mayor:.6f}',
             transform=ax3.transAxes, fontsize=10, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Gráfico 4: Resumen de resultados y fórmula
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Mostrar la fórmula
    formula_text = (
        f"Fórmula Hipergeométrica:\n\n"
        f"P(X = k) = C(K,k) · C(N-K, n-k) / C(N, n)\n\n"
        f"P(X = {k}) = C({K},{k}) · C({N-K},{n-k}) / C({N},{n})\n\n"
        f"P(X = {k}) = {comb(K, k):.0f} · {comb(N-K, n-k):.0f} / {comb(N, n):.0f}\n"
        f"P(X = {k}) = {prob_exacta:.6f}\n"
        f"({prob_exacta*100:.4f}%)"
    )

    ax4.text(0.5, 0.75, formula_text, transform=ax4.transAxes,
             fontsize=11, verticalalignment='center', horizontalalignment='center',
             bbox=dict(boxstyle='round,pad=0.5', facecolor='#F0F8FF', edgecolor='#2C3E50'),
             family='monospace')

    # Tabla de resultados
    resultados = [
        ['Concepto', 'Valor', 'Porcentaje'],
        ['P(X = k)', f'{prob_exacta:.6f}', f'{prob_exacta*100:.4f}%'],
        ['P(X ≤ k)', f'{prob_acumulada_menor:.6f}', f'{prob_acumulada_menor*100:.4f}%'],
        ['P(X ≥ k)', f'{prob_acumulada_mayor:.6f}', f'{prob_acumulada_mayor*100:.4f}%'],
        ['Media (μ)', f'{media:.4f}', ''],
        ['Varianza (σ²)', f'{varianza:.4f}', ''],
        ['Desv. Estándar (σ)', f'{desviacion:.4f}', ''],
        ['Moda', f'{moda}', '']
    ]

    # Crear tabla en la parte inferior
    table = ax4.table(cellText=resultados, loc='bottom', cellLoc='center',
                      colWidths=[0.3, 0.35, 0.35], bbox=[0, 0, 1, 0.4])

    # Estilo de la tabla
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.3)

    # Colores de la tabla
    for i in range(len(resultados)):
        for j in range(len(resultados[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            elif i in [1, 2, 3]:
                cell.set_facecolor('#E8F4F8')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('#F5F5F5')

    # Ajustar layout
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # 5. GRÁFICO ADICIONAL: FUNCIÓN DE DISTRIBUCIÓN ACUMULADA
    #==========================================================

    # Crear una figura separada para la CDF
    fig2, ax5 = plt.subplots(figsize=(10, 6))

    # Calcular CDF
    cdf = hypergeom.cdf(x, N, K, n)

    # Graficar CDF
    ax5.step(x, cdf, where='mid', linewidth=2.5, color='#2E86AB', label='CDF')
    ax5.fill_between(x, 0, cdf, step='mid', alpha=0.3, color='#2E86AB')

    # Punto en k
    ax5.scatter(k, cdf[x.tolist().index(k)], color='#E63946', s=150, zorder=5,
               label=f'F({k}) = {cdf[x.tolist().index(k)]:.4f}')

    # Líneas de referencia
    ax5.axhline(y=cdf[x.tolist().index(k)], color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.axvline(x=k, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles
    cuantiles = [0.25, 0.5, 0.75]
    colores_cuantiles = ['#F39C12', '#2ECC71', '#3498DB']
    for i, q in enumerate(cuantiles):
        try:
            q_val = hypergeom.ppf(q, N, K, n)
            ax5.axvline(x=q_val, color=colores_cuantiles[i], linestyle=':',
                       linewidth=1.5, alpha=0.5)
            ax5.text(q_val, -0.02, f'Q{int(q*100)}={q_val}',
                    ha='center', fontsize=9, color=colores_cuantiles[i])
        except:
            pass

    ax5.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax5.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax5.set_title(f'Función de Distribución Acumulada\nN={N}, K={K}, n={n}',
                  fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.set_xticks(np.arange(min_k, max_k+1, max(1, (max_k-min_k)//15)))
    ax5.set_ylim(-0.05, 1.05)
    ax5.legend(loc='best')

    # Texto con la probabilidad
    ax5.text(0.02, 0.98,
             f'P(X ≤ {k}) = {cdf[x.tolist().index(k)]:.6f}\nP(X > {k}) = {1-cdf[x.tolist().index(k)]:.6f}\n'
             f'Media = {media:.2f}',
             transform=ax5.transAxes, fontsize=11, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()

    #==========================================================
    # 6. GRÁFICO DE COMPARACIÓN CON DISTRIBUCIÓN BINOMIAL
    #==========================================================

    # Crear figura para comparación
    fig3, ax6 = plt.subplots(figsize=(12, 6))

    # Comparar hipergeométrica con binomial (con reemplazo)
    from scipy.stats import binom

    p_binom = K / N
    pmf_binom = binom.pmf(x, n, p_binom)

    # Graficar ambas
    width = 0.35
    ax6.bar(x - width/2, pmf, width, color='#2E86AB', alpha=0.8,
            label=f'Hipergeométrica (N={N}, K={K}, n={n})', edgecolor='black')
    ax6.bar(x + width/2, pmf_binom, width, color='#E63946', alpha=0.8,
            label=f'Binomial (n={n}, p={p_binom:.3f})', edgecolor='black')

    ax6.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax6.set_ylabel('Probabilidad', fontsize=12)
    ax6.set_title('Comparación: Hipergeométrica vs Binomial', fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3, linestyle='--')
    ax6.legend(loc='best')
    ax6.set_xticks(np.arange(min_k, max_k+1, max(1, (max_k-min_k)//15)))

    # Mostrar diferencia
    diff = np.max(np.abs(pmf - pmf_binom))
    ax6.text(0.98, 0.98, f'Máxima diferencia: {diff:.4f}',
             transform=ax6.transAxes, fontsize=10, verticalalignment='top',
             horizontalalignment='right',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()

    #==========================================================
    # 7. GRÁFICO DE COMPARACIÓN CON DIFERENTES TAMAÑOS DE MUESTRA
    #==========================================================

    # Mostrar cómo cambia la distribución con diferentes n
    fig4, ax7 = plt.subplots(figsize=(12, 6))

    tamanos_muestra = [2, 5, 10, 15, 20]
    tamanos_muestra = [t for t in tamanos_muestra if t <= N]
    colores = ['#2E86AB', '#E63946', '#F39C12', '#2ECC71', '#9B59B6']

    for i, n_comp in enumerate(tamanos_muestra):
        min_k_comp = max(0, n_comp - (N - K))
        max_k_comp = min(n_comp, K)
        x_comp = np.arange(min_k_comp, max_k_comp + 1)
        pmf_comp = hypergeom.pmf(x_comp, N, K, n_comp)

        ax7.plot(x_comp, pmf_comp, 'o-', color=colores[i],
                linewidth=2, markersize=6, label=f'n = {n_comp}')

    ax7.set_xlabel('Número de éxitos (X)', fontsize=12)
    ax7.set_ylabel('Probabilidad', fontsize=12)
    ax7.set_title(f'Comparación para diferentes tamaños de muestra\nN={N}, K={K}',
                  fontsize=13, fontweight='bold')
    ax7.grid(True, alpha=0.3, linestyle='--')
    ax7.legend(loc='best')

    plt.tight_layout()

    #==========================================================
    # 8. MOSTRAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # 9. OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Hipergeometrica_N{N}_K{K}_n{n}_k{k}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_cdf.png", dpi=300, bbox_inches='tight')
        fig3.savefig(f"{nombre_base}_comparacion_binomial.png", dpi=300, bbox_inches='tight')
        fig4.savefig(f"{nombre_base}_comparacion_muestras.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_cdf.png")
        print(f"   • {nombre_base}_comparacion_binomial.png")
        print(f"   • {nombre_base}_comparacion_muestras.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

    return {
        'N': N,
        'K': K,
        'n': n,
        'k': k,
        'prob_exacta': prob_exacta,
        'prob_menor': prob_acumulada_menor,
        'prob_mayor': prob_acumulada_mayor,
        'media': media,
        'varianza': varianza,
        'desviacion': desviacion,
        'moda': moda,
        'fpc': fpc
    }

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_hipergeometrica():
    """
    Ejecuta ejemplos clásicos de la distribución hipergeométrica
    """

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN HIPERGEOMÉTRICA")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Control de calidad',
            'descripcion': 'Un lote de 50 piezas contiene 5 defectuosas. Se seleccionan 10 piezas al azar sin reemplazo. ¿Cuál es la probabilidad de encontrar exactamente 2 defectuosas?',
            'N': 50,
            'K': 5,
            'n': 10,
            'k': 2
        },
        {
            'titulo': 'Ejemplo 2: Urna con bolas',
            'descripcion': 'Una urna contiene 10 bolas rojas y 20 azules. Se extraen 5 bolas sin reemplazo. ¿Cuál es la probabilidad de obtener exactamente 3 rojas?',
            'N': 30,
            'K': 10,
            'n': 5,
            'k': 3
        },
        {
            'titulo': 'Ejemplo 3: Encuesta de satisfacción',
            'descripcion': 'En una empresa de 200 empleados, 60 están insatisfechos. Se seleccionan 15 empleados al azar. ¿Cuál es la probabilidad de que al menos 5 estén insatisfechos?',
            'N': 200,
            'K': 60,
            'n': 15,
            'k': 5
        },
        {
            'titulo': 'Ejemplo 4: Cartas de una baraja',
            'descripcion': 'De una baraja de 52 cartas (13 de cada palo), se extraen 7 cartas sin reemplazo. ¿Cuál es la probabilidad de obtener exactamente 3 cartas de corazones?',
            'N': 52,
            'K': 13,
            'n': 7,
            'k': 3
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")
        print(f"   N={ej['N']}, K={ej['K']}, n={ej['n']}, k={ej['k']}")

        prob = hypergeom.pmf(ej['k'], ej['N'], ej['K'], ej['n'])
        prob_acum = hypergeom.cdf(ej['k'], ej['N'], ej['K'], ej['n'])
        prob_mayor = 1 - hypergeom.cdf(ej['k'] - 1, ej['N'], ej['K'], ej['n']) if ej['k'] > 0 else 1

        media = ej['n'] * ej['K'] / ej['N']

        print(f"   P(X = {ej['k']}) = {prob:.6f} ({prob*100:.2f}%)")
        print(f"   P(X ≤ {ej['k']}) = {prob_acum:.6f} ({prob_acum*100:.2f}%)")
        print(f"   P(X ≥ {ej['k']}) = {prob_mayor:.6f} ({prob_mayor*100:.2f}%)")
        print(f"   Media = {media:.4f} éxitos esperados")

    print("\n" + "="*70)
    print("¿Deseas analizar alguno de estos ejemplos en detalle?")
    print("Selecciona 1, 2, 3, 4 o ejecuta la calculadora principal.")

#==========================================================
# FUNCIÓN PARA COMPARAR CON Y SIN REEMPLAZO
#==========================================================

def comparar_con_binomial():
    """
    Compara la distribución hipergeométrica con la binomial
    """

    print("\n" + "="*70)
    print("📊 COMPARACIÓN: HIPERGEOMÉTRICA vs BINOMIAL")
    print("="*70)
    print("\nLa distribución hipergeométrica (sin reemplazo) se aproxima")
    print("a la distribución binomial (con reemplazo) cuando:")
    print("  • La población es grande (N → ∞)")
    print("  • La muestra es pequeña comparada con la población (n/N → 0)\n")

    try:
        print("Ingresa los parámetros para la comparación:")
        N = int(input("  Tamaño de la población (N): "))
        K = int(input(f"  Número de éxitos (K, 0 ≤ K ≤ {N}): "))
        n = int(input(f"  Tamaño de la muestra (n, 0 ≤ n ≤ {N}): "))

        # Calcular ambas distribuciones
        min_k = max(0, n - (N - K))
        max_k = min(n, K)
        x = np.arange(min_k, max_k + 1)

        pmf_hyper = hypergeom.pmf(x, N, K, n)
        p_binom = K / N
        from scipy.stats import binom
        pmf_binom = binom.pmf(x, n, p_binom)

        # Crear gráfico
        fig, ax = plt.subplots(figsize=(12, 6))

        width = 0.35
        ax.bar(x - width/2, pmf_hyper, width, color='#2E86AB', alpha=0.8,
               label='Hipergeométrica (sin reemplazo)', edgecolor='black')
        ax.bar(x + width/2, pmf_binom, width, color='#E63946', alpha=0.8,
               label='Binomial (con reemplazo)', edgecolor='black')

        ax.set_xlabel('Número de éxitos (X)', fontsize=12)
        ax.set_ylabel('Probabilidad', fontsize=12)
        ax.set_title(f'Comparación: Hipergeométrica vs Binomial\nN={N}, K={K}, n={n}, p={p_binom:.3f}',
                    fontsize=13, fontweight='bold')
        ax.grid(True, alpha=0.3, linestyle='--')
        ax.legend(loc='best')

        # Estadísticas de diferencia
        diff = np.max(np.abs(pmf_hyper - pmf_binom))
        diff_prom = np.mean(np.abs(pmf_hyper - pmf_binom))

        ax.text(0.98, 0.95, f'Diferencia máxima: {diff:.4f}\nDiferencia promedio: {diff_prom:.4f}',
               transform=ax.transAxes, fontsize=10, verticalalignment='top',
               horizontalalignment='right',
               bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

        plt.tight_layout()
        plt.show()

        print("\nResultados de la comparación:")
        print(f"  • Diferencia máxima: {diff:.6f}")
        print(f"  • Diferencia promedio: {diff_prom:.6f}")
        if n/N < 0.1:
            print("  ✅ La muestra es pequeña comparada con la población (n/N < 10%)")
            print("  → La aproximación binomial es buena")
        else:
            print("  ⚠️ La muestra es grande comparada con la población (n/N ≥ 10%)")
            print("  → Usar la distribución hipergeométrica es más apropiado")

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA CALCULAR TAMAÑO DE MUESTRA
#==========================================================

def calcular_tamano_muestra():
    """
    Calcula el tamaño de muestra necesario para un nivel de precisión dado
    """

    print("\n" + "="*70)
    print("📏 CÁLCULO DE TAMAÑO DE MUESTRA")
    print("="*70)
    print("\nEsta función ayuda a determinar el tamaño de muestra necesario")
    print("para estimar la proporción de éxitos en una población finita.\n")

    try:
        N = int(input("  Tamaño de la población (N): "))
        p_esperada = float(input("  Proporción esperada de éxitos (0-1): "))
        error = float(input("  Error máximo permitido (±, en puntos porcentuales, ej: 5): ")) / 100
        nivel_confianza = float(input("  Nivel de confianza (ej: 95): ")) / 100

        # Valor z para el nivel de confianza
        from scipy.stats import norm
        z = norm.ppf(1 - (1 - nivel_confianza) / 2)

        # Tamaño de muestra inicial (sin corrección por población finita)
        n0 = (z**2 * p_esperada * (1 - p_esperada)) / (error**2)

        # Corrección por población finita
        n = n0 / (1 + (n0 - 1) / N)

        print(f"\nResultados:")
        print(f"  • Tamaño de muestra requerido: {np.ceil(n):.0f}")
        print(f"  • Tamaño de muestra sin corrección: {np.ceil(n0):.0f}")
        print(f"  • Proporción de la población a muestrear: {n/N*100:.2f}%")

        if n > N:
            print("  ⚠️ El tamaño de muestra calculado supera el tamaño de la población")
            print("  → Se recomienda usar toda la población")

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN HIPERGEOMÉTRICA")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado de una distribución")
        print("  2. Ver ejemplos clásicos")
        print("  3. Comparar con distribución binomial")
        print("  4. Calcular tamaño de muestra")
        print("  5. Salir")

        opcion = input("\nSelecciona una opción (1-5): ")

        if opcion == '1':
            resultados = calculadora_hipergeometrica()
        elif opcion == '2':
            ejemplos_hipergeometrica()
        elif opcion == '3':
            comparar_con_binomial()
        elif opcion == '4':
            calcular_tamano_muestra()
        elif opcion == '5':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3, 4 o 5.")
```

# **32. Distribución Uniforme Discreta**
+ *La distribución uniforme discreta modela una situación donde todos los valores enteros posibles entre dos límites $a$ y $b$ tienen la misma probabilidad.*

- **Función de probabilidad**: $$P(X = x) = \frac{1}{b - a + 1}$$ para $x = a, a+1, \dots, b$
- **Parámetros**: $a$ es el límite inferior, $b$ es el límite superior.

### **Distribución Uniforme Discreta - Momentos**
- **Valor esperado (Esperanza)**: $$E(X) = \frac{a + b}{2}$$
- **Varianza**: $$\text{Var}(X) = \frac{(b - a + 1)^2 - 1}{12}$$
- **Desviación estándar**: $$\sigma(X) = \sqrt{\text{Var}(X)}$$


```python
# ==========================================================
# @title **A. Gráfica de la Distribución Uniforme Discreta**
#
# $$P(X=x)=\frac{1}{b-a+1},\qquad x=a,a+1,\ldots,b$$
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt

# Parámetros
a = 1
b = 6

# Valores de la variable aleatoria
x = np.arange(a, b + 1)

# Función de probabilidad
pmf = np.repeat(1 / (b - a + 1), len(x))

# Mostrar probabilidades
print("=" * 45)
print("DISTRIBUCIÓN UNIFORME DISCRETA")
print("=" * 45)

for xi, pi in zip(x, pmf):
    print(f"P(X = {xi}) = {pi:.4f}")

print("=" * 45)

# Gráfico
plt.figure(figsize=(8,5))

plt.bar(
    x,
    pmf,
    color="skyblue",
    edgecolor="black",
    width=0.7
)

# Etiquetas sobre las barras
for xi, pi in zip(x, pmf):
    plt.text(
        xi,
        pi + 0.005,
        f"{pi:.3f}",
        ha="center",
        fontsize=11
    )

plt.title("Distribución Uniforme Discreta", fontsize=14, fontweight="bold")
plt.suptitle(r"$X \sim U\{1,\ldots,6\}$", fontsize=14, y=0.8)

plt.xlabel("Valores de X")
plt.ylabel("Probabilidad")

plt.xticks(x)
plt.ylim(0, max(pmf) + 0.05)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

```python
# ==========================================================
# @title **B. Gráfica de la Distribución Uniforme Discreta**
#
# $$P(X=x)=\frac{1}{b-a+1},\qquad x=a,a+1,\ldots,b$$
# ==========================================================
%%R
library(ggplot2)

# Parámetros
a <- 1
b <- 6

# Valores de la variable aleatoria
x <- a:b

# Función de probabilidad
pmf <- rep(1 / (b - a + 1), length(x))

# Data frame
datos <- data.frame(
  x = factor(x),
  Probabilidad = pmf
)

# Mostrar probabilidades
cat("=============================================\n")
cat("DISTRIBUCIÓN UNIFORME DISCRETA\n")
cat("=============================================\n")

for(i in 1:length(x)){
  cat(sprintf("P(X = %d) = %.4f\n",
              x[i], pmf[i]))
}

cat("=============================================\n")

# Gráfico
ggplot(datos,
       aes(x = x,
           y = Probabilidad)) +

  geom_col(fill = "skyblue",
           color = "black",
           width = 0.7) +

  geom_text(aes(label = sprintf("%.3f", Probabilidad)),
            vjust = -0.5,
            size = 4) +

  labs(
    title = "Distribución Uniforme Discreta",
    subtitle = paste("X ~ U{", a, ", ...,", b, "}"),
    x = "Valores de X",
    y = "Probabilidad"
  ) +

  theme_minimal(base_size = 13) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

# **33. Variables aleatorias continuas y funciones de densidad de Probabilidad**

## **Definición de variable aleatoria continua**
+ Una variable $X$ es llamada continua si toma todos sus valores sobre un intervalo de la recta real.  Esto es, el conjunto $\{X(w):w∈Ω\}$ es un intervalo sobre $\mathbb{R}$.  
+ Como ejemplo este caso se pueden considerar variables aleatorias continuas: _tiempo, edad y estatura_.

+ **Función densidad**. Sea $X$ una variable aleatoria definida sobre $(Ω,A,P)$.  La función de densidad $f(x)$ es dada por alguna función integrable sobre $\mathbb{R}$ tal que

 -  $f(x)\geq0$, para  todo  $x∈\mathbb{R}$
 -  $P_{X}(A)=∫_{A}f(x)dx$ para todo evento $A\subseteq\mathbb{R}$


+ **Teorema** Sea $X$ una variable aleatoria definida sobre $(Ω,A,P)$. Toda función $f(x)$ que es integrable sobre $\mathbb{R}$ y satisface:

 -  $f(x)\geq0$, para  todo  $x∈\mathbb{R}$
 -  $ ∫_{-∞}^{∞}f(x)dx=1$  
  es la función de densidad de alguna variable aleatoria continua $X$.
+ **Nota**. Este teorema nos sirve para determinar cuando una función integrable sobre $\mathbb{R}$ es una función de densidad de alguna variable aleatoria continua $X$.

# **34. Ejemplo 1. De una variable aleatoria continua**.
+ Sea  $f(x)$ una función sobre $\mathbb{R}$ dada por


$$f(x)= \left\{ \begin{array}{lcc}
             x	&   si	&   0≤x≤1 , \\
             \\ 2-x	&   si	&   1≤x≤2, \\
             \\ 0 &   en & otro & caso
             \end{array}
   \right.$$


 - Esta es una función integrable que satisface que  $f(x)$ para  todo  $x∈\mathbb{R}$ como se puede observar en la figura 1.


```python
# @title **A. Ejemplo. De una variable aleatoria continua**
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 3,1000)
y = np.piecewise(x, [x<0, (x>=0)&(x<1),(x>=1)&(x<2),x>=2], [lambda x: 0, lambda x: x,lambda x: 2-x, lambda x: 0])

plt.plot(x, y)
plt.grid()
plt.text(-0.5,0.1, 'y = 0')
plt.text(0.0,0.6, 'y = x')
plt.text(1.6,0.6, 'y = 2-x')
plt.text(2.5,0.1, 'y = 0')
plt.title('Figura 1. Funcion densidad de probabilidad')
plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo. Función de Densidad de una Variable Aleatoria Continua**
# ==========================================================
%%R
library(ggplot2)

# Generar valores de x
x <- seq(-1, 3, length.out = 1000)

# Definir la función por tramos
y <- ifelse(x < 0, 0,
            ifelse(x < 1, x,
                   ifelse(x < 2, 2 - x, 0)))

# Crear data frame
datos <- data.frame(
  x = x,
  y = y
)

# Graficar
ggplot(datos, aes(x = x, y = y)) +
  geom_line(color = "blue", linewidth = 1.2) +
  geom_hline(yintercept = 0, color = "black") +
  annotate("text", x = -0.5, y = 0.08, label = "y = 0", color = "black", size = 4) +
  annotate("text", x = 0.35, y = 0.65, label = "y = x", color = "blue", size = 4) +
  annotate("text", x = 1.55, y = 0.65, label = "y = 2 - x", color = "blue", size = 4) +
  annotate("text", x = 2.5, y = 0.08, label = "y = 0", color = "black", size = 4) +
  labs(
    title = "Figura 1. Función de Densidad de Probabilidad",
    x = "x",
    y = "f(x)"
  ) +
  theme_minimal(base_size = 14) +
  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

## **b. Probando que es una función densidad**
+ Además

	$$∫_{-∞}^{∞}f(x)dx=∫_{-∞}^{0}0dx+∫_{0}^{1}xdx+∫_{1}^{2}(2-x)dx+∫_{2}^{∞}0dx=1.0$$



```python
#@title **c. Probando que es una función densidad $∫_{-∞}^{∞}f(x)dx$ usando pyhton**
from sympy import integrate, init_printing
from sympy.abc import x

print( r"∫_{-∞}^{∞}f(x)dx = ",integrate(x, (x,0,1))+integrate(2-x, (x,1,2)))
```

```python
# ==========================================================
# @title **C. Probando que es una función de densidad**
# usando integrate() de R
# ==========================================================
%%R
cat("=========================================\n")
cat("COMPROBACIÓN DE LA FUNCIÓN DE DENSIDAD\n")
cat("=========================================\n\n")

# Definir la función de densidad
f <- function(x){
  ifelse(x >= 0 & x <= 1, x,
         ifelse(x > 1 & x <= 2, 2 - x, 0))
}

# Integral en todo el soporte
resultado <- integrate(f, lower = -Inf, upper = Inf)

cat(sprintf("∫₋∞^∞ f(x) dx = %.6f\n", resultado$value))

if(abs(resultado$value - 1) < 1e-10){
  cat("\n✅ La función es una función de densidad de probabilidad.\n")
} else {
  cat("\n❌ La función NO es una función de densidad.\n")
}
```

```python
#@title **d. Probando que es una función densidad $$p(A)=∫_{0.5}^{1}xdx+∫_{1}^{1.8}(2-x)dx$$**
from sympy import integrate, init_printing
from sympy.abc import x

print('Calculando la integral p(A)=∫_{0.5}^{1}xdx+∫_{1}^{1.8}(2-x)dx',integrate(x, (x,0.5,1))+integrate(2-x, (x,1,1.8)))
```

```python
# ==========================================================
# @title **D. Probando que es una función de densidad**
# $$P(A)=\int_{0.5}^{1}x\,dx+\int_{1}^{1.8}(2-x)\,dx$$
# ==========================================================
%%R
cat("=========================================\n")
cat("CÁLCULO DE LA PROBABILIDAD P(A)\n")
cat("=========================================\n\n")

# Primera parte: ∫_{0.5}^{1} x dx
I1 <- integrate(function(x) x,
                lower = 0.5,
                upper = 1)

# Segunda parte: ∫_{1}^{1.8} (2-x) dx
I2 <- integrate(function(x) 2 - x,
                lower = 1,
                upper = 1.8)

# Probabilidad total
PA <- I1$value + I2$value

cat("∫₀.₅¹ x dx             =", round(I1$value, 5), "\n")
cat("∫₁¹.⁸ (2 - x) dx       =", round(I2$value, 5), "\n")
cat("-----------------------------------------\n")
cat("P(A) = ", round(PA, 5), "\n")
```

## **e. Sea el evento $A=(0.5,1.8]$ entonces la probabilidad de $A$ puede ser calculada como**


+ $$p(A)=∫_{0.5}^{1}xdx+∫_{1}^{1.8}(2-x)dx=0.855$$

# **35. Ejemplo 2. Función de probabilidad acumulada continua**

+ La función de probabilidad acumulada o función de distribución de una variable aleatoria $X$ sobre $Ω$, denotada por $F(x)$, es definida por la relación
$$F(x)=P\{w:-∞<X(w)≤x\}=P(X≤x)=∫_{-∞}^{x}f(x)dx$$

# **a. Ejemplo de la distribución acumulada**
+ Para el ejemplo tratado anteriormente la función de distribución es determinada como sigue:

 -  Divida el rango de la variable en subintervalos: $x\leq0$,  $0<x\leq1$, $1<x\leq2$ y $x\geq2$. esta división es realizada de acuerdo a la partición de la recta real dada en la función de probabilidad.  cambie la notación de f(x) por f(t) para evitar confuciones.

 -  Calcule la función de probabilidad acumulada para un un valor x que se encuentre en el intervalo como la suma de las probabilidades de los valores de la variable menores a x.

 -  Para $x\leq0$, $$	F(x)=P(X≤x)=∫_{-∞}^{x}f(t)dt=∫_{-∞}^{x}0dt=0$$
ya que según la definición de la función de probabilidad $f(x)=0$
  -  Para $0≤x<1$ $$	F(x)  =P(X≤x)=∫_{-∞}^{x}f(t)dt
	 =∫_{-∞}^{0}0dt+∫_{0}^{x}tdt=t^2/2|_{0}^{x}=x^2/2$$

 -  Para $1<x\leq2$, $$
	F(x)  =P(X≤x)=∫_{-∞}^{x}f(t)dt
	 =∫_{-∞}^{0}0dt+∫_{0}^{1}tdt+∫_{1}^{x}(2-t)dt
	 =0+((t^2)/2)|_{0}^{1}+(2t-((t^2)/2))|_{1}^{x}
	 =((1^2)/2)+[(2x-((x^2)/2))-(2-((1^2)/2))]
	 =2x-x^2/2-1$$


+ Para $x\geq2$, $$
	F(x)  =P(X≤x)=∫_{-∞}^{x}f(t)dt
	 =∫_{-∞}^{0}0dt+∫_{0}^{1}tdt+∫_{1}^{2}(2-t)dt+∫_{0}^{x}0dt
	 =0+((t^2)/2)|_{0}^{1}+(2t-((t^2)/2))|_{1}^{2}+0
	 =1$$


+ Luego la función distribución acumulada $F(x)$ de la variable aleatoria continua X es dada por


$$F(x)= \left\{ \begin{array}{lcc}
             0	&   si	&   x≤0 , \\
             \\ \frac{x^2}{2}	&   si	&   0≤x<1, \\
             \\ 2x-\frac{x^2}{2}-1	&   si	&   1≤x≤2, \\
             \\ 1 &   si	&   x\geq2
             \end{array}
   \right.$$

y el gráfico $F(x)$ es dado

```python
# @title **b. Ejemplo. Gráfico de la Distribución acumulada de probabilidad**
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 3,1000)
y = np.piecewise(x, [x<0, (x>=0)&(x<1),(x>=1)&(x<2),x>=2], [lambda x: 0, lambda x: x**2/2,lambda x: 2*x-x**2/2-1, lambda x: 1])

plt.plot(x, y)
plt.grid()
plt.text(-0.5,0.1, 'y = 0')
plt.text(0.4,0.4, '$y = x^2/2$')
plt.text(1.2,0.5, '$y = 2x-x^2/2-1$')
plt.text(2.5,0.95, '$y = 1$')
plt.title('Figura 1. Funcion Distribución Acumulada de probabilidad')

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo. Gráfico de la Función de Distribución Acumulada**
#
#            0,                    x < 0
# F(x) =     x²/2,               0 ≤ x < 1
#            2x - x²/2 - 1,      1 ≤ x < 2
#            1,                    x ≥ 2
# ==========================================================
%%R
library(ggplot2)

# Valores de x
x <- seq(-1, 3, length.out = 1000)

# Función de distribución acumulada
F_x <- ifelse(
  x < 0, 0,
  ifelse(
    x < 1, x^2 / 2,
    ifelse(
      x < 2,
      2 * x - x^2 / 2 - 1,
      1
    )
  )
)

# Data frame
datos <- data.frame(
  x = x,
  F = F_x
)

# Gráfico
ggplot(datos, aes(x = x, y = F)) +
  geom_line(color = "blue", linewidth = 1.2) +
  labs(
    title = "Figura 1. Función de Distribución Acumulada de Probabilidad",
    x = "x",
    y = "F(x)"
  ) +
  annotate("text", x = -0.55, y = 0.08,
           label = "F(x)=0", size = 4) +
  annotate("text", x = 0.45, y = 0.35,
           label = "F(x)=x²/2", color = "blue", size = 4) +
  annotate("text", x = 1.45, y = 0.55,
           label = "F(x)=2x-x²/2-1", color = "blue", size = 4) +
  annotate("text", x = 2.55, y = 0.96,
           label = "F(x)=1", size = 4) +
  scale_x_continuous(breaks = seq(-1, 3, 0.5)) +
  scale_y_continuous(limits = c(0, 1.05),
                     breaks = seq(0, 1, 0.1)) +
  theme_minimal(base_size = 14) +
  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

# **36.  Distribuciones continuas - La Distribución Uniforme Continua**
# **a. La Definición**
+ Una variable aleatoria continua $X$ sigue una distribución Uniforme de parámetros $a$ y $b$, lo cual denotamos por $$X∼U(a , b)$$ si tiene funcion FDP dada por la formula:
$$ f(X=x)= f(x,μ , σ) = \frac{1}{b-a}, \ para\ a\leq x \leq b$$
+ Tambien $$E(X)=\mu=\frac{a+b}{2}, \ \ \ \ V(X)=σ^2=\frac{(b-a)^2}{12}$$

+ En dónde $ \{X\leq x\}$ representa al evento de todos resultados menores que $x$.
+ $\mu$ representa al número esperado.
+ $V(X)=σ^2$ representa a la varianza.


# **b. Valor Esperado y Varianza de una variable aleatoria continua $X$**

 - **Media de $X$**
 $$\mu = E(X)=∫_{-∞}^{∞}xf(x)dx=∫_{a}^{b}x\frac{1}{b-a}dx=\frac{x^2}{2(b-a)}|_{a}^{b}$$

 $$\mu = E(X)=\frac{b^2-a^2}{2(b-a)}=\frac{b+a}{2}$$

 - **Varianza  de $X$** $$\sigma^2 = E(X-μ)^2=∫_{-∞}^{∞}(x-μ)^2f(x)dx=E(X^2)-μ^2$$
 -  $$ E(X^2)=∫_{-∞}^{∞}x^2f(x)dx=∫_{a}^{b}x^2\frac{1}{b-a}dx=\frac{x^3}{3(b-a)}|_{a}^{b}$$

 $$ E(X^2)=\frac{b^3-a^3}{3(b-a)}=\frac{b^2+ab+a^2}{3}$$

 - **Varianza  de $X$** $$\sigma^2 =E(X^2)-μ^2$$

 $$\sigma^2 =\frac{b^2+ab+a^2}{3}-(\frac{b+a}{2})^2$$
 $$\sigma^2 =\frac{b^2+ab+a^2}{3}-\frac{b^2+2ab+a^2}{4}$$
 $$\sigma^2 =\frac{b^2-2ab+a^2}{12}$$

# **37. Ejemplo 1. La Distribución Uniforme Continua**

```python
# @title **A. Ejemplo 1. La distribución Uniforme Continua usando Numpy**
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de la distribución uniforme continua
a = 5
b = 8

# Generar muestras
uniform_samples = np.random.uniform(a, b, size=10000)

# Calcular la media y varianza teóricas
theoretical_mean = (a + b) / 2
theoretical_variance = ((b - a)**2) / 12

# Calcular la media y varianza de la muestra
sample_mean = np.mean(uniform_samples)
sample_variance = np.var(uniform_samples)

print(f"Parámetros de la distribución: a={a}, b={b}")
print(f"Media Teórica: {theoretical_mean:.2f}")
print(f"Varianza Teórica: {theoretical_variance:.2f}")
print(f"Media Muestral: {sample_mean:.2f}")
print(f"Varianza Muestral: {sample_variance:.2f}")

fig, ax = plt.subplots(figsize=(10, 6))
plt.hist(uniform_samples, density=True, bins=6, edgecolor='black', alpha=0.7)
plt.title('Distribución Uniforme Continua', size=14)
plt.xlabel('Muestras')
plt.ylabel('Densidad')

# Añadir línea para la densidad teórica
plt.axhline(1 / (b - a), color='#3793ef', linestyle='--', label=f'Densidad teórica = {1/(b-a):.2f}')

# Añadir líneas para la media teórica y muestral
plt.axvline(theoretical_mean, color='red', linestyle=':', label=f'Media teórica = {theoretical_mean:.2f}')
plt.axvline(sample_mean, color='green', linestyle=':', label=f'Media muestral = {sample_mean:.2f}')

plt.legend()
plt.grid(True)
plt.show()
```

```python
# ==========================================================
# @title **A. Ejemplo 1. La Distribución Uniforme Continua**
# ==========================================================
%%R
library(ggplot2)

# Fijar semilla para reproducibilidad
set.seed(123)

# Parámetros de la distribución
a <- 5
b <- 8

# Generar muestras
uniform_samples <- runif(10000, min = a, max = b)

# Media y varianza teóricas
media_teorica <- (a + b) / 2
varianza_teorica <- (b - a)^2 / 12

# Media y varianza muestrales
media_muestral <- mean(uniform_samples)
varianza_muestral <- var(uniform_samples)

# Mostrar resultados
cat("=============================================\n")
cat("DISTRIBUCIÓN UNIFORME CONTINUA\n")
cat("=============================================\n")
cat(sprintf("Parámetros: a = %.0f, b = %.0f\n", a, b))
cat(sprintf("Media teórica     = %.4f\n", media_teorica))
cat(sprintf("Varianza teórica  = %.4f\n", varianza_teorica))
cat(sprintf("Media muestral    = %.4f\n", media_muestral))
cat(sprintf("Varianza muestral = %.4f\n", varianza_muestral))
cat("=============================================\n")

# Data frame
datos <- data.frame(Muestras = uniform_samples)

# Histograma
ggplot(datos, aes(x = Muestras)) +

  geom_histogram(aes(y = after_stat(density)),
                 bins = 6,
                 fill = "skyblue",
                 color = "black",
                 alpha = 0.7) +

  # Densidad teórica
  geom_hline(yintercept = 1/(b-a),
             color = "blue",
             linewidth = 1.2,
             linetype = "dashed") +

  # Media teórica
  geom_vline(xintercept = media_teorica,
             color = "red",
             linewidth = 1.2,
             linetype = "dotted") +

  # Media muestral
  geom_vline(xintercept = media_muestral,
             color = "darkgreen",
             linewidth = 1.2,
             linetype = "dotted") +

  annotate("text",
           x = media_teorica,
           y = 0.40,
           label = paste("Media teórica =", round(media_teorica,2)),
           color = "red",
           angle = 90,
           vjust = -0.5) +

  annotate("text",
           x = media_muestral,
           y = 0.40,
           label = paste("Media muestral =", round(media_muestral,2)),
           color = "darkgreen",
           angle = 90,
           vjust = 1.5) +

  labs(
    title = "Distribución Uniforme Continua",
    subtitle = paste("U(", a, ",", b, ")", sep = ""),
    x = "Muestras",
    y = "Densidad"
  ) +

  theme_minimal(base_size = 14) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

# **39.  Ejemplo 1. Construye la grafica de la distribucion normal con media 10 y desviacion 2**

```python
# @title **A. Ejemplo 1. Construye la grafica de la distribucion normal con media 10 y desviacion 2**

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parámetros de la distribución normal
media = 10
desviacion = 2

# Generar datos para la gráfica
x = np.linspace(media - 4 * desviacion, media + 4 * desviacion, 100)
y = norm.pdf(x, media, desviacion)

# Crear la gráfica
plt.plot(x, y)
plt.title('Distribución Normal')
plt.xlabel('Valores')
plt.ylabel('Probabilidad')
plt.grid(True)
plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 1. Construye la gráfica de la Distribución Normal con media 10 y desviación estándar 2**
# ==========================================================

# Instalar si es necesario
# install.packages("ggplot2")
%%R
library(ggplot2)

# Parámetros
media <- 10
desviacion <- 2

# Generar valores de x
x <- seq(media - 4 * desviacion,
         media + 4 * desviacion,
         length.out = 1000)

# Función de densidad
y <- dnorm(x,
           mean = media,
           sd = desviacion)

# Data frame
datos <- data.frame(
  x = x,
  Densidad = y
)

# Mostrar parámetros
cat("=====================================\n")
cat("DISTRIBUCIÓN NORMAL\n")
cat("=====================================\n")
cat(sprintf("Media (μ) = %.2f\n", media))
cat(sprintf("Desviación estándar (σ) = %.2f\n", desviacion))
cat("=====================================\n")

# Gráfico
ggplot(datos, aes(x = x, y = Densidad)) +

  geom_line(color = "blue",
            linewidth = 1.3) +

  geom_vline(xintercept = media,
             color = "red",
             linetype = "dashed",
             linewidth = 1) +

  annotate("text",
           x = media,
           y = max(y)*0.95,
           label = expression(mu==10),
           color = "red",
           size = 5) +

  labs(
    title = "Distribución Normal",
    subtitle = expression(X %~% N(mu==10, sigma==2)),
    x = "Valores",
    y = "Densidad"
  ) +

  theme_minimal(base_size = 14) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

# **42.  Distribución Normal Estándar $N(0,1)$**

+ La distribución normal estándar es un caso particular de la distribución normal en el que:

$$
\mu = 0
\qquad \text{y} \qquad
\sigma = 1
$$

+ Por tanto,

$$
Z \sim N(0,1)
$$

---

## **Estandarización de una Variable Normal**

+ Podemos transformar una variable aleatoria

$$
X \sim N(\mu,\sigma^2)
$$

+ en una variable normal estándar

$$
Z \sim N(0,1)
$$

+ mediante la transformación:

$$
Z=\frac{X-\mu}{\sigma}
$$

+ donde:

  - $X$ es la variable original.
  - $\mu$ es la media poblacional.
  - $\sigma$ es la desviación estándar poblacional.
  - $Z$ es la variable estandarizada.

---

## **Función de Densidad de la Normal General**

+ Si

$$
X \sim N(\mu,\sigma^2),
$$

+ su función de densidad es

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}
$$

---

## **Función de Densidad de la Normal Estándar**

+ Después de la transformación

$$
Z=\frac{X-\mu}{\sigma},
$$

+ obtenemos

$$
Z \sim N(0,1)
$$

con función de densidad

$$
f(z)=\frac{1}{\sqrt{2\pi}}
e^{-\frac{z^2}{2}}
$$

---

## **Transformación Inversa**

+ Si conocemos un valor estandarizado

$$
Z \sim N(0,1),
$$

+ podemos recuperar el valor original de la variable normal mediante:

$$
X=\mu+\sigma Z
$$

+ Esta expresión permite convertir valores de la normal estándar a cualquier distribución normal.

---

## **Resumen de Transformaciones**

### **De Normal General a Normal Estándar**

$$
X \sim N(\mu,\sigma^2)
\quad \Longrightarrow \quad
Z=\frac{X-\mu}{\sigma}
\quad \Longrightarrow \quad
Z \sim N(0,1)
$$

### **De Normal Estándar a Normal General**

$$
Z \sim N(0,1)
\quad \Longrightarrow \quad
X=\mu+\sigma Z
\quad \Longrightarrow \quad
X \sim N(\mu,\sigma^2)
$$

---

## **Ventaja de la Estandarización**

+ La transformación a la normal estándar permite utilizar una única tabla de probabilidades para calcular áreas bajo la curva normal, independientemente de los valores de $\mu$ y $\sigma$.

$$
P(X\leq x)
=
P\left(
Z\leq
\frac{x-\mu}{\sigma}
\right)
$$

+ y, en general,

$$
P(a\leq X\leq b)
=
P\left(
\frac{a-\mu}{\sigma}
\leq Z \leq
\frac{b-\mu}{\sigma}
\right).
$$

```python
# @title **Distribución Normal estandard**

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parámetros de la distribución normal estándar
media = 0
desviacion_estandar = 1

# Generar datos para la gráfica
x = np.linspace(media - 4 * desviacion_estandar, media + 4 * desviacion_estandar, 100)
y = norm.pdf(x, media, desviacion_estandar)

# Crear la gráfica
plt.plot(x, y)
plt.title('Curva Normal Estándar')
plt.xlabel('X')
plt.ylabel('Probabilidad')
plt.grid(True)
plt.show()
```

# **43.  Ejemplo 1. Distribución Normal. Calcula el area a la izquierda de 1.85, dibuja y sombrea la region**

```python
# @title **A. Ejemplo 1. Distribución Normal. Calcula el area a la izquierda de 1.85, dibuja y sombrea la region**

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parámetros de la distribución normal estándar
media = 0
desviacion_estandar = 1

# Generar datos para la gráfica
x = np.linspace(media - 6 * desviacion_estandar, media + 6 * desviacion_estandar, 100)
y = norm.pdf(x, media, desviacion_estandar)

# Crear la gráfica
plt.plot(x, y)
plt.title('Curva Normal Estándar')
plt.xlabel('X')
plt.ylabel('Probabilidad')
plt.grid(True)

# Calcular el área a la izquierda de 1.85
z = 1.85
area_izquierda = norm.cdf(z)

# Sombrear la región a la izquierda de 1.85
x_sombreado = np.linspace(media - 5 * desviacion_estandar, z, 100)
y_sombreado = norm.pdf(x_sombreado, media, desviacion_estandar)
plt.fill_between(x_sombreado, y_sombreado, color='skyblue')

# Mostrar el valor del área
plt.text(z - 1.5, 0.05, f'Área pedida {z:.2f}: {area_izquierda:.5f}', fontsize=12)

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 1. Distribución Normal Estándar**
# **Calcular el área a la izquierda de z = 1.85**
# ==========================================================

# Instalar si es necesario
# install.packages("ggplot2")
%%R
library(ggplot2)

# Parámetros
media <- 0
desviacion <- 1
z <- 1.85

# Calcular el área a la izquierda
area_izquierda <- pnorm(z, mean = media, sd = desviacion)

cat("=========================================\n")
cat("DISTRIBUCIÓN NORMAL ESTÁNDAR\n")
cat("=========================================\n")
cat(sprintf("Media (μ) = %.0f\n", media))
cat(sprintf("Desviación estándar (σ) = %.0f\n", desviacion))
cat(sprintf("Valor de z = %.2f\n", z))
cat("-----------------------------------------\n")
cat(sprintf("P(X ≤ %.2f) = %.5f\n", z, area_izquierda))
cat("=========================================\n")

# Datos para la curva
x <- seq(-4, 4, length.out = 1000)
y <- dnorm(x, mean = media, sd = desviacion)

datos <- data.frame(x = x, y = y)

# Datos para sombrear
datos_sombra <- subset(datos, x <= z)

# Gráfico
ggplot(datos, aes(x = x, y = y)) +

  geom_line(color = "blue", linewidth = 1.3) +

  geom_area(data = datos_sombra,
            aes(x = x, y = y),
            fill = "skyblue",
            alpha = 0.6) +

  geom_vline(xintercept = z,
             color = "red",
             linetype = "dashed",
             linewidth = 1.1) +

  annotate("text",
           x = -2.2,
           y = 0.08,
           label = paste0("P(X ≤ ", z, ") = ",
                          round(area_izquierda,5)),
           color = "red",
           size = 5) +

  labs(
    title = "Distribución Normal Estándar",
    subtitle = paste("Área a la izquierda de z =", z),
    x = "Z",
    y = "Densidad"
  ) +

  theme_minimal(base_size = 14) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

# **44.  Ejemplo 2. Distribución Normal.**

```python
# ==========================================================
# @title **A. Ejemplo 2. El volumen de hojas que una oficina de abogados deposita en un juzgados de sus casos atendidos es una variable aleatoria normal para la cual μ = 1200  y σ = 250. Determine la probabilidad de que el volumen depositado sea menor a 1050 hojas y mayor a 980**

#
# El volumen de hojas que una oficina de abogados deposita
# en un juzgado sigue una distribución Normal con:
#
# μ = 1200 hojas
# σ = 250 hojas
#
# Calcular:
# a) P(X < 1050)
# b) P(X > 980)
# c) P(980 < X < 1050)
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# ----------------------------------------------------------
# Parámetros
# ----------------------------------------------------------
mu = 1200
sigma = 250

x1 = 980
x2 = 1050

# ----------------------------------------------------------
# Puntajes Z
# ----------------------------------------------------------
z1 = (x1 - mu) / sigma
z2 = (x2 - mu) / sigma

# ----------------------------------------------------------
# Probabilidades
# ----------------------------------------------------------
P_menor_1050 = norm.cdf(x2, mu, sigma)

P_mayor_980 = 1 - norm.cdf(x1, mu, sigma)

P_entre = norm.cdf(x2, mu, sigma) - norm.cdf(x1, mu, sigma)

# ----------------------------------------------------------
# Resultados
# ----------------------------------------------------------
print("="*60)
print("DISTRIBUCIÓN NORMAL")
print("="*60)

print(f"Media (μ) = {mu}")
print(f"Desviación estándar (σ) = {sigma}")

print("\nValores Z")
print(f"Z(980)  = {z1:.3f}")
print(f"Z(1050) = {z2:.3f}")

print("\nProbabilidades")
print(f"P(X < 1050)        = {P_menor_1050:.5f}")
print(f"P(X > 980)         = {P_mayor_980:.5f}")
print(f"P(980 < X < 1050)  = {P_entre:.5f}")

# ----------------------------------------------------------
# Curva Normal
# ----------------------------------------------------------
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, mu, sigma)

plt.figure(figsize=(11,6))

plt.plot(x, y,
         color="navy",
         linewidth=2.5,
         label="Distribución Normal")

# Región sombreada
xs = np.linspace(x1, x2, 500)
ys = norm.pdf(xs, mu, sigma)

plt.fill_between(xs,
                 ys,
                 color="skyblue",
                 alpha=0.7,
                 label=f"P(980 < X < 1050) = {P_entre:.5f}")

# Líneas verticales
plt.axvline(x1,
            color="red",
            linestyle="--",
            linewidth=2)

plt.axvline(x2,
            color="red",
            linestyle="--",
            linewidth=2)

# Etiquetas
plt.text(x1,
         norm.pdf(x1,mu,sigma)+0.00015,
         "980",
         color="red",
         ha="center",
         fontsize=11)

plt.text(x2,
         norm.pdf(x2,mu,sigma)+0.00015,
         "1050",
         color="red",
         ha="center",
         fontsize=11)

# Información del gráfico
plt.title("Distribución Normal del Volumen de Hojas", fontsize=15)
plt.xlabel("Número de hojas")
plt.ylabel("Densidad")
plt.grid(alpha=0.30)

# Caja de resultados
texto = (
    f"P(X < 1050) = {P_menor_1050:.5f}\n"
    f"P(X > 980) = {P_mayor_980:.5f}\n"
    f"P(980 < X < 1050) = {P_entre:.5f}"
)

plt.text(mu + 220,
         0.0011,
         texto,
         fontsize=11,
         bbox=dict(facecolor="white",
                   edgecolor="black",
                   alpha=0.85))

plt.legend()

plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 2. Distribución Normal**
#
# El volumen de hojas que una oficina de abogados deposita
# en un juzgado sigue una distribución Normal con:
#
# μ = 1200 hojas
# σ = 250 hojas
#
# Calcular:
# a) P(X < 1050)
# b) P(X > 980)
# c) P(980 < X < 1050)
# ==========================================================
%%R
library(ggplot2)

# ----------------------------------------------------------
# Parámetros
# ----------------------------------------------------------
mu <- 1200
sigma <- 250

x1 <- 980
x2 <- 1050

# ----------------------------------------------------------
# Puntajes Z
# ----------------------------------------------------------
z1 <- (x1 - mu) / sigma
z2 <- (x2 - mu) / sigma

# ----------------------------------------------------------
# Probabilidades
# ----------------------------------------------------------
P_menor_1050 <- pnorm(x2, mean = mu, sd = sigma)

P_mayor_980 <- 1 - pnorm(x1, mean = mu, sd = sigma)

P_entre <- pnorm(x2, mean = mu, sd = sigma) -
            pnorm(x1, mean = mu, sd = sigma)

# ----------------------------------------------------------
# Resultados
# ----------------------------------------------------------
cat("====================================================\n")
cat("         DISTRIBUCIÓN NORMAL\n")
cat("====================================================\n")

cat(sprintf("Media (μ) = %.0f\n", mu))
cat(sprintf("Desviación estándar (σ) = %.0f\n", sigma))

cat("\nValores Z\n")
cat(sprintf("Z(980)  = %.3f\n", z1))
cat(sprintf("Z(1050) = %.3f\n", z2))

cat("\nProbabilidades\n")
cat(sprintf("P(X < 1050)       = %.5f\n", P_menor_1050))
cat(sprintf("P(X > 980)        = %.5f\n", P_mayor_980))
cat(sprintf("P(980 < X <1050)  = %.5f\n", P_entre))

# ----------------------------------------------------------
# Datos para la curva normal
# ----------------------------------------------------------
x <- seq(mu - 4*sigma,
         mu + 4*sigma,
         length.out = 1000)

y <- dnorm(x,
           mean = mu,
           sd = sigma)

datos <- data.frame(
  x = x,
  y = y
)

# Región sombreada
xs <- seq(x1,
          x2,
          length.out = 300)

ys <- dnorm(xs,
            mean = mu,
            sd = sigma)

datos_sombra <- data.frame(
  x = xs,
  y = ys
)

# Texto con resultados
texto <- paste0(
  "P(X < 1050) = ", sprintf("%.5f", P_menor_1050),
  "\nP(X > 980) = ", sprintf("%.5f", P_mayor_980),
  "\nP(980 < X < 1050) = ", sprintf("%.5f", P_entre)
)

# ----------------------------------------------------------
# Gráfico
# ----------------------------------------------------------
ggplot(datos, aes(x, y)) +

  geom_line(color = "blue",
            linewidth = 1.3) +

  geom_area(data = datos_sombra,
            aes(x, y),
            fill = "skyblue",
            alpha = 0.6) +

  geom_vline(xintercept = x1,
             colour = "red",
             linetype = "dashed",
             linewidth = 1) +

  geom_vline(xintercept = x2,
             colour = "red",
             linetype = "dashed",
             linewidth = 1) +

  annotate("text",
           x = x1,
           y = dnorm(x1, mu, sigma) + 0.00015,
           label = "980",
           colour = "red",
           fontface = "bold") +

  annotate("text",
           x = x2,
           y = dnorm(x2, mu, sigma) + 0.00015,
           label = "1050",
           colour = "red",
           fontface = "bold") +

  annotate("label",
           x = mu + 350,
           y = 0.0012,
           label = texto,
           size = 4,
           fill = "white") +

  labs(
    title = "Distribución Normal del Volumen de Hojas",
    subtitle = expression(X %~% N(mu == 1200, sigma == 250)),
    x = "Número de hojas",
    y = "Densidad"
  ) +

  theme_minimal(base_size = 13) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

# **45.  Ejemplo 3. Distribución Normal.**

```python
# @title **A. Ejemplo 3. Una máquina despachadora de nieve está ajustada para servir un promedio de 30 ml por cono. Si la cantidad de nieve está normalmente distribuida con una varianza de 225 ml. ¿Cuál es la probabilidad de de que un cono contenga entre 25 ml y 35 ml?**

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parámetros de la distribución normal
media = 30
desviacion_estandar = np.sqrt(225)  # Raíz cuadrada de la varianza

# Calcular la probabilidad de que la cantidad de nieve esté entre 25 ml y 35 ml
z_25 = (25 - media) / desviacion_estandar
z_35 = (35 - media) / desviacion_estandar

probabilidad_entre_25_35 = norm.cdf(z_35) - norm.cdf(z_25)

# Crear la gráfica
x = np.linspace(media - 4 * desviacion_estandar, media + 4 * desviacion_estandar, 100)
y = norm.pdf(x, media, desviacion_estandar)

plt.plot(x, y)
plt.title('Distribución de la Cantidad de Nieve')
plt.xlabel('Cantidad de Nieve (ml)')
plt.ylabel('Probabilidad')
plt.grid(True)

# Sombrear la región entre 25 ml y 35 ml
x_sombreado = np.linspace(25, 35, 100)
y_sombreado = norm.pdf(x_sombreado, media, desviacion_estandar)
plt.fill_between(x_sombreado, y_sombreado, color='skyblue')

# Mostrar el valor del área
plt.text(25, 0.01, f'P(25 < X < 35): {probabilidad_entre_25_35:.4f}', fontsize=12)

plt.show()

print(f"La probabilidad de que un cono contenga entre 25 ml y 35 ml de nieve es: {probabilidad_entre_25_35:.4f}")
```

```python
# ==========================================================
# @title **B. Ejemplo 3. Distribución Normal**
#
# Una máquina despachadora de nieve está ajustada para servir
# un promedio de 30 ml por cono. La cantidad de nieve está
# normalmente distribuida con una varianza de 225 ml².
#
# ¿Cuál es la probabilidad de que un cono contenga
# entre 25 ml y 35 ml?
# ==========================================================
%%R
library(ggplot2)

# ----------------------------------------------------------
# Parámetros
# ----------------------------------------------------------
media <- 30
varianza <- 225
desviacion <- sqrt(varianza)

x1 <- 25
x2 <- 35

# ----------------------------------------------------------
# Puntajes Z
# ----------------------------------------------------------
z1 <- (x1 - media) / desviacion
z2 <- (x2 - media) / desviacion

# ----------------------------------------------------------
# Probabilidad
# ----------------------------------------------------------
prob_entre <- pnorm(x2,
                    mean = media,
                    sd = desviacion) -
              pnorm(x1,
                    mean = media,
                    sd = desviacion)

# ----------------------------------------------------------
# Mostrar resultados
# ----------------------------------------------------------
cat("====================================================\n")
cat("         DISTRIBUCIÓN NORMAL\n")
cat("====================================================\n")

cat(sprintf("Media (μ) = %.0f ml\n", media))
cat(sprintf("Varianza (σ²) = %.0f ml²\n", varianza))
cat(sprintf("Desviación estándar (σ) = %.2f ml\n\n", desviacion))

cat(sprintf("Z(25) = %.3f\n", z1))
cat(sprintf("Z(35) = %.3f\n\n", z2))

cat(sprintf("P(25 < X < 35) = %.5f\n", prob_entre))
cat(sprintf("P(25 < X < 35) = %.2f%%\n", prob_entre * 100))

# ----------------------------------------------------------
# Datos para la curva normal
# ----------------------------------------------------------
x <- seq(media - 4*desviacion,
         media + 4*desviacion,
         length.out = 1000)

y <- dnorm(x,
           mean = media,
           sd = desviacion)

datos <- data.frame(
  x = x,
  y = y
)

# Región sombreada
xs <- seq(x1,
          x2,
          length.out = 300)

ys <- dnorm(xs,
            mean = media,
            sd = desviacion)

datos_sombra <- data.frame(
  x = xs,
  y = ys
)

# Texto para el gráfico
texto <- paste0(
  "P(25 < X < 35)\n=",
  sprintf("%.5f", prob_entre),
  "\n(",
  sprintf("%.2f", prob_entre*100),
  "%)"
)

# ----------------------------------------------------------
# Gráfico
# ----------------------------------------------------------
ggplot(datos, aes(x, y)) +

  geom_line(color = "blue",
            linewidth = 1.3) +

  geom_area(data = datos_sombra,
            aes(x, y),
            fill = "skyblue",
            alpha = 0.65) +

  geom_vline(xintercept = x1,
             colour = "red",
             linetype = "dashed",
             linewidth = 1) +

  geom_vline(xintercept = x2,
             colour = "red",
             linetype = "dashed",
             linewidth = 1) +

  annotate("text",
           x = x1,
           y = dnorm(x1, media, desviacion) + 0.002,
           label = "25 ml",
           colour = "red",
           fontface = "bold") +

  annotate("text",
           x = x2,
           y = dnorm(x2, media, desviacion) + 0.002,
           label = "35 ml",
           colour = "red",
           fontface = "bold") +

  annotate("label",
           x = media + 18,
           y = 0.020,
           label = texto,
           fill = "white",
           size = 4) +

  labs(
    title = "Distribución Normal de la Cantidad de Nieve",
    subtitle = expression(X %~% N(mu == 30, sigma^2 == 225)),
    x = "Cantidad de nieve (ml)",
    y = "Densidad"
  ) +

  theme_minimal(base_size = 13) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

```python
# ==========================================================
# @title **B. Ejemplo 4. Distribución Normal**
#
# La cantidad real de café instantáneo que una máquina vierte
# en frascos de 4 onzas sigue una distribución Normal con:
#
# μ = 4.082 onzas
# σ = 0.040 onzas
#
# ¿Qué porcentaje de frascos contienen más de 4.090 onzas
# y son desechados por control de calidad?
# ==========================================================
%%R
library(ggplot2)

# ----------------------------------------------------------
# Parámetros
# ----------------------------------------------------------
media <- 4.082
desviacion <- 0.040

x0 <- 4.090

# ----------------------------------------------------------
# Puntaje Z
# ----------------------------------------------------------
z <- (x0 - media) / desviacion

# ----------------------------------------------------------
# Probabilidad
# ----------------------------------------------------------
prob_mayor <- 1 - pnorm(x0,
                        mean = media,
                        sd = desviacion)

# ----------------------------------------------------------
# Resultados
# ----------------------------------------------------------
cat("=====================================================\n")
cat("          DISTRIBUCIÓN NORMAL\n")
cat("=====================================================\n")

cat(sprintf("Media (μ) = %.3f onzas\n", media))
cat(sprintf("Desviación estándar (σ) = %.3f onzas\n\n", desviacion))

cat(sprintf("Z(4.090) = %.3f\n\n", z))

cat(sprintf("P(X > %.3f) = %.5f\n", x0, prob_mayor))
cat(sprintf("Porcentaje = %.2f%%\n", prob_mayor * 100))

# ----------------------------------------------------------
# Datos para la curva normal
# ----------------------------------------------------------
x <- seq(media - 4*desviacion,
         media + 4*desviacion,
         length.out = 1000)

y <- dnorm(x,
           mean = media,
           sd = desviacion)

datos <- data.frame(
  x = x,
  y = y
)

# Región sombreada
xs <- seq(x0,
          max(x),
          length.out = 300)

ys <- dnorm(xs,
            mean = media,
            sd = desviacion)

datos_sombra <- data.frame(
  x = xs,
  y = ys
)

texto <- paste0(
  "P(X > 4.090)\n=",
  sprintf("%.5f", prob_mayor),
  "\n(",
  sprintf("%.2f%%", prob_mayor*100),
  ")"
)

# ----------------------------------------------------------
# Gráfico
# ----------------------------------------------------------
ggplot(datos, aes(x, y)) +

  geom_line(color = "blue",
            linewidth = 1.3) +

  geom_area(data = datos_sombra,
            aes(x, y),
            fill = "red",
            alpha = 0.65) +

  geom_vline(xintercept = x0,
             colour = "red",
             linetype = "dashed",
             linewidth = 1) +

  annotate("text",
           x = x0,
           y = dnorm(x0, media, desviacion) + 1,
           label = "4.090",
           colour = "red",
           fontface = "bold") +

  annotate("label",
           x = media + 0.10,
           y = max(y)*0.90,
           label = texto,
           fill = "white",
           size = 4) +

  labs(
    title = "Distribución Normal del Volumen de Café Instantáneo",
    subtitle = expression(X %~% N(mu == 4.082, sigma == 0.040)),
    x = "Cantidad de café (onzas)",
    y = "Densidad"
  ) +

  theme_minimal(base_size = 13) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

# **47.  Ejemplo 5. Distribución Normal.**

```python
# @title **A. Ejemplo 5. Se sabe que la prueba de auto estima sigue una distribución normal con una media de 60, con una desviación estándar de 5.83. Si se aplica una prueba de autoestima a 25 personas quienes obtienen una calificación promedio de 62.1 ¿Cuál es la probabilidad de que la media muestral sea no mas que 62.1?**

# Parámetros de la distribución normal
media_poblacional = 60
desviacion_estandar_poblacional = 5.83
n = 25  # Tamaño de la muestra
media_muestral = 62.1

# Calcular la desviación estándar de la media muestral
desviacion_estandar_muestral = desviacion_estandar_poblacional / (n ** 0.5)

# Calcular la probabilidad de que la media muestral sea menor o igual a 62.1
probabilidad = norm.cdf(media_muestral, media_poblacional, desviacion_estandar_muestral)

print(f"La probabilidad de que la media muestral sea menor o igual a 62.1 es: {probabilidad:.4f}")


# Generar datos para la gráfica
x = np.linspace(media_poblacional - 4 * desviacion_estandar_muestral, media_poblacional + 4 * desviacion_estandar_muestral, 100)
y = norm.pdf(x, media_poblacional, desviacion_estandar_muestral)

# Crear la figura y los ejes
fig, ax = plt.subplots(figsize=(10, 6))

# Dibujar la curva normal
ax.plot(x, y)

# Sombrear el área menor o igual a 62.1
ax.fill_between(x, y, where=(x <= media_muestral), color='blue', alpha=0.5)

# Configurar la gráfica
ax.set_title('Distribución Normal de la Media Muestral')
ax.set_xlabel('Media Muestral')
ax.set_ylabel('Probabilidad')
ax.grid(True)

# Mostrar el porcentaje en la gráfica
ax.text(58, 0.05, f'Probabilidad: {probabilidad * 100:.2f}%', fontsize=12)


# Mostrar la gráfica
plt.show()
```

```python
# ==========================================================
# @title **B. Ejemplo 5. Distribución de la Media Muestral**
#
# Se sabe que la prueba de autoestima sigue una distribución
# normal con:
#
# μ = 60
# σ = 5.83
#
# Se selecciona una muestra de n = 25 personas y se obtiene
# una media muestral de 62.1.
#
# ¿Cuál es la probabilidad de que la media muestral sea
# menor o igual que 62.1?
# ==========================================================
%%R
library(ggplot2)

#-----------------------------------------------------------
# Parámetros
#-----------------------------------------------------------
mu <- 60
sigma <- 5.83
n <- 25
xbarra <- 62.1

#-----------------------------------------------------------
# Distribución de la media muestral
#-----------------------------------------------------------
sigma_xbarra <- sigma / sqrt(n)

# Puntaje Z
z <- (xbarra - mu) / sigma_xbarra

# Probabilidad
probabilidad <- pnorm(xbarra,
                      mean = mu,
                      sd = sigma_xbarra)

#-----------------------------------------------------------
# Resultados
#-----------------------------------------------------------
cat("=====================================================\n")
cat("      DISTRIBUCIÓN DE LA MEDIA MUESTRAL\n")
cat("=====================================================\n")

cat(sprintf("Media poblacional (μ) = %.2f\n", mu))
cat(sprintf("Desviación estándar poblacional (σ) = %.2f\n", sigma))
cat(sprintf("Tamaño de la muestra (n) = %d\n", n))
cat(sprintf("Error estándar = %.3f\n\n", sigma_xbarra))

cat(sprintf("Media muestral = %.2f\n", xbarra))
cat(sprintf("Valor Z = %.3f\n\n", z))

cat(sprintf("P(X̄ ≤ %.2f) = %.5f\n", xbarra, probabilidad))
cat(sprintf("Probabilidad = %.2f%%\n", probabilidad*100))

#-----------------------------------------------------------
# Datos para la curva
#-----------------------------------------------------------
x <- seq(mu - 4*sigma_xbarra,
         mu + 4*sigma_xbarra,
         length.out = 1000)

y <- dnorm(x,
           mean = mu,
           sd = sigma_xbarra)

datos <- data.frame(
  x = x,
  y = y
)

# Área sombreada
xs <- seq(min(x),
          xbarra,
          length.out = 400)

ys <- dnorm(xs,
            mean = mu,
            sd = sigma_xbarra)

datos_sombra <- data.frame(
  x = xs,
  y = ys
)

texto <- paste0(
  "P(X̄ ≤ ", xbarra, ") = ",
  sprintf("%.5f", probabilidad),
  "\n(",
  sprintf("%.2f%%", probabilidad*100),
  ")"
)

#-----------------------------------------------------------
# Gráfico
#-----------------------------------------------------------
ggplot(datos, aes(x, y)) +

  geom_line(color = "blue",
            linewidth = 1.3) +

  geom_area(data = datos_sombra,
            aes(x, y),
            fill = "steelblue",
            alpha = 0.60) +

  geom_vline(xintercept = xbarra,
             colour = "red",
             linetype = "dashed",
             linewidth = 1) +

  annotate("text",
           x = xbarra,
           y = dnorm(xbarra, mu, sigma_xbarra) + 0.02,
           label = "62.1",
           colour = "red",
           fontface = "bold",
           size = 5) +

  annotate("label",
           x = mu + 3.5,
           y = max(y)*0.90,
           label = texto,
           size = 4,
           fill = "white") +

  labs(
    title = "Distribución Normal de la Media Muestral",
    subtitle = expression(bar(X) %~% N(mu == 60,
                                       sigma[bar(X)] == frac(5.83,sqrt(25)))),
    x = expression(bar(X)),
    y = "Densidad"
  ) +

  theme_minimal(base_size = 13) +

  theme(
    plot.title = element_text(face = "bold",
                              hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

# **48.  CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN NORMAL.**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN NORMAL**
#
# La distribución Normal (Gaussiana) es una de las distribuciones
# más importantes en estadística. Se caracteriza por su forma
# de campana y está definida por dos parámetros:
#   μ (media) - centro de la distribución
#   σ (desviación estándar) - dispersión de los datos
#
# Fórmula: f(x) = (1/(σ√(2π))) * e^(-(x-μ)²/(2σ²))
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo (corregida)
#==========================================================
# Verificar estilos disponibles y usar uno que exista
try:
    # Intentar con estilo moderno
    plt.style.use('seaborn-v0_8')
except:
    try:
        # Si no existe, intentar con seaborn
        plt.style.use('seaborn')
    except:
        # Si no hay seaborn, usar estilo por defecto
        plt.style.use('default')
        print("ℹ️ Usando estilo por defecto de matplotlib")

plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_normal():
    """
    Calculadora interactiva para la distribución Normal
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN NORMAL")
    print("="*70)
    print("\nLa distribución Normal (campana de Gauss) es la distribución")
    print("de probabilidad continua más importante en estadística.")
    print("\nEjemplos típicos:")
    print("  • Altura y peso de una población")
    print("  • Puntajes en exámenes estandarizados")
    print("  • Errores de medición en experimentos")
    print("  • Características físicas y biológicas\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            mu = float(input("  Media (μ): "))
            sigma = float(input("  Desviación estándar (σ > 0): "))
            if sigma <= 0:
                print("  ❌ σ debe ser mayor que 0. Intenta de nuevo.")
                continue

            print("\nTipo de cálculo:")
            print("  1. P(X = a)  - Probabilidad puntual")
            print("  2. P(X ≤ a)  - Probabilidad acumulada a la izquierda")
            print("  3. P(X ≥ a)  - Probabilidad acumulada a la derecha")
            print("  4. P(a ≤ X ≤ b) - Probabilidad entre dos valores")
            print("  5. Cuantil - Encontrar X dado un percentil")

            tipo = input("\nSelecciona el tipo de cálculo (1-5): ")

            if tipo == '1':
                a = float(input("  Valor de a: "))
                prob = 0
                visualizar_normal(mu, sigma, a, None, prob, tipo)
                return {'mu': mu, 'sigma': sigma, 'a': a, 'prob': prob}

            elif tipo == '2':
                a = float(input("  Valor de a: "))
                prob = norm.cdf(a, mu, sigma)
                z = (a - mu) / sigma

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≤ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Z = ({a} - {mu}) / {sigma} = {z:.4f}")
                print(f"P(Z ≤ {z:.4f}) = {prob:.6f}")

                visualizar_normal(mu, sigma, a, None, prob, tipo)
                return {'mu': mu, 'sigma': sigma, 'a': a, 'prob': prob}

            elif tipo == '3':
                a = float(input("  Valor de a: "))
                prob = 1 - norm.cdf(a, mu, sigma)
                z = (a - mu) / sigma

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≥ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Z = ({a} - {mu}) / {sigma} = {z:.4f}")
                print(f"P(Z ≥ {z:.4f}) = {prob:.6f}")

                visualizar_normal(mu, sigma, a, None, prob, tipo)
                return {'mu': mu, 'sigma': sigma, 'a': a, 'prob': prob}

            elif tipo == '4':
                a = float(input("  Límite inferior (a): "))
                b = float(input("  Límite superior (b): "))
                if a > b:
                    a, b = b, a
                    print("  ↳ Los límites se han intercambiado para que a < b")
                prob = norm.cdf(b, mu, sigma) - norm.cdf(a, mu, sigma)
                z_a = (a - mu) / sigma
                z_b = (b - mu) / sigma

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP({a} ≤ X ≤ {b}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Z₁ = ({a} - {mu}) / {sigma} = {z_a:.4f}")
                print(f"Z₂ = ({b} - {mu}) / {sigma} = {z_b:.4f}")
                print(f"P({z_a:.4f} ≤ Z ≤ {z_b:.4f}) = {prob:.6f}")

                visualizar_normal(mu, sigma, a, b, prob, tipo)
                return {'mu': mu, 'sigma': sigma, 'a': a, 'b': b, 'prob': prob}

            elif tipo == '5':
                p = float(input("  Percentil (0-1): "))
                if p <= 0 or p >= 1:
                    print("  ❌ El percentil debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                x = norm.ppf(p, mu, sigma)
                z = (x - mu) / sigma

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nPara el percentil {p*100:.1f}%:")
                print(f"X = {x:.6f}")
                print(f"Z = {z:.6f}")
                print(f"Interpretación: El {p*100:.1f}% de los datos están por debajo de {x:.4f}")

                visualizar_cuantil(mu, sigma, x, p)
                return {'mu': mu, 'sigma': sigma, 'percentil': p, 'x': x}

            else:
                print("  ❌ Opción no válida. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

#==========================================================
# FUNCIÓN PARA VISUALIZAR LA DISTRIBUCIÓN NORMAL
#==========================================================

def visualizar_normal(mu, sigma, a, b, prob, tipo):
    """Crea visualizaciones completas de la distribución Normal"""

    # Rango de valores para la gráfica
    x_min = mu - 4*sigma
    x_max = mu + 4*sigma
    x = np.linspace(x_min, x_max, 1000)
    y = norm.pdf(x, mu, sigma)

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    #==========================================================
    # GRÁFICO 1: Distribución completa con área sombreada
    #==========================================================
    ax1 = plt.subplot(2, 2, 1)
    ax1.plot(x, y, 'b-', linewidth=2.5, label='Distribución Normal')

    # Sombrear el área según el tipo de cálculo
    if tipo == '2':  # P(X ≤ a)
        mask = x <= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≤ {a})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a}')

    elif tipo == '3':  # P(X ≥ a)
        mask = x >= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≥ {a})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a}')

    elif tipo == '4':  # P(a ≤ X ≤ b)
        mask = (x >= a) & (x <= b)
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P({a} ≤ X ≤ {b})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)
        ax1.axvline(x=b, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    # Líneas de referencia
    ax1.axvline(x=mu, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {mu}')
    ax1.axvline(x=mu + sigma, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)
    ax1.axvline(x=mu - sigma, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Normal\nμ={mu:.2f}, σ={sigma:.2f}', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', fontsize=10)

    # Agregar información de probabilidad
    if prob is not None:
        ax1.text(0.98, 0.95, f'Probabilidad = {prob:.6f}\n({prob*100:.2f}%)',
                transform=ax1.transAxes, fontsize=11, verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 2: Función de densidad con áreas en la cola
    #==========================================================
    ax2 = plt.subplot(2, 2, 2)
    ax2.plot(x, y, 'b-', linewidth=2.5)

    # Áreas en las colas (más allá de ±2σ)
    mask_cola_izq = x <= mu - 2*sigma
    mask_cola_der = x >= mu + 2*sigma
    ax2.fill_between(x[mask_cola_izq], 0, y[mask_cola_izq], color='#FF6B6B', alpha=0.3, label='Cola izquierda (2σ)')
    ax2.fill_between(x[mask_cola_der], 0, y[mask_cola_der], color='#FF6B6B', alpha=0.3, label='Cola derecha (2σ)')

    # Líneas en ±2σ y ±3σ
    for z, color in [(-2, '#FF6B6B'), (2, '#FF6B6B'), (-3, '#FFB347'), (3, '#FFB347')]:
        x_val = mu + z*sigma
        ax2.axvline(x=x_val, color=color, linestyle='--', linewidth=1.5, alpha=0.5)
        ax2.text(x_val, 0.01, f'{z}σ', ha='center', fontsize=9, color=color)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax2.set_title('Áreas en las colas (Regla empírica)', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best', fontsize=10)

    # Agregar información de la regla empírica
    info_text = (
        f"Regla empírica:\n"
        f"μ ± 1σ: 68.27%\n"
        f"μ ± 2σ: 95.45%\n"
        f"μ ± 3σ: 99.73%"
    )
    ax2.text(0.02, 0.98, info_text, transform=ax2.transAxes, fontsize=10,
            verticalalignment='top', bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 3: Función de distribución acumulada
    #==========================================================
    ax3 = plt.subplot(2, 2, 3)
    cdf = norm.cdf(x, mu, sigma)
    ax3.plot(x, cdf, 'g-', linewidth=2.5, label='CDF')
    ax3.fill_between(x, 0, cdf, alpha=0.2, color='green')

    # Puntos de interés
    if tipo == '2':  # P(X ≤ a)
        ax3.scatter(a, prob, color='#E63946', s=100, zorder=5, label=f'F({a}) = {prob:.4f}')
        ax3.axhline(y=prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '3':  # P(X ≥ a)
        ax3.scatter(a, 1-prob, color='#E63946', s=100, zorder=5, label=f'F({a}) = {1-prob:.4f}')
        ax3.axhline(y=1-prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '4':  # P(a ≤ X ≤ b)
        ax3.scatter(a, norm.cdf(a, mu, sigma), color='#E63946', s=100, zorder=5, label=f'F({a})')
        ax3.scatter(b, norm.cdf(b, mu, sigma), color='#E63946', s=100, zorder=5, label=f'F({b})')
        ax3.axhline(y=norm.cdf(a, mu, sigma), color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axhline(y=norm.cdf(b, mu, sigma), color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuartiles
    for q in [0.25, 0.5, 0.75]:
        q_val = norm.ppf(q, mu, sigma)
        ax3.axvline(x=q_val, color='orange', linestyle=':', linewidth=1.5, alpha=0.5)
        ax3.text(q_val, 0.01, f'Q{int(q*100)}', ha='center', fontsize=9, color='orange')

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax3.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', fontsize=10)
    ax3.set_ylim(-0.05, 1.05)

    #==========================================================
    # GRÁFICO 4: Tabla de valores Z y percentiles
    #==========================================================
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Crear tabla de valores Z comunes
    z_values = [-3, -2.5, -2, -1.5, -1, -0.5, 0, 0.5, 1, 1.5, 2, 2.5, 3]
    percentiles = [norm.cdf(z) for z in z_values]
    x_values = [mu + z*sigma for z in z_values]

    # Mostrar tabla
    table_data = []
    table_data.append(['Z', 'Percentil', 'X'])
    for z, p, x_val in zip(z_values, percentiles, x_values):
        table_data.append([f'{z:.1f}', f'{p*100:.2f}%', f'{x_val:.2f}'])

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.2, 0.25, 0.3])

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.5)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            elif i in [5, 6, 7]:  # Valores centrales
                cell.set_facecolor('#E8F4F8')
                cell.set_text_props(weight='bold')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title('Tabla de Valores Z Comunes', fontsize=13, fontweight='bold', pad=20)

    #==========================================================
    # Ajustar layout
    #==========================================================
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # GRÁFICO ADICIONAL: Distribución en diferentes σ
    #==========================================================
    fig2, ax5 = plt.subplots(figsize=(12, 6))

    # Mostrar cómo cambia la distribución con diferentes σ
    sigmas = [0.5*sigma, sigma, 1.5*sigma, 2*sigma]
    colores = ['#2E86AB', '#E63946', '#F39C12', '#2ECC71']

    for s, color in zip(sigmas, colores):
        y_temp = norm.pdf(x, mu, s)
        ax5.plot(x, y_temp, color=color, linewidth=2,
                label=f'σ = {s:.1f}')

    ax5.axvline(x=mu, color='#2C3E50', linestyle='--', linewidth=2, alpha=0.7)
    ax5.set_xlabel('X', fontsize=12)
    ax5.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax5.set_title(f'Efecto de σ en la distribución Normal (μ={mu})',
                 fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.legend(loc='best')

    plt.tight_layout()

    #==========================================================
    # MOSTRAR GRÁFICOS
    #==========================================================
    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Normal_mu{mu:.2f}_sigma{sigma:.2f}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_comparacion_sigmas.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_comparacion_sigmas.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

def visualizar_cuantil(mu, sigma, x, p):
    """Visualiza el cuantil en la distribución Normal"""

    # Rango de valores
    x_min = mu - 4*sigma
    x_max = mu + 4*sigma
    x_vals = np.linspace(x_min, x_max, 1000)
    y = norm.pdf(x_vals, mu, sigma)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Distribución con cuantil marcado
    ax1.plot(x_vals, y, 'b-', linewidth=2.5)
    mask = x_vals <= x
    ax1.fill_between(x_vals[mask], 0, y[mask], color='#E63946', alpha=0.5,
                     label=f'Área = {p*100:.1f}%')
    ax1.axvline(x=x, color='#E63946', linestyle='--', linewidth=2, label=f'X = {x:.2f}')
    ax1.axvline(x=mu, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {mu}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Percentil {p*100:.1f}%', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best')

    # Gráfico 2: Función de distribución acumulada
    cdf = norm.cdf(x_vals, mu, sigma)
    ax2.plot(x_vals, cdf, 'g-', linewidth=2.5)
    ax2.scatter(x, p, color='#E63946', s=150, zorder=5, label=f'F({x:.2f}) = {p:.4f}')
    ax2.axhline(y=p, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax2.axvline(x=x, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax2.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best')
    ax2.set_ylim(-0.05, 1.05)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_normal():
    """
    Ejecuta ejemplos clásicos de la distribución Normal
    """

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN NORMAL")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Altura de estudiantes',
            'descripcion': 'La altura de los estudiantes de una universidad sigue una distribución Normal con μ=170 cm y σ=10 cm. ¿Cuál es la probabilidad de que un estudiante mida menos de 180 cm?',
            'mu': 170,
            'sigma': 10,
            'tipo': 2,
            'a': 180
        },
        {
            'titulo': 'Ejemplo 2: Puntajes de examen',
            'descripcion': 'Los puntajes de un examen siguen una distribución Normal con μ=75 y σ=12. ¿Cuál es la probabilidad de obtener un puntaje entre 80 y 90?',
            'mu': 75,
            'sigma': 12,
            'tipo': 4,
            'a': 80,
            'b': 90
        },
        {
            'titulo': 'Ejemplo 3: Tiempo de espera',
            'descripcion': 'El tiempo de espera en un banco sigue una distribución Normal con μ=8 minutos y σ=2.5 minutos. ¿Cuál es la probabilidad de esperar más de 10 minutos?',
            'mu': 8,
            'sigma': 2.5,
            'tipo': 3,
            'a': 10
        },
        {
            'titulo': 'Ejemplo 4: Peso de productos',
            'descripcion': 'El peso de un producto sigue una distribución Normal con μ=500g y σ=20g. ¿Cuál es el peso que supera el 95% de los productos?',
            'mu': 500,
            'sigma': 20,
            'tipo': 5,
            'p': 0.95
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")

        if ej['tipo'] == 2:
            prob = norm.cdf(ej['a'], ej['mu'], ej['sigma'])
            print(f"   P(X ≤ {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")
        elif ej['tipo'] == 3:
            prob = 1 - norm.cdf(ej['a'], ej['mu'], ej['sigma'])
            print(f"   P(X ≥ {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")
        elif ej['tipo'] == 4:
            prob = norm.cdf(ej['b'], ej['mu'], ej['sigma']) - norm.cdf(ej['a'], ej['mu'], ej['sigma'])
            print(f"   P({ej['a']} ≤ X ≤ {ej['b']}) = {prob:.6f} ({prob*100:.2f}%)")
        elif ej['tipo'] == 5:
            x = norm.ppf(ej['p'], ej['mu'], ej['sigma'])
            print(f"   Percentil {ej['p']*100:.0f}% = {x:.4f}")

    print("\n" + "="*70)
    print("¿Deseas analizar alguno de estos ejemplos en detalle?")
    print("Selecciona 1, 2, 3, 4 o ejecuta la calculadora principal.")

#==========================================================
# FUNCIÓN PARA VALIDAR NORMALIDAD
#==========================================================

def validar_normalidad():
    """
    Herramientas para validar la normalidad de datos
    """

    print("\n" + "="*70)
    print("🔍 VALIDACIÓN DE NORMALIDAD")
    print("="*70)
    print("\nPruebas para verificar si los datos siguen una distribución Normal:")
    print("\n1. Prueba de Shapiro-Wilk")
    print("   • H₀: Los datos son normales")
    print("   • H₁: Los datos no son normales")
    print("   • Si p-value > 0.05, no se rechaza normalidad")

    print("\n2. Prueba de Kolmogorov-Smirnov")
    print("   • Compara la distribución empírica con la Normal")
    print("   • Se usa para muestras grandes")

    print("\n3. Gráfico Q-Q (Quantile-Quantile)")
    print("   • Compara cuantiles observados vs esperados")
    print("   • Si los puntos están en línea recta → normalidad")

    print("\n4. Histograma con curva Normal")
    print("   • Superposición visual de datos y distribución teórica")

    print("\n5. Coeficientes de asimetría y curtosis")
    print("   • Asimetría ≈ 0 → distribución simétrica")
    print("   • Curtosis ≈ 3 → distribución mesocúrtica (Normal)")

    print("\n" + "="*70)
    print("📊 EJEMPLO DE ANÁLISIS")
    print("="*70)

    try:
        # Generar datos de ejemplo
        np.random.seed(42)
        datos = np.random.normal(100, 15, 100)

        # Prueba de Shapiro-Wilk
        stat, p_value = stats.shapiro(datos)
        print(f"\nPrueba de Shapiro-Wilk:")
        print(f"  Estadístico = {stat:.6f}")
        print(f"  p-value = {p_value:.6f}")
        if p_value > 0.05:
            print("  ✅ No se rechaza normalidad (p-value > 0.05)")
        else:
            print("  ❌ Se rechaza normalidad (p-value < 0.05)")

        # Estadísticas
        print(f"\nEstadísticas descriptivas:")
        print(f"  Media = {np.mean(datos):.4f}")
        print(f"  Desviación = {np.std(datos):.4f}")
        print(f"  Asimetría = {stats.skew(datos):.4f}")
        print(f"  Curtosis = {stats.kurtosis(datos):.4f} (exceso)")

        # Crear gráfico Q-Q
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

        # Q-Q plot
        stats.probplot(datos, dist="norm", plot=ax1)
        ax1.set_title('Gráfico Q-Q de Normalidad', fontsize=12, fontweight='bold')
        ax1.grid(True, alpha=0.3)

        # Histograma con curva Normal
        ax2.hist(datos, bins=20, density=True, alpha=0.6, color='#2E86AB', edgecolor='black')
        x = np.linspace(min(datos), max(datos), 100)
        y = norm.pdf(x, np.mean(datos), np.std(datos))
        ax2.plot(x, y, 'r-', linewidth=2, label='Normal teórica')
        ax2.set_xlabel('Datos')
        ax2.set_ylabel('Densidad')
        ax2.set_title('Histograma vs Curva Normal', fontsize=12, fontweight='bold')
        ax2.grid(True, alpha=0.3)
        ax2.legend()

        plt.tight_layout()
        plt.show()

        print("\n✅ Análisis de normalidad completado")

    except Exception as e:
        print(f"❌ Error en el análisis: {e}")

#==========================================================
# FUNCIÓN PARA INTERVALOS DE CONFIANZA
#==========================================================

def intervalos_confianza_normal():
    """
    Calcula intervalos de confianza para la media usando distribución Normal
    """

    print("\n" + "="*70)
    print("📈 INTERVALOS DE CONFIANZA PARA LA MEDIA")
    print("="*70)
    print("\nEsta función calcula intervalos de confianza para la media")
    print("de una población Normal con varianza conocida.\n")

    try:
        print("Ingresa los datos:")
        mu = float(input("  Media muestral (x̄): "))
        sigma = float(input("  Desviación estándar poblacional (σ): "))
        n = int(input("  Tamaño de muestra (n): "))

        nivel = float(input("  Nivel de confianza (ej: 95): ")) / 100

        z = norm.ppf(1 - (1 - nivel) / 2)
        error = z * sigma / np.sqrt(n)

        li = mu - error
        ls = mu + error

        print(f"\nResultados:")
        print(f"  • Error estándar = {sigma/np.sqrt(n):.4f}")
        print(f"  • Valor crítico z = {z:.4f}")
        print(f"  • Margen de error = {error:.4f}")
        print(f"  • Intervalo de confianza al {nivel*100:.0f}%: ({li:.4f}, {ls:.4f})")
        print(f"  • Interpretación: Con un {nivel*100:.0f}% de confianza, la media poblacional")
        print(f"    se encuentra entre {li:.4f} y {ls:.4f}")

        # Visualización
        x = np.linspace(mu - 4*error, mu + 4*error, 1000)
        y = norm.pdf(x, mu, error)

        fig, ax = plt.subplots(figsize=(10, 5))
        ax.plot(x, y, 'b-', linewidth=2)

        # Área del intervalo de confianza
        x_fill = np.linspace(li, ls, 100)
        y_fill = norm.pdf(x_fill, mu, error)
        ax.fill_between(x_fill, 0, y_fill, color='#E63946', alpha=0.5)

        ax.axvline(x=li, color='#E63946', linestyle='--', linewidth=2)
        ax.axvline(x=ls, color='#E63946', linestyle='--', linewidth=2)
        ax.axvline(x=mu, color='#2C3E50', linestyle=':', linewidth=2)

        ax.set_xlabel('Media muestral', fontsize=12)
        ax.set_ylabel('Densidad de probabilidad', fontsize=12)
        ax.set_title(f'Intervalo de Confianza al {nivel*100:.0f}%',
                    fontsize=13, fontweight='bold')
        ax.grid(True, alpha=0.3, linestyle='--')

        ax.text(0.98, 0.95, f'IC {nivel*100:.0f}%\n({li:.4f}, {ls:.4f})',
               transform=ax.transAxes, fontsize=11, verticalalignment='top',
               horizontalalignment='right',
               bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN NORMAL")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado de una distribución")
        print("  2. Ver ejemplos clásicos")
        print("  3. Validar normalidad de datos")
        print("  4. Calcular intervalos de confianza")
        print("  5. Salir")

        opcion = input("\nSelecciona una opción (1-5): ")

        if opcion == '1':
            resultados = calculadora_normal()
        elif opcion == '2':
            ejemplos_normal()
        elif opcion == '3':
            validar_normalidad()
        elif opcion == '4':
            intervalos_confianza_normal()
        elif opcion == '5':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3, 4 o 5.")
```

```python
# ==========================================================
# @title **B. Ejemplo 6. Distribución de la Media Muestral**
#
# Se sabe que la prueba de autoestima sigue una distribución
# normal con media μ = 60 y desviación estándar σ = 5.83.
# Si se aplica la prueba a una muestra de n = 25 personas,
# ¿cuál es la probabilidad de que la media muestral esté
# entre 59 y 61?
# ==========================================================
%%R
library(ggplot2)

# Parámetros
media_poblacional <- 60
desviacion_poblacional <- 5.83
n <- 25

# Desviación estándar de la media muestral
error_estandar <- desviacion_poblacional / sqrt(n)

# Intervalo solicitado
a <- 59
b <- 61

# Probabilidad
probabilidad <- pnorm(b,
                      mean = media_poblacional,
                      sd = error_estandar) -
                pnorm(a,
                      mean = media_poblacional,
                      sd = error_estandar)

cat("=============================================\n")
cat("DISTRIBUCIÓN NORMAL DE LA MEDIA MUESTRAL\n")
cat("=============================================\n")
cat(sprintf("Media poblacional (μ)              = %.2f\n", media_poblacional))
cat(sprintf("Desviación estándar poblacional    = %.2f\n", desviacion_poblacional))
cat(sprintf("Tamaño de la muestra (n)           = %d\n", n))
cat(sprintf("Error estándar                     = %.4f\n", error_estandar))
cat("---------------------------------------------\n")
cat(sprintf("P(59 < X̄ < 61) = %.5f\n", probabilidad))
cat(sprintf("P(59 < X̄ < 61) = %.2f%%\n", probabilidad*100))
cat("=============================================\n")

# Datos para la curva
x <- seq(media_poblacional - 4*error_estandar,
         media_poblacional + 4*error_estandar,
         length.out = 1000)

y <- dnorm(x,
           mean = media_poblacional,
           sd = error_estandar)

datos <- data.frame(x, y)

# Región sombreada
datos_sombreado <- subset(datos,
                          x >= a & x <= b)

# Gráfico
ggplot(datos, aes(x, y)) +

  geom_line(color = "blue",
            linewidth = 1.3) +

  geom_area(data = datos_sombreado,
            aes(y = y),
            fill = "forestgreen",
            alpha = 0.5) +

  geom_vline(xintercept = c(a, b),
             colour = "red",
             linetype = "dashed",
             linewidth = 1) +

  annotate("text",
           x = media_poblacional,
           y = max(y)*0.92,
           label = paste0(
             "P(59 < X̄ < 61) = ",
             sprintf("%.4f", probabilidad)
           ),
           colour = "darkgreen",
           size = 5,
           fontface = "bold") +

  labs(
    title = "Distribución Normal de la Media Muestral",
    subtitle = paste0(
      "μ = ", media_poblacional,
      ", σ = ", desviacion_poblacional,
      ", n = ", n
    ),
    x = "Media muestral (X̄)",
    y = "Densidad"
  ) +

  theme_minimal(base_size = 14) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    panel.grid.minor = element_blank()
  )
```

```python
# @title
#**a. Construya el gráfico de la Distribución dada si los parámetros son μ=85,σ=3.8**
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from scipy import stats
plt.figure(figsize=(5, 5))
#x = np.linspace(0, 20, 200)
media = 85 # parametro
sigma = 3.8 # parametro
normal = stats.norm(media,sigma )
x = np.linspace(normal.ppf(0.0001),
                normal.ppf(0.9999), 100)
fp = normal.pdf(x) # Función de Probabilidad
plt.plot(x, fp)
plt.title('Distribución Normal')
plt.ylabel('Probabilidad')
plt.xlabel('Duración')
plt.show()
```

## **Animacion de la distribución normal**

```python
# @title
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from scipy import stats
from IPython.display import HTML

# Parámetros de la distribución
mu, sd = 85, 3.8
x = np.linspace(70, 100, 450)
y = stats.norm(mu, sd).pdf(x)

# Valores objetivo
a = 70
b = 100
c = stats.norm(mu, sd).cdf(b) - stats.norm(mu, sd).cdf(a) # Probabilidad pedida

# Crear la figura y el eje
fig, ax = plt.subplots(figsize=[6, 4])
ax.set_xlabel('Valores de x')
ax.set_ylabel('Densidad')
ax.set_title(f'P{{70 <= X <= 100}} = {c:.4f}, D. Normal con $μ=85,σ=3.8$')

# Inicialización de las líneas y áreas
line, = ax.plot(x, y, c='g')
fill = ax.fill_between([], [], color="r")
text_cdf = ax.text(87.25, 0.04, '', horizontalalignment='center')

# Función de inicialización para la animación
def init():
    line.set_data(x, y)
    return line,

# Número de cuadros necesarios para alcanzar el valor objetivo
num_frames = 100
incremento = (b - a) / num_frames

# Función de animación
def animate(i):
    global fill
    fill.remove()
    x1 = np.linspace(a, a + i * incremento, 401)
    fill = ax.fill_between(x1, 0, stats.norm(mu, sd).pdf(x1), color="r")
    probabilidad_actual = stats.norm(mu, sd).cdf(a + i * incremento) - stats.norm(mu, sd).cdf(a-0.25)
    text_cdf.set_text(f'{probabilidad_actual:.4f}')
    return line, fill, text_cdf

# Crear la animación
ani = animation.FuncAnimation(fig, animate, init_func=init, frames=num_frames, interval=100, blit=True)

# Mostrar la animación
HTML(ani.to_html5_video())
```

#**k. ¿Cuál es la probabilidad de que una batería dure almenos dos desviaciones estandard de su media?**
+ $μ=85,σ=3.8$
+ $$P\{|X-μ|\geq 2σ\}=P\{X-μ\geq 2σ\}+P\{-2σ\geq(X-μ)\} \ \
=P\{ X\geq μ+2σ\}+P\{X\leq μ-2σ\}=P\{ X\geq 92.6\}+P\{X\leq 77.4\}$$

```python
# @title
#**k. ¿Cuál es la probabilidad de que una batería dure más a lo más dos desviaciones estandard de su media?**
fig, ax = plt.subplots(figsize=[6, 4])
x = np.linspace(70, 100, 401)
mu, sd = 85, 3.8
y = stats.norm(mu, sd).pdf(x)
a = 77.4 # Valor de la variable a calcular a la izquierda
b = 92.6 # Valor de la variable a calcular a la izquierda
c = stats.norm(mu, sd).cdf(a) # Probabilidad acumulada
d = stats.norm(mu, sd).cdf(b) # Probabilidad acumulada
x1 = np.linspace(70, a, 401)
x2 = np.linspace(b, 100, 401)

ax.plot(x, y, c='g')

ax.fill_between(x1, 0,stats.norm(mu, sd).pdf(x1) , color="r")
ax.fill_between(x2, 0,stats.norm(mu, sd).pdf(x2) , color="b")

ax.text(b+0.1, 0.02, f'{1-d:.4f}', horizontalalignment='center')
ax.text(a-0.1, 0.02, f'{c:.4f}', horizontalalignment='center')

ax.set_xlabel('Valores de x')
ax.set_ylabel('Densidad')
plt.title('Histograma Distribución Normal con $μ=85,σ=3.8$')
print('la probabilidad de que una batería dure más a lo más dos desviaciones estandard de su media es')
print('P{88=<X8} =',c+1-d)

plt.show()
```

# **50.  La distribución Exponencial**

# **a. Generalidades de la Distribución Exponencial**

+ **Wikipedia**. En Teoría de Probabilidad y Estadística, la distribución exponencial es una distribución continua que se utiliza para modelar tiempos de espera para la ocurrencia de un cierto evento.
+ Esta distribución al igual que la distribución geométrica tiene la propiedad de pérdida de memoria.
+ La distribución exponencial es un caso particular de la distribución gamma.


+ **Ejemplos para la distribución exponencial**:
 - $X=:$ *El tiempo transcurrido entre dos llamadas en un centro de atención de llamadas.*
 - $X=:$ *El tiempo que debe esperar una persona hasta que pasa un taxi libre por una calle concreta.*
 - $X=:$ *El tiempo de espera hasta que entra un nuevo cliente en una tienda.*
 - $X=:$ *es la distribución de la longitud de los intervalos de una variable continua que transcurren entre dos sucesos, que se distribuyen según la distribución de Poisson.*

 - $X=:$ *El tiempo transcurrido en un centro de llamadas hasta recibir la primera llamada del día se podría modelar como una exponencial.*
  - $X=:$ *El intervalo de tiempo entre terremotos (de una determinada magnitud) sigue una distribución exponencial.*
 - $X=:$ *Supongamos una máquina que produce hilo de alambre, la cantidad de metros de alambre hasta encontrar una falla en el alambre se podría modelar como una exponencial.
  - $X=:$ *En fiabilidad de sistemas, un dispositivo con tasa de fallo constante sigue una distribución exponencial.*


# **b. Distribuciones continuas - La Distribución Exponencial**

+ Una variable aleatoria continua $X$ sigue una distribución *Exponencial* de parámetros $β$, lo cual denotamos por $$X∼exp(β)$$ si tiene funcion FDP dada por la formula:
$$ P(X=x)= f(x,β) = \frac{1}{β}exp{[\frac{-x}{β}]}, \ x\geq 0$$
+ Tambien $$E(X)=\mu=β,  \ \ \ \ V(X)=σ^2=β^2$$

+ Si $ \lambda$ representa al parámetro de una distribución de Poisson, entoces  $ \lambda= \frac{1}{β}$.
+ $\mu$ representa al número esperado.
+ $V(X)$ representa a la varianza.
+ Se puede mostrar


# **c. Valor Esperado de una Variable aleatoria normal**

+  $$E(X)=\mu$$
+ $$E(X) = ∫_{0}^{∞}xf(x)dx= ∫_{0}^{∞}\frac{x}{β}exp{[\frac{-x}{β}]}dx$$
+ **integrando por partes**
 - Hagamos $u=x →du=dx$ y $dv=\frac{1}{β}exp{[\frac{-x}{β}]}dx→v=-exp{[\frac{-x}{β}]}$
 - Aplicando la formula de integral por partes $$E(X) = ∫_{0}^{∞}xf(x)dx=-xexp{\{\frac{-x}{β}\}}|_{0}^{∞}+ ∫_{0}^{∞}exp{\{\frac{-x}{β}\}}dx$$
 $$E(X) =  ∫_{0}^{∞}exp{\{\frac{-x}{β}\}}dx=-βexp{\{\frac{-x}{β}\}}|_{0}^{∞}=β=\mu$$



# **d. Varianza de una Variable aleatoria normal**

+  $$V(X)=σ^2$$
 -  $$E(X^2) = ∫_{0}^{∞}x^2f(x)dx=  ∫_{0}^{∞}\frac{x^2}{β}exp{[\frac{-x}{β}]}dx$$


+ **integrando por partes**
 - Hagamos $u=x^2 →du=2xdx$ y $dv=\frac{1}{β}exp{[\frac{-x}{β}]}dx→v=-exp{\{\frac{-x}{β}\}}$
 - Aplicando la formula de integral por partes $$E(X^2) = ∫_{0}^{∞}x^2f(x)dx=-x^2exp{\{\frac{-x}{β}\}}|_{0}^{∞}+ 2∫_{0}^{∞}xexp{\{\frac{-x}{β}\}}dx$$
 $$E(X^2) = 2β ∫_{0}^{∞}\frac{x}{β}exp{\{\frac{-x}{β}\}}dx=2β(β)=2β^2$$

+ **Aplicando la formula de varianza**
 +  $$V(X)=σ^2 = E(X^2)-E(X)^2=2β^2-β^2=β^2$$

# **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN EXPONENCIAL**

[⬆️ Volver al inicio](#inicio)

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN EXPONENCIAL**
#
# La distribución Exponencial modela el tiempo entre eventos
# en un proceso de Poisson. Se utiliza para:
#   • Tiempo de vida de componentes
#   • Tiempo de espera entre llegadas
#   • Duración de llamadas telefónicas
#   • Tiempo hasta el fallo de sistemas
#
# Parámetros:
#   λ (tasa) - tasa de ocurrencia de eventos (λ > 0)
#   β = 1/λ - escala (tiempo medio entre eventos)
#
# Fórmula: f(x) = λ * e^(-λx) para x ≥ 0
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon, norm
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
try:
    plt.style.use('seaborn-v0_8')
except:
    try:
        plt.style.use('seaborn')
    except:
        plt.style.use('default')

plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_exponencial():
    """
    Calculadora interactiva para la distribución Exponencial
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN EXPONENCIAL")
    print("="*70)
    print("\nLa distribución Exponencial modela el tiempo entre eventos")
    print("en un proceso de Poisson. Se utiliza para:")
    print("  • Tiempo de vida de componentes electrónicos")
    print("  • Tiempo de espera entre llegadas")
    print("  • Duración de servicios")
    print("  • Tiempo hasta el fallo de sistemas\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            print("Puedes ingresar λ (tasa) o β (escala):")
            print("  λ = 1/β (tasa de ocurrencia)")
            print("  β = 1/λ (tiempo medio entre eventos)")

            parametro = input("¿Qué parámetro deseas ingresar? (λ/β): ").lower()

            if parametro == 'λ':
                lam = float(input("  Tasa (λ > 0): "))
                if lam <= 0:
                    print("  ❌ λ debe ser mayor que 0. Intenta de nuevo.")
                    continue
                beta = 1 / lam
            elif parametro == 'β':
                beta = float(input("  Escala (β > 0): "))
                if beta <= 0:
                    print("  ❌ β debe ser mayor que 0. Intenta de nuevo.")
                    continue
                lam = 1 / beta
            else:
                print("  ❌ Parámetro no válido. Usa 'λ' o 'β'.")
                continue

            print(f"\n  λ = {lam:.4f} (tasa)")
            print(f"  β = {beta:.4f} (tiempo medio)")

            print("\nTipo de cálculo:")
            print("  1. P(X = a)  - Probabilidad puntual (siempre 0)")
            print("  2. P(X ≤ a)  - Probabilidad acumulada a la izquierda")
            print("  3. P(X ≥ a)  - Probabilidad acumulada a la derecha")
            print("  4. P(a ≤ X ≤ b) - Probabilidad entre dos valores")
            print("  5. Cuantil - Encontrar X dado un percentil")
            print("  6. Propiedad de falta de memoria")
            print("  7. Comparar con diferentes tasas")

            tipo = input("\nSelecciona el tipo de cálculo (1-7): ")

            if tipo == '1':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 0
                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X = {a}) = 0")
                print("En distribuciones continuas, la probabilidad puntual es siempre 0.")
                print(f"Densidad en X={a}: {expon.pdf(a, scale=beta):.6f}")

                visualizar_exponencial(lam, beta, a, None, prob, tipo)
                return {'lam': lam, 'beta': beta, 'a': a}

            elif tipo == '2':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = expon.cdf(a, scale=beta)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≤ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con λ = {lam:.4f}, β = {beta:.4f}")

                visualizar_exponencial(lam, beta, a, None, prob, tipo)
                return {'lam': lam, 'beta': beta, 'a': a, 'prob': prob}

            elif tipo == '3':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 1 - expon.cdf(a, scale=beta)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≥ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con λ = {lam:.4f}, β = {beta:.4f}")

                visualizar_exponencial(lam, beta, a, None, prob, tipo)
                return {'lam': lam, 'beta': beta, 'a': a, 'prob': prob}

            elif tipo == '4':
                a = float(input("  Límite inferior (a ≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                b = float(input("  Límite superior (b > a): "))
                if b <= a:
                    print("  ❌ b debe ser mayor que a. Intenta de nuevo.")
                    continue
                prob = expon.cdf(b, scale=beta) - expon.cdf(a, scale=beta)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP({a} ≤ X ≤ {b}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con λ = {lam:.4f}, β = {beta:.4f}")

                visualizar_exponencial(lam, beta, a, b, prob, tipo)
                return {'lam': lam, 'beta': beta, 'a': a, 'b': b, 'prob': prob}

            elif tipo == '5':
                p = float(input("  Percentil (0-1): "))
                if p <= 0 or p >= 1:
                    print("  ❌ El percentil debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                x = expon.ppf(p, scale=beta)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nPara el percentil {p*100:.1f}%:")
                print(f"X = {x:.6f}")
                print(f"Con λ = {lam:.4f}, β = {beta:.4f}")
                print(f"Interpretación: El {p*100:.1f}% de los datos están por debajo de {x:.4f}")

                visualizar_cuantil_exponencial(lam, beta, x, p)
                return {'lam': lam, 'beta': beta, 'percentil': p, 'x': x}

            elif tipo == '6':
                propiedad_memoria(lam, beta)
                return {'lam': lam, 'beta': beta}

            elif tipo == '7':
                comparar_tasas(lam, beta)
                return {'lam': lam, 'beta': beta}

            else:
                print("  ❌ Opción no válida. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

#==========================================================
# FUNCIÓN PARA VISUALIZAR LA DISTRIBUCIÓN EXPONENCIAL
#==========================================================

def visualizar_exponencial(lam, beta, a, b, prob, tipo):
    """Crea visualizaciones completas de la distribución Exponencial"""

    # Rango de valores para la gráfica
    x_max = max(expon.ppf(0.999, scale=beta), a * 1.5 if a > 0 else expon.ppf(0.999, scale=beta))
    x_max = max(x_max, 5)
    x = np.linspace(0, x_max, 1000)
    y = expon.pdf(x, scale=beta)

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    #==========================================================
    # GRÁFICO 1: Distribución con área sombreada
    #==========================================================
    ax1 = plt.subplot(2, 2, 1)
    ax1.plot(x, y, 'b-', linewidth=2.5, label=f'Exponencial (λ={lam:.2f})')

    # Sombrear el área según el tipo de cálculo
    if tipo == '2':  # P(X ≤ a)
        mask = x <= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≤ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '3':  # P(X ≥ a)
        mask = x >= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≥ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '4':  # P(a ≤ X ≤ b)
        mask = (x >= a) & (x <= b)
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P({a:.2f} ≤ X ≤ {b:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)
        ax1.axvline(x=b, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    # Estadísticas
    media = beta
    desv = beta
    mediana = beta * np.log(2)
    moda = 0

    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.2f}')
    ax1.axvline(x=mediana, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Mediana = {mediana:.2f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Exponencial (λ={lam:.3f}, β={beta:.3f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', fontsize=10)
    ax1.set_xlim(0, x_max)

    # Agregar información de probabilidad
    if prob is not None:
        ax1.text(0.98, 0.95, f'Probabilidad = {prob:.6f}\n({prob*100:.2f}%)',
                transform=ax1.transAxes, fontsize=11, verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 2: Comparación con diferentes tasas
    #==========================================================
    ax2 = plt.subplot(2, 2, 2)

    tasas = [0.5, 1, 2, 3, 5]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    x2 = np.linspace(0, 5, 1000)
    for tasa, color in zip(tasas, colores):
        y_temp = expon.pdf(x2, scale=1/tasa)
        ax2.plot(x2, y_temp, color=color, linewidth=2,
                label=f'λ = {tasa}')

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax2.set_title('Comparación de diferentes tasas λ', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best', fontsize=9)
    ax2.set_xlim(0, 5)

    #==========================================================
    # GRÁFICO 3: Función de distribución acumulada
    #==========================================================
    ax3 = plt.subplot(2, 2, 3)
    cdf = expon.cdf(x, scale=beta)
    ax3.plot(x, cdf, 'g-', linewidth=2.5, label='CDF')
    ax3.fill_between(x, 0, cdf, alpha=0.2, color='green')

    # Puntos de interés
    if tipo == '2':  # P(X ≤ a)
        ax3.scatter(a, prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {prob:.4f}')
        ax3.axhline(y=prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '3':  # P(X ≥ a)
        ax3.scatter(a, 1-prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {1-prob:.4f}')
        ax3.axhline(y=1-prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles comunes
    for q in [0.25, 0.5, 0.75, 0.9, 0.95]:
        q_val = expon.ppf(q, scale=beta)
        if q_val <= x_max:
            ax3.axvline(x=q_val, color='orange', linestyle=':', linewidth=1.5, alpha=0.5)
            ax3.text(q_val, 0.01, f'Q{int(q*100)}', ha='center', fontsize=8, color='orange')

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax3.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', fontsize=10)
    ax3.set_ylim(-0.05, 1.05)
    ax3.set_xlim(0, x_max)

    #==========================================================
    # GRÁFICO 4: Tabla de valores importantes
    #==========================================================
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Crear tabla con estadísticas y percentiles
    table_data = [
        ['Estadística', 'Valor', 'Interpretación'],
        ['Media (μ)', f'{beta:.4f}', 'Tiempo medio'],
        ['Desv. Estándar (σ)', f'{beta:.4f}', 'Igual a la media'],
        ['Mediana', f'{beta*np.log(2):.4f}', '50% de los datos'],
        ['Moda', '0', 'Máximo de la densidad'],
        ['Varianza', f'{beta**2:.4f}', 'β²'],
        ['Asimetría', '2', 'Siempre positiva'],
        ['Curtosis', '6', 'Cola pesada'],
        ['λ (tasa)', f'{lam:.4f}', '1/β']
    ]

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.3, 0.3, 0.4])

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.5)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title('Propiedades de la Distribución Exponencial',
                  fontsize=13, fontweight='bold', pad=20)

    #==========================================================
    # Ajustar layout
    #==========================================================
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # GRÁFICO ADICIONAL: Función de supervivencia
    #==========================================================
    fig2, (ax5, ax6) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 5: Función de supervivencia (cola)
    surv = 1 - cdf
    ax5.plot(x, surv, 'r-', linewidth=2.5, label='S(x) = P(X > x)')
    ax5.fill_between(x, 0, surv, alpha=0.2, color='red')
    ax5.axhline(y=0.5, color='gray', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.axvline(x=beta*np.log(2), color='gray', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.text(beta*np.log(2), 0.52, 'Mediana', ha='center', fontsize=9)

    ax5.set_xlabel('X', fontsize=12)
    ax5.set_ylabel('Supervivencia S(X)', fontsize=12)
    ax5.set_title('Función de Supervivencia', fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.legend(loc='best')
    ax5.set_xlim(0, x_max)
    ax5.set_ylim(0, 1.05)

    # Gráfico 6: Tasa de fallo (constante)
    hazard = np.ones_like(x) * lam
    ax6.plot(x, hazard, 'purple', linewidth=2.5, label='h(x) = λ')
    ax6.fill_between(x, 0, hazard, alpha=0.2, color='purple')

    ax6.set_xlabel('X', fontsize=12)
    ax6.set_ylabel('Tasa de fallo h(x)', fontsize=12)
    ax6.set_title('Tasa de Fallo Constante', fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3, linestyle='--')
    ax6.legend(loc='best')
    ax6.set_xlim(0, x_max)

    plt.tight_layout()

    #==========================================================
    # MOSTRAR GRÁFICOS
    #==========================================================
    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Exponencial_lam{lam:.2f}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_supervivencia.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_supervivencia.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

def visualizar_cuantil_exponencial(lam, beta, x, p):
    """Visualiza el cuantil en la distribución Exponencial"""

    x_max = max(expon.ppf(0.999, scale=beta), x * 1.5)
    x_vals = np.linspace(0, x_max, 1000)
    y = expon.pdf(x_vals, scale=beta)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Distribución con cuantil marcado
    ax1.plot(x_vals, y, 'b-', linewidth=2.5)
    mask = x_vals <= x
    ax1.fill_between(x_vals[mask], 0, y[mask], color='#E63946', alpha=0.5,
                     label=f'Área = {p*100:.1f}%')
    ax1.axvline(x=x, color='#E63946', linestyle='--', linewidth=2, label=f'X = {x:.2f}')

    # Media y mediana
    media = beta
    mediana = beta * np.log(2)
    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.2f}')
    ax1.axvline(x=mediana, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Mediana = {mediana:.2f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Percentil {p*100:.1f}% (λ={lam:.2f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best')
    ax1.set_xlim(0, x_max)

    # Gráfico 2: Función de distribución acumulada
    cdf = expon.cdf(x_vals, scale=beta)
    ax2.plot(x_vals, cdf, 'g-', linewidth=2.5)
    ax2.scatter(x, p, color='#E63946', s=150, zorder=5, label=f'F({x:.2f}) = {p:.4f}')
    ax2.axhline(y=p, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax2.axvline(x=x, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax2.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best')
    ax2.set_ylim(-0.05, 1.05)
    ax2.set_xlim(0, x_max)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA LA PROPIEDAD DE FALTA DE MEMORIA
#==========================================================

def propiedad_memoria(lam, beta):
    """Demuestra la propiedad de falta de memoria de la exponencial"""

    print("\n" + "="*70)
    print("🔍 PROPIEDAD DE FALTA DE MEMORIA")
    print("="*70)
    print("\nLa distribución Exponencial tiene la propiedad de falta de memoria:")
    print("P(X > s + t | X > s) = P(X > t)")
    print("Esto significa que el proceso 'no recuerda' el tiempo transcurrido.\n")

    try:
        print("Ingresa los valores para demostrar la propiedad:")
        s = float(input("  Tiempo ya transcurrido (s ≥ 0): "))
        if s < 0:
            print("  ❌ s debe ser ≥ 0. Usando s=0.")
            s = 0
        t = float(input("  Tiempo adicional (t > 0): "))
        if t <= 0:
            print("  ❌ t debe ser > 0. Usando t=1.")
            t = 1

        # Calcular P(X > s + t | X > s)
        prob_cond = (1 - expon.cdf(s + t, scale=beta)) / (1 - expon.cdf(s, scale=beta))
        prob_simple = 1 - expon.cdf(t, scale=beta)

        print(f"\nResultados (λ={lam:.4f}, β={beta:.4f}):")
        print("-"*70)
        print(f"P(X > {s} + {t} | X > {s}) = {prob_cond:.6f} ({prob_cond*100:.2f}%)")
        print(f"P(X > {t}) = {prob_simple:.6f} ({prob_simple*100:.2f}%)")
        print(f"Diferencia: {abs(prob_cond - prob_simple):.10f}")

        if abs(prob_cond - prob_simple) < 1e-10:
            print("\n✅ La propiedad de falta de memoria se cumple perfectamente")
        else:
            print("\n⚠️ La diferencia se debe a errores numéricos")

        # Visualización
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

        x = np.linspace(0, max(s + t + 2, 5), 1000)
        y = expon.pdf(x, scale=beta)

        # Gráfico 1: Distribución condicional
        ax1.plot(x, y, 'b-', linewidth=2, label='Distribución original')

        # Área condicional
        mask_cond = x >= s + t
        ax1.fill_between(x[mask_cond], 0, y[mask_cond], color='#E63946', alpha=0.5,
                        label=f'P(X > {s+t} | X > {s})')

        ax1.axvline(x=s, color='#2C3E50', linestyle='--', linewidth=2, label=f'X > {s}')
        ax1.axvline(x=s+t, color='#E63946', linestyle='--', linewidth=2, label=f'X > {s+t}')

        ax1.set_xlabel('X', fontsize=12)
        ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
        ax1.set_title('Propiedad de Falta de Memoria', fontsize=13, fontweight='bold')
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='best')

        # Gráfico 2: Comparación de distribuciones
        # Distribución original desplazada
        x_shifted = x + s
        y_shifted = expon.pdf(x, scale=beta)

        ax2.plot(x, y, 'b-', linewidth=2, label='Original')
        ax2.plot(x_shifted, y_shifted, 'r-', linewidth=2, label='Desplazada (misma forma)')
        ax2.axvline(x=0, color='gray', linestyle=':', linewidth=1)
        ax2.axvline(x=s, color='gray', linestyle=':', linewidth=1, alpha=0.5)

        ax2.set_xlabel('X', fontsize=12)
        ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
        ax2.set_title('Invarianza por desplazamiento', fontsize=13, fontweight='bold')
        ax2.grid(True, alpha=0.3)
        ax2.legend(loc='best')
        ax2.set_xlim(0, max(s + t + 2, 5))

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA COMPARAR DIFERENTES TASAS
#==========================================================

def comparar_tasas(lam_actual, beta_actual):
    """Compara diferentes tasas de la distribución Exponencial"""

    print("\n" + "="*70)
    print("📊 COMPARACIÓN DE DIFERENTES TASAS")
    print("="*70)

    tasas = [0.5, 1, 2, 3, 5]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    x = np.linspace(0, 5, 1000)

    # Gráfico 1: Densidades
    ax1 = axes[0, 0]
    for tasa, color in zip(tasas, colores):
        y = expon.pdf(x, scale=1/tasa)
        ax1.plot(x, y, color=color, linewidth=2,
                label=f'λ = {tasa}', alpha=0.8 if tasa != lam_actual else 1.0)
        if tasa == lam_actual:
            ax1.fill_between(x, 0, y, alpha=0.2, color=color)

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad', fontsize=12)
    ax1.set_title('Densidades para diferentes λ', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='best')
    ax1.set_xlim(0, 5)

    # Gráfico 2: CDF
    ax2 = axes[0, 1]
    for tasa, color in zip(tasas, colores):
        y = expon.cdf(x, scale=1/tasa)
        ax2.plot(x, y, color=color, linewidth=2,
                label=f'λ = {tasa}', alpha=0.8 if tasa != lam_actual else 1.0)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('CDF', fontsize=12)
    ax2.set_title('Funciones de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')
    ax2.set_xlim(0, 5)
    ax2.set_ylim(0, 1.05)

    # Gráfico 3: Supervivencia
    ax3 = axes[1, 0]
    for tasa, color in zip(tasas, colores):
        y = 1 - expon.cdf(x, scale=1/tasa)
        ax3.plot(x, y, color=color, linewidth=2,
                label=f'λ = {tasa}', alpha=0.8 if tasa != lam_actual else 1.0)

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Supervivencia S(X)', fontsize=12)
    ax3.set_title('Funciones de Supervivencia', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3)
    ax3.legend(loc='best')
    ax3.set_xlim(0, 5)
    ax3.set_ylim(0, 1.05)

    # Gráfico 4: Media y mediana en función de λ
    ax4 = axes[1, 1]
    tasas_vals = np.linspace(0.1, 5, 50)
    medias = 1 / tasas_vals
    medianas = np.log(2) / tasas_vals

    ax4.plot(tasas_vals, medias, 'b-', linewidth=2, label='Media (1/λ)')
    ax4.plot(tasas_vals, medianas, 'r-', linewidth=2, label='Mediana (ln(2)/λ)')
    ax4.axvline(x=lam_actual, color='purple', linestyle='--', linewidth=2, alpha=0.5)

    ax4.set_xlabel('λ', fontsize=12)
    ax4.set_ylabel('Valor', fontsize=12)
    ax4.set_title('Media y Mediana vs λ', fontsize=13, fontweight='bold')
    ax4.grid(True, alpha=0.3)
    ax4.legend(loc='best')

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA GENERAR DATOS EXPONENCIALES
#==========================================================

def generar_datos_exponencial():
    """Genera y visualiza datos de una distribución Exponencial"""

    print("\n" + "="*70)
    print("📊 GENERACIÓN DE DATOS EXPONENCIALES")
    print("="*70)

    try:
        print("Ingresa los parámetros:")
        lam = float(input("  Tasa (λ > 0): "))
        if lam <= 0:
            print("  ❌ λ debe ser mayor que 0.")
            return

        n = int(input("  Número de observaciones a generar: "))
        if n <= 0:
            print("  ❌ n debe ser mayor que 0.")
            return

        # Generar datos
        np.random.seed(42)
        datos = np.random.exponential(scale=1/lam, size=n)

        print(f"\nEstadísticas de los datos generados:")
        print("-"*70)
        print(f"  Media muestral: {np.mean(datos):.4f}")
        print(f"  Mediana muestral: {np.median(datos):.4f}")
        print(f"  Desv. estándar muestral: {np.std(datos):.4f}")
        print(f"  Mínimo: {np.min(datos):.4f}")
        print(f"  Máximo: {np.max(datos):.4f}")

        # Visualización
        fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

        # Histograma con curva teórica
        ax1.hist(datos, bins=30, density=True, alpha=0.6, color='#2E86AB',
                edgecolor='black', label='Datos generados')

        x = np.linspace(0, max(datos), 1000)
        y_teorica = expon.pdf(x, scale=1/lam)
        ax1.plot(x, y_teorica, 'r-', linewidth=2, label='Teórica')

        ax1.set_xlabel('X', fontsize=12)
        ax1.set_ylabel('Densidad', fontsize=12)
        ax1.set_title('Histograma vs Curva Teórica', fontsize=13, fontweight='bold')
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='best')

        # Q-Q plot
        stats.probplot(datos, dist="expon", plot=ax2)
        ax2.set_title('Gráfico Q-Q Exponencial', fontsize=13, fontweight='bold')
        ax2.grid(True, alpha=0.3)

        # CDF empírica vs teórica
        from statsmodels.distributions.empirical_distribution import ECDF
        ecdf = ECDF(datos)

        x_ecdf = np.linspace(0, max(datos), 1000)
        ax3.plot(x_ecdf, ecdf(x_ecdf), 'b-', linewidth=2, label='Empírica')
        ax3.plot(x_ecdf, expon.cdf(x_ecdf, scale=1/lam), 'r--', linewidth=2, label='Teórica')

        ax3.set_xlabel('X', fontsize=12)
        ax3.set_ylabel('CDF', fontsize=12)
        ax3.set_title('CDF Empírica vs Teórica', fontsize=13, fontweight='bold')
        ax3.grid(True, alpha=0.3)
        ax3.legend(loc='best')
        ax3.set_ylim(0, 1.05)

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_exponencial():
    """Ejecuta ejemplos clásicos de la distribución Exponencial"""

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN EXPONENCIAL")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Tiempo de vida de componentes',
            'descripcion': 'Un componente electrónico tiene una vida útil exponencial con λ=0.02 (fallos por hora). ¿Cuál es la probabilidad de que dure más de 100 horas?',
            'lam': 0.02,
            'a': 100,
            'tipo': 'mayor'
        },
        {
            'titulo': 'Ejemplo 2: Tiempo de espera',
            'descripcion': 'En una estación de autobuses, el tiempo entre llegadas sigue una exponencial con media de 15 minutos (β=15). ¿Cuál es la probabilidad de esperar menos de 10 minutos?',
            'beta': 15,
            'a': 10,
            'tipo': 'menor'
        },
        {
            'titulo': 'Ejemplo 3: Tiempo de servicio',
            'descripcion': 'El tiempo de servicio en un banco sigue una exponencial con λ=0.1 clientes por minuto. ¿Cuál es el percentil 90 del tiempo de servicio?',
            'lam': 0.1,
            'p': 0.90,
            'tipo': 'cuantil'
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")

        if 'lam' in ej:
            lam = ej['lam']
            beta = 1/lam
        else:
            beta = ej['beta']
            lam = 1/beta

        if ej['tipo'] == 'mayor':
            prob = 1 - expon.cdf(ej['a'], scale=beta)
            print(f"   λ={lam:.3f}, β={beta:.3f}")
            print(f"   P(X > {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'menor':
            prob = expon.cdf(ej['a'], scale=beta)
            print(f"   λ={lam:.3f}, β={beta:.3f}")
            print(f"   P(X ≤ {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'cuantil':
            x = expon.ppf(ej['p'], scale=beta)
            print(f"   λ={lam:.3f}, β={beta:.3f}")
            print(f"   Percentil {ej['p']*100:.0f}% = {x:.4f}")

    print("\n" + "="*70)

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN EXPONENCIAL")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado")
        print("  2. Ver ejemplos clásicos")
        print("  3. Generar datos exponenciales")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_exponencial()
        elif opcion == '2':
            ejemplos_exponencial()
        elif opcion == '3':
            generar_datos_exponencial()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```

#**a. Construya el gráfico de la Distribución dada si  $μ=15$**

```python
# @title
# **a. Grafico manual de la función
from numpy.ma.core import log
# nuestra libreria
import numpy as np
import matplotlib.pyplot as plt
# Definimos una funcion
media = 15 # parametro

def lognormal1(x):
  return np.exp(-x/media)/media
# los valores que va a tomar x
x = np.linspace(0,100,1000) # espacio donde vamos a graficar la funcion
fig, ax = plt.subplots() #lugar donde vamos a estar graficando, ax objeto y subplot porque vamos hacer varios graficos
ax.plot(x,lognormal1(x)) # variable independiente y la funcion
plt.title('Distribución Exponencial β=15')
plt.ylabel('Probabilidad')
plt.xlabel('Dias de antelacion')
ax.grid() # le da mas elegancia al grafico
```

```python
# @title **a. Construya el gráfico de la Distribución dada si  $μ=15$**
plt.figure(figsize=(8, 8))
scale = 15 # parametro
exponencial = stats.expon(scale=15)
x = np.linspace(exponencial.ppf(0.01),
                exponencial.ppf(0.99), 1000)
fp = exponencial.pdf(x) # Función de Probabilidad
plt.plot(x, fp)
plt.title('Distribución Exponencial β=15')
plt.ylabel('Probabilidad')
plt.xlabel('Dias de antelacion')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **52. La Función Gamma**

# **a. Generalidades de la Distribución Gamma**

+ La distribución **Gamma** debe su nombre a la *función Gamma*. Ésta es muy
conocida en diferentes ramas de las matemáticas desde la Teoría de Ecuaciones
Diferenciales hasta la Estadística. Sin embargo, el origen de esta función se encuentra en la unión de un problema de interpolación con otro de cálculo integral. La función Gamma fue descubierta por Leonhard Euler en 1729 entre la correspondencia que mantenía con Goldbach.
+


+ **Ejemplos para la distribución exponencial**:Veamos algunos casos donde esta distribución es aplicable:
 - $X=:$ *Tiempos de espera en la Teoría de Colas**.
 - $X=:$ *En la meteorología para ajustar los datos de las precipitaciones.
 - $X=:$ *Análisis de supervivencia: el estudio estadístico del tiempo que pasa hasta que ocurre un evento.
 - $X=:$ *Ajusta la distribución de la renta.
 - $X=:$ *Modeliza los datos de consumo de productos a granel*
  - $X=:$ *La distribución gamma se suele utilizar en: Intervalos de tiempos entre dos fallos de un motor*
 - $X=:$ *Intervalos de tiempos entre dos llegadas de automóviles a una gasolinera*
 - $X=:$ *Tiempos de vida de sistemas electrónicos, etc.*


# **b. La función Gamma**
+ La *función Gamma se define*  

$$
\Gamma(α) = \int_0^\infty x^{α-1}e^{-x}dx\,.
$$
# **c. Propiedades de la función Gamma**
+ **Propiedad 1**. $\Gamma(1) =\Gamma(2) =1$
 - **Prueba**
 $$
\Gamma(1) = ∫_{0}^{∞} x^{1-1}e^{-x}dx= ∫_{0}^{∞} e^{-x}dx=-e^{-x}|_{0}^{∞}=1
$$
$$
\Gamma(2) = ∫_{0}^{∞} x^{2-1}e^{-x}dx= ∫_{0}^{∞} xe^{-x}dx=-xe^{-x}|_{0}^{∞}+∫_{0}^{∞} e^{-x}dx=1
$$
+ **Propiedad 2**. $\Gamma(α+1) = α!$, si $α\in\mathbb N$
+ **Prueba**
  $$
\Gamma(α+1) = ∫_{0}^{∞} x^{α-1+1}e^{-x}dx= ∫_{0}^{∞} x^{α}e^{-x}dx=
 \ \ \ -x^{α}e^{-x}|_{0}^{∞}+α∫_{0}^{∞} x^{α-1}e^{-x}dx=α\Gamma(α)
$$
+   $$
\Gamma(α+1) =α\Gamma(α)=α(α-1)\Gamma(α-1)=α(α-1)⋯2\cdot\Gamma(1)=α!,α\in\mathbb N
$$

+ **Propiedad 3**. $\Gamma(1/2) = \sqrt{π}$
$$
\Gamma(1/2) = ∫_{0}^{∞} x^{1/2-1}e^{-x}dx= ∫_{0}^{∞}  x^{-1/2}e^{-x}dx
$$
+ **Hagamos** $u^2=2x$ entonces $2udu=2dx$ y $x=\frac{u^2}{2}$ y $x^{-1/2}=(\frac{u^2}{2})^{-1/2}=\frac{\sqrt{2}}{u}$
+ **Reemplazando todo en la integral arriba**

$$
\Gamma(1/2) = ∫_{0}^{∞} x^{1/2-1}e^{-x}dx= ∫_{0}^{∞}  x^{-1/2}e^{-x}dx=∫_{0}^{∞}\frac{\sqrt{2}}{u}e^{-\frac{u^2}{2}}udu=∫_{0}^{∞}\sqrt{2}e^{-\frac{u^2}{2}}du
$$
$$
\Gamma(1/2) = ∫_{0}^{∞}\sqrt{2}e^{-\frac{u^2}{2}}du=\frac{1}{2}∫_{-∞}^{∞}\sqrt{2}e^{-\frac{u^2}{2}}du=∫_{-∞}^{∞}\frac{1}{\sqrt{2}}e^{-\frac{u^2}{2}}du =\sqrt{π}∫_{-∞}^{∞}\frac{1}{\sqrt{2π}}e^{-\frac{u^2}{2}}du=\sqrt{π}*1=\sqrt{π}
$$

# **d. Grafico manual de la función gamma**

```python
#@title **La función gamma**
import numpy as np
import matplotlib.pyplot as plt
# Definimos una funcion
alpha = 5 # parametro

def fgamma(x):
  return np.exp(-x)*(x**(alpha-1))
# los valores que va a tomar x
x = np.linspace(0,20,1000) # espacio donde vamos a graficar la funcion
fig, ax = plt.subplots() #lugar donde vamos a estar graficando, ax objeto y subplot porque vamos hacer varios graficos
ax.plot(x,fgamma(x)) # variable independiente y la funcion
plt.title('funcion gamma α=5')
plt.ylabel('frecuencias')
plt.xlabel('valores de x')
ax.grid() # le da mas elegancia al grafico
```

# **e. Propiedades de la función Gamma**
$$
\Gamma(α) = \int_0^\infty x^{α-1}e^{-x}dx\,.
$$

```python
# ==========================================================
# @title **e. Propiedades de la Función Gamma**
#
#              ∞
# Γ(a) = ∫ e^(-x)x^(a-1) dx
#             0
#
# Propiedades importantes:
# Γ(1)   = 1
# Γ(n)   = (n-1)!
# Γ(1/2) = √π
# ==========================================================

import numpy as np
from scipy.integrate import quad
from scipy.special import gamma

#------------------------------------------------------------
# Definición del integrando
#------------------------------------------------------------
def integrando(x, a):
    return np.exp(-x) * x**(a - 1)

#------------------------------------------------------------
# Función Gamma mediante integración numérica
#------------------------------------------------------------
def funcion_gamma(a):
    resultado, error = quad(integrando, 0, np.inf, args=(a,))
    return resultado

#------------------------------------------------------------
# Valores a evaluar
#------------------------------------------------------------
valores = [1, 5, 0.5]

print("="*60)
print("        PROPIEDADES DE LA FUNCIÓN GAMMA")
print("="*60)

for a in valores:

    gamma_integral = funcion_gamma(a)
    gamma_scipy = gamma(a)

    print(f"\na = {a}")
    print(f"Γ({a}) por integración : {gamma_integral:.8f}")
    print(f"Γ({a}) usando scipy    : {gamma_scipy:.8f}")

print("\n" + "="*60)
print("VERIFICACIÓN DE PROPIEDADES")
print("="*60)

print(f"Γ(1)      = {funcion_gamma(1):.8f}")
print(f"Γ(5)      = {funcion_gamma(5):.8f} = 4! = 24")
print(f"Γ(1/2)    = {funcion_gamma(0.5):.8f}")
print(f"√π        = {np.sqrt(np.pi):.8f}")

print("\n¿Coinciden Γ(1/2) y √π?")
print(np.isclose(funcion_gamma(0.5), np.sqrt(np.pi)))

print("="*60)
```

# **53. La Distribución Gamma**
# **a. La Ditribución Gamma - Definición**

+ Una variable aleatoria continua $X$ sigue una distribución *Gamma* de parámetros $β$, lo cual denotamos por $$X∼gamma(α,β)$$ si tiene funcion FDP dada por la formula:
$$ f(x,α,β) = \frac{1}{β^αΓ(α)}x^{α-1}exp{\{\frac{-x}{β}\}}, \ x\geq 0, 0<α,β<∞, $$
+ Si hacemos $λ=β^{-1}$ Se tendría que:

$$ f(x,α,λ) = \frac{(λx)^{α-1}}{Γ(α)}λe^{-λx}, \ x\geq 0$$

+ Tambien $$E(X)=\mu=βα,  \ \ \ \ V(X)=σ^2=αβ^2$$

+ En dónde $ \{X\leq x\}$ representa al evento de todos resultados menores que $x$.
+ $\mu$ representa al número esperado.
+ $V(X)$ representa a la varianza.
+ Se puede mostrar



# **b. Veamos que es una función de densidad**
+ $$ ∫_{-∞}^{∞}f(x)dx= ∫_{0}^{∞}\frac{1}{β^αΓ(α)}x^{α-1}exp{\{\frac{-x}{β}\}}dx= \frac{1}{β^αΓ(α)}∫_{0}^{∞}x^{α-1}exp{\{\frac{-x}{β}\}}dx$$

+ **hacemos** $y = \frac{x}{β}$ esto es $x = βy ⇒dx= βdy$
+  $$ ∫_{-∞}^{∞}f(x)dx=\frac{1}{β^αΓ(α)}∫_{0}^{∞}(βy)^{α-1}e^{-y}βdy
=\frac{β^α}{β^αΓ(α)}∫_{0}^{∞}y^{α-1}e^{-y}dy
$$
+ **Pero sabemos que**
$$
\Gamma(α) = \int_0^\infty x^{α-1}e^{-x}dx\,.
$$
+ **De esta forma se tiene**
 $$ ∫_{-∞}^{∞}f(x)dx=\frac{1}{β^αΓ(α)}∫_{0}^{∞}(βy)^{α-1}e^{-y}βdy
=\frac{β^α}{β^αΓ(α)}\Gamma(α) =1
$$


# **c. Valor Esperado**
+ $$E(X)=μ= ∫_{-∞}^{∞}xf(x)dx= ∫_{0}^{∞}\frac{x}{β^αΓ(α)}x^{α-1}exp{\{\frac{-x}{β}\}}dx$$

+ $$E(X)= ∫_{0}^{∞}\frac{αβ}{β^{α+1}αΓ(α)}x^{(α+1)-1}exp{\{\frac{-x}{β}\}}dx
\ =αβ ∫_{0}^{∞}\frac{1}{β^{α+1}Γ(α+1)}x^{(α+1)-1}exp{\{\frac{-x}{β}\}}dx
$$

+ **Por la parte arriba se tiene**
$$ ∫_{-∞}^{∞}f(x)dx= ∫_{0}^{∞}\frac{1}{β^{α+1}Γ(α+1)}x^{(α+1)-1}exp{\{\frac{-x}{β}\}}dx=1$$

+ **De manera que**

+ $$E(X)=αβ $$


# **d. Varianza de una Variable aleatoria normal**

+  $$V(X)=σ^2$$
 - $$E(X^2)= ∫_{-∞}^{∞}x^2f(x)dx= ∫_{0}^{∞}\frac{x^2}{β^αΓ(α)}x^{α-1}exp{\{\frac{-x}{β}\}}dx$$

+ $$E(X^2)= ∫_{0}^{∞}\frac{α(α+1)β^2}{β^{α+2}α(α+1)Γ(α)}x^{(α+2)-1}exp{\{\frac{-x}{β}\}}dx
\ =α(α+1)β^2 ∫_{0}^{∞}\frac{1}{β^{α+2}Γ(α+2)}x^{(α+2)-1}exp{\{\frac{-x}{β}\}}dx
$$

+ **Por la parte arriba se tiene**
$$ ∫_{-∞}^{∞}f(x)dx= ∫_{0}^{∞}\frac{1}{β^{α+2}Γ(α+2)}x^{(α+2)-1}exp{\{\frac{-x}{β}\}}dx=1$$

+ **De manera que**

+ $$E(X^2)=α(α+1)β^2 $$

+ **Aplicando la formula de varianza**
 +  $$V(X)=σ^2 = E(X^2)-E(X)^2=α(α+1)β^2-α^2β^2=αβ^2$$

# **54. Calculadora de la Distribución Gamma**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN GAMMA**
#
# La distribución Gamma es una familia de distribuciones continuas
# que generaliza la distribución Exponencial. Se utiliza para:
#   • Modelar tiempos de espera (suma de exponenciales)
#   • Tiempo hasta que ocurren k eventos
#   • Modelar cantidades positivas y sesgadas
#   • Distribución previa en estadística bayesiana
#
# Parámetros:
#   α (forma) - parámetro de forma (α > 0)
#   β (tasa) - parámetro de tasa (β > 0) o θ (escala) = 1/β
#
# Fórmula: f(x) = β^α * x^(α-1) * e^(-βx) / Γ(α)
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gamma, expon, norm, chi2
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
try:
    plt.style.use('seaborn-v0_8')
except:
    try:
        plt.style.use('seaborn')
    except:
        plt.style.use('default')

plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_gamma():
    """
    Calculadora interactiva para la distribución Gamma
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN GAMMA")
    print("="*70)
    print("\nLa distribución Gamma generaliza la distribución Exponencial.")
    print("Se utiliza para modelar tiempos de espera y cantidades positivas.")
    print("\nCasos especiales:")
    print("  • α = 1 → Distribución Exponencial")
    print("  • α = n/2, β = 1/2 → Distribución χ² con n grados de libertad")
    print("  • α = k (entero) → Distribución Erlang (tiempo hasta k eventos)\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            print("Puedes ingresar usando (α, β) o (α, θ):")
            print("  β = 1/θ (tasa)")
            print("  θ = 1/β (escala)")

            parametro = input("¿Qué parametrización deseas usar? (β/θ): ").lower()

            alpha = float(input("  Parámetro de forma (α > 0): "))
            if alpha <= 0:
                print("  ❌ α debe ser mayor que 0. Intenta de nuevo.")
                continue

            if parametro == 'β':
                beta = float(input("  Parámetro de tasa (β > 0): "))
                if beta <= 0:
                    print("  ❌ β debe ser mayor que 0. Intenta de nuevo.")
                    continue
                theta = 1 / beta
            elif parametro == 'θ':
                theta = float(input("  Parámetro de escala (θ > 0): "))
                if theta <= 0:
                    print("  ❌ θ debe ser mayor que 0. Intenta de nuevo.")
                    continue
                beta = 1 / theta
            else:
                print("  ❌ Parámetro no válido. Usa 'β' o 'θ'.")
                continue

            print(f"\n  α = {alpha:.4f} (forma)")
            print(f"  β = {beta:.4f} (tasa)")
            print(f"  θ = {theta:.4f} (escala)")

            print("\nTipo de cálculo:")
            print("  1. P(X = a)  - Probabilidad puntual (siempre 0)")
            print("  2. P(X ≤ a)  - Probabilidad acumulada a la izquierda")
            print("  3. P(X ≥ a)  - Probabilidad acumulada a la derecha")
            print("  4. P(a ≤ X ≤ b) - Probabilidad entre dos valores")
            print("  5. Cuantil - Encontrar X dado un percentil")
            print("  6. Comparar con diferentes α")
            print("  7. Relación con otras distribuciones")

            tipo = input("\nSelecciona el tipo de cálculo (1-7): ")

            if tipo == '1':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 0
                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X = {a}) = 0")
                print("En distribuciones continuas, la probabilidad puntual es siempre 0.")
                print(f"Densidad en X={a}: {gamma.pdf(a, alpha, scale=theta):.6f}")

                visualizar_gamma(alpha, beta, theta, a, None, prob, tipo)
                return {'alpha': alpha, 'beta': beta, 'theta': theta, 'a': a}

            elif tipo == '2':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = gamma.cdf(a, alpha, scale=theta)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≤ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con α={alpha:.4f}, β={beta:.4f}, θ={theta:.4f}")

                visualizar_gamma(alpha, beta, theta, a, None, prob, tipo)
                return {'alpha': alpha, 'beta': beta, 'theta': theta, 'a': a, 'prob': prob}

            elif tipo == '3':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 1 - gamma.cdf(a, alpha, scale=theta)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≥ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con α={alpha:.4f}, β={beta:.4f}, θ={theta:.4f}")

                visualizar_gamma(alpha, beta, theta, a, None, prob, tipo)
                return {'alpha': alpha, 'beta': beta, 'theta': theta, 'a': a, 'prob': prob}

            elif tipo == '4':
                a = float(input("  Límite inferior (a ≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                b = float(input("  Límite superior (b > a): "))
                if b <= a:
                    print("  ❌ b debe ser mayor que a. Intenta de nuevo.")
                    continue
                prob = gamma.cdf(b, alpha, scale=theta) - gamma.cdf(a, alpha, scale=theta)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP({a} ≤ X ≤ {b}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con α={alpha:.4f}, β={beta:.4f}, θ={theta:.4f}")

                visualizar_gamma(alpha, beta, theta, a, b, prob, tipo)
                return {'alpha': alpha, 'beta': beta, 'theta': theta, 'a': a, 'b': b, 'prob': prob}

            elif tipo == '5':
                p = float(input("  Percentil (0-1): "))
                if p <= 0 or p >= 1:
                    print("  ❌ El percentil debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                x = gamma.ppf(p, alpha, scale=theta)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nPara el percentil {p*100:.1f}%:")
                print(f"X = {x:.6f}")
                print(f"Con α={alpha:.4f}, β={beta:.4f}, θ={theta:.4f}")
                print(f"Interpretación: El {p*100:.1f}% de los datos están por debajo de {x:.4f}")

                visualizar_cuantil_gamma(alpha, beta, theta, x, p)
                return {'alpha': alpha, 'beta': beta, 'theta': theta, 'percentil': p, 'x': x}

            elif tipo == '6':
                comparar_alpha_gamma(alpha, beta, theta)
                return {'alpha': alpha, 'beta': beta, 'theta': theta}

            elif tipo == '7':
                relacion_distribuciones(alpha, beta, theta)
                return {'alpha': alpha, 'beta': beta, 'theta': theta}

            else:
                print("  ❌ Opción no válida. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

#==========================================================
# FUNCIÓN PARA VISUALIZAR LA DISTRIBUCIÓN GAMMA
#==========================================================

def visualizar_gamma(alpha, beta, theta, a, b, prob, tipo):
    """Crea visualizaciones completas de la distribución Gamma"""

    # Rango de valores para la gráfica
    x_max = max(gamma.ppf(0.999, alpha, scale=theta), a * 1.5 if a > 0 else gamma.ppf(0.999, alpha, scale=theta))
    x_max = max(x_max, 5)
    x = np.linspace(0.001, x_max, 1000)
    y = gamma.pdf(x, alpha, scale=theta)

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    #==========================================================
    # GRÁFICO 1: Distribución con área sombreada
    #==========================================================
    ax1 = plt.subplot(2, 2, 1)
    ax1.plot(x, y, 'b-', linewidth=2.5, label=f'Gamma(α={alpha:.2f}, β={beta:.2f})')

    # Sombrear el área según el tipo de cálculo
    if tipo == '2':  # P(X ≤ a)
        mask = x <= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≤ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '3':  # P(X ≥ a)
        mask = x >= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≥ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '4':  # P(a ≤ X ≤ b)
        mask = (x >= a) & (x <= b)
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P({a:.2f} ≤ X ≤ {b:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)
        ax1.axvline(x=b, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    # Estadísticas
    media = alpha * theta
    varianza = alpha * theta**2
    desv = np.sqrt(varianza)
    moda = max(0, (alpha - 1) * theta) if alpha >= 1 else 0

    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.2f}')
    if moda > 0:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Moda = {moda:.2f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Gamma (α={alpha:.2f}, β={beta:.2f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', fontsize=10)
    ax1.set_xlim(0, x_max)

    # Agregar información de probabilidad
    if prob is not None:
        ax1.text(0.98, 0.95, f'Probabilidad = {prob:.6f}\n({prob*100:.2f}%)',
                transform=ax1.transAxes, fontsize=11, verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 2: Comparación con diferentes α
    #==========================================================
    ax2 = plt.subplot(2, 2, 2)

    alphas = [0.5, 1, 2, 3, 5]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    x2 = np.linspace(0.001, 10, 1000)
    for alpha_val, color in zip(alphas, colores):
        y_temp = gamma.pdf(x2, alpha_val, scale=theta)
        ax2.plot(x2, y_temp, color=color, linewidth=2,
                label=f'α = {alpha_val}')

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax2.set_title(f'Efecto de α (β={beta:.2f} fijo)', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best', fontsize=9)
    ax2.set_xlim(0, 10)

    #==========================================================
    # GRÁFICO 3: Función de distribución acumulada
    #==========================================================
    ax3 = plt.subplot(2, 2, 3)
    cdf = gamma.cdf(x, alpha, scale=theta)
    ax3.plot(x, cdf, 'g-', linewidth=2.5, label='CDF')
    ax3.fill_between(x, 0, cdf, alpha=0.2, color='green')

    # Puntos de interés
    if tipo == '2':  # P(X ≤ a)
        ax3.scatter(a, prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {prob:.4f}')
        ax3.axhline(y=prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '3':  # P(X ≥ a)
        ax3.scatter(a, 1-prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {1-prob:.4f}')
        ax3.axhline(y=1-prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles comunes
    for q in [0.25, 0.5, 0.75, 0.9, 0.95]:
        q_val = gamma.ppf(q, alpha, scale=theta)
        if q_val <= x_max:
            ax3.axvline(x=q_val, color='orange', linestyle=':', linewidth=1.5, alpha=0.5)
            ax3.text(q_val, 0.01, f'Q{int(q*100)}', ha='center', fontsize=8, color='orange')

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax3.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', fontsize=10)
    ax3.set_ylim(-0.05, 1.05)
    ax3.set_xlim(0, x_max)

    #==========================================================
    # GRÁFICO 4: Tabla de estadísticas
    #==========================================================
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Calcular estadísticas
    media = alpha * theta
    varianza = alpha * theta**2
    desv = np.sqrt(varianza)
    asimetria = 2 / np.sqrt(alpha)
    curtosis = 6 / alpha

    # Crear tabla
    table_data = [
        ['Estadística', 'Valor', 'Fórmula'],
        ['Media (μ)', f'{media:.4f}', 'α·θ'],
        ['Varianza (σ²)', f'{varianza:.4f}', 'α·θ²'],
        ['Desv. Estándar (σ)', f'{desv:.4f}', '√(α·θ²)'],
        ['Moda', f'{moda:.4f}' if moda > 0 else '0', '(α-1)·θ'],
        ['Asimetría', f'{asimetria:.4f}', '2/√α'],
        ['Curtosis', f'{curtosis:.4f}', '6/α'],
        ['Coef. Variación', f'{desv/media:.4f}', '1/√α']
    ]

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.3, 0.3, 0.4])

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.5)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title('Propiedades de la Distribución Gamma',
                  fontsize=13, fontweight='bold', pad=20)

    #==========================================================
    # Ajustar layout
    #==========================================================
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # GRÁFICO ADICIONAL: Media y varianza vs α
    #==========================================================
    fig2, (ax5, ax6) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 5: Media y varianza en función de α
    alpha_vals = np.linspace(0.1, 10, 100)
    medias = alpha_vals * theta
    varianzas = alpha_vals * theta**2

    ax5.plot(alpha_vals, medias, 'b-', linewidth=2, label='Media')
    ax5.plot(alpha_vals, varianzas, 'r-', linewidth=2, label='Varianza')
    ax5.axvline(x=alpha, color='purple', linestyle='--', linewidth=2, alpha=0.5)

    ax5.set_xlabel('α', fontsize=12)
    ax5.set_ylabel('Valor', fontsize=12)
    ax5.set_title('Media y Varianza vs α', fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3)
    ax5.legend(loc='best')

    # Gráfico 6: Densidad para diferentes α
    x6 = np.linspace(0.001, 10, 1000)
    alphas_mostrar = [0.5, 1, 2, 3, 5]
    colores_mostrar = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    for alpha_val, color in zip(alphas_mostrar, colores_mostrar):
        y_temp = gamma.pdf(x6, alpha_val, scale=theta)
        ax6.plot(x6, y_temp, color=color, linewidth=2,
                label=f'α = {alpha_val}', alpha=0.8 if alpha_val != alpha else 1.0)
        if alpha_val == alpha:
            ax6.fill_between(x6, 0, y_temp, alpha=0.2, color=color)

    ax6.set_xlabel('X', fontsize=12)
    ax6.set_ylabel('Densidad', fontsize=12)
    ax6.set_title('Densidades para diferentes α', fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3)
    ax6.legend(loc='best')
    ax6.set_xlim(0, 10)

    plt.tight_layout()

    #==========================================================
    # MOSTRAR GRÁFICOS
    #==========================================================
    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Gamma_alpha{alpha:.2f}_beta{beta:.2f}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_propiedades.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_propiedades.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

def visualizar_cuantil_gamma(alpha, beta, theta, x, p):
    """Visualiza el cuantil en la distribución Gamma"""

    x_max = max(gamma.ppf(0.999, alpha, scale=theta), x * 1.5)
    x_vals = np.linspace(0.001, x_max, 1000)
    y = gamma.pdf(x_vals, alpha, scale=theta)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Distribución con cuantil marcado
    ax1.plot(x_vals, y, 'b-', linewidth=2.5)
    mask = x_vals <= x
    ax1.fill_between(x_vals[mask], 0, y[mask], color='#E63946', alpha=0.5,
                     label=f'Área = {p*100:.1f}%')
    ax1.axvline(x=x, color='#E63946', linestyle='--', linewidth=2, label=f'X = {x:.2f}')

    # Media y moda
    media = alpha * theta
    moda = max(0, (alpha - 1) * theta) if alpha >= 1 else 0
    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.2f}')
    if moda > 0:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Moda = {moda:.2f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Percentil {p*100:.1f}% (α={alpha:.2f}, β={beta:.2f})',
                  fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best')
    ax1.set_xlim(0, x_max)

    # Gráfico 2: Función de distribución acumulada
    cdf = gamma.cdf(x_vals, alpha, scale=theta)
    ax2.plot(x_vals, cdf, 'g-', linewidth=2.5)
    ax2.scatter(x, p, color='#E63946', s=150, zorder=5, label=f'F({x:.2f}) = {p:.4f}')
    ax2.axhline(y=p, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax2.axvline(x=x, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax2.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best')
    ax2.set_ylim(-0.05, 1.05)
    ax2.set_xlim(0, x_max)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA COMPARAR DIFERENTES α
#==========================================================

def comparar_alpha_gamma(alpha_actual, beta, theta):
    """Compara diferentes valores de α en la distribución Gamma"""

    print("\n" + "="*70)
    print("📊 COMPARACIÓN DE DIFERENTES α")
    print("="*70)

    alphas = [0.5, 1, 2, 3, 5, 10]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#9B59B6', '#1ABC9C']

    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    x = np.linspace(0.001, 15, 1000)

    # Gráfico 1: Densidades
    ax1 = axes[0, 0]
    for alpha_val, color in zip(alphas, colores):
        y = gamma.pdf(x, alpha_val, scale=theta)
        ax1.plot(x, y, color=color, linewidth=2,
                label=f'α = {alpha_val}', alpha=0.8 if alpha_val != alpha_actual else 1.0)
        if alpha_val == alpha_actual:
            ax1.fill_between(x, 0, y, alpha=0.2, color=color)

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad', fontsize=12)
    ax1.set_title('Densidades para diferentes α', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='best')
    ax1.set_xlim(0, 10)

    # Gráfico 2: CDF
    ax2 = axes[0, 1]
    for alpha_val, color in zip(alphas, colores):
        y = gamma.cdf(x, alpha_val, scale=theta)
        ax2.plot(x, y, color=color, linewidth=2,
                label=f'α = {alpha_val}', alpha=0.8 if alpha_val != alpha_actual else 1.0)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('CDF', fontsize=12)
    ax2.set_title('Funciones de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')
    ax2.set_xlim(0, 10)
    ax2.set_ylim(0, 1.05)

    # Gráfico 3: Media y varianza
    ax3 = axes[1, 0]
    alpha_vals = np.linspace(0.1, 10, 100)
    medias = alpha_vals * theta
    varianzas = alpha_vals * theta**2

    ax3.plot(alpha_vals, medias, 'b-', linewidth=2, label='Media')
    ax3.plot(alpha_vals, varianzas, 'r-', linewidth=2, label='Varianza')
    ax3.axvline(x=alpha_actual, color='purple', linestyle='--', linewidth=2, alpha=0.5)

    ax3.set_xlabel('α', fontsize=12)
    ax3.set_ylabel('Valor', fontsize=12)
    ax3.set_title('Media y Varianza en función de α', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3)
    ax3.legend(loc='best')

    # Gráfico 4: Coeficiente de variación
    ax4 = axes[1, 1]
    cv = 1 / np.sqrt(alpha_vals)
    ax4.plot(alpha_vals, cv, 'g-', linewidth=2, label='CV = 1/√α')
    ax4.axvline(x=alpha_actual, color='purple', linestyle='--', linewidth=2, alpha=0.5)

    ax4.set_xlabel('α', fontsize=12)
    ax4.set_ylabel('Coeficiente de Variación', fontsize=12)
    ax4.set_title('Coeficiente de Variación vs α', fontsize=13, fontweight='bold')
    ax4.grid(True, alpha=0.3)
    ax4.legend(loc='best')

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA RELACIONES CON OTRAS DISTRIBUCIONES
#==========================================================

def relacion_distribuciones(alpha, beta, theta):
    """Muestra la relación de la Gamma con otras distribuciones"""

    print("\n" + "="*70)
    print("🔗 RELACIÓN CON OTRAS DISTRIBUCIONES")
    print("="*70)

    print("\nLa distribución Gamma está relacionada con:")
    print("  1. Exponencial: Gamma(α=1, β) = Exponencial(β)")
    print("  2. Chi-cuadrado: Gamma(α=n/2, β=1/2) = χ²(n)")
    print("  3. Erlang: Gamma(α=k entero, β) = Erlang(k, β)")
    print("  4. Normal: Gamma(α grande) → Normal")
    print("  5. Poisson: Relación en procesos de Poisson\n")

    # Crear visualización de las relaciones
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    x = np.linspace(0.001, 15, 1000)

    # 1. Comparación con Exponencial
    ax1 = axes[0, 0]
    y_gamma = gamma.pdf(x, alpha, scale=theta)
    y_exp = expon.pdf(x, scale=theta)  # α=1
    ax1.plot(x, y_gamma, 'b-', linewidth=2, label=f'Gamma(α={alpha:.1f})')
    ax1.plot(x, y_exp, 'r--', linewidth=2, label='Exponencial(θ)')
    ax1.fill_between(x, 0, y_exp, alpha=0.2, color='red')
    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad', fontsize=12)
    ax1.set_title('Gamma vs Exponencial', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='best')
    ax1.set_xlim(0, 10)

    # 2. Comparación con Chi-cuadrado
    ax2 = axes[0, 1]
    if alpha * 2 == int(alpha * 2):  # Si 2α es entero
        chi2_df = 2 * alpha
        y_chi2 = chi2.pdf(x, chi2_df)
        ax2.plot(x, y_gamma, 'b-', linewidth=2, label=f'Gamma(α={alpha:.1f})')
        ax2.plot(x, y_chi2, 'g--', linewidth=2, label=f'χ²({chi2_df:.0f})')
        ax2.fill_between(x, 0, y_chi2, alpha=0.2, color='green')
    else:
        ax2.text(0.5, 0.5, f'α={alpha:.1f} no es múltiplo de 0.5\npara comparar con χ²',
                transform=ax2.transAxes, ha='center', va='center', fontsize=12)
    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad', fontsize=12)
    ax2.set_title('Gamma vs Chi-cuadrado', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')
    ax2.set_xlim(0, 10)

    # 3. Convergencia a Normal (α grande)
    ax3 = axes[1, 0]
    if alpha > 5:
        media = alpha * theta
        desv = np.sqrt(alpha * theta**2)
        y_norm = norm.pdf(x, media, desv)
        ax3.plot(x, y_gamma, 'b-', linewidth=2, label=f'Gamma(α={alpha:.1f})')
        ax3.plot(x, y_norm, 'm--', linewidth=2, label='Aprox. Normal')
        ax3.fill_between(x, 0, y_norm, alpha=0.2, color='magenta')
    else:
        ax3.text(0.5, 0.5, f'α={alpha:.1f} es pequeño\npara mostrar convergencia a Normal',
                transform=ax3.transAxes, ha='center', va='center', fontsize=12)
    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Densidad', fontsize=12)
    ax3.set_title('Convergencia a la Normal', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3)
    ax3.legend(loc='best')
    ax3.set_xlim(0, 10)

    # 4. Diferentes α y su interpretación
    ax4 = axes[1, 1]
    ax4.axis('off')

    # Crear tabla de casos especiales
    table_data = [
        ['Caso', 'Distribución', 'Aplicación'],
        ['α = 1', 'Exponencial', 'Tiempo entre eventos'],
        ['α = n/2, β=1/2', 'χ²(n)', 'Pruebas de hipótesis'],
        ['α = k (entero)', 'Erlang', 'Tiempo hasta k eventos'],
        ['α → ∞', 'Normal', 'Teorema central del límite'],
        ['α < 1', 'Decreciente', 'Tiempos de falla temprana'],
        ['α > 1', 'Acampanada', 'Tiempos de vida']
    ]

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.25, 0.35, 0.4])

    table.auto_set_font_size(False)
    table.set_fontsize(11)
    table.scale(1, 2)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title('Casos Especiales de la Gamma', fontsize=13, fontweight='bold', pad=20)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA GENERAR DATOS GAMMA
#==========================================================

def generar_datos_gamma():
    """Genera y visualiza datos de una distribución Gamma"""

    print("\n" + "="*70)
    print("📊 GENERACIÓN DE DATOS GAMMA")
    print("="*70)

    try:
        print("Ingresa los parámetros:")
        alpha = float(input("  Parámetro de forma (α > 0): "))
        if alpha <= 0:
            print("  ❌ α debe ser mayor que 0.")
            return

        beta = float(input("  Parámetro de tasa (β > 0): "))
        if beta <= 0:
            print("  ❌ β debe ser mayor que 0.")
            return

        n = int(input("  Número de observaciones a generar: "))
        if n <= 0:
            print("  ❌ n debe ser mayor que 0.")
            return

        theta = 1 / beta

        # Generar datos
        np.random.seed(42)
        datos = np.random.gamma(shape=alpha, scale=theta, size=n)

        print(f"\nEstadísticas de los datos generados:")
        print("-"*70)
        print(f"  Media teórica: {alpha*theta:.4f}")
        print(f"  Media muestral: {np.mean(datos):.4f}")
        print(f"  Varianza teórica: {alpha*theta**2:.4f}")
        print(f"  Varianza muestral: {np.var(datos):.4f}")
        print(f"  Mediana muestral: {np.median(datos):.4f}")
        print(f"  Mínimo: {np.min(datos):.4f}")
        print(f"  Máximo: {np.max(datos):.4f}")

        # Visualización
        fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

        # Histograma con curva teórica
        ax1.hist(datos, bins=30, density=True, alpha=0.6, color='#2E86AB',
                edgecolor='black', label='Datos generados')

        x = np.linspace(0, max(datos), 1000)
        y_teorica = gamma.pdf(x, alpha, scale=theta)
        ax1.plot(x, y_teorica, 'r-', linewidth=2, label='Teórica')

        ax1.set_xlabel('X', fontsize=12)
        ax1.set_ylabel('Densidad', fontsize=12)
        ax1.set_title('Histograma vs Curva Teórica', fontsize=13, fontweight='bold')
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='best')

        # Q-Q plot
        from scipy import stats
        stats.probplot(datos, dist="gamma", sparams=(alpha, 0, theta), plot=ax2)
        ax2.set_title('Gráfico Q-Q Gamma', fontsize=13, fontweight='bold')
        ax2.grid(True, alpha=0.3)

        # CDF empírica vs teórica
        from statsmodels.distributions.empirical_distribution import ECDF
        ecdf = ECDF(datos)

        x_ecdf = np.linspace(0, max(datos), 1000)
        ax3.plot(x_ecdf, ecdf(x_ecdf), 'b-', linewidth=2, label='Empírica')
        ax3.plot(x_ecdf, gamma.cdf(x_ecdf, alpha, scale=theta), 'r--', linewidth=2, label='Teórica')

        ax3.set_xlabel('X', fontsize=12)
        ax3.set_ylabel('CDF', fontsize=12)
        ax3.set_title('CDF Empírica vs Teórica', fontsize=13, fontweight='bold')
        ax3.grid(True, alpha=0.3)
        ax3.legend(loc='best')
        ax3.set_ylim(0, 1.05)

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_gamma():
    """Ejecuta ejemplos clásicos de la distribución Gamma"""

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN GAMMA")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Tiempo de espera (Erlang)',
            'descripcion': 'En promedio llegan 2 clientes por hora (β=2). ¿Cuál es la probabilidad de que el tiempo hasta que lleguen 3 clientes sea mayor a 2 horas? (α=3)',
            'alpha': 3,
            'beta': 2,
            'a': 2,
            'tipo': 'mayor'
        },
        {
            'titulo': 'Ejemplo 2: Tiempo de vida',
            'descripcion': 'Un componente tiene vida Gamma con α=2, β=0.5. ¿Cuál es la probabilidad de que dure menos de 3 horas?',
            'alpha': 2,
            'beta': 0.5,
            'a': 3,
            'tipo': 'menor'
        },
        {
            'titulo': 'Ejemplo 3: Chi-cuadrado como caso especial',
            'descripcion': 'Para α=5, β=0.5 (χ² con 10 df), ¿cuál es el percentil 95?',
            'alpha': 5,
            'beta': 0.5,
            'p': 0.95,
            'tipo': 'cuantil'
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")

        alpha = ej['alpha']
        beta = ej['beta']
        theta = 1/beta

        if ej['tipo'] == 'mayor':
            prob = 1 - gamma.cdf(ej['a'], alpha, scale=theta)
            print(f"   α={alpha:.1f}, β={beta:.2f}, θ={theta:.2f}")
            print(f"   P(X > {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'menor':
            prob = gamma.cdf(ej['a'], alpha, scale=theta)
            print(f"   α={alpha:.1f}, β={beta:.2f}, θ={theta:.2f}")
            print(f"   P(X ≤ {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'cuantil':
            x = gamma.ppf(ej['p'], alpha, scale=theta)
            print(f"   α={alpha:.1f}, β={beta:.2f}, θ={theta:.2f}")
            print(f"   Percentil {ej['p']*100:.0f}% = {x:.4f}")
            if alpha == 5 and beta == 0.5:
                print(f"   (χ² con 10 grados de libertad)")

    print("\n" + "="*70)

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN GAMMA")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado")
        print("  2. Ver ejemplos clásicos")
        print("  3. Generar datos Gamma")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_gamma()
        elif opcion == '2':
            ejemplos_gamma()
        elif opcion == '3':
            generar_datos_gamma()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```

#**a. Construya el gráfico de la Distribución dada si  $α=5$ y $β=8$**
+ Una variable aleatoria continua $X$ sigue una distribución *Gamma* de parámetros $β$, lo cual denotamos por $$X∼gamma(α,β)$$ si tiene funcion FDP dada por la formula:
$$ f(x,α,β) = \frac{1}{β^αΓ(α)}x^{α-1}exp{\{\frac{-x}{β}\}}, \ x\geq 0, 0<α,β<∞, $$


```python
# **Librerias para usar en esta parte del cursor**
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
from sympy.functions.combinatorial.factorials import factorial
print("Setup Complete")

#@title **a. Construya el gráfico de la Distribución dada si  $α=5$ y $β=8$**
from numpy.ma.core import log
from sympy import integrate, init_printing
from sympy.abc import x
# nuestra libreria
import numpy as np
import matplotlib.pyplot as plt
# Definimos una funcion
a = 5 # parametro alpha
b = 8 # parametro beta

def dis_gamma(x):
  return np.exp(-x/b)*(x**(a-1))/((b**a)*factorial(a-1))
# los valores que va a tomar x
x = np.linspace(0,200,1000)
fig, ax = plt.subplots() #lugar donde vamos a estar graficando, ax objeto y subplot porque vamos hacer varios graficos
ax.plot(x,dis_gamma(x)) # variable independiente y la funcion
plt.title('Distribución Gamma α=5 y β=8')
plt.ylabel('Probabilidad')
plt.xlabel('Consumo de Energia Kw/hora')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **55. La Distribución Weibull**

# **a. Generalidades de la Ditribución Weibull**
+ **Wikipedia**. La distribución de Weibull es una distribución de probabilidad continua. Recibe su nombre de Waloddi Weibull, que la describió detalladamente en 1951, aunque fue descubierta inicialmente por Fréchet (1927) y aplicada por primera vez por Rosin y Rammler (1933) para describir la distribución de los tamaños de determinadas partículas.

+ **Ejemplos para la distribución Weibull**: Veamos algunos casos donde esta distribución es aplicable:

 - Análisis de la supervivencia.
 - En ingeniería, para modelar procesos estocásticos relacionados con el tiempo de fabricación y distribución de bienes.

 - Aplicación de la distribución de probabilidad acumulada de Weibull a lluvias diárias máximas.
 - Teoría de valores extremos
 - Meteorología
 - Para modelar la distribución de la velocidad del viento (frecuencia con la que se dan diferentes velocidades de viento)
 - En telecomunicaciones
 - En sistemas de radar para simular la dispersión de la señal recibida
 - En energía solar, para modelar la distribución de irradiación solar anual
 - En seguros, para modelar el tamaño de las pérdidas
 - En la hidrología, se utiliza la distribución de Weibull para analizar variables aleatorias como valores máximos de la precipitación y la descarga de ríos,​ y además para describir épocas de sequía.

# **b. La Distribución Weibull - Definición**

+ **Función de Densidad**. Si $X$ es una variable aleatoria continua, se dice que $X$ tiene una distribución Weibull con parámetros ${\displaystyle \alpha ,β >0}$ y escribimos $X∼Weibull(α,β)$ si su función de densidad es

$${\displaystyle f(x;\alpha, \beta)=\frac{αx^{α-1}}{β^α}e^{-(\frac{x}{β})^{\alpha }}x\geq 0}$$


+ Tambien $$E(X)=\mu=β\Gamma(1+\frac{1}{α}),  \ \ \ \ V(X)=σ^2=β^2[\Gamma(1+\frac{2}{α})-\Gamma^2(1+\frac{1}{α})]$$

+ En dónde $ \{X\leq x\}$ representa al evento de todos resultados menores que $x$.
+ $\mu$ representa al número esperado.
+ $V(X)$ representa a la varianza.
+ Se puede mostrar

+ La *función Gamma* satisface $\Gamma(n) = (n-1)!\quad\forall n\in\mathbb N$ a través de la integral de Euler

$$
\Gamma(α) = \int_0^\infty x^{α-1}e^{-x}dx\,.
$$


# **c. Veamos que es una función de densidad**
+ $$ ∫_{-∞}^{∞}f(x)dx= ∫_{0}^{∞}\frac{α}{β^α}x^{α-1}exp{\{\frac{-x}{β}\}^α}dx$$

+ **hacemos** $y = (\frac{x}{β})^α$ esto es $x = βy^{1/α} ⇒dx= (β/α)y^{1/α-1}dy$
+  $$ ∫_{-∞}^{∞}f(x)dx=\frac{α}{β^α}∫_{0}^{∞}(βy^{1/α})^{α-1}e^{-y}(β/α)y^{1/α-1}dy$$
+ **Simplificando**

+  $$ ∫_{-∞}^{∞}f(x)dx=\frac{1}{1}∫_{0}^{∞}(y^{1/α})^{α-1}e^{-y}y^{1/α-1}dy$$

+  $$ ∫_{-∞}^{∞}f(x)dx=∫_{0}^{∞}(y^{1-1/α})e^{-y}y^{1/α-1}dy=∫_{0}^{∞}e^{-y}dy=1$$

# **d. Valor Esperado**
+ $$E(X)=μ= ∫_{-∞}^{∞}xf(x)dx= ∫_{0}^{∞}x\frac{α}{β^α}x^{α-1}exp{\{\frac{-x}{β}\}^α}dx$$

+ $$E(X)= ∫_{0}^{∞}\frac{α}{β^α}x^{α}exp{\{\frac{-x}{β}\}^α}dx$$

+ **hacemos** $y = (\frac{x}{β})^α$ esto es $x = βy^{1/α} ⇒dx= (β/α)y^{1/α-1}dy$
+ $$E(X)=μ=\frac{α}{β^α}∫_{0}^{∞}(βy^{1/α})^{α}e^{-y}(β/α)y^{1/α-1}dy$$
+ **Simplificando**

+  $$E(X)=\frac{β}{1}∫_{0}^{∞}(y^{1/α})^{α}e^{-y}y^{1/α-1}dy$$

+  $$E(X)=β∫_{0}^{∞}(y^{1})e^{-y}y^{1/α-1}dy=β∫_{0}^{∞}y^{(1/α+1)-1}e^{-y}dy$$

+ **De la definición**
$$
\Gamma(α) = \int_0^\infty x^{α-1}e^{-x}dx\,.
$$


+  $$E(X)=β\Gamma(\frac{1}{α}+1)$$



# **e. Varianza de una Variable aleatoria Weibull**
+ **Usando el mismo procedimiento arriba, pruebe que** $$E(X^2)=μ= ∫_{-∞}^{∞}x^2f(x)dx= ∫_{0}^{∞}x^2\frac{α}{β^α}x^{α-1}exp{\{\frac{-x}{β}\}^α}dx=β^2\Gamma(\frac{2}{α}+1)$$
+ Por tanto $$V(X)=β^2\Gamma(\frac{2}{α}+1)-β^2\Gamma^2(\frac{1}{α}+1)$$

# **56.  Calculadora Distribución Weibull**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN WEIBULL**
#
# La distribución Weibull es una de las distribuciones más
# utilizadas en ingeniería de confiabilidad y análisis de
# tiempos de vida. Se caracteriza por su flexibilidad para
# modelar diferentes tasas de fallo.
#
# Parámetros:
#   λ (escala) - parámetro de escala (λ > 0)
#   k (forma) - parámetro de forma (k > 0)
#
# Fórmula: f(x) = (k/λ) * (x/λ)^(k-1) * e^(-(x/λ)^k)
#
# Casos especiales:
#   k = 1 → Distribución Exponencial
#   k = 2 → Distribución Rayleigh
#   k ≈ 3.4 → Aproximación Normal
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import weibull_min, expon, norm, rayleigh
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
try:
    plt.style.use('seaborn-v0_8')
except:
    try:
        plt.style.use('seaborn')
    except:
        plt.style.use('default')

plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_weibull():
    """
    Calculadora interactiva para la distribución Weibull
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN WEIBULL")
    print("="*70)
    print("\nLa distribución Weibull es fundamental en:")
    print("  • Análisis de confiabilidad y tiempos de vida")
    print("  • Ingeniería de mantenimiento")
    print("  • Análisis de supervivencia")
    print("  • Control de calidad\n")

    print("Casos especiales:")
    print("  • k = 1 → Distribución Exponencial (tasa de fallo constante)")
    print("  • k = 2 → Distribución Rayleigh (tasa de fallo creciente)")
    print("  • k < 1 → Tasa de fallo decreciente (fallos tempranos)")
    print("  • k > 1 → Tasa de fallo creciente (desgaste)")
    print("  • k ≈ 3.4 → Aproximación Normal\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            lam = float(input("  Parámetro de escala (λ > 0): "))
            if lam <= 0:
                print("  ❌ λ debe ser mayor que 0. Intenta de nuevo.")
                continue

            k = float(input("  Parámetro de forma (k > 0): "))
            if k <= 0:
                print("  ❌ k debe ser mayor que 0. Intenta de nuevo.")
                continue

            print(f"\n  λ = {lam:.4f} (escala)")
            print(f"  k = {k:.4f} (forma)")

            print("\nTipo de cálculo:")
            print("  1. P(X = a)  - Probabilidad puntual (siempre 0)")
            print("  2. P(X ≤ a)  - Probabilidad acumulada a la izquierda")
            print("  3. P(X ≥ a)  - Probabilidad acumulada a la derecha")
            print("  4. P(a ≤ X ≤ b) - Probabilidad entre dos valores")
            print("  5. Cuantil - Encontrar X dado un percentil")
            print("  6. Tasa de fallo - h(x)")
            print("  7. Comparar con diferentes k")

            tipo = input("\nSelecciona el tipo de cálculo (1-7): ")

            if tipo == '1':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 0
                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X = {a}) = 0")
                print("En distribuciones continuas, la probabilidad puntual es siempre 0.")
                print(f"Densidad en X={a}: {weibull_min.pdf(a, k, scale=lam):.6f}")

                visualizar_weibull(lam, k, a, None, prob, tipo)
                return {'lam': lam, 'k': k, 'a': a}

            elif tipo == '2':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = weibull_min.cdf(a, k, scale=lam)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≤ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con λ={lam:.4f}, k={k:.4f}")

                visualizar_weibull(lam, k, a, None, prob, tipo)
                return {'lam': lam, 'k': k, 'a': a, 'prob': prob}

            elif tipo == '3':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 1 - weibull_min.cdf(a, k, scale=lam)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≥ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con λ={lam:.4f}, k={k:.4f}")

                visualizar_weibull(lam, k, a, None, prob, tipo)
                return {'lam': lam, 'k': k, 'a': a, 'prob': prob}

            elif tipo == '4':
                a = float(input("  Límite inferior (a ≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                b = float(input("  Límite superior (b > a): "))
                if b <= a:
                    print("  ❌ b debe ser mayor que a. Intenta de nuevo.")
                    continue
                prob = weibull_min.cdf(b, k, scale=lam) - weibull_min.cdf(a, k, scale=lam)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP({a} ≤ X ≤ {b}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con λ={lam:.4f}, k={k:.4f}")

                visualizar_weibull(lam, k, a, b, prob, tipo)
                return {'lam': lam, 'k': k, 'a': a, 'b': b, 'prob': prob}

            elif tipo == '5':
                p = float(input("  Percentil (0-1): "))
                if p <= 0 or p >= 1:
                    print("  ❌ El percentil debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                x = weibull_min.ppf(p, k, scale=lam)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nPara el percentil {p*100:.1f}%:")
                print(f"X = {x:.6f}")
                print(f"Con λ={lam:.4f}, k={k:.4f}")
                print(f"Interpretación: El {p*100:.1f}% de los datos están por debajo de {x:.4f}")

                visualizar_cuantil_weibull(lam, k, x, p)
                return {'lam': lam, 'k': k, 'percentil': p, 'x': x}

            elif tipo == '6':
                tasa_fallo_weibull(lam, k)
                return {'lam': lam, 'k': k}

            elif tipo == '7':
                comparar_k_weibull(lam, k)
                return {'lam': lam, 'k': k}

            else:
                print("  ❌ Opción no válida. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

#==========================================================
# FUNCIÓN PARA VISUALIZAR LA DISTRIBUCIÓN WEIBULL
#==========================================================

def visualizar_weibull(lam, k, a, b, prob, tipo):
    """Crea visualizaciones completas de la distribución Weibull"""

    # Rango de valores para la gráfica
    x_max = max(weibull_min.ppf(0.999, k, scale=lam), a * 1.5 if a > 0 else weibull_min.ppf(0.999, k, scale=lam))
    x_max = max(x_max, 5)
    x = np.linspace(0.001, x_max, 1000)
    y = weibull_min.pdf(x, k, scale=lam)

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    #==========================================================
    # GRÁFICO 1: Distribución con área sombreada
    #==========================================================
    ax1 = plt.subplot(2, 2, 1)
    ax1.plot(x, y, 'b-', linewidth=2.5, label=f'Weibull(λ={lam:.2f}, k={k:.2f})')

    # Sombrear el área según el tipo de cálculo
    if tipo == '2':  # P(X ≤ a)
        mask = x <= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≤ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '3':  # P(X ≥ a)
        mask = x >= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≥ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '4':  # P(a ≤ X ≤ b)
        mask = (x >= a) & (x <= b)
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P({a:.2f} ≤ X ≤ {b:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)
        ax1.axvline(x=b, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    # Estadísticas
    from scipy.special import gamma as gamma_func
    media = lam * gamma_func(1 + 1/k)
    varianza = lam**2 * (gamma_func(1 + 2/k) - gamma_func(1 + 1/k)**2)
    desv = np.sqrt(varianza)
    mediana = lam * (np.log(2))**(1/k)
    moda = lam * ((k-1)/k)**(1/k) if k > 1 else 0

    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.2f}')
    if moda > 0:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Moda = {moda:.2f}')
    ax1.axvline(x=mediana, color='#2ECC71', linestyle='--', linewidth=1.5, alpha=0.7, label=f'Mediana = {mediana:.2f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Weibull (λ={lam:.2f}, k={k:.2f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', fontsize=10)
    ax1.set_xlim(0, x_max)

    # Agregar información de probabilidad
    if prob is not None:
        ax1.text(0.98, 0.95, f'Probabilidad = {prob:.6f}\n({prob*100:.2f}%)',
                transform=ax1.transAxes, fontsize=11, verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 2: Comparación con diferentes k
    #==========================================================
    ax2 = plt.subplot(2, 2, 2)

    ks = [0.5, 1, 2, 3, 5]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    x2 = np.linspace(0.001, x_max, 1000)
    for k_val, color in zip(ks, colores):
        y_temp = weibull_min.pdf(x2, k_val, scale=lam)
        ax2.plot(x2, y_temp, color=color, linewidth=2,
                label=f'k = {k_val}')

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax2.set_title(f'Efecto de k (λ={lam:.2f} fijo)', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best', fontsize=9)
    ax2.set_xlim(0, x_max)

    #==========================================================
    # GRÁFICO 3: Función de distribución acumulada
    #==========================================================
    ax3 = plt.subplot(2, 2, 3)
    cdf = weibull_min.cdf(x, k, scale=lam)
    ax3.plot(x, cdf, 'g-', linewidth=2.5, label='CDF')
    ax3.fill_between(x, 0, cdf, alpha=0.2, color='green')

    # Puntos de interés
    if tipo == '2':  # P(X ≤ a)
        ax3.scatter(a, prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {prob:.4f}')
        ax3.axhline(y=prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '3':  # P(X ≥ a)
        ax3.scatter(a, 1-prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {1-prob:.4f}')
        ax3.axhline(y=1-prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles comunes
    for q in [0.25, 0.5, 0.75, 0.9, 0.95]:
        q_val = weibull_min.ppf(q, k, scale=lam)
        if q_val <= x_max:
            ax3.axvline(x=q_val, color='orange', linestyle=':', linewidth=1.5, alpha=0.5)
            ax3.text(q_val, 0.01, f'Q{int(q*100)}', ha='center', fontsize=8, color='orange')

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax3.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', fontsize=10)
    ax3.set_ylim(-0.05, 1.05)
    ax3.set_xlim(0, x_max)

    #==========================================================
    # GRÁFICO 4: Tabla de estadísticas
    #==========================================================
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    from scipy.special import gamma as gamma_func

    # Calcular estadísticas
    media = lam * gamma_func(1 + 1/k)
    varianza = lam**2 * (gamma_func(1 + 2/k) - gamma_func(1 + 1/k)**2)
    desv = np.sqrt(varianza)
    mediana = lam * (np.log(2))**(1/k)
    moda = lam * ((k-1)/k)**(1/k) if k > 1 else 0

    # Crear tabla
    table_data = [
        ['Estadística', 'Valor', 'Fórmula'],
        ['Media (μ)', f'{media:.4f}', 'λ·Γ(1+1/k)'],
        ['Varianza (σ²)', f'{varianza:.4f}', 'λ²[Γ(1+2/k)-Γ(1+1/k)²]'],
        ['Desv. Estándar (σ)', f'{desv:.4f}', '√varianza'],
        ['Mediana', f'{mediana:.4f}', 'λ·(ln2)^(1/k)'],
        ['Moda', f'{moda:.4f}' if moda > 0 else '0', 'λ·((k-1)/k)^(1/k)'],
        ['Tasa de fallo', 'h(x)', 'k/λ · (x/λ)^(k-1)'],
        ['Confiabilidad', 'R(x)', 'e^(-(x/λ)^k)']
    ]

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.3, 0.3, 0.4])

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.5)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title('Propiedades de la Distribución Weibull',
                  fontsize=13, fontweight='bold', pad=20)

    #==========================================================
    # Ajustar layout
    #==========================================================
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # GRÁFICO ADICIONAL: Función de supervivencia y tasa de fallo
    #==========================================================
    fig2, (ax5, ax6) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 5: Función de supervivencia (confiabilidad)
    surv = 1 - cdf
    ax5.plot(x, surv, 'r-', linewidth=2.5, label='R(x) = P(X > x)')
    ax5.fill_between(x, 0, surv, alpha=0.2, color='red')

    # Línea de la mediana
    ax5.axhline(y=0.5, color='gray', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.axvline(x=mediana, color='gray', linestyle='--', linewidth=1.5, alpha=0.5)
    ax5.text(mediana, 0.52, f'Mediana = {mediana:.2f}', ha='center', fontsize=9)

    ax5.set_xlabel('X', fontsize=12)
    ax5.set_ylabel('Confiabilidad R(X)', fontsize=12)
    ax5.set_title('Función de Confiabilidad (Supervivencia)', fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.legend(loc='best')
    ax5.set_xlim(0, x_max)
    ax5.set_ylim(0, 1.05)

    # Gráfico 6: Tasa de fallo
    hazard = (k/lam) * (x/lam)**(k-1)
    ax6.plot(x, hazard, 'purple', linewidth=2.5, label='h(x) = (k/λ)(x/λ)^(k-1)')
    ax6.fill_between(x, 0, hazard, alpha=0.2, color='purple')

    # Interpretación de la tasa de fallo
    if k < 1:
        ax6.text(0.7, 0.9, 'Tasa decreciente\n(Fallos tempranos)',
                transform=ax6.transAxes, fontsize=11, ha='center',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
    elif k == 1:
        ax6.text(0.7, 0.9, 'Tasa constante\n(Exponencial)',
                transform=ax6.transAxes, fontsize=11, ha='center',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
    else:
        ax6.text(0.7, 0.9, 'Tasa creciente\n(Desgaste)',
                transform=ax6.transAxes, fontsize=11, ha='center',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    ax6.set_xlabel('X', fontsize=12)
    ax6.set_ylabel('Tasa de fallo h(x)', fontsize=12)
    ax6.set_title('Tasa de Fallo', fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3, linestyle='--')
    ax6.legend(loc='best')
    ax6.set_xlim(0, x_max)

    plt.tight_layout()

    #==========================================================
    # MOSTRAR GRÁFICOS
    #==========================================================
    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Weibull_lam{lam:.2f}_k{k:.2f}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_confiabilidad.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_confiabilidad.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

def visualizar_cuantil_weibull(lam, k, x, p):
    """Visualiza el cuantil en la distribución Weibull"""

    x_max = max(weibull_min.ppf(0.999, k, scale=lam), x * 1.5)
    x_vals = np.linspace(0.001, x_max, 1000)
    y = weibull_min.pdf(x_vals, k, scale=lam)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Distribución con cuantil marcado
    ax1.plot(x_vals, y, 'b-', linewidth=2.5)
    mask = x_vals <= x
    ax1.fill_between(x_vals[mask], 0, y[mask], color='#E63946', alpha=0.5,
                     label=f'Área = {p*100:.1f}%')
    ax1.axvline(x=x, color='#E63946', linestyle='--', linewidth=2, label=f'X = {x:.2f}')

    # Media y mediana
    from scipy.special import gamma as gamma_func
    media = lam * gamma_func(1 + 1/k)
    mediana = lam * (np.log(2))**(1/k)

    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.2f}')
    ax1.axvline(x=mediana, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Mediana = {mediana:.2f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Percentil {p*100:.1f}% (λ={lam:.2f}, k={k:.2f})',
                  fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best')
    ax1.set_xlim(0, x_max)

    # Gráfico 2: Función de distribución acumulada
    cdf = weibull_min.cdf(x_vals, k, scale=lam)
    ax2.plot(x_vals, cdf, 'g-', linewidth=2.5)
    ax2.scatter(x, p, color='#E63946', s=150, zorder=5, label=f'F({x:.2f}) = {p:.4f}')
    ax2.axhline(y=p, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax2.axvline(x=x, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax2.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best')
    ax2.set_ylim(-0.05, 1.05)
    ax2.set_xlim(0, x_max)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA TASA DE FALLO
#==========================================================

def tasa_fallo_weibull(lam, k):
    """Analiza la tasa de fallo de la distribución Weibull"""

    print("\n" + "="*70)
    print("📈 ANÁLISIS DE TASA DE FALLO")
    print("="*70)

    print(f"\nPara Weibull con λ={lam:.4f}, k={k:.4f}:")
    print(f"  h(x) = (k/λ) * (x/λ)^(k-1)")

    # Interpretación según k
    print("\nInterpretación de la tasa de fallo:")
    if k < 1:
        print("  • k < 1: Tasa de fallo DECRECIENTE")
        print("  → Fallos tempranos (mortalidad infantil)")
        print("  → Componentes mejoran con el tiempo")
    elif k == 1:
        print("  • k = 1: Tasa de fallo CONSTANTE")
        print("  → Distribución Exponencial")
        print("  → Fallos aleatorios (vida útil)")
    else:
        print("  • k > 1: Tasa de fallo CRECIENTE")
        print("  → Fallos por desgaste")
        print("  → Componentes se degradan con el tiempo")

    # Calcular tasa de fallo en algunos puntos
    puntos = [0.5 * lam, lam, 1.5 * lam, 2 * lam, 3 * lam]

    print("\nTasa de fallo en puntos específicos:")
    print("-"*70)
    print(f"{'X':^10} | {'h(x)':^12} | {'Interpretación'}")
    print("-"*70)
    for x in puntos:
        hazard = (k/lam) * (x/lam)**(k-1)
        if k < 1:
            interpre = 'Decreciente' if x < lam else 'Continúa'
        elif k == 1:
            interpre = 'Constante'
        else:
            interpre = 'Creciente' if x > lam else 'Inicio'
        print(f"{x:^10.2f} | {hazard:^12.4f} | {interpre}")

    # Visualización
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    x = np.linspace(0.001, 4*lam, 1000)

    # Gráfico 1: Tasa de fallo
    hazard = (k/lam) * (x/lam)**(k-1)
    ax1.plot(x, hazard, 'purple', linewidth=2.5, label='h(x)')
    ax1.fill_between(x, 0, hazard, alpha=0.2, color='purple')
    ax1.axvline(x=lam, color='gray', linestyle='--', linewidth=1.5, alpha=0.5, label='λ')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Tasa de fallo h(x)', fontsize=12)
    ax1.set_title('Tasa de Fallo de Weibull', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='best')

    # Gráfico 2: Comparación de tasas para diferentes k
    ks = [0.5, 1, 1.5, 2, 3]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    for k_val, color in zip(ks, colores):
        hazard_k = (k_val/lam) * (x/lam)**(k_val-1)
        ax2.plot(x, hazard_k, color=color, linewidth=2,
                label=f'k = {k_val}')

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Tasa de fallo', fontsize=12)
    ax2.set_title('Comparación de tasas para diferentes k', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')
    ax2.set_xlim(0, 4*lam)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA COMPARAR DIFERENTES k
#==========================================================

def comparar_k_weibull(lam, k_actual):
    """Compara diferentes valores de k en la distribución Weibull"""

    print("\n" + "="*70)
    print("📊 COMPARACIÓN DE DIFERENTES k")
    print("="*70)

    ks = [0.5, 1, 1.5, 2, 3, 5]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#9B59B6', '#1ABC9C']

    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    x = np.linspace(0.001, 4*lam, 1000)

    # Gráfico 1: Densidades
    ax1 = axes[0, 0]
    for k_val, color in zip(ks, colores):
        y = weibull_min.pdf(x, k_val, scale=lam)
        ax1.plot(x, y, color=color, linewidth=2,
                label=f'k = {k_val}', alpha=0.8 if k_val != k_actual else 1.0)
        if k_val == k_actual:
            ax1.fill_between(x, 0, y, alpha=0.2, color=color)

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad', fontsize=12)
    ax1.set_title('Densidades para diferentes k', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='best')
    ax1.set_xlim(0, 4*lam)

    # Gráfico 2: CDF
    ax2 = axes[0, 1]
    for k_val, color in zip(ks, colores):
        y = weibull_min.cdf(x, k_val, scale=lam)
        ax2.plot(x, y, color=color, linewidth=2,
                label=f'k = {k_val}', alpha=0.8 if k_val != k_actual else 1.0)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('CDF', fontsize=12)
    ax2.set_title('Funciones de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')
    ax2.set_xlim(0, 4*lam)
    ax2.set_ylim(0, 1.05)

    # Gráfico 3: Tasa de fallo
    ax3 = axes[1, 0]
    for k_val, color in zip(ks, colores):
        hazard = (k_val/lam) * (x/lam)**(k_val-1)
        ax3.plot(x, hazard, color=color, linewidth=2,
                label=f'k = {k_val}', alpha=0.8 if k_val != k_actual else 1.0)

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Tasa de fallo', fontsize=12)
    ax3.set_title('Tasas de Fallo', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3)
    ax3.legend(loc='best')
    ax3.set_xlim(0, 4*lam)

    # Gráfico 4: Confiabilidad
    ax4 = axes[1, 1]
    for k_val, color in zip(ks, colores):
        surv = 1 - weibull_min.cdf(x, k_val, scale=lam)
        ax4.plot(x, surv, color=color, linewidth=2,
                label=f'k = {k_val}', alpha=0.8 if k_val != k_actual else 1.0)

    ax4.set_xlabel('X', fontsize=12)
    ax4.set_ylabel('Confiabilidad R(X)', fontsize=12)
    ax4.set_title('Funciones de Confiabilidad', fontsize=13, fontweight='bold')
    ax4.grid(True, alpha=0.3)
    ax4.legend(loc='best')
    ax4.set_xlim(0, 4*lam)
    ax4.set_ylim(0, 1.05)

    plt.tight_layout()
    plt.show()

    # Mostrar resumen de características
    print("\nResumen de características según k:")
    print("-"*70)
    print(f"{'k':^10} | {'Comportamiento':^30} | {'Aplicación típica'}")
    print("-"*70)
    print(f"{'k < 1':^10} | {'Tasa decreciente':^30} | {'Fallos tempranos'}")
    print(f"{'k = 1':^10} | {'Tasa constante':^30} | {'Vida útil (Exponencial)'}")
    print(f"{'1 < k < 2':^10} | {'Tasa creciente cóncava':^30} | {'Desgaste inicial'}")
    print(f"{'k = 2':^10} | {'Tasa creciente lineal':^30} | {'Rayleigh'}")
    print(f"{'k > 2':^10} | {'Tasa creciente convexa':^30} | {'Desgaste acelerado'}")
    print("-"*70)

#==========================================================
# FUNCIÓN PARA GENERAR DATOS WEIBULL
#==========================================================

def generar_datos_weibull():
    """Genera y visualiza datos de una distribución Weibull"""

    print("\n" + "="*70)
    print("📊 GENERACIÓN DE DATOS WEIBULL")
    print("="*70)

    try:
        print("Ingresa los parámetros:")
        lam = float(input("  Parámetro de escala (λ > 0): "))
        if lam <= 0:
            print("  ❌ λ debe ser mayor que 0.")
            return

        k = float(input("  Parámetro de forma (k > 0): "))
        if k <= 0:
            print("  ❌ k debe ser mayor que 0.")
            return

        n = int(input("  Número de observaciones a generar: "))
        if n <= 0:
            print("  ❌ n debe ser mayor que 0.")
            return

        # Generar datos
        np.random.seed(42)
        datos = weibull_min.rvs(k, scale=lam, size=n)

        print(f"\nEstadísticas de los datos generados:")
        print("-"*70)

        from scipy.special import gamma as gamma_func
        media_teorica = lam * gamma_func(1 + 1/k)
        varianza_teorica = lam**2 * (gamma_func(1 + 2/k) - gamma_func(1 + 1/k)**2)

        print(f"  Media teórica: {media_teorica:.4f}")
        print(f"  Media muestral: {np.mean(datos):.4f}")
        print(f"  Varianza teórica: {varianza_teorica:.4f}")
        print(f"  Varianza muestral: {np.var(datos):.4f}")
        print(f"  Mediana muestral: {np.median(datos):.4f}")
        print(f"  Mínimo: {np.min(datos):.4f}")
        print(f"  Máximo: {np.max(datos):.4f}")

        # Visualización
        fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

        # Histograma con curva teórica
        ax1.hist(datos, bins=30, density=True, alpha=0.6, color='#2E86AB',
                edgecolor='black', label='Datos generados')

        x = np.linspace(0, max(datos), 1000)
        y_teorica = weibull_min.pdf(x, k, scale=lam)
        ax1.plot(x, y_teorica, 'r-', linewidth=2, label='Teórica')

        ax1.set_xlabel('X', fontsize=12)
        ax1.set_ylabel('Densidad', fontsize=12)
        ax1.set_title('Histograma vs Curva Teórica', fontsize=13, fontweight='bold')
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='best')

        # Q-Q plot
        stats.probplot(datos, dist="weibull_min", sparams=(k, 0, lam), plot=ax2)
        ax2.set_title('Gráfico Q-Q Weibull', fontsize=13, fontweight='bold')
        ax2.grid(True, alpha=0.3)

        # CDF empírica vs teórica
        from statsmodels.distributions.empirical_distribution import ECDF
        ecdf = ECDF(datos)

        x_ecdf = np.linspace(0, max(datos), 1000)
        ax3.plot(x_ecdf, ecdf(x_ecdf), 'b-', linewidth=2, label='Empírica')
        ax3.plot(x_ecdf, weibull_min.cdf(x_ecdf, k, scale=lam), 'r--', linewidth=2, label='Teórica')

        ax3.set_xlabel('X', fontsize=12)
        ax3.set_ylabel('CDF', fontsize=12)
        ax3.set_title('CDF Empírica vs Teórica', fontsize=13, fontweight='bold')
        ax3.grid(True, alpha=0.3)
        ax3.legend(loc='best')
        ax3.set_ylim(0, 1.05)

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_weibull():
    """Ejecuta ejemplos clásicos de la distribución Weibull"""

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN WEIBULL")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Confiabilidad de componentes',
            'descripcion': 'Un componente tiene vida Weibull con λ=1000 horas y k=1.5. ¿Cuál es la probabilidad de que dure más de 800 horas?',
            'lam': 1000,
            'k': 1.5,
            'a': 800,
            'tipo': 'mayor'
        },
        {
            'titulo': 'Ejemplo 2: Tiempo de falla',
            'descripcion': 'Un sistema tiene Weibull con λ=500 y k=0.8. ¿Cuál es la probabilidad de que falle antes de 300 horas?',
            'lam': 500,
            'k': 0.8,
            'a': 300,
            'tipo': 'menor'
        },
        {
            'titulo': 'Ejemplo 3: Vida útil garantizada',
            'descripcion': 'Un producto tiene Weibull con λ=2000 y k=2. ¿Cuál es el tiempo para el cual el 90% de los productos aún funcionan?',
            'lam': 2000,
            'k': 2,
            'p': 0.10,  # 10% de fallos, 90% de confiabilidad
            'tipo': 'cuantil'
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")

        lam = ej['lam']
        k = ej['k']

        if ej['tipo'] == 'mayor':
            prob = 1 - weibull_min.cdf(ej['a'], k, scale=lam)
            print(f"   λ={lam:.0f}, k={k:.2f}")
            print(f"   P(X > {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'menor':
            prob = weibull_min.cdf(ej['a'], k, scale=lam)
            print(f"   λ={lam:.0f}, k={k:.2f}")
            print(f"   P(X ≤ {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'cuantil':
            # Para confiabilidad, usamos el percentil complementario
            x = weibull_min.ppf(ej['p'], k, scale=lam)
            print(f"   λ={lam:.0f}, k={k:.2f}")
            print(f"   Tiempo para {ej['p']*100:.0f}% de fallos: {x:.2f}")
            print(f"   → {100-ej['p']*100:.0f}% de los productos duran más de {x:.2f} horas")

    print("\n" + "="*70)

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN WEIBULL")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado")
        print("  2. Ver ejemplos clásicos")
        print("  3. Generar datos Weibull")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_weibull()
        elif opcion == '2':
            ejemplos_weibull()
        elif opcion == '3':
            generar_datos_weibull()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```

#**a. Construya el gráfico de la Distribución para la Región Montañosa si  $α=11.7$ y $β=7.24$**
+ Una variable aleatoria continua $X$ sigue una distribución *Gamma* de parámetros $β$, lo cual denotamos por $$X∼Weibull(α,β)$$ si tiene funcion FDP dada por la formula:
$${\displaystyle f(x;\alpha, \beta)=\frac{αx^{α-1}}{β^α}e^{-(\frac{x}{β})^{\alpha }}x\geq 0}$$


```python
#@title **a. Manual - Construya el gráfico de la Distribución dada si  $α=11.7$ y $β=7.24$**
from numpy.ma.core import log
from sympy import integrate, init_printing
from sympy.abc import x
# nuestra libreria
import numpy as np
import matplotlib.pyplot as plt
# Definimos una funcion
a = 11.7 # parametro alpha
b = 7.24 # parametro beta

def dis_weibull(x):
  return np.exp(-(x/b)**a)*a*(x**(a-1))/(b**a)
# los valores que va a tomar x
x = np.linspace(0,10,1000)
fig, ax = plt.subplots() #lugar donde vamos a estar graficando, ax objeto y subplot porque vamos hacer varios graficos
ax.plot(x,dis_weibull(x)) # variable independiente y la funcion
plt.title('Región Montañosa - Distribución Weibull α=11.7 y β=7.24')
plt.ylabel('Probabilidad')
plt.xlabel('Carga Acumulada ESAL (millones)')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

```python
#@title **a. Construya el gráfico de la Distribución dada si  $α=11.7$ y $β=7.24$**
plt.figure(figsize=(8, 8))
scale = 15 # parametro
weibull = stats.weibull_min(c = 11.7, scale=7.24)  # alfa 11.7, beta = 7.24
x = np.linspace(weibull.ppf(0.0001),
                weibull.ppf(0.99999), 10000)
fp = weibull.pdf(x) # Función de Probabilidad
plt.plot(x, fp)
plt.title('Región Montañosa - Distribución Weibull c=α=11.7 y β=7.24')
plt.ylabel('Probabilidad')
plt.xlabel('Carga Acumulada ESAL (millones)')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **57.  La Distribución Beta**
# **a. Generalidades de la Distribución Beta**
+ **Wikipedia**. La distribución beta es una familia de distribuciones continuas de probabilidad definidas en el intervalo $(0,1)$
parametrizada por dos parámetros positivos de forma, denotados por
$\alpha$  y $\beta$ , que aparecen como exponentes de la variable aleatoria y controlan la forma de la distribución. La generalización de esta distribución a varias variables es conocida como la distribución de Dirichlet..

+ **Ejemplos para la distribución Beta**: Veamos algunos casos donde esta distribución es aplicable:

 -La distribución beta se usa generalmente para estudiar las variaciones, a través de varias muestras.

 - De un porcentaje que representa algún fenómeno, por ejemplo, el tiempo diario que la gente dedica a mirar televisión.
 - La distribución beta suele utilizarse para modelar la distribución de estadísticos de orden (por ejemplo, el estadístico de orden késimo de una muestra de variables n uniformes (0, 1) tiene una distribución $beta (k, n + 1 – k))$
 - para modelar eventos que se definen por valos mínimos y máximos.
 - Una distribución Beta es una forma versátil de representar resultados para porcentajes o proporciones. Por ejemplo, ¿cuál es la probabilidad de que un candidato deshonesto gane las próximas elecciones presidenciales? Se podría pensar que la probabilidad es 0,2. Tu amigo podría pensar que es 0,15. La distribución beta te da una manera de describir esto.

# **b. La Función Beta**
- **Definición** $${B(α,β)} = \int_0^{1} x^{α-1}(1-x)^{β-1}dx$$
- **La función tiene la forma de convolución.** $$
f(t)*g(t) = \int_0^{t} f(x)g(t-x)dx$$
- **Si en la función beta cambiamos $1$ por $t$ y la reescribimos como una convolución se tiene**: $$t^{α-1}*t^{β-1} = \int_0^{t} x^{α-1}(t-x)^{β-1}dx$$
- **Usaremos el Teorema de la Transformada inversa de Laplace que dice**: $$ℒ^{-1}[F(s)G(s)]=ℒ^{-1}[F(s)]*ℒ^{-1}[G(s)]$$
+ **También sabemos que**: $$ℒ[t^α]=\frac{Γ(α+1)}{s^{α+1}}$$
+ **De manera que:** $$t^α=ℒ^{-1}[\frac{Γ(α+1)}{s^{α+1}}]$$
+ **Replanteando la integral**
$$t^{α-1}*t^{β-1} = \int_0^{t} x^{α-1}(t-x)^{β-1}dx$$
+ **Nos queda como:**

$$ℒ^{-1}[\frac{Γ(α)}{s^{α}}]*ℒ^{-1}[\frac{Γ(β)}{s^{β}}] = \int_0^{t} x^{α-1}(t-x)^{β-1}dx$$

- **Usaremos el Teorema**
$$ℒ^{-1}[\frac{Γ(α)}{s^{α}}\frac{Γ(β)}{s^{β}}] = \int_0^{t} x^{α-1}(t-x)^{β-1}dx$$
- **Por la Linealidad**
$$Γ(α)Γ(β)ℒ^{-1}[\frac{1}{s^{α+β}}] = \int_0^{t} x^{α-1}(t-x)^{β-1}dx$$

- **Multiplicando y dividiendo por $Γ(α+β)$**
$$\frac{Γ(α)Γ(β)}{Γ(α+β)}ℒ^{-1}[\frac{Γ(α+β)}{s^{α+β}}] = \int_0^{t} x^{α-1}(t-x)^{β-1}dx$$

+ **Aplicando el resultado** $$t^a=ℒ^{-1}[\frac{Γ(a+1)}{s^{a+1}}]$$ **cambiando** $a = α+β-1$

+ **Se tiene al final que**

$$\frac{Γ(α)Γ(β)}{Γ(α+β)}t^{α+β-1} = \int_0^{t} x^{α-1}(t-x)^{β-1}dx$$
+ **Evaluando $t=1$ se tiene la función Beta**

 $$
{B(α,β)} = \int_0^{1} x^{α-1}(1-x)^{β-1}dx=\frac{\Gamma(α)\Gamma(β)}{\Gamma(α+β)}
$$

# **c. Propiedades de la función Beta**
+ $B(α,β)=B(β,α)$
+ $B(α,1)=\frac{\Gamma(α)\Gamma(1)}{\Gamma(α+1)}=\frac{\Gamma(α)(1)}{α\Gamma(α)}=\frac{1}{α}$

+ $B(α+1,β)=\frac{\Gamma(α+1)\Gamma(β)}{\Gamma(α+β+1)}=\frac{α\Gamma(α)\Gamma(β)}{(α+β)\Gamma(α+β)}=\frac{α}{(α+β)}B(α,β)$


# **d. Distribuciones continuas - La Distribución Beta**

+ **Función de Densidad**. Si $X$ es una variable aleatoria continua, se dice que $X$ tiene una distribución Beta con parámetros ${\displaystyle \alpha ,β >0}$ y escribimos $X∼Beta(α,β)$ si su función de densidad es

$$f(x;\alpha, \beta)=\frac{1}{B(α,β)}x^{α-1}(1-x)^{β-1}, \ \ 0<x<1$$

+ Tambien $$E(X)=\mu=\frac{α}{α+β},  \ \ \ \ V(X)=σ^2=\frac{αβ}{(α+β)^2(α+β+1)}$$




# **e. Distribución Beta (4 parámetros)**.
- Uso común: modelo para una variable con ambos límites inferior y superior. Frecuentemente usada como una distribución previa para análisis Bayesiano
- Rango de $X: a \leq X \leq b$

- Función densidad de probabilidad

$${\displaystyle f_{X}(x)={\frac {(x-a)^{\alpha -1}(b-x)^{\beta -1}}{\mathrm {B} (\alpha ,\beta )(b-a)^{\alpha +\beta-1}}}}$$
- Media: La media de la variable aleatoria $X$ es

$${\displaystyle {\text{E}}[X]={a+\frac {b\alpha }{\alpha +\beta }}}$$

- La varianza de $x$ es $$
{\displaystyle {\text{Var}}(X)={\frac {\alpha \beta(b-a)^2 }{(\alpha +\beta +1)(\alpha +\beta )^{2}}}}$$








# **f. Veamos que es una función de densidad**
+ **Sabemos que**
$$f(x;\alpha, \beta)=\frac{1}{B(α,β)}x^{α-1}(1-x)^{β-1}, \ \ 0<x<1$$
+ $$ ∫_{-∞}^{∞}f(x)dx= ∫_{0}^{1}\frac{1}{B(α,β)}x^{α-1}(1-x)^{β-1}dx$$
+ $$ ∫_{-∞}^{∞}f(x)dx= \frac{1}{B(α,β)} ∫_{0}^{1}x^{α-1}(1-x)^{β-1}dx$$
+ **Usando la definición de la función beta se tiene**$$ ∫_{-∞}^{∞}f(x)dx= \frac{1}{B(α,β)} B(α,β)=1$$


# **g. Valor Esperado**
+ $$E(X)=μ= ∫_{-∞}^{∞}xf(x)dx= ∫_{0}^{1}x\frac{1}{B(α,β)}x^{α-1}(1-x)^{β-1}dx$$

+ **Reorganzando la integral**
+ $$E(X)= \frac{1}{B(α,β)}∫_{0}^{1}x^{(α+1)-1}(1-x)^{β-1}dx$$

+ **Por las propiedades y la definición de Beta**
+ $$E(X)= \frac{1}{B(α,β)}∫_{0}^{1}x^{(α+1)-1}(1-x)^{β-1}dx=\frac{1}{B(α,β)}B(α+1,β)$$
+ **Reemplazando se tiene que**

 $B(α+1,β)=\frac{α}{(α+β)}B(α,β)$

 $$E(X)=\frac{α}{α+β}$$




# **h. Varianza**
+ $$E(X^2)= ∫_{-∞}^{∞}xf(x)dx= ∫_{0}^{1}x^2\frac{1}{B(α,β)}x^{α-1}(1-x)^{β-1}dx$$

+ **Reorganzando la integral**
+ $$E(X^2)= \frac{1}{B(α,β)}∫_{0}^{1}x^{(α+2)-1}(1-x)^{β-1}dx$$

+ **Por las propiedades y la definición de Beta**
+ $$E(X^2)= \frac{1}{B(α,β)}∫_{0}^{1}x^{(α+2)-1}(1-x)^{β-1}dx=\frac{1}{B(α,β)}B(α+2,β)$$
+ **Reemplazando se tiene que**
+ $B(α+1,β)=\frac{α}{(α+β)}B(α,β)$

-  $B(α+2,β)=\frac{α+1}{(α+1+β)}B(α+1,β)=\frac{α+1}{(α+1+β)}\frac{α}{(α+β)}B(α,β)$

 $$E(X^2)=\frac{α+1}{(α+1+β)}\frac{α}{(α+β)}$$

+ **Verifique que**
 + $$σ^2 = E(X^2)-E(X)^2=\frac{α+1}{(α+1+β)}\frac{α}{(α+β)}-(\frac{α}{(α+β)})^2$$
 + $$σ^2 =\frac{αβ}{(α+1+β)(α+β)^2}$$

# **58.  Calculadora de la Distribución Beta**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BETA**
#
# La distribución Beta es una distribución continua definida
# en el intervalo [0, 1]. Es muy flexible y se utiliza para
# modelar proporciones, probabilidades y porcentajes.
#
# Parámetros:
#   α (alfa) - parámetro de forma 1 (α > 0)
#   β (beta) - parámetro de forma 2 (β > 0)
#
# Fórmula: f(x) = x^(α-1) * (1-x)^(β-1) / B(α,β)
# donde B(α,β) = Γ(α)*Γ(β)/Γ(α+β)
#
# Casos especiales:
#   α = β = 1 → Distribución Uniforme(0,1)
#   α = β = 0.5 → Distribución Arcoseno
#   α = 1, β = 1 → Uniforme
#   α > 1, β > 1 → Unimodal
#   α < 1, β < 1 → Bimodal (U-shaped)
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta, uniform
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
try:
    plt.style.use('seaborn-v0_8')
except:
    try:
        plt.style.use('seaborn')
    except:
        plt.style.use('default')

plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_beta():
    """
    Calculadora interactiva para la distribución Beta
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BETA")
    print("="*70)
    print("\nLa distribución Beta es ideal para modelar:")
    print("  • Proporciones y porcentajes")
    print("  • Probabilidades de éxito")
    print("  • Distribuciones previas en estadística bayesiana")
    print("  • Variabilidad en estudios de opinión\n")

    print("Casos especiales:")
    print("  • α = β = 1 → Uniforme(0,1)")
    print("  • α = β = 0.5 → Distribución Arcoseno")
    print("  • α = 1, β = 1 → Uniforme")
    print("  • α > 1, β > 1 → Unimodal (en forma de campana)")
    print("  • α < 1, β < 1 → Bimodal (forma de U)")
    print("  • α > 1, β = 1 → Creciente")
    print("  • α = 1, β > 1 → Decreciente\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            alpha = float(input("  Parámetro α (alfa > 0): "))
            if alpha <= 0:
                print("  ❌ α debe ser mayor que 0. Intenta de nuevo.")
                continue

            beta_param = float(input("  Parámetro β (beta > 0): "))
            if beta_param <= 0:
                print("  ❌ β debe ser mayor que 0. Intenta de nuevo.")
                continue

            print(f"\n  α = {alpha:.4f}")
            print(f"  β = {beta_param:.4f}")

            print("\nTipo de cálculo:")
            print("  1. P(X = a)  - Probabilidad puntual (siempre 0)")
            print("  2. P(X ≤ a)  - Probabilidad acumulada a la izquierda")
            print("  3. P(X ≥ a)  - Probabilidad acumulada a la derecha")
            print("  4. P(a ≤ X ≤ b) - Probabilidad entre dos valores")
            print("  5. Cuantil - Encontrar X dado un percentil")
            print("  6. Comparar con diferentes parámetros")
            print("  7. Relación con otras distribuciones")

            tipo = input("\nSelecciona el tipo de cálculo (1-7): ")

            if tipo == '1':
                a = float(input("  Valor de a (0 ≤ a ≤ 1): "))
                if a < 0 or a > 1:
                    print("  ❌ a debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                prob = 0
                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X = {a}) = 0")
                print("En distribuciones continuas, la probabilidad puntual es siempre 0.")
                print(f"Densidad en X={a}: {beta.pdf(a, alpha, beta_param):.6f}")

                visualizar_beta(alpha, beta_param, a, None, prob, tipo)
                return {'alpha': alpha, 'beta': beta_param, 'a': a}

            elif tipo == '2':
                a = float(input("  Valor de a (0 ≤ a ≤ 1): "))
                if a < 0 or a > 1:
                    print("  ❌ a debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                prob = beta.cdf(a, alpha, beta_param)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≤ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con α={alpha:.4f}, β={beta_param:.4f}")

                visualizar_beta(alpha, beta_param, a, None, prob, tipo)
                return {'alpha': alpha, 'beta': beta_param, 'a': a, 'prob': prob}

            elif tipo == '3':
                a = float(input("  Valor de a (0 ≤ a ≤ 1): "))
                if a < 0 or a > 1:
                    print("  ❌ a debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                prob = 1 - beta.cdf(a, alpha, beta_param)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≥ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con α={alpha:.4f}, β={beta_param:.4f}")

                visualizar_beta(alpha, beta_param, a, None, prob, tipo)
                return {'alpha': alpha, 'beta': beta_param, 'a': a, 'prob': prob}

            elif tipo == '4':
                a = float(input("  Límite inferior (0 ≤ a ≤ 1): "))
                if a < 0 or a > 1:
                    print("  ❌ a debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                b = float(input("  Límite superior (b > a, ≤ 1): "))
                if b <= a or b > 1:
                    print("  ❌ b debe ser mayor que a y ≤ 1. Intenta de nuevo.")
                    continue
                prob = beta.cdf(b, alpha, beta_param) - beta.cdf(a, alpha, beta_param)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP({a} ≤ X ≤ {b}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con α={alpha:.4f}, β={beta_param:.4f}")

                visualizar_beta(alpha, beta_param, a, b, prob, tipo)
                return {'alpha': alpha, 'beta': beta_param, 'a': a, 'b': b, 'prob': prob}

            elif tipo == '5':
                p = float(input("  Percentil (0-1): "))
                if p <= 0 or p >= 1:
                    print("  ❌ El percentil debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                x = beta.ppf(p, alpha, beta_param)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nPara el percentil {p*100:.1f}%:")
                print(f"X = {x:.6f}")
                print(f"Con α={alpha:.4f}, β={beta_param:.4f}")
                print(f"Interpretación: El {p*100:.1f}% de los datos están por debajo de {x:.4f}")

                visualizar_cuantil_beta(alpha, beta_param, x, p)
                return {'alpha': alpha, 'beta': beta_param, 'percentil': p, 'x': x}

            elif tipo == '6':
                comparar_parametros_beta(alpha, beta_param)
                return {'alpha': alpha, 'beta': beta_param}

            elif tipo == '7':
                relacion_beta_distribuciones(alpha, beta_param)
                return {'alpha': alpha, 'beta': beta_param}

            else:
                print("  ❌ Opción no válida. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

#==========================================================
# FUNCIÓN PARA VISUALIZAR LA DISTRIBUCIÓN BETA
#==========================================================

def visualizar_beta(alpha, beta_param, a, b, prob, tipo):
    """Crea visualizaciones completas de la distribución Beta"""

    # Rango de valores para la gráfica
    x = np.linspace(0.001, 0.999, 1000)
    y = beta.pdf(x, alpha, beta_param)

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    #==========================================================
    # GRÁFICO 1: Distribución con área sombreada
    #==========================================================
    ax1 = plt.subplot(2, 2, 1)
    ax1.plot(x, y, 'b-', linewidth=2.5, label=f'Beta(α={alpha:.2f}, β={beta_param:.2f})')

    # Sombrear el área según el tipo de cálculo
    if tipo == '2':  # P(X ≤ a)
        mask = x <= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≤ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '3':  # P(X ≥ a)
        mask = x >= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≥ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '4':  # P(a ≤ X ≤ b)
        mask = (x >= a) & (x <= b)
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P({a:.2f} ≤ X ≤ {b:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)
        ax1.axvline(x=b, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    # Estadísticas
    media = alpha / (alpha + beta_param)
    varianza = (alpha * beta_param) / ((alpha + beta_param)**2 * (alpha + beta_param + 1))
    desv = np.sqrt(varianza)
    moda = (alpha - 1) / (alpha + beta_param - 2) if alpha > 1 and beta_param > 1 else None

    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.3f}')
    if moda is not None and 0 < moda < 1:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Moda = {moda:.3f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Distribución Beta (α={alpha:.2f}, β={beta_param:.2f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', fontsize=10)
    ax1.set_xlim(0, 1)

    # Agregar información de probabilidad
    if prob is not None:
        ax1.text(0.98, 0.95, f'Probabilidad = {prob:.6f}\n({prob*100:.2f}%)',
                transform=ax1.transAxes, fontsize=11, verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 2: Efecto de los parámetros
    #==========================================================
    ax2 = plt.subplot(2, 2, 2)

    # Mostrar diferentes combinaciones de α y β
    combinaciones = [(0.5, 0.5), (1, 1), (2, 2), (2, 5), (5, 2), (0.5, 2)]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#9B59B6', '#1ABC9C']

    for (a_val, b_val), color in zip(combinaciones, colores):
        y_temp = beta.pdf(x, a_val, b_val)
        label = f'α={a_val}, β={b_val}'
        if a_val == alpha and b_val == beta_param:
            label += ' ★'
            ax2.plot(x, y_temp, color=color, linewidth=3, label=label)
        else:
            ax2.plot(x, y_temp, color=color, linewidth=2, alpha=0.6, label=label)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax2.set_title('Comparación de diferentes (α, β)', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best', fontsize=8)
    ax2.set_xlim(0, 1)

    #==========================================================
    # GRÁFICO 3: Función de distribución acumulada
    #==========================================================
    ax3 = plt.subplot(2, 2, 3)
    cdf = beta.cdf(x, alpha, beta_param)
    ax3.plot(x, cdf, 'g-', linewidth=2.5, label='CDF')
    ax3.fill_between(x, 0, cdf, alpha=0.2, color='green')

    # Puntos de interés
    if tipo == '2':  # P(X ≤ a)
        ax3.scatter(a, prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {prob:.4f}')
        ax3.axhline(y=prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '3':  # P(X ≥ a)
        ax3.scatter(a, 1-prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {1-prob:.4f}')
        ax3.axhline(y=1-prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles comunes
    for q in [0.25, 0.5, 0.75, 0.9, 0.95]:
        q_val = beta.ppf(q, alpha, beta_param)
        ax3.axvline(x=q_val, color='orange', linestyle=':', linewidth=1.5, alpha=0.5)
        ax3.text(q_val, 0.01, f'Q{int(q*100)}', ha='center', fontsize=8, color='orange')

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax3.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', fontsize=10)
    ax3.set_ylim(-0.05, 1.05)
    ax3.set_xlim(0, 1)

    #==========================================================
    # GRÁFICO 4: Tabla de estadísticas
    #==========================================================
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Calcular estadísticas
    media = alpha / (alpha + beta_param)
    varianza = (alpha * beta_param) / ((alpha + beta_param)**2 * (alpha + beta_param + 1))
    desv = np.sqrt(varianza)
    moda = (alpha - 1) / (alpha + beta_param - 2) if alpha > 1 and beta_param > 1 else None

    # Crear tabla
    table_data = [
        ['Estadística', 'Valor', 'Fórmula'],
        ['Media (μ)', f'{media:.6f}', 'α/(α+β)'],
        ['Varianza (σ²)', f'{varianza:.6f}', 'αβ/((α+β)²(α+β+1))'],
        ['Desv. Estándar (σ)', f'{desv:.6f}', '√varianza'],
        ['Moda', f'{moda:.6f}' if moda is not None and 0 < moda < 1 else 'No definida', '(α-1)/(α+β-2)'],
        ['Asimetría', f'{2*(beta_param-alpha)*np.sqrt(alpha+beta_param+1)/((alpha+beta_param+2)*np.sqrt(alpha*beta_param)):.4f}', '2(β-α)√(α+β+1)/((α+β+2)√(αβ))'],
        ['Curtosis', 'Depende', '6(α³+...)/...']
    ]

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.3, 0.3, 0.4])

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.5)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title('Propiedades de la Distribución Beta',
                  fontsize=13, fontweight='bold', pad=20)

    #==========================================================
    # Ajustar layout
    #==========================================================
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # GRÁFICO ADICIONAL: Función de densidad para diferentes valores
    #==========================================================
    fig2, (ax5, ax6) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 5: Efecto de α (β fijo)
    alphas = [0.5, 1, 2, 5, 10]
    colores2 = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    for a_val, color in zip(alphas, colores2):
        y_temp = beta.pdf(x, a_val, beta_param)
        ax5.plot(x, y_temp, color=color, linewidth=2,
                label=f'α = {a_val}', alpha=0.8 if a_val != alpha else 1.0)
        if a_val == alpha:
            ax5.fill_between(x, 0, y_temp, alpha=0.2, color=color)

    ax5.set_xlabel('X', fontsize=12)
    ax5.set_ylabel('Densidad', fontsize=12)
    ax5.set_title(f'Efecto de α (β={beta_param:.1f} fijo)', fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3)
    ax5.legend(loc='best')
    ax5.set_xlim(0, 1)

    # Gráfico 6: Efecto de β (α fijo)
    betas = [0.5, 1, 2, 5, 10]
    colores3 = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    for b_val, color in zip(betas, colores3):
        y_temp = beta.pdf(x, alpha, b_val)
        ax6.plot(x, y_temp, color=color, linewidth=2,
                label=f'β = {b_val}', alpha=0.8 if b_val != beta_param else 1.0)
        if b_val == beta_param:
            ax6.fill_between(x, 0, y_temp, alpha=0.2, color=color)

    ax6.set_xlabel('X', fontsize=12)
    ax6.set_ylabel('Densidad', fontsize=12)
    ax6.set_title(f'Efecto de β (α={alpha:.1f} fijo)', fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3)
    ax6.legend(loc='best')
    ax6.set_xlim(0, 1)

    plt.tight_layout()

    #==========================================================
    # MOSTRAR GRÁFICOS
    #==========================================================
    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Beta_alpha{alpha:.2f}_beta{beta_param:.2f}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_parametros.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_parametros.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

def visualizar_cuantil_beta(alpha, beta_param, x, p):
    """Visualiza el cuantil en la distribución Beta"""

    x_vals = np.linspace(0.001, 0.999, 1000)
    y = beta.pdf(x_vals, alpha, beta_param)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Distribución con cuantil marcado
    ax1.plot(x_vals, y, 'b-', linewidth=2.5)
    mask = x_vals <= x
    ax1.fill_between(x_vals[mask], 0, y[mask], color='#E63946', alpha=0.5,
                     label=f'Área = {p*100:.1f}%')
    ax1.axvline(x=x, color='#E63946', linestyle='--', linewidth=2, label=f'X = {x:.3f}')

    # Media y moda
    media = alpha / (alpha + beta_param)
    moda = (alpha - 1) / (alpha + beta_param - 2) if alpha > 1 and beta_param > 1 else None

    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.3f}')
    if moda is not None and 0 < moda < 1:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Moda = {moda:.3f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Percentil {p*100:.1f}% (α={alpha:.2f}, β={beta_param:.2f})',
                  fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best')
    ax1.set_xlim(0, 1)

    # Gráfico 2: Función de distribución acumulada
    cdf = beta.cdf(x_vals, alpha, beta_param)
    ax2.plot(x_vals, cdf, 'g-', linewidth=2.5)
    ax2.scatter(x, p, color='#E63946', s=150, zorder=5, label=f'F({x:.3f}) = {p:.4f}')
    ax2.axhline(y=p, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax2.axvline(x=x, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax2.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best')
    ax2.set_ylim(-0.05, 1.05)
    ax2.set_xlim(0, 1)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA COMPARAR PARÁMETROS
#==========================================================

def comparar_parametros_beta(alpha_actual, beta_actual):
    """Compara diferentes combinaciones de parámetros Beta"""

    print("\n" + "="*70)
    print("📊 COMPARACIÓN DE PARÁMETROS")
    print("="*70)

    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    x = np.linspace(0.001, 0.999, 1000)

    # Gráfico 1: Diferentes α (β fijo)
    ax1 = axes[0, 0]
    alphas = [0.5, 1, 2, 5, 10]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    for a_val, color in zip(alphas, colores):
        y = beta.pdf(x, a_val, beta_actual)
        ax1.plot(x, y, color=color, linewidth=2,
                label=f'α = {a_val}', alpha=0.8 if a_val != alpha_actual else 1.0)
        if a_val == alpha_actual:
            ax1.fill_between(x, 0, y, alpha=0.2, color=color)

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad', fontsize=12)
    ax1.set_title(f'Variando α (β={beta_actual:.1f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='best')
    ax1.set_xlim(0, 1)

    # Gráfico 2: Diferentes β (α fijo)
    ax2 = axes[0, 1]
    betas = [0.5, 1, 2, 5, 10]

    for b_val, color in zip(betas, colores):
        y = beta.pdf(x, alpha_actual, b_val)
        ax2.plot(x, y, color=color, linewidth=2,
                label=f'β = {b_val}', alpha=0.8 if b_val != beta_actual else 1.0)
        if b_val == beta_actual:
            ax2.fill_between(x, 0, y, alpha=0.2, color=color)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad', fontsize=12)
    ax2.set_title(f'Variando β (α={alpha_actual:.1f})', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')
    ax2.set_xlim(0, 1)

    # Gráfico 3: Media vs α+β
    ax3 = axes[1, 0]
    x_vals = np.linspace(0.1, 10, 100)

    # Mantener relación α/β constante
    relacion = alpha_actual / beta_actual
    alphas_constant = x_vals * relacion / (1 + relacion)
    betas_constant = x_vals / (1 + relacion)

    medias = alphas_constant / (alphas_constant + betas_constant)
    varianzas = (alphas_constant * betas_constant) / ((alphas_constant + betas_constant)**2 * (alphas_constant + betas_constant + 1))

    ax3.plot(x_vals, medias, 'b-', linewidth=2, label='Media')
    ax3.plot(x_vals, varianzas, 'r-', linewidth=2, label='Varianza')
    ax3.axvline(x=alpha_actual + beta_actual, color='purple', linestyle='--', linewidth=2, alpha=0.5)

    ax3.set_xlabel('α + β', fontsize=12)
    ax3.set_ylabel('Valor', fontsize=12)
    ax3.set_title('Media y Varianza vs α+β', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3)
    ax3.legend(loc='best')

    # Gráfico 4: Mapa de calor de la moda
    ax4 = axes[1, 1]
    alphas_grid = np.linspace(0.1, 5, 50)
    betas_grid = np.linspace(0.1, 5, 50)
    Alphas, Betas = np.meshgrid(alphas_grid, betas_grid)

    # Calcular moda (si existe)
    Moda = np.where((Alphas > 1) & (Betas > 1),
                    (Alphas - 1) / (Alphas + Betas - 2),
                    np.nan)

    im = ax4.contourf(Alphas, Betas, Moda, levels=20, cmap='viridis')
    ax4.axhline(y=beta_actual, color='red', linestyle='--', linewidth=2, alpha=0.7)
    ax4.axvline(x=alpha_actual, color='red', linestyle='--', linewidth=2, alpha=0.7)
    ax4.scatter(alpha_actual, beta_actual, color='red', s=100, zorder=5)

    ax4.set_xlabel('α', fontsize=12)
    ax4.set_ylabel('β', fontsize=12)
    ax4.set_title('Mapa de calor de la Moda', fontsize=13, fontweight='bold')
    plt.colorbar(im, ax=ax4, label='Moda')

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA RELACIONES CON OTRAS DISTRIBUCIONES
#==========================================================

def relacion_beta_distribuciones(alpha, beta_param):
    """Muestra la relación de la Beta con otras distribuciones"""

    print("\n" + "="*70)
    print("🔗 RELACIÓN CON OTRAS DISTRIBUCIONES")
    print("="*70)

    print("\nLa distribución Beta está relacionada con:")
    print("  1. Uniforme: Beta(1,1) = Uniforme(0,1)")
    print("  2. Arcoseno: Beta(0.5,0.5) = Arcoseno")
    print("  3. Distribución de probabilidades en procesos de Bernoulli")
    print("  4. Distribución previa conjugada en estadística bayesiana")
    print("  5. Relación con la distribución F\n")

    # Crear visualización de las relaciones
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    x = np.linspace(0.001, 0.999, 1000)

    # 1. Comparación con Uniforme
    ax1 = axes[0, 0]
    y_beta = beta.pdf(x, alpha, beta_param)
    y_uniform = uniform.pdf(x, 0, 1)

    ax1.plot(x, y_beta, 'b-', linewidth=2.5, label=f'Beta(α={alpha:.1f}, β={beta_param:.1f})')
    ax1.plot(x, y_uniform, 'r--', linewidth=2, label='Uniforme(0,1)')
    ax1.fill_between(x, 0, y_uniform, alpha=0.2, color='red')
    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad', fontsize=12)
    ax1.set_title('Beta vs Uniforme', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='best')
    ax1.set_xlim(0, 1)

    # 2. Comparación con Arcoseno
    ax2 = axes[0, 1]
    y_arcoseno = 1 / (np.pi * np.sqrt(x * (1-x)))
    y_arcoseno = np.where(np.isfinite(y_arcoseno), y_arcoseno, 0)

    ax2.plot(x, y_beta, 'b-', linewidth=2.5, label=f'Beta(α={alpha:.1f}, β={beta_param:.1f})')
    if alpha == 0.5 and beta_param == 0.5:
        ax2.plot(x, y_arcoseno, 'g--', linewidth=2, label='Arcoseno(0.5,0.5)')
        ax2.fill_between(x, 0, y_arcoseno, alpha=0.2, color='green')
    else:
        ax2.text(0.5, 0.5, f'No es Arcoseno\n(α={alpha:.1f}, β={beta_param:.1f})',
                transform=ax2.transAxes, ha='center', va='center', fontsize=12)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad', fontsize=12)
    ax2.set_title('Beta vs Arcoseno', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')
    ax2.set_xlim(0, 1)

    # 3. Interpretación bayesiana
    ax3 = axes[1, 0]
    ax3.axis('off')

    # Tabla de interpretación
    table_data = [
        ['α, β', 'Forma', 'Interpretación'],
        ['α < 1, β < 1', 'U-shaped', 'Incertidumbre alta en extremos'],
        ['α = 1, β = 1', 'Uniforme', 'Sin información previa'],
        ['α > 1, β > 1', 'Campana', 'Información concentrada'],
        ['α > 1, β = 1', 'Creciente', 'Éxitos probables'],
        ['α = 1, β > 1', 'Decreciente', 'Fallos probables'],
        ['α = β > 1', 'Simétrica', 'Simetría en la creencia']
    ]

    table = ax3.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.2, 0.3, 0.5])

    table.auto_set_font_size(False)
    table.set_fontsize(11)
    table.scale(1, 1.8)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax3.set_title('Interpretación de Parámetros en Bayesiana',
                  fontsize=13, fontweight='bold', pad=20)

    # 4. Relación con distribución F
    ax4 = axes[1, 1]

    # Mostrar transformación
    if alpha > 0 and beta_param > 0:
        # La Beta se relaciona con F: Si X ~ Beta(α,β), entonces
        # (α/β) * X/(1-X) ~ F(2α, 2β)
        from scipy.stats import f

        # Crear valores para la transformación
        y_vals = np.linspace(0, 5, 1000)
        f_pdf = f.pdf(y_vals, 2*alpha, 2*beta_param)

        ax4.plot(y_vals, f_pdf, 'purple', linewidth=2.5)
        ax4.fill_between(y_vals, 0, f_pdf, alpha=0.2, color='purple')

        ax4.set_xlabel('Y', fontsize=12)
        ax4.set_ylabel('Densidad', fontsize=12)
        ax4.set_title(f'Transformación: F(2α={2*alpha:.1f}, 2β={2*beta_param:.1f})',
                     fontsize=13, fontweight='bold')
        ax4.grid(True, alpha=0.3)

        ax4.text(0.98, 0.95, f'Y = (α/β)·X/(1-X)',
                transform=ax4.transAxes, fontsize=10,
                verticalalignment='top', horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA GENERAR DATOS BETA
#==========================================================

def generar_datos_beta():
    """Genera y visualiza datos de una distribución Beta"""

    print("\n" + "="*70)
    print("📊 GENERACIÓN DE DATOS BETA")
    print("="*70)

    try:
        print("Ingresa los parámetros:")
        alpha = float(input("  Parámetro α (alfa > 0): "))
        if alpha <= 0:
            print("  ❌ α debe ser mayor que 0.")
            return

        beta_param = float(input("  Parámetro β (beta > 0): "))
        if beta_param <= 0:
            print("  ❌ β debe ser mayor que 0.")
            return

        n = int(input("  Número de observaciones a generar: "))
        if n <= 0:
            print("  ❌ n debe ser mayor que 0.")
            return

        # Generar datos
        np.random.seed(42)
        datos = beta.rvs(alpha, beta_param, size=n)

        # Calcular estadísticas
        media_teorica = alpha / (alpha + beta_param)
        varianza_teorica = (alpha * beta_param) / ((alpha + beta_param)**2 * (alpha + beta_param + 1))

        print(f"\nEstadísticas de los datos generados:")
        print("-"*70)
        print(f"  Media teórica: {media_teorica:.6f}")
        print(f"  Media muestral: {np.mean(datos):.6f}")
        print(f"  Varianza teórica: {varianza_teorica:.6f}")
        print(f"  Varianza muestral: {np.var(datos):.6f}")
        print(f"  Mediana muestral: {np.median(datos):.6f}")
        print(f"  Mínimo: {np.min(datos):.6f}")
        print(f"  Máximo: {np.max(datos):.6f}")

        # Visualización
        fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

        # Histograma con curva teórica
        ax1.hist(datos, bins=30, density=True, alpha=0.6, color='#2E86AB',
                edgecolor='black', label='Datos generados')

        x = np.linspace(0.001, 0.999, 1000)
        y_teorica = beta.pdf(x, alpha, beta_param)
        ax1.plot(x, y_teorica, 'r-', linewidth=2, label='Teórica')

        ax1.set_xlabel('X', fontsize=12)
        ax1.set_ylabel('Densidad', fontsize=12)
        ax1.set_title('Histograma vs Curva Teórica', fontsize=13, fontweight='bold')
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='best')
        ax1.set_xlim(0, 1)

        # Q-Q plot
        stats.probplot(datos, dist="beta", sparams=(alpha, beta_param), plot=ax2)
        ax2.set_title('Gráfico Q-Q Beta', fontsize=13, fontweight='bold')
        ax2.grid(True, alpha=0.3)

        # CDF empírica vs teórica
        from statsmodels.distributions.empirical_distribution import ECDF
        ecdf = ECDF(datos)

        x_ecdf = np.linspace(0, 1, 1000)
        ax3.plot(x_ecdf, ecdf(x_ecdf), 'b-', linewidth=2, label='Empírica')
        ax3.plot(x_ecdf, beta.cdf(x_ecdf, alpha, beta_param), 'r--', linewidth=2, label='Teórica')

        ax3.set_xlabel('X', fontsize=12)
        ax3.set_ylabel('CDF', fontsize=12)
        ax3.set_title('CDF Empírica vs Teórica', fontsize=13, fontweight='bold')
        ax3.grid(True, alpha=0.3)
        ax3.legend(loc='best')
        ax3.set_ylim(0, 1.05)
        ax3.set_xlim(0, 1)

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_beta():
    """Ejecuta ejemplos clásicos de la distribución Beta"""

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN BETA")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Proporción de éxito',
            'descripcion': 'En una encuesta, se estima que la proporción de éxito sigue Beta(4, 6). ¿Cuál es la probabilidad de que la proporción sea mayor a 0.5?',
            'alpha': 4,
            'beta': 6,
            'a': 0.5,
            'tipo': 'mayor'
        },
        {
            'titulo': 'Ejemplo 2: Probabilidad de fraude',
            'descripcion': 'La probabilidad de fraude en una empresa sigue Beta(2, 8). ¿Cuál es la probabilidad de que sea menor a 0.3?',
            'alpha': 2,
            'beta': 8,
            'a': 0.3,
            'tipo': 'menor'
        },
        {
            'titulo': 'Ejemplo 3: Percentil en estudios de opinión',
            'descripcion': 'En un estudio de opinión, la proporción de aprobación sigue Beta(5, 3). ¿Cuál es el percentil 90?',
            'alpha': 5,
            'beta': 3,
            'p': 0.90,
            'tipo': 'cuantil'
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")

        alpha = ej['alpha']
        beta_param = ej['beta']

        if ej['tipo'] == 'mayor':
            prob = 1 - beta.cdf(ej['a'], alpha, beta_param)
            print(f"   α={alpha:.1f}, β={beta_param:.1f}")
            print(f"   P(X > {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'menor':
            prob = beta.cdf(ej['a'], alpha, beta_param)
            print(f"   α={alpha:.1f}, β={beta_param:.1f}")
            print(f"   P(X ≤ {ej['a']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'cuantil':
            x = beta.ppf(ej['p'], alpha, beta_param)
            print(f"   α={alpha:.1f}, β={beta_param:.1f}")
            print(f"   Percentil {ej['p']*100:.0f}% = {x:.6f}")

    print("\n" + "="*70)

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN BETA")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado")
        print("  2. Ver ejemplos clásicos")
        print("  3. Generar datos Beta")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_beta()
        elif opcion == '2':
            ejemplos_beta()
        elif opcion == '3':
            generar_datos_beta()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```

#**a. Construya el gráfico de la Distribución para   $α=50$ y $β=49$**
+ **Función de Densidad**. Si $X$ es una variable aleatoria continua, se dice que $X$ tiene una distribución Beta con parámetros ${\displaystyle \alpha ,β >0}$ y escribimos $X∼Beta(α,β)$ si su función de densidad es

$$f(x;\alpha, \beta)=\frac{1}{B(α,β)}x^{α-1}(1-x)^{β-1}, \ \ 0<x<1$$

- Donde $$
{B(α,β)} = \int_0^{1} x^{α-1}(1-x)^{β-1}dx=\frac{\Gamma(α)\Gamma(β)}{\Gamma(α+β)},.
$$

```python
#@title **a. Construya el gráfico de la Distribución dada si  $α=50$ y $β=49$**
from numpy.ma.core import log
from sympy import integrate, init_printing
from sympy.abc import x
# nuestra libreria
import numpy as np
import matplotlib.pyplot as plt
# Definimos una funcion
a = 50 # parametro alpha
b = 49 # parametro beta
beta = factorial(a-1)*factorial(b-1)/factorial(a+b-1)

def dis_beta(x):
  return x**(a-1)*(1-x)**(b-1)/(beta)
# los valores que va a tomar x
x = np.linspace(0,1,1000)
fig, ax = plt.subplots() #lugar donde vamos a estar graficando, ax objeto y subplot porque vamos hacer varios graficos
ax.plot(x,dis_beta(x)) # variable independiente y la funcion
plt.title('Fabrica A - Distribución Beta α=50 y β=49')
plt.ylabel('Probabilidad')
plt.xlabel('Produccion perdida (Porcentajes)')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

```python
#@title **a. Construya el gráfico de la Distribución dada si  α=11.7 y β=7.24**
plt.figure(figsize=(8, 8))
a = 50 # parametro alpha
b = 49 # parametro beta

beta = stats.beta(a, b)  # alfa 50, beta = 49
x = np.linspace(beta.ppf(0.0001),
                beta.ppf(0.99999), 10000)
fp = beta.pdf(x) # Función de Probabilidad
plt.plot(x, fp)

plt.title('Fabrica A - Distribución Beta α=50 y β=49')
plt.ylabel('Probabilidad')
plt.xlabel('Produccion perdida (Porcentajes)')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

```python
#@title **m. Dada la Fábrica A. Parámetros de la Distribución de Beta   $α=11.7$ y $β=10.52$, calcule el valor de $k$  tal que $P (k < X < 0.55) = 0.4$**
fig, ax = plt.subplots(figsize=[6, 4])
x = np.linspace(0, 1, 401)

y = stats.beta(11.7,10.52).pdf(x)
a = 0.55 # Valor de la variable a calcular a la izquierda
b = stats.beta(11.7,10.52).cdf(a)
c = stats.beta(11.7,10.52).cdf(a) - 0.42
d = round(stats.beta(11.7,10.52).ppf(c),2) # valor critico pedido
e = stats.beta(11.7,10.52).cdf(d) # Probabilidad acumulada
x1 = np.linspace(0, d, 401)
x2 = np.linspace(d, a, 401)
x3 = np.linspace(a, 1, 401)

ax.plot(x, y, c='g')

ax.fill_between(x1, 0,stats.beta(11.7,10.52).pdf(x1) , color="r", alpha = 0.3)
ax.fill_between(x2, 0,stats.beta(11.7,10.52).pdf(x2) , color="b", alpha =0.3)
ax.fill_between(x3, 0,stats.beta(11.7,10.52).pdf(x3) , color="y", alpha =0.3)

ax.text(d, 0.00001, d)
ax.text(a, 0.00001, a)

ax.text(a, 1, f'Area1= {1-b:.4f}')
ax.text((a+d)/2, 1.2, f'Area2= {c:.4f}')
ax.text(d, 0.3, f'Area3= {e:.4f}')


print(f'Area solicitada = {c:.4f}')

print(f'El valor critico solicitado es k ={d}')
print(f'P(X<={d})={e:.4f}')
print(f'P({d}=<X<={a})={b-e:.4f}')
print(f'P(X<={a})={b:.4f}')

plt.title('Fabrica A - Distribución beta(11.7,10.52)')
plt.ylabel('Probabilidad')
plt.xlabel('Produccion perdida (Porcentajes)')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **La Distribución Chi Cuadrado**

# **a. Generalidades de la Distribución Chi Cuadrado**
+ **Wikipedia**. En teoría de la probabilidad y en estadística, la distribución ji al cuadrado (también llamada distribución de Pearson o distribución $χ^2$ con $k∈\mathbb{N}$ grados de libertad es la distribución de la suma del cuadrado de $k$  variables aleatorias independientes con distribución normal estándar. La distribución chi cuadrada es un caso especial de la distribución gamma y es una de las distribuciones de probabilidad más usadas en Inferencia Estadística, principalmente en pruebas de hipótesis y en la construcción de intervalos de confianza.

+ **Ejemplos para la distribución  $χ^2$** tiene muchas aplicaciones en inferencia estadística. La más conocida es la denominada prueba $χ^2$, utilizada como prueba de independencia y como prueba de buen ajuste y en la estimación de varianzas.
+  Pero también está involucrada en el problema de estimar la media de una población normalmente distribuida y en el problema de estimar la pendiente de una recta de regresión lineal, a través de su papel en la distribución t de Student.

+ Aparece también en todos los problemas de análisis de varianza por su relación con la distribución F de Snedecor, que es la distribución del cociente de dos variables aleatorias independientes con distribución $χ^2$
- Uso común: distribución de la varianza muestral $s^2$ de una población normal.

- La distribución chi cuadrada es un caso especial de la distribución gamma y es una de las distribuciones de probabilidad más usadas en Inferencia Estadística, principalmente en pruebas de hipótesis y en la construcción de intervalos de confianza.




# **b. Definición - Distribución Chi-Cuadrada.**
### **Como la suma de normales estándar**
- Sean ${\displaystyle Z_{1},\dots ,Z_{k}}$ variables aleatorias independientes tales que ${\displaystyle Z_{i}\sim N(0,1)}$ para ${\displaystyle i=1,2,\dots ,k}$ entonces la variable aleatoria $X$ definida por

$${\displaystyle {\begin{aligned}X&=Z_{1}^{2}+Z_{2}^{2}+\cdots +Z_{k}^{2}\\&=\sum _{i=1}^{k}Z_{i}^{2}\end{aligned}}}$$
- tiene una distribución chi cuadrada con $k$ grados de libertad.

### **Notación**.
+  Si la variable aleatoria continua $X$ tiene una distribución Chi Cuadrada con $k$ grados de libertad entonces escribiremos ${\displaystyle X\sim \chi _{k}^{2}}$.

###  **Función de Densidad**
+ Si $X$ es una variable aleatoria continua, se dice que $X$ tiene una distribución Chi-cuadrado con parámetro $n$ y escribimos $X∼Χ^2(n)$ si su función de densidad es

$${\displaystyle f(x;n)=\frac{{(\frac{x}{2})^{\frac{n}{2}-1}}e^{-(\frac{x}{2})}}{2\Gamma(\frac{n}{2})},\dots ,x\geq 0}$$


+ Tambien $$E(X)=\mu=n,  \ \ \ \ V(X)=σ^2=2n$$





# **c. Distribuciones continuas - La Distribución Chi-Cuadrado**



## **Variable aleatoria con Distribución $χ^2$**

- Sea ${\displaystyle X_{1},\dots ,X_{n}}$ una muestra aleatoria proveniente de una población con distribución ${\displaystyle N(\mu ,\sigma ^{2})}$ entonces

  - ${\displaystyle {\overline {X}}}$ y el vector ${\displaystyle \left(X_{1}-{\overline {X}},\dots ,X_{n}-{\overline {X}}\right)}$ son independientes.
  - ${\displaystyle {\overline {X}}}$ y ${\displaystyle S^{2}}$ son independientes.
  - Entonces, la variable aleatoria $${\displaystyle {\frac {(n-1)S^{2}}{\sigma ^{2}}}\sim \chi _{n-1}^{2}}$$
  - ${\displaystyle \operatorname {E} [S^{2}]=\sigma ^{2}}$ y ${\displaystyle \operatorname {Var} (S^{2})={\frac {2\sigma ^{4}}{n-1}}}$.
- donde ${\displaystyle {\overline {X}}={\frac {1}{n}}\sum _{i=1}^{n}X_{i}}$ y ${\displaystyle S^{2}={\frac {1}{n-1}}\sum _{i=1}^{n}\left(X_{i}-{\overline {X}}\right)^{2}}$ son la media y varianza de la muestra aleatoria respectivamente


## **Intervalo para la varianza con Distribución $χ^2$**
- Sean ${\displaystyle X_{1},\dots ,X_{n}}$ una muestra aleatoria proveniente de una población con distribución ${\displaystyle N(\mu ,\sigma ^{2})}$ donde ${\displaystyle \mu }$ y $\sigma^2$ son desconocidos. Se tiene que ${\displaystyle {\frac {(n-1)S^{2}}{\sigma ^{2}}}\sim \chi _{n-1}^{2}}$
- Sean ${\displaystyle \chi _{n-1,\alpha /2}, \ \chi _{n-1,1-\alpha /2}\in \mathbb {R} }$ tales que ${\displaystyle \operatorname {P} [\chi _{n-1,\alpha /2}<Y<\chi _{n-1,1-\alpha /2}]=1-\alpha }$. siendo ${\displaystyle Y\sim \chi _{n-1}^{2}}$.
$${\displaystyle {\begin{aligned}&\operatorname {P} \left[\chi _{n-1,\alpha /2}<{\frac {(n-1)S^{2}}{\sigma ^{2}}}<\chi _{n-1,1-\alpha /2}\right]=1-\alpha \\&\operatorname {P} \left[{\frac {1}{\chi _{n-1,\alpha /2}}}>{\frac {\sigma ^{2}}{(n-1)S^{2}}}>{\frac {1}{\chi _{n-1,1-\alpha /2}}}\right]=1-\alpha \\&\operatorname {P} \left[{\frac {(n-1)S^{2}}{\chi _{n-1,1-\alpha /2}}}<\sigma ^{2}<{\frac {(n-1)S^{2}}{\chi _{n-1,\alpha /2}}}\right]=1-\alpha \end{aligned}}}$$

- por lo tanto un intervalo de ${\displaystyle (1-\alpha )100\%}$ de confianza para $\sigma ^{2}$ está dado por

$${\displaystyle \left({\frac {(n-1)S^{2}}{\chi _{n-1,1-\alpha /2}}},{\frac {(n-1)S^{2}}{\chi _{n-1,\alpha /2}}}\right)}$$





# **d. Veamos que es una función de densidad**

+ **Densidad Distribución Chi Cuadrado**
$${\displaystyle f(x;n)=\frac{{(\frac{x}{2})^{\frac{n}{2}-1}}e^{-(\frac{x}{2})}}{2\Gamma(\frac{n}{2})},\dots ,x\geq 0}$$

+ **Densidad Distribución Función Gamma**
$$ f(x,α,β) = \frac{1}{β^αΓ(α)}x^{α-1}exp{\{\frac{-x}{β}\}}, \ x\geq 0, 0<α,β<∞, $$


+ **Hagamos las comparaciones en Gamma**
 - $α = \frac{n}{2}$, y $β = 2$
 - $$ f(x,\frac{n}{2},2) = \frac{1}{2^{\frac{n}{2}}Γ(\frac{n}{2})}x^{\frac{n}{2}-1}exp{\{\frac{-x}{2}\}}$$
 - $$ f(x,\frac{n}{2},2) = \frac{1}{2*2^{\frac{n}{2}-1}Γ(\frac{n}{2})}x^{\frac{n}{2}-1}exp{\{\frac{-x}{2}\}}=f(x;n)$$



+ **Conclusión:** $χ^2$ es un caso especial de la función Gamma, reemplazando $α = \frac{n}{2}$, y $β = 2$


# **e. Valor Esperado y varianza**
+ **Valor Esperado de la Gamma es**
 $$E(X)=\mu=βα,  \ \ \ \ V(X)=σ^2=αβ^2$$
+ **Valor Esperado de  $χ^2$**  reemplazando $α = \frac{n}{2}$, y $β = 2$
 $$E(χ^2)=n$$

+ **Varianza de  $χ^2$**  reemplazando $α = \frac{n}{2}$, y $β = 2$
 $$V(χ^2)=2n$$

# **60. Calculadora de la Distribución Chi cuadrado**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN CHI-CUADRADO (χ²)**
#
# La distribución Chi-Cuadrado es utilizada principalmente en:
#   • Pruebas de bondad de ajuste
#   • Pruebas de independencia
#   • Análisis de varianza
#   • Estimación de varianzas poblacionales
#
# Parámetros:
#   k (grados de libertad) - define la forma de la distribución
#
# Fórmula: f(x) = x^(k/2-1) * e^(-x/2) / (2^(k/2) * Γ(k/2))
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2, norm
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
try:
    plt.style.use('seaborn-v0_8')
except:
    try:
        plt.style.use('seaborn')
    except:
        plt.style.use('default')

plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_chi2():
    """
    Calculadora interactiva para la distribución Chi-Cuadrado
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN CHI-CUADRADO (χ²)")
    print("="*70)
    print("\nLa distribución Chi-Cuadrado se utiliza para:")
    print("  • Pruebas de bondad de ajuste")
    print("  • Pruebas de independencia en tablas de contingencia")
    print("  • Intervalos de confianza para la varianza")
    print("  • Pruebas de homogeneidad\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            df = float(input("  Grados de libertad (k > 0): "))
            if df <= 0:
                print("  ❌ k debe ser mayor que 0. Intenta de nuevo.")
                continue

            print("\nTipo de cálculo:")
            print("  1. P(X = a)  - Probabilidad puntual")
            print("  2. P(X ≤ a)  - Probabilidad acumulada a la izquierda")
            print("  3. P(X ≥ a)  - Probabilidad acumulada a la derecha")
            print("  4. P(a ≤ X ≤ b) - Probabilidad entre dos valores")
            print("  5. Cuantil - Encontrar X dado un percentil")
            print("  6. Comparar con diferentes grados de libertad")

            tipo = input("\nSelecciona el tipo de cálculo (1-6): ")

            if tipo == '1':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 0  # En distribuciones continuas, P(X=a)=0
                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X = {a}) = 0")
                print("En distribuciones continuas, la probabilidad puntual es siempre 0.")
                print(f"Densidad en X={a}: {chi2.pdf(a, df):.6f}")

                visualizar_chi2(df, a, None, prob, tipo)
                return {'df': df, 'a': a}

            elif tipo == '2':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = chi2.cdf(a, df)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≤ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con k = {df} grados de libertad")

                visualizar_chi2(df, a, None, prob, tipo)
                return {'df': df, 'a': a, 'prob': prob}

            elif tipo == '3':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 1 - chi2.cdf(a, df)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≥ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con k = {df} grados de libertad")

                visualizar_chi2(df, a, None, prob, tipo)
                return {'df': df, 'a': a, 'prob': prob}

            elif tipo == '4':
                a = float(input("  Límite inferior (a ≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                b = float(input("  Límite superior (b > a): "))
                if b <= a:
                    print("  ❌ b debe ser mayor que a. Intenta de nuevo.")
                    continue
                prob = chi2.cdf(b, df) - chi2.cdf(a, df)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP({a} ≤ X ≤ {b}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con k = {df} grados de libertad")

                visualizar_chi2(df, a, b, prob, tipo)
                return {'df': df, 'a': a, 'b': b, 'prob': prob}

            elif tipo == '5':
                p = float(input("  Percentil (0-1): "))
                if p <= 0 or p >= 1:
                    print("  ❌ El percentil debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                x = chi2.ppf(p, df)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nPara el percentil {p*100:.1f}%:")
                print(f"X = {x:.6f}")
                print(f"Con k = {df} grados de libertad")
                print(f"Interpretación: El {p*100:.1f}% de los datos están por debajo de {x:.4f}")

                visualizar_cuantil_chi2(df, x, p)
                return {'df': df, 'percentil': p, 'x': x}

            elif tipo == '6':
                comparar_df_chi2(df)
                return {'df': df}

            else:
                print("  ❌ Opción no válida. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

#==========================================================
# FUNCIÓN PARA VISUALIZAR LA DISTRIBUCIÓN CHI-CUADRADO
#==========================================================

def visualizar_chi2(df, a, b, prob, tipo):
    """Crea visualizaciones completas de la distribución Chi-Cuadrado"""

    # Rango de valores para la gráfica
    x_max = max(chi2.ppf(0.999, df), a * 1.5 if a > 0 else chi2.ppf(0.999, df))
    x_max = max(x_max, 10)
    x = np.linspace(0, x_max, 1000)
    y = chi2.pdf(x, df)

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    #==========================================================
    # GRÁFICO 1: Distribución con área sombreada
    #==========================================================
    ax1 = plt.subplot(2, 2, 1)
    ax1.plot(x, y, 'b-', linewidth=2.5, label=f'χ² (k={df:.1f})')

    # Sombrear el área según el tipo de cálculo
    if tipo == '2':  # P(X ≤ a)
        mask = x <= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≤ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '3':  # P(X ≥ a)
        mask = x >= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≥ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '4':  # P(a ≤ X ≤ b)
        mask = (x >= a) & (x <= b)
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P({a:.2f} ≤ X ≤ {b:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)
        ax1.axvline(x=b, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    # Estadísticas
    media = df
    varianza = 2 * df
    desv = np.sqrt(varianza)
    moda = max(0, df - 2)

    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.1f}')
    if moda > 0:
        ax1.axvline(x=moda, color='#F39C12', linestyle='-.', linewidth=1.5, label=f'Moda = {moda:.1f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Distribución χ² (k={df:.1f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', fontsize=10)
    ax1.set_xlim(0, x_max)

    # Agregar información de probabilidad
    if prob is not None:
        ax1.text(0.98, 0.95, f'Probabilidad = {prob:.6f}\n({prob*100:.2f}%)',
                transform=ax1.transAxes, fontsize=11, verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 2: Comparación con diferentes grados de libertad
    #==========================================================
    ax2 = plt.subplot(2, 2, 2)

    dfs = [1, 2, 3, 5, 10]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    x2 = np.linspace(0, 20, 1000)
    for d, color in zip(dfs, colores):
        y_temp = chi2.pdf(x2, d)
        ax2.plot(x2, y_temp, color=color, linewidth=2,
                label=f'k = {d}')

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax2.set_title('Comparación: Diferentes grados de libertad', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best', fontsize=9)
    ax2.set_xlim(0, 20)

    #==========================================================
    # GRÁFICO 3: Función de distribución acumulada
    #==========================================================
    ax3 = plt.subplot(2, 2, 3)
    cdf = chi2.cdf(x, df)
    ax3.plot(x, cdf, 'g-', linewidth=2.5, label='CDF')
    ax3.fill_between(x, 0, cdf, alpha=0.2, color='green')

    # Puntos de interés
    if tipo == '2':  # P(X ≤ a)
        ax3.scatter(a, prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {prob:.4f}')
        ax3.axhline(y=prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '3':  # P(X ≥ a)
        ax3.scatter(a, 1-prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {1-prob:.4f}')
        ax3.axhline(y=1-prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles comunes
    for q in [0.25, 0.5, 0.75, 0.9, 0.95]:
        q_val = chi2.ppf(q, df)
        if q_val <= x_max:
            ax3.axvline(x=q_val, color='orange', linestyle=':', linewidth=1.5, alpha=0.5)
            ax3.text(q_val, 0.01, f'Q{int(q*100)}', ha='center', fontsize=8, color='orange')

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax3.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', fontsize=10)
    ax3.set_ylim(-0.05, 1.05)
    ax3.set_xlim(0, x_max)

    #==========================================================
    # GRÁFICO 4: Tabla de valores críticos
    #==========================================================
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Crear tabla de valores críticos comunes
    niveles = [0.90, 0.95, 0.975, 0.99, 0.995, 0.999]
    valores_chi2 = [chi2.ppf(nivel, df) for nivel in niveles]

    table_data = []
    table_data.append(['Nivel', 'χ² crítico', 'Interpretación'])
    for nivel, chi_val in zip(niveles, valores_chi2):
        if chi_val < x_max * 1.5:
            table_data.append([
                f'{nivel*100:.1f}%',
                f'{chi_val:.4f}',
                f'P(X ≤ {chi_val:.2f}) = {nivel:.3f}'
            ])

    # Si hay demasiados datos, mostrar solo los más relevantes
    if len(table_data) > 8:
        table_data = table_data[:1] + table_data[1::2]

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.25, 0.3, 0.4])

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.8)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title(f'Valores Críticos χ² (k={df:.1f})', fontsize=13, fontweight='bold', pad=20)

    #==========================================================
    # Ajustar layout
    #==========================================================
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # GRÁFICO ADICIONAL: Propiedades de la distribución
    #==========================================================
    fig2, (ax5, ax6) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 5: Media y varianza en función de k
    k_vals = np.linspace(0.5, 20, 100)
    media_vals = k_vals
    var_vals = 2 * k_vals
    desv_vals = np.sqrt(var_vals)

    ax5.plot(k_vals, media_vals, 'b-', linewidth=2, label='Media = k')
    ax5.plot(k_vals, var_vals, 'r-', linewidth=2, label='Varianza = 2k')
    ax5.plot(k_vals, desv_vals, 'g-', linewidth=2, label='Desv. Estándar = √(2k)')
    ax5.axvline(x=df, color='purple', linestyle='--', linewidth=2, alpha=0.5, label=f'k = {df}')
    ax5.set_xlabel('Grados de libertad (k)', fontsize=12)
    ax5.set_ylabel('Valor', fontsize=12)
    ax5.set_title('Media y Varianza de χ²', fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3)
    ax5.legend(loc='best')

    # Gráfico 6: Convergencia a la Normal
    ax6.plot(x, y, 'b-', linewidth=2.5, label=f'χ² (k={df:.1f})')

    # Aproximación Normal para k grande
    if df >= 30:
        mu_norm = df
        sigma_norm = np.sqrt(2*df)
        y_norm = norm.pdf(x, mu_norm, sigma_norm)
        ax6.plot(x, y_norm, 'r--', linewidth=2, label='Aprox. Normal')

    ax6.set_xlabel('X', fontsize=12)
    ax6.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax6.set_title('Convergencia a la Normal (teorema central)', fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3)
    ax6.legend(loc='best')
    ax6.set_xlim(0, x_max)

    plt.tight_layout()

    #==========================================================
    # MOSTRAR GRÁFICOS
    #==========================================================
    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"Chi2_df{df:.1f}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_propiedades.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_propiedades.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

def visualizar_cuantil_chi2(df, x, p):
    """Visualiza el cuantil en la distribución Chi-Cuadrado"""

    x_max = max(chi2.ppf(0.999, df), x * 1.5)
    x_vals = np.linspace(0, x_max, 1000)
    y = chi2.pdf(x_vals, df)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Distribución con cuantil marcado
    ax1.plot(x_vals, y, 'b-', linewidth=2.5)
    mask = x_vals <= x
    ax1.fill_between(x_vals[mask], 0, y[mask], color='#E63946', alpha=0.5,
                     label=f'Área = {p*100:.1f}%')
    ax1.axvline(x=x, color='#E63946', linestyle='--', linewidth=2, label=f'X = {x:.2f}')

    media = df
    ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.1f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Percentil {p*100:.1f}% (k={df:.1f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best')
    ax1.set_xlim(0, x_max)

    # Gráfico 2: Función de distribución acumulada
    cdf = chi2.cdf(x_vals, df)
    ax2.plot(x_vals, cdf, 'g-', linewidth=2.5)
    ax2.scatter(x, p, color='#E63946', s=150, zorder=5, label=f'F({x:.2f}) = {p:.4f}')
    ax2.axhline(y=p, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax2.axvline(x=x, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax2.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best')
    ax2.set_ylim(-0.05, 1.05)
    ax2.set_xlim(0, x_max)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA COMPARAR DIFERENTES GRADOS DE LIBERTAD
#==========================================================

def comparar_df_chi2(df_actual):
    """Compara la distribución actual con diferentes grados de libertad"""

    print("\n" + "="*70)
    print("📊 COMPARACIÓN DE GRADOS DE LIBERTAD")
    print("="*70)

    dfs = [1, 2, 3, 5, 10, 20, 30]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#9B59B6', '#1ABC9C', '#2C3E50']

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Densidades
    x = np.linspace(0, 40, 1000)
    for d, color in zip(dfs, colores):
        y = chi2.pdf(x, d)
        axes[0].plot(x, y, color=color, linewidth=2,
                    label=f'k = {d}', alpha=0.8 if d != df_actual else 1.0)
        if d == df_actual:
            axes[0].fill_between(x, 0, y, alpha=0.2, color=color)

    axes[0].set_xlabel('X', fontsize=12)
    axes[0].set_ylabel('Densidad de probabilidad', fontsize=12)
    axes[0].set_title('Distribuciones χ² con diferentes k', fontsize=13, fontweight='bold')
    axes[0].grid(True, alpha=0.3)
    axes[0].legend(loc='best', ncol=2)
    axes[0].set_xlim(0, 30)

    # Gráfico 2: CDF
    for d, color in zip(dfs, colores):
        y = chi2.cdf(x, d)
        axes[1].plot(x, y, color=color, linewidth=2,
                    label=f'k = {d}', alpha=0.8 if d != df_actual else 1.0)

    axes[1].set_xlabel('X', fontsize=12)
    axes[1].set_ylabel('Probabilidad acumulada', fontsize=12)
    axes[1].set_title('Funciones de Distribución Acumulada', fontsize=13, fontweight='bold')
    axes[1].grid(True, alpha=0.3)
    axes[1].legend(loc='best', ncol=2)
    axes[1].set_xlim(0, 30)
    axes[1].set_ylim(0, 1.05)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA INTERVALOS DE CONFIANZA DE VARIANZA
#==========================================================

def intervalo_confianza_varianza():
    """Calcula intervalos de confianza para la varianza usando χ²"""

    print("\n" + "="*70)
    print("📈 INTERVALO DE CONFIANZA PARA LA VARIANZA")
    print("="*70)
    print("\nEsta función calcula intervalos de confianza para la")
    print("varianza poblacional usando la distribución χ².\n")

    try:
        print("Ingresa los datos:")
        varianza = float(input("  Varianza muestral (s²): "))
        n = int(input("  Tamaño de muestra (n): "))
        nivel = float(input("  Nivel de confianza (ej: 95): ")) / 100

        # Grados de libertad
        df = n - 1

        # Valores críticos χ²
        chi2_inf = chi2.ppf((1 - nivel) / 2, df)
        chi2_sup = chi2.ppf(1 - (1 - nivel) / 2, df)

        # Intervalo de confianza
        li = df * varianza / chi2_sup
        ls = df * varianza / chi2_inf

        print(f"\nResultados:")
        print(f"  • Grados de libertad: {df}")
        print(f"  • χ² inferior: {chi2_inf:.4f}")
        print(f"  • χ² superior: {chi2_sup:.4f}")
        print(f"  • Intervalo de confianza al {nivel*100:.0f}% para σ²: ({li:.4f}, {ls:.4f})")
        print(f"  • Intervalo de confianza para σ: ({np.sqrt(li):.4f}, {np.sqrt(ls):.4f})")

        # Visualización
        fig, ax = plt.subplots(figsize=(10, 6))

        x = np.linspace(0, max(chi2_sup * 1.5, 30), 1000)
        y = chi2.pdf(x, df)

        ax.plot(x, y, 'b-', linewidth=2, label=f'χ² (k={df})')

        # Área del intervalo de confianza
        x_fill = np.linspace(chi2_inf, chi2_sup, 100)
        y_fill = chi2.pdf(x_fill, df)
        ax.fill_between(x_fill, 0, y_fill, color='#E63946', alpha=0.5,
                        label=f'IC {nivel*100:.0f}%')

        ax.axvline(x=chi2_inf, color='#E63946', linestyle='--', linewidth=2)
        ax.axvline(x=chi2_sup, color='#E63946', linestyle='--', linewidth=2)
        ax.axvline(x=df, color='#2C3E50', linestyle=':', linewidth=2, label=f'k = {df}')

        ax.set_xlabel('χ²', fontsize=12)
        ax.set_ylabel('Densidad de probabilidad', fontsize=12)
        ax.set_title(f'Intervalo de Confianza para Varianza\nn={n}, {nivel*100:.0f}% de confianza',
                    fontsize=13, fontweight='bold')
        ax.grid(True, alpha=0.3, linestyle='--')
        ax.legend(loc='best')

        ax.text(0.98, 0.95, f'IC {nivel*100:.0f}%\n({li:.4f}, {ls:.4f})',
               transform=ax.transAxes, fontsize=11, verticalalignment='top',
               horizontalalignment='right',
               bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA PRUEBA DE BONDAD DE AJUSTE
#==========================================================

def prueba_bondad_ajuste():
    """Realiza una prueba de bondad de ajuste con χ²"""

    print("\n" + "="*70)
    print("📊 PRUEBA DE BONDAD DE AJUSTE (χ²)")
    print("="*70)
    print("\nEsta función realiza una prueba de bondad de ajuste")
    print("para verificar si los datos observados siguen una")
    print("distribución esperada.\n")

    try:
        print("Ingresa los datos:")
        n_categorias = int(input("  Número de categorías: "))

        observados = []
        for i in range(n_categorias):
            obs = float(input(f"  Frecuencia observada para categoría {i+1}: "))
            observados.append(obs)

        esperados = []
        for i in range(n_categorias):
            esp = float(input(f"  Frecuencia esperada para categoría {i+1}: "))
            esperados.append(esp)

        # Calcular estadístico χ²
        chi2_stat = sum((obs - esp)**2 / esp for obs, esp in zip(observados, esperados))
        df = n_categorias - 1
        p_value = 1 - chi2.cdf(chi2_stat, df)

        print(f"\nResultados:")
        print(f"  • Estadístico χ² = {chi2_stat:.4f}")
        print(f"  • Grados de libertad = {df}")
        print(f"  • p-value = {p_value:.6f} ({p_value*100:.4f}%)")

        if p_value > 0.05:
            print("  ✅ No se rechaza H₀: Los datos siguen la distribución esperada (p > 0.05)")
        else:
            print("  ❌ Se rechaza H₀: Los datos NO siguen la distribución esperada (p ≤ 0.05)")

        # Visualización
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

        # Gráfico 1: Comparación de frecuencias
        x = np.arange(len(observados))
        width = 0.35
        ax1.bar(x - width/2, observados, width, label='Observados', color='#2E86AB', alpha=0.8)
        ax1.bar(x + width/2, esperados, width, label='Esperados', color='#E63946', alpha=0.8)
        ax1.set_xlabel('Categoría', fontsize=12)
        ax1.set_ylabel('Frecuencia', fontsize=12)
        ax1.set_title('Comparación de Frecuencias', fontsize=13, fontweight='bold')
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='best')

        # Gráfico 2: Distribución χ²
        x_chi2 = np.linspace(0, max(chi2_stat * 2, 10), 1000)
        y_chi2 = chi2.pdf(x_chi2, df)

        ax2.plot(x_chi2, y_chi2, 'b-', linewidth=2, label=f'χ² (k={df})')
        ax2.axvline(x=chi2_stat, color='#E63946', linestyle='--', linewidth=2,
                   label=f'χ² = {chi2_stat:.2f}')

        # Área de rechazo (p-value)
        mask = x_chi2 >= chi2_stat
        ax2.fill_between(x_chi2[mask], 0, y_chi2[mask], color='#E63946', alpha=0.3,
                        label=f'p-value = {p_value:.4f}')

        ax2.set_xlabel('χ²', fontsize=12)
        ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
        ax2.set_title('Distribución χ² y Estadístico de Prueba', fontsize=13, fontweight='bold')
        ax2.grid(True, alpha=0.3)
        ax2.legend(loc='best')

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_chi2():
    """Ejecuta ejemplos clásicos de la distribución Chi-Cuadrado"""

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN CHI-CUADRADO")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Valor crítico χ²',
            'descripcion': 'Con 10 grados de libertad, ¿cuál es el valor crítico para un nivel de significancia de 0.05 (prueba de una cola)?',
            'tipo': 'critico',
            'df': 10,
            'alfa': 0.05
        },
        {
            'titulo': 'Ejemplo 2: Probabilidad en χ²',
            'descripcion': 'Con 8 grados de libertad, ¿cuál es la probabilidad de que X sea mayor que 15?',
            'tipo': 'prob',
            'df': 8,
            'valor': 15
        },
        {
            'titulo': 'Ejemplo 3: Intervalo de confianza para varianza',
            'descripcion': 'Una muestra de 20 elementos tiene varianza 25. ¿Cuál es el intervalo de confianza al 95% para la varianza poblacional?',
            'tipo': 'IC_varianza',
            'varianza': 25,
            'n': 20,
            'nivel': 95
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")

        if ej['tipo'] == 'critico':
            chi_crit = chi2.ppf(1 - ej['alfa'], ej['df'])
            print(f"   χ² crítico (α={ej['alfa']}, k={ej['df']}) = {chi_crit:.4f}")

        elif ej['tipo'] == 'prob':
            prob = 1 - chi2.cdf(ej['valor'], ej['df'])
            print(f"   P(X > {ej['valor']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'IC_varianza':
            df = ej['n'] - 1
            chi2_inf = chi2.ppf((1 - ej['nivel']/100) / 2, df)
            chi2_sup = chi2.ppf(1 - (1 - ej['nivel']/100) / 2, df)
            li = df * ej['varianza'] / chi2_sup
            ls = df * ej['varianza'] / chi2_inf
            print(f"   IC al {ej['nivel']}% para σ²: ({li:.4f}, {ls:.4f})")
            print(f"   IC para σ: ({np.sqrt(li):.4f}, {np.sqrt(ls):.4f})")

    print("\n" + "="*70)

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN CHI-CUADRADO")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado")
        print("  2. Ver ejemplos clásicos")
        print("  3. Intervalo de confianza para varianza")
        print("  4. Prueba de bondad de ajuste")
        print("  5. Salir")

        opcion = input("\nSelecciona una opción (1-5): ")

        if opcion == '1':
            resultados = calculadora_chi2()
        elif opcion == '2':
            ejemplos_chi2()
        elif opcion == '3':
            intervalo_confianza_varianza()
        elif opcion == '4':
            prueba_bondad_ajuste()
        elif opcion == '5':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3, 4 o 5.")
```

#**a. Construya el gráfico de la Distribución dada**
+ **Densidad Distribución Chi Cuadrado**
$${\displaystyle f(x;n)=\frac{{(\frac{x}{2})^{\frac{n}{2}-1}}e^{-(\frac{x}{2})}}{2\Gamma(\frac{n}{2})},\dots ,x\geq 0}$$

```python
#@title **a. Construya el gráfico de la Distribución dada si  $α=11.7 y β=7.24$**
plt.figure(figsize=(8, 8))
df = 50 # parametro de forma.
chi2 = stats.chi2(df)

x = np.linspace(chi2.ppf(0.0001),
                chi2.ppf(0.99999), 10000)
fp = chi2.pdf(x) # Función de Probabilidad
plt.plot(x, fp)

plt.title('Fabrica A - Distribución Chi_Cuadrado')
plt.ylabel('Probabilidad')
plt.xlabel('Variabilidad contenido en contenido de las bolsas')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **61.  La Distribución *t* de Student**

# **a. Generalidades de la Distribución *t* de Student**
+ **Wikipedia**. En probabilidad y estadística, la distribución $t$ de Student es una distribución de probabilidad que surge del problema de estimar la media de una población normalmente distribuida cuando el tamaño de la muestra es pequeño y la desviación estándar poblacional es desconocida.

- Fue desarrollada por William Sealy Gosset bajo el pseudónimo “Student”.
- Aparece de manera natural al realizar la prueba t de Student para la determinación de las diferencias entre dos varianzas muestrales y para la construcción del intervalo de confianza para la diferencia entre las partes de dos poblaciones cuando se desconoce la desviación típica de una población y esta debe ser estimada a partir de los datos de una muestra.


- *Historia*. La distribución de Student fue descrita en el año 1908 por *William Sealy Gosset*. Gosset trabajaba en una fábrica de cerveza, *Guinness*, que prohibía a sus empleados la publicación de artículos científicos debido a una difusión previa de secretos industriales. De ahí que Gosset publicase sus resultados bajo el pseudónimo de *“Student”*




#  **b. Distribución t de Student a partir de una muestra aleatoria**
- Sea ${\displaystyle X_{1},\cdots ,X_{n}}$ variables aleatorias independientes distribuidas ${\displaystyle N(\mu ,\sigma ^{2})}$, esto es, ${\displaystyle X_{1},\dots ,X_{n}}$ es una muestra aleatoria de tamaño ${\displaystyle n}$ proveniente de una población con distribución normal con media $\mu$ y varianza $\sigma ^{2}$.

- **Consideremos los Estadísticos: la media muestral y la varianza muestral**
$$
{\displaystyle {\overline {X}}={\frac {1}{n}}\sum _{i=1}^{n}X_{i}}$$


$${\displaystyle S^{2}={\frac {1}{n-1}}\sum _{i=1}^{n}\left(X_{i}-{\overline {X}}\right)^{2}}$$

- **Entonces, la variable aleatoria**

$${\displaystyle {\frac {{\overline {X}}-\mu }{\sigma /{\sqrt {n}}}}}$$
- sigue una distribución normal estándar (es decir, una distribución normal con media 0 y varianza 1)
- y la variable aleatoria

$${\displaystyle {\frac {{\overline {X}}-\mu }{S/{\sqrt {n}}}}}$$
- donde ${\displaystyle S}$ ha sido sustituido por ${\displaystyle \sigma }$, tiene una distribución $t$ de student con $n-1$ grados de libertad.




# **c. Función de densidad  t de Student**
- Sean $X$ una variable aleatoria continua y ${\displaystyle n>0}$, si ${\displaystyle X}$ tiene una distribución $t$ con $v$ grados de libertad entonces escribiremos ${\displaystyle X\sim t_{n}}$ o ${\displaystyle X\sim t(n)}$ si tiene  **función de densidad de probabilidad**
$${\displaystyle f_{X}(x)={\frac {\Gamma \left({\frac {n+1}{2}}\right)}{{\sqrt {v\pi }}\;\Gamma \left({\frac {n}{2}}\right)}}\left(1+{\frac {x^{2}}{n}}\right)^{-{\frac {n+1}{2}}}}$$
para ${\displaystyle x\in \mathbb {R} }$, donde ${\displaystyle v}$ denota los grados de libertad y ${\displaystyle \Gamma }$ es la función gamma.

- **La expresión anterior también suele escribirse como:**
$$f_{X}(x)={\frac {1}{{\sqrt {n}}\;\operatorname {B} \left({\frac {1}{2}},{\frac {n}{2}}\right)}}\left(1+{\frac {x^{2}}{n}}\right)^{-{\frac {n+1}{2}}}$$
donde $B$ es la función beta.


+ **d. Propiedades**

- Si ${\displaystyle X}$ es una variable aleatoria tal que ${\displaystyle X\sim t_{n}}$ entonces $X$ satisface:

- **Media**. La media de $X$ para valores ${\displaystyle n>1}$ es
$${\displaystyle \operatorname {E} [X]=0}$$
- **Varianza**. La varianza de $X$ para valores ${\displaystyle n>2}$ es

$${\displaystyle \operatorname {Var} (X)={\frac {n}{n-2}}}$$



# **d. Caracterización**
+ La distribución $t$ de Student con $v$ grados de libertad puede definirse como la distribución de la variable aleatoria $T$ definida por:

$${\displaystyle T:={\frac {Z}{\sqrt {\frac {X}{v}}}}\sim t_{v}}$$
- donde ${\displaystyle Z\sim N(0,1)}$, es decir, ${\displaystyle Z}$ es una variable aleatoria con distribución normal estándar (distribución normal con media 0 y varianza 1).

- ${\displaystyle X\sim \chi _{v}^{2}}$, es decir $X$ es una variable aleatoria que sigue una distribución chi-cuadrada con $v$ grados de libertad.
- $Z$ y $X$  son variables aleatorias independientes.
- Para una constante {\displaystyle \mu }\mu no nula, el cociente

- ${\displaystyle (Z+\mu ){\sqrt {\frac {v}{X}}}}$ es una variable aleatoria que sigue la distribución no central $t$  de Student con parámetro de no-centralidad $\mu$.

# **e. Intervalo para la media**
- $\sigma^2$ es desconocida

- Sean ${\displaystyle X_{1},\dots ,X_{n}}$ una muestra aleatoria proveniente de una población con distribución $N(\mu ,\sigma ^{2})$ donde $\mu$ y $\sigma$ son desconocidos.

- Se tiene que
$${\displaystyle {\frac {{\overline {X}}-\mu }{\sigma /{\sqrt {n}}}}\sim N(0,1)}$$
- y
$${\displaystyle {\frac {(n-1)S^{2}}{\sigma ^{2}}}\sim \chi _{n-1}^{2}}$$
- son independientes entonces el cociente
$${\displaystyle {\frac {\frac {{\overline {X}}-\mu }{\sigma /{\sqrt {n}}}}{\sqrt {\frac {\frac {(n-1)S^{2}}{\sigma ^{2}}}{n-1}}}}\sim t_{n-1}}$$
- esto es
$${\displaystyle {\frac {{\overline {X}}-\mu }{S/{\sqrt {n}}}}\sim t_{n-1}}$$
- Sea ${\displaystyle t_{n-1,1-\alpha /2}\in \mathbb {R} }$ tal que

- ${\displaystyle \operatorname {P} [Y\leq t_{n-1,1-\alpha /2}]=1-{\frac {\alpha }{2}}}$ siendo ${\displaystyle Y\sim t_{n-1}}$ entonces

 $${\displaystyle {\begin{aligned}&\operatorname {P} \left[-t_{n-1,1-\alpha /2}\leq {\frac {{\overline {X}}-\mu }{S/{\sqrt {n}}}}\leq t_{n-1,1-\alpha /2}\right]=1-\alpha \\&\operatorname {P} \left[-t_{n-1,1-\alpha /2}\;{\frac {S}{\sqrt {n}}}\leq {\overline {X}}-\mu \leq t_{n-1,1-\alpha /2}\;{\frac {S}{\sqrt {n}}}\right]=1-\alpha \\&\operatorname {P} \left[-{\overline {X}}-t_{n-1,1-\alpha /2}\;{\frac {S}{\sqrt {n}}}\leq -\mu \leq -{\overline {X}}+t_{n-1,1-\alpha /2}\;{\frac {S}{\sqrt {n}}}\right]=1-\alpha \\&\operatorname {P} \left[{\overline {X}}-t_{n-1,1-\alpha /2}\;{\frac {S}{\sqrt {n}}}\leq \mu \leq {\overline {X}}+t_{n-1,1-\alpha /2}\;{\frac {S}{\sqrt {n}}}\right]=1-\alpha \\\end{aligned}}}$$
- por lo tanto un intervalo de ${\displaystyle (1-\alpha )100\%}$ de confianza para $\mu$ cuando $\sigma ^{2}$  es desconocida es

$${\displaystyle \left({\overline {X}}-t_{n-1,1-\alpha /2}\;{\frac {S}{\sqrt {n}}},{\overline {X}}+t_{n-1,1-\alpha /2}\;{\frac {S}{\sqrt {n}}}\right)}$$





# **d. Veamos que es una función de densidad**

+ **Densidad Distribución t de Student**
 $$f_X(x)=\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}(1+\frac{x^2}{n})^{-\frac{n+1}{2}}$$

+ **Integrando**.
$$ ∫_{-∞}^{∞}f(x)dx= ∫_{-∞}^{∞}\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}(1+\frac{x^2}{n})^{-\frac{n+1}{2}}dx$$
$$ ∫_{-∞}^{∞}f(x)dx= 2\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}∫_{0}^{∞}(1+\frac{x^2}{n})^{-\frac{n+1}{2}}dx$$


+ **hacemos** $y = (1+\frac{x^2}{n})^{-1}$ esto es $x = \sqrt {n}(y^{-1}-1)^{1/2} ⇒dx=-\frac{1}{2} \sqrt {n}(1-y)^{-1/2}y^{-3/2}dy$

+ **Sustituyendo en la Integral**.
$$ ∫_{-∞}^{∞}f(x)dx= -2\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}∫_{1}^{0}(y)^{\frac{n+1}{2}}\frac{1}{2} \sqrt {n}(1-y)^{-1/2}y^{-3/2}dy$$

+ **Reorganizando la Integral**.
$$ ∫_{-∞}^{∞}f(x)dx= \frac{Γ(\frac{n+1}{2})}{\sqrt {\pi }Γ(\frac{n}{2})}∫_{0}^{1}y^{\frac{n}{2}-1} (1-y)^{\frac{1}{2}-1}dy$$

+ **Recordando la función Beta**

 $$
{B(α,β)} = \int_0^{1} x^{α-1}(1-x)^{β-1}dx=\frac{\Gamma(α)\Gamma(β)}{\Gamma(α+β)}
$$

+ **Reemplazando en la  Integral**.
$$ ∫_{-∞}^{∞}f(x)dx= \frac{Γ(\frac{n+1}{2})}{\sqrt {\pi }Γ(\frac{n}{2})}∫_{0}^{1}y^{\frac{n}{2}-1} (1-y)^{\frac{1}{2}-1}dy=\frac{Γ(\frac{n+1}{2})}{\sqrt {\pi }Γ(\frac{n}{2})}B(\frac{n}{2},\frac{1}{2})$$

+ **Usando la propiedad de la funcion Beta arriba**.
$$ ∫_{-∞}^{∞}f(x)dx= \frac{Γ(\frac{n+1}{2})}{\sqrt {\pi }Γ(\frac{n}{2})}\frac{\Gamma(\frac{n}{2})\Gamma(\frac{1}{2})}{\Gamma(\frac{n+1}{2})}=1$$


# **e. Valor Esperado y varianza**

+ **Valor Esperado**.
$$ E(X) = ∫_{-∞}^{∞}xf(x)dx= ∫_{-∞}^{∞}\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}x(1+\frac{x^2}{n})^{-\frac{n+1}{2}}dx = 0$$

+ **Valor Esperado de $X^2$**.
$$ E(X^2) = ∫_{-∞}^{∞}x^2f(x)dx= ∫_{-∞}^{∞}\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}x^2(1+\frac{x^2}{n})^{-\frac{n+1}{2}}dx = 0$$
$$ E(X^2) =∫_{-∞}^{∞}x^2f(x)dx= 2\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}∫_{0}^{∞}x^2(1+\frac{x^2}{n})^{-\frac{n+1}{2}}dx$$

+ **hacemos** $y = (1+\frac{x^2}{n})^{-1}$ esto es $x = \sqrt {n}(y^{-1}-1)^{1/2}  ⇒dx=-\frac{1}{2} \sqrt {n}(1-y)^{-1/2}y^{-3/2}dy$

 - **Donde** $$x^2 = n(1-y)y^{-1}$$

+ **Sustituyendo en la Integral**.
$$ E(X^2) == -2\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}∫_{1}^{0} n(1-y)y^{-1}(y)^{\frac{n+1}{2}}\frac{1}{2} \sqrt {n}(1-y)^{-1/2}y^{-3/2}dy$$


+ **Reorganizando en la Integral**.
$$ E(X^2) = n\frac{Γ(\frac{n+1}{2})}{\sqrt {\pi }Γ(\frac{n}{2})}∫_{0}^{1} (1-y)y^{-1}(y)^{\frac{n+1}{2}} (1-y)^{-1/2}y^{-3/2}dy$$

$$E(X^2) = n\frac{Γ(\frac{n+1}{2})}{\sqrt {\pi }Γ(\frac{n}{2})}∫_{0}^{1} y^{\frac{n-2}{2}-1}(1-y)^{3/2-1}dy$$

+ **Recordando la función Beta**

 $$
{B(α,β)} = \int_0^{1} x^{α-1}(1-x)^{β-1}dx=\frac{\Gamma(α)\Gamma(β)}{\Gamma(α+β)}
$$

+ **Usando la función Beta**

$$ E(X^2) =n\frac{Γ(\frac{n+1}{2})}{\sqrt {\pi }Γ(\frac{n}{2})}B(\frac{n-2}{2},\frac{3}{2})$$

+ **Usando la función Beta**
$$ E(X^2) = nB(\frac{n}{2},\frac{1}{2})B(\frac{n-2}{2},\frac{3}{2})$$



+ **c. Propiedades de la función Beta y gamma**
+ $Γ(α+1)=αΓ(α)$
+ $B(α,β)=B(β,α)$
+ $B(α,1)=\frac{\Gamma(α)\Gamma(1)}{\Gamma(α+1)}=\frac{\Gamma(α)(1)}{α\Gamma(α)}=\frac{1}{α}$

+ $B(α+1,β)=\frac{\Gamma(α+1)\Gamma(β)}{\Gamma(α+β+1)}=\frac{α\Gamma(α)\Gamma(β)}{(α+β)\Gamma(α+β)}=\frac{α}{(α+β)}B(α,β)$


+ **Usando las propiedades de la función Beta**

$$ E(X^2) =n\frac{Γ(\frac{n+1}{2})}{\sqrt {\pi }Γ(\frac{n}{2})}\frac{Γ(\frac{n-2}{2})Γ(\frac{3}{2})}{Γ(\frac{n+1}{2})}$$

 -  $Γ(\frac{3}{2})=Γ(\frac{1}{2}+1)=\frac{1}{2}Γ(\frac{1}{2})=\frac{1}{2}\sqrt {\pi }$

 + **Usando las propiedades de la función Gamma y simplificando**

$$ E(X^2) =\frac{n}{2}\frac{1}{Γ(\frac{n}{2})}\frac{Γ(\frac{n-2}{2})}{1}$$
-  $Γ(\frac{n}{2})=Γ(\frac{n-2}{2}+1)=\frac{n-2}{2}Γ(\frac{n-2}{2})$

+ **Usando las propiedades de la función Gamma y simplificando**

$$ E(X^2) = \frac{n}{2}\frac{1}{\frac{n-2}{2}Γ(\frac{n-2}{2})}\frac{Γ(\frac{n-2}{2})}{1}=\frac{n}{n-2}$$

+ **Por tanto**
$$Var(X)=E(X^2)- E(X)^2 = =\frac{n}{n-2}$$

# **62.  Calculadora de la Distribución t Student**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN t DE STUDENT**
#
# La distribución t de Student es utilizada principalmente
# para estimar la media de una población cuando el tamaño
# de muestra es pequeño y la varianza poblacional es desconocida.
#
# Parámetros:
#   ν (grados de libertad) - define la forma de la distribución
#
# Fórmula: f(t) = Γ((ν+1)/2) / (√(νπ) * Γ(ν/2)) * (1 + t²/ν)^(-(ν+1)/2)
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
# Verificar estilos disponibles
try:
    plt.style.use('seaborn-v0_8')
except:
    try:
        plt.style.use('seaborn')
    except:
        plt.style.use('default')

plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_t_student():
    """
    Calculadora interactiva para la distribución t de Student
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN t DE STUDENT")
    print("="*70)
    print("\nLa distribución t de Student se utiliza para:")
    print("  • Pruebas de hipótesis con muestras pequeñas")
    print("  • Intervalos de confianza cuando σ es desconocida")
    print("  • Comparación de medias entre dos grupos")
    print("  • Análisis de regresión\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            df = float(input("  Grados de libertad (ν > 0): "))
            if df <= 0:
                print("  ❌ ν debe ser mayor que 0. Intenta de nuevo.")
                continue

            print("\nTipo de cálculo:")
            print("  1. P(t = a)  - Probabilidad puntual")
            print("  2. P(t ≤ a)  - Probabilidad acumulada a la izquierda")
            print("  3. P(t ≥ a)  - Probabilidad acumulada a la derecha")
            print("  4. P(a ≤ t ≤ b) - Probabilidad entre dos valores")
            print("  5. Cuantil - Encontrar t dado un percentil")
            print("  6. Comparar con distribución Normal")

            tipo = input("\nSelecciona el tipo de cálculo (1-6): ")

            if tipo == '1':
                a = float(input("  Valor de a: "))
                prob = 0  # En distribuciones continuas, P(t=a)=0
                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(t = {a}) = 0")
                print("En distribuciones continuas, la probabilidad puntual es siempre 0.")
                print(f"Densidad en t={a}: {t.pdf(a, df):.6f}")

                visualizar_t_student(df, a, None, prob, tipo)
                return {'df': df, 'a': a}

            elif tipo == '2':
                a = float(input("  Valor de a: "))
                prob = t.cdf(a, df)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(t ≤ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con ν = {df} grados de libertad")

                visualizar_t_student(df, a, None, prob, tipo)
                return {'df': df, 'a': a, 'prob': prob}

            elif tipo == '3':
                a = float(input("  Valor de a: "))
                prob = 1 - t.cdf(a, df)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(t ≥ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con ν = {df} grados de libertad")

                visualizar_t_student(df, a, None, prob, tipo)
                return {'df': df, 'a': a, 'prob': prob}

            elif tipo == '4':
                a = float(input("  Límite inferior (a): "))
                b = float(input("  Límite superior (b): "))
                if a > b:
                    a, b = b, a
                    print("  ↳ Los límites se han intercambiado para que a < b")
                prob = t.cdf(b, df) - t.cdf(a, df)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP({a} ≤ t ≤ {b}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con ν = {df} grados de libertad")

                visualizar_t_student(df, a, b, prob, tipo)
                return {'df': df, 'a': a, 'b': b, 'prob': prob}

            elif tipo == '5':
                p = float(input("  Percentil (0-1): "))
                if p <= 0 or p >= 1:
                    print("  ❌ El percentil debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                x = t.ppf(p, df)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nPara el percentil {p*100:.1f}%:")
                print(f"t = {x:.6f}")
                print(f"Con ν = {df} grados de libertad")
                print(f"Interpretación: El {p*100:.1f}% de los datos están por debajo de {x:.4f}")

                visualizar_cuantil_t(df, x, p)
                return {'df': df, 'percentil': p, 'x': x}

            elif tipo == '6':
                comparar_con_normal(df)
                return {'df': df}

            else:
                print("  ❌ Opción no válida. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

#==========================================================
# FUNCIÓN PARA VISUALIZAR LA DISTRIBUCIÓN t
#==========================================================

def visualizar_t_student(df, a, b, prob, tipo):
    """Crea visualizaciones completas de la distribución t de Student"""

    # Rango de valores para la gráfica
    x_min = -5
    x_max = 5
    x = np.linspace(x_min, x_max, 1000)
    y = t.pdf(x, df)

    # Comparar con Normal estándar
    y_norm = norm.pdf(x, 0, 1)

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    #==========================================================
    # GRÁFICO 1: Distribución t con área sombreada
    #==========================================================
    ax1 = plt.subplot(2, 2, 1)
    ax1.plot(x, y, 'b-', linewidth=2.5, label=f't (ν={df})')
    ax1.plot(x, y_norm, 'gray', linewidth=1.5, linestyle='--',
            label='Normal estándar', alpha=0.7)

    # Sombrear el área según el tipo de cálculo
    if tipo == '2':  # P(t ≤ a)
        mask = x <= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(t ≤ {a})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f't = {a}')

    elif tipo == '3':  # P(t ≥ a)
        mask = x >= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(t ≥ {a})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f't = {a}')

    elif tipo == '4':  # P(a ≤ t ≤ b)
        mask = (x >= a) & (x <= b)
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P({a} ≤ t ≤ {b})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)
        ax1.axvline(x=b, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    # Línea central
    ax1.axvline(x=0, color='#2C3E50', linestyle=':', linewidth=2, label='t = 0')

    ax1.set_xlabel('t', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Distribución t de Student (ν={df:.1f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', fontsize=10)

    # Agregar información de probabilidad
    if prob is not None:
        ax1.text(0.98, 0.95, f'Probabilidad = {prob:.6f}\n({prob*100:.2f}%)',
                transform=ax1.transAxes, fontsize=11, verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 2: Comparación con diferentes grados de libertad
    #==========================================================
    ax2 = plt.subplot(2, 2, 2)

    dfs = [1, 2, 5, 10, 30]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    for d, color in zip(dfs, colores):
        y_temp = t.pdf(x, d)
        ax2.plot(x, y_temp, color=color, linewidth=2,
                label=f'ν = {d}')

    ax2.plot(x, y_norm, 'k--', linewidth=2, label='Normal(0,1)', alpha=0.7)

    ax2.set_xlabel('t', fontsize=12)
    ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax2.set_title('Comparación: Diferentes grados de libertad', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best', fontsize=9)

    #==========================================================
    # GRÁFICO 3: Función de distribución acumulada
    #==========================================================
    ax3 = plt.subplot(2, 2, 3)
    cdf = t.cdf(x, df)
    ax3.plot(x, cdf, 'g-', linewidth=2.5, label='CDF')
    ax3.fill_between(x, 0, cdf, alpha=0.2, color='green')

    # Puntos de interés
    if tipo == '2':  # P(t ≤ a)
        ax3.scatter(a, prob, color='#E63946', s=100, zorder=5, label=f'F({a}) = {prob:.4f}')
        ax3.axhline(y=prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '3':  # P(t ≥ a)
        ax3.scatter(a, 1-prob, color='#E63946', s=100, zorder=5, label=f'F({a}) = {1-prob:.4f}')
        ax3.axhline(y=1-prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles
    for q in [0.25, 0.5, 0.75]:
        q_val = t.ppf(q, df)
        ax3.axvline(x=q_val, color='orange', linestyle=':', linewidth=1.5, alpha=0.5)
        ax3.text(q_val, 0.01, f'Q{int(q*100)}', ha='center', fontsize=9, color='orange')

    ax3.set_xlabel('t', fontsize=12)
    ax3.set_ylabel('Probabilidad acumulada F(t)', fontsize=12)
    ax3.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', fontsize=10)
    ax3.set_ylim(-0.05, 1.05)

    #==========================================================
    # GRÁFICO 4: Tabla de valores t críticos
    #==========================================================
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Crear tabla de valores t críticos comunes
    niveles = [0.90, 0.95, 0.975, 0.99, 0.995]
    valores_t = [t.ppf(nivel, df) for nivel in niveles]

    table_data = []
    table_data.append(['Nivel', 't crítico', 'Interpretación'])
    for nivel, t_val in zip(niveles, valores_t):
        table_data.append([
            f'{(1-nivel)*100:.1f}%',
            f'{t_val:.4f}',
            f'P(t > {t_val:.3f}) = {1-nivel:.3f}'
        ])

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.25, 0.3, 0.4])

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.8)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title(f'Valores t Críticos (ν={df:.1f})', fontsize=13, fontweight='bold', pad=20)

    #==========================================================
    # Ajustar layout
    #==========================================================
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # GRÁFICO ADICIONAL: Efecto de los grados de libertad
    #==========================================================
    fig2, ax5 = plt.subplots(figsize=(12, 6))

    # Mostrar cómo la distribución se acerca a la Normal
    dfs_effect = [1, 3, 5, 10, 30]
    colores_effect = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    for d, color in zip(dfs_effect, colores_effect):
        y_temp = t.pdf(x, d)
        ax5.plot(x, y_temp, color=color, linewidth=2.5,
                label=f'ν = {d}')

    ax5.plot(x, y_norm, 'k--', linewidth=3, label='Normal(0,1)', alpha=0.6)
    ax5.axvline(x=0, color='gray', linestyle=':', linewidth=1, alpha=0.5)

    ax5.set_xlabel('t', fontsize=12)
    ax5.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax5.set_title('Convergencia a la Normal Estándar', fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.legend(loc='best', fontsize=10)
    ax5.set_xlim(-4, 4)

    plt.tight_layout()

    #==========================================================
    # MOSTRAR GRÁFICOS
    #==========================================================
    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"TStudent_df{df:.1f}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_convergencia.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_convergencia.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

def visualizar_cuantil_t(df, x, p):
    """Visualiza el cuantil en la distribución t"""

    x_vals = np.linspace(-5, 5, 1000)
    y = t.pdf(x_vals, df)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Distribución con cuantil marcado
    ax1.plot(x_vals, y, 'b-', linewidth=2.5)
    mask = x_vals <= x
    ax1.fill_between(x_vals[mask], 0, y[mask], color='#E63946', alpha=0.5,
                     label=f'Área = {p*100:.1f}%')
    ax1.axvline(x=x, color='#E63946', linestyle='--', linewidth=2, label=f't = {x:.2f}')
    ax1.axvline(x=0, color='#2C3E50', linestyle=':', linewidth=2, label='t = 0')

    ax1.set_xlabel('t', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Percentil {p*100:.1f}% (ν={df:.1f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best')

    # Gráfico 2: Función de distribución acumulada
    cdf = t.cdf(x_vals, df)
    ax2.plot(x_vals, cdf, 'g-', linewidth=2.5)
    ax2.scatter(x, p, color='#E63946', s=150, zorder=5, label=f'F({x:.2f}) = {p:.4f}')
    ax2.axhline(y=p, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax2.axvline(x=x, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax2.set_xlabel('t', fontsize=12)
    ax2.set_ylabel('Probabilidad acumulada F(t)', fontsize=12)
    ax2.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best')
    ax2.set_ylim(-0.05, 1.05)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA COMPARAR CON LA DISTRIBUCIÓN NORMAL
#==========================================================

def comparar_con_normal(df):
    """Compara la distribución t con la Normal estándar"""

    print("\n" + "="*70)
    print("📊 COMPARACIÓN: t-STUDENT vs NORMAL ESTÁNDAR")
    print("="*70)
    print(f"\nComparando t con {df} grados de libertad vs Normal(0,1)")

    # Calcular diferencias en cuantiles
    niveles = [0.90, 0.95, 0.975, 0.99, 0.995, 0.999]

    print("\nComparación de cuantiles:")
    print("-"*70)
    print(f"{'Nivel':^10} | {'t-Student':^12} | {'Normal':^12} | {'Diferencia':^12}")
    print("-"*70)

    for nivel in niveles:
        t_val = t.ppf(nivel, df)
        norm_val = norm.ppf(nivel, 0, 1)
        diff = abs(t_val - norm_val)
        print(f"{nivel*100:>5.1f}%   | {t_val:>12.4f} | {norm_val:>12.4f} | {diff:>12.4f}")

    print("\n" + "="*70)
    print("Interpretación:")
    if df >= 30:
        print("✅ Con ν ≥ 30, la distribución t se aproxima bien a la Normal")
    elif df >= 10:
        print("✅ Con ν ≥ 10, la aproximación es razonable")
    else:
        print("⚠️ Con ν < 10, la distribución t tiene colas más pesadas que la Normal")

    # Crear gráfico comparativo
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    x = np.linspace(-4, 4, 1000)
    y_t = t.pdf(x, df)
    y_norm = norm.pdf(x, 0, 1)

    # Gráfico 1: Densidades
    ax1.plot(x, y_t, 'b-', linewidth=2.5, label=f't (ν={df})')
    ax1.plot(x, y_norm, 'r--', linewidth=2, label='Normal(0,1)')
    ax1.fill_between(x, 0, np.minimum(y_t, y_norm), alpha=0.3, color='purple')
    ax1.set_xlabel('x', fontsize=12)
    ax1.set_ylabel('Densidad', fontsize=12)
    ax1.set_title('Comparación de Densidades', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='best')

    # Gráfico 2: Diferencia en colas
    x_colas = np.linspace(2, 4, 500)
    diff_colas = t.pdf(x_colas, df) - norm.pdf(x_colas, 0, 1)
    ax2.plot(x_colas, diff_colas, 'g-', linewidth=2)
    ax2.axhline(y=0, color='black', linestyle='-', linewidth=0.5)
    ax2.fill_between(x_colas, 0, diff_colas, where=(diff_colas>0), alpha=0.3, color='red', label='Colas más pesadas')
    ax2.fill_between(x_colas, 0, diff_colas, where=(diff_colas<0), alpha=0.3, color='blue', label='Colas más ligeras')
    ax2.set_xlabel('x', fontsize=12)
    ax2.set_ylabel('Diferencia (t - Normal)', fontsize=12)
    ax2.set_title('Diferencia en las Colas', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA INTERVALOS DE CONFIANZA CON t
#==========================================================

def intervalo_confianza_t():
    """Calcula intervalos de confianza usando la distribución t"""

    print("\n" + "="*70)
    print("📈 INTERVALO DE CONFIANZA CON DISTRIBUCIÓN t")
    print("="*70)
    print("\nEsta función calcula intervalos de confianza para la media")
    print("cuando la varianza poblacional es desconocida.\n")

    try:
        print("Ingresa los datos:")
        media = float(input("  Media muestral (x̄): "))
        desv = float(input("  Desviación estándar muestral (s): "))
        n = int(input("  Tamaño de muestra (n): "))
        nivel = float(input("  Nivel de confianza (ej: 95): ")) / 100

        # Grados de libertad
        df = n - 1

        # Valor crítico t
        t_crit = t.ppf(1 - (1 - nivel) / 2, df)

        # Error estándar
        error_estandar = desv / np.sqrt(n)

        # Margen de error
        margen_error = t_crit * error_estandar

        li = media - margen_error
        ls = media + margen_error

        print(f"\nResultados:")
        print(f"  • Grados de libertad: {df}")
        print(f"  • Error estándar: {error_estandar:.4f}")
        print(f"  • Valor crítico t: {t_crit:.4f}")
        print(f"  • Margen de error: {margen_error:.4f}")
        print(f"  • Intervalo de confianza al {nivel*100:.0f}%: ({li:.4f}, {ls:.4f})")

        # Comparar con intervalo usando Normal
        z_crit = norm.ppf(1 - (1 - nivel) / 2, 0, 1)
        margen_error_z = z_crit * error_estandar
        li_z = media - margen_error_z
        ls_z = media + margen_error_z

        print(f"\nComparación con Normal (σ conocida):")
        print(f"  • Intervalo usando Normal: ({li_z:.4f}, {ls_z:.4f})")
        print(f"  • Diferencia en ancho: {abs((ls-li) - (ls_z-li_z)):.4f}")

        if n < 30:
            print(f"\n⚠️ Nota: Con n={n} < 30, es apropiado usar la distribución t")
        else:
            print(f"\n✅ Con n={n} ≥ 30, la distribución t y Normal dan resultados similares")

        # Visualización
        fig, ax = plt.subplots(figsize=(10, 6))

        x = np.linspace(media - 4*margen_error, media + 4*margen_error, 1000)
        y = t.pdf((x - media) / error_estandar, df) / error_estandar

        ax.plot(x, y, 'b-', linewidth=2, label=f't (ν={df})')

        # Área del intervalo de confianza
        x_fill = np.linspace(li, ls, 100)
        y_fill = t.pdf((x_fill - media) / error_estandar, df) / error_estandar
        ax.fill_between(x_fill, 0, y_fill, color='#E63946', alpha=0.5,
                        label=f'IC {nivel*100:.0f}%')

        ax.axvline(x=li, color='#E63946', linestyle='--', linewidth=2)
        ax.axvline(x=ls, color='#E63946', linestyle='--', linewidth=2)
        ax.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'x̄ = {media}')

        ax.set_xlabel('Media muestral', fontsize=12)
        ax.set_ylabel('Densidad de probabilidad', fontsize=12)
        ax.set_title(f'Intervalo de Confianza con t-Student\nn={n}, ν={df}, {nivel*100:.0f}% de confianza',
                    fontsize=13, fontweight='bold')
        ax.grid(True, alpha=0.3, linestyle='--')
        ax.legend(loc='best')

        ax.text(0.98, 0.95, f'IC {nivel*100:.0f}%\n({li:.4f}, {ls:.4f})',
               transform=ax.transAxes, fontsize=11, verticalalignment='top',
               horizontalalignment='right',
               bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_t_student():
    """Ejecuta ejemplos clásicos de la distribución t"""

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN t DE STUDENT")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Prueba t de una muestra',
            'descripcion': 'Se toma una muestra de 25 estudiantes. La media muestral es 75 y la desviación estándar es 10. ¿Cuál es el intervalo de confianza al 95% para la media poblacional?',
            'tipo': 'IC',
            'media': 75,
            'desv': 10,
            'n': 25,
            'nivel': 95
        },
        {
            'titulo': 'Ejemplo 2: Valor crítico t',
            'descripcion': 'Para una prueba t con 15 grados de libertad, ¿cuál es el valor crítico para un nivel de significancia de 0.05 (prueba de dos colas)?',
            'tipo': 'critico',
            'df': 15,
            'alfa': 0.05
        },
        {
            'titulo': 'Ejemplo 3: Probabilidad en t',
            'descripcion': 'Con 20 grados de libertad, ¿cuál es la probabilidad de que t sea mayor que 2.5?',
            'tipo': 'prob',
            'df': 20,
            'valor': 2.5
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")

        if ej['tipo'] == 'IC':
            df = ej['n'] - 1
            t_crit = t.ppf(1 - (1 - ej['nivel']/100) / 2, df)
            error = t_crit * ej['desv'] / np.sqrt(ej['n'])
            li = ej['media'] - error
            ls = ej['media'] + error
            print(f"   Intervalo de confianza al {ej['nivel']}%: ({li:.4f}, {ls:.4f})")

        elif ej['tipo'] == 'critico':
            t_crit = t.ppf(1 - ej['alfa']/2, ej['df'])
            print(f"   t crítico (α={ej['alfa']}, ν={ej['df']}) = {t_crit:.4f}")

        elif ej['tipo'] == 'prob':
            prob = 1 - t.cdf(ej['valor'], ej['df'])
            print(f"   P(t > {ej['valor']}) = {prob:.6f} ({prob*100:.2f}%)")

    print("\n" + "="*70)

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN t DE STUDENT")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado")
        print("  2. Ver ejemplos clásicos")
        print("  3. Intervalo de confianza con t")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_t_student()
        elif opcion == '2':
            ejemplos_t_student()
        elif opcion == '3':
            intervalo_confianza_t()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```

#**a. Construya el gráfico de la Distribución dada**
+ **Densidad Distribución t - Student**

$$f_X(x)=\frac{Γ(\frac{n+1}{2})}{\sqrt {n\pi }Γ(\frac{n}{2})}(1+\frac{x^2}{n})^{-\frac{n+1}{2}}$$


$$f_X(x)=q_0(1+\frac{x^2}{n})^{-\frac{n+1}{2}}$$

```python

#@title **a. Construya el gráfico de la Distribución dada si  $α=50$ y $β=49$**

# Definimos una funcion
n = 25 # parametro alpha
a = n/2
b = (n+1)/2

gamma_a = math.gamma(a)
gamma_b = math.gamma(b)
c = sqrt(n*np.pi)
q_0 = gamma_b/(c*gamma_a)

def dis_t(x):
  return q_0*(1+x**2/n)**(-b)
# los valores que va a tomar x
x = np.linspace(-5,5,1000)
fig, ax = plt.subplots() #lugar donde vamos a estar graficando, ax objeto y subplot porque vamos hacer varios graficos
ax.plot(x,dis_t(x)) # variable independiente y la funcion

plt.title('Fabrica A - Distribución t Stdent')
plt.ylabel('Probabilidad')
plt.xlabel('X :contenido faltante de las bolsas')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

```python
#@title **a. Construya el gráfico de la Distribución dada si  α=11.7 y β=7.24**
plt.figure(figsize=(8, 8))
df = 25 # parametro de forma.
t = stats.t(df)

x = np.linspace(t.ppf(0.0001),
                t.ppf(0.99999), 10000)
fp = t.pdf(x) # Función de Probabilidad
plt.plot(x, fp)

plt.title('Fabrica A - Distribución t Stdent')
plt.ylabel('Probabilidad')
plt.xlabel('X :contenido faltante de las bolsas')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **i. Para Fábrica B, calcule la probabilidad de que X sea mayor a 1.5 o menor a -1.5**

```python
#@title **i. Para Fábrica B, calcule la probabilidad de que X sea mayor a 1.5 o menor a -1.5**
fig, ax = plt.subplots(figsize=[6, 4])
x = np.linspace(-4, 4, 401)

y = stats.t(15).pdf(x)

a = -1.5 # Valor de la variable a calcular a la izquierda
b =1.5 # Valor de la variable a calcular a la izquierda
c = stats.t(15).cdf(a) # Probabilidad pedida
d = 1-stats.t(15).cdf(b)# Probabilidad pedida
e = c+d # Probabilidad pedida
x1 = np.linspace(-4, a, 401)
x2 = np.linspace(b, 4, 401)

ax.plot(x, y, c='g')

ax.fill_between(x1, 0,stats.t(15).pdf(x1) , color="r")
ax.fill_between(x2, 0,stats.t(15).pdf(x2) , color="b")


ax.text(a, 0.00001, a)
ax.text(b, 0.00001, b)

ax.text(a, 0.05, f'Prob1 = {c:.4f}')
ax.text(b, 0.05, f'Prob2 = {d:.4f}')

ax.text((a+b)/2, 0.1, f'Area solicitada = {e:.4f}')
print('la probabilidad pedida es')
print(f'P(X>={b}) +P(X<={a})=',f'Area solicitada = {e:.4f}')


plt.title('Fabrica B - Distribución t Stdent')
plt.ylabel('Probabilidad')
plt.xlabel('X :contenido faltante de las bolsas')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **j. Dada la Fábrica A, calcule el valor de k tal que $P(X>k) = 0.301$**


```python
#@title **j. Dada la Fábrica A, calcule el valor de k tal que $P(X>k) = 0.301$**
x = np.linspace(-4, 4, 401)
y = stats.t(25).pdf(x)
a = round(stats.t(25).ppf(0.699),3)
fig, ax = plt.subplots(figsize=[6, 4])
c = 1 - stats.t(25).cdf(a)
x1 = np.linspace(a, 4, 401)

ax.plot(x, y, c='g')

ax.fill_between(x1, 0,stats.t(25).pdf(x1) , color="r")

ax.text(a, 0.05, f'Área pedida ={c:.4f}', horizontalalignment='center')

ax.text(a, 0.00001, f'k = {a}')

print('El tiempo mínimo de compra del paquete es a',a,'dias')
print(f'El área a la derecha de P(X>={a})=',f'{c:.4f}')


plt.title('Fabrica A - Distribución t Stdent')
plt.ylabel('Probabilidad')
plt.xlabel('X :contenido faltante de las bolsas')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

#  **k. Dada la Fábrica A, calcule el valor de $k$  tal que $P (k < X < 1.5) = 0.42$**

```python
#@title  **k. Dada la Fábrica A, calcule el valor de $k$  tal que $P (k < X < 1.5) = 0.42$**
fig, ax = plt.subplots(figsize=[6, 4])
x = np.linspace(-5, 5, 401)

y = stats.t(25).pdf(x)
a = 1.0 # Valor de la variable a calcular a la izquierda
b = stats.t(25).cdf(a)
c = stats.t(25).cdf(a) - 0.42
d = round(stats.t(25).ppf(c),2) # valor critico pedido
e = stats.t(25).cdf(d) # Probabilidad acumulada
x1 = np.linspace(-5, d, 401)
x2 = np.linspace(d, a, 401)
x3 = np.linspace(a, 5, 401)

ax.plot(x, y, c='g')

ax.fill_between(x1, 0,stats.t(25).pdf(x1) , color="r", alpha = 0.3)
ax.fill_between(x2, 0,stats.t(25).pdf(x2) , color="b", alpha =0.3)
ax.fill_between(x3, 0,stats.t(25).pdf(x3) , color="y", alpha =0.3)

ax.text(d, 0.00001, d)
ax.text(a, 0.00001, a)

ax.text(a, 0.1, f'Area1=P(X>{a}) = {1-b:.4f}')
ax.text((a+d)/2, 0.05, f'Area2=P({d}=<X<={a})= {c:.4f}')
ax.text(d-4, 0.15, f'Area3= P(X<={d}) = {e:.4f}')


print(f'Area solicitada = {c:.4f}')

print(f'El valor critico solicitado es k ={d}')
print(f'P(X<={d})={e:.4f}')
print(f'P({d}=<X<={a})={b-e:.4f}')
print(f'P(X<={a})={b:.4f}')


plt.title('Fabrica A - Distribución t Stdent')
plt.ylabel('Probabilidad')
plt.xlabel('X :contenido faltante de las bolsas')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **65.  Distribución Birnbaum-Saunders**.
- Uso común: modelo para el número de ciclos necesarios para que una fisura crezca a un tamaño para que ocurra una fractura.

- La distribución de Birnbaum-Saunders , también conocida como distribución de vida útil a la fatiga, es una distribución de probabilidad utilizada ampliamente en aplicaciones de confiabilidad para modelar los tiempos de falla. + Hay varias formulaciones alternativas de esta distribución en la literatura. Lleva el nombre de *ZW Birnbaum* y *SC Saunders*.


- La fórmula general para la función de densidad de probabilidad (pdf) es

$${\displaystyle f(x)={\frac {{\sqrt {\frac {x-\mu }{\beta }}}+{\sqrt {\frac {\beta}{x-\mu }}}} {2\gamma \left(x-\mu \right)}}\phi \left({\frac {{\sqrt {\frac {x-\mu }{\beta }}}-{\sqrt {\frac {\beta }{x-\mu }}}}{\gamma }}\right)\quad x>\mu ;\gamma ,\beta >0}$$

- La forma más habitual de esta distribución es:

$${\displaystyle F(x;\alpha ,\beta )=\Phi \left({\frac {1}{\alpha }}\left[\left({\frac {x}{\beta }}\right) ^{0.5}-\left({\frac {\beta}{x}}\right)^{0.5}\right]\right)}$$
- Aquí $\alpha$ es el parámetro de forma y $\beta$ es el parámetro de escala. $\phi$ es la función de distribución acumulativa de la distribución normal.


- La media $\mu$, la varianza $\sigma^2$, la asimetría $\gamma$ y la curtosis $\kappa$ son las siguientes:

$${\displaystyle \mu =\beta \left(1+{\frac {\alpha ^{2}}{2}}\right)}$$
 $${\displaystyle \sigma ^{2}=(\alpha \beta )^{2}\left(1+{\frac {5\alpha ^{2}}{4}}\right)}$$

$${\displaystyle \gamma ={\frac {4\alpha (11\alpha ^{2}+6)}{(5\alpha ^{2}+4)^{\frac {3}{2}}}} }$$

$${\displaystyle \kappa =3+{\frac {6\alpha ^{2}(93\alpha ^{2}+40)}{(5\alpha ^{2}+4)^{2}}}}$$

# **66.  Distribución Cauchy.**
- Uso común: modelo para datos de mediciones con colas más largas y aplanadas que la distribución normal.
- La distribución Cauchy-Lorentz, llamada en honor a Augustin Cauchy y Hendrik Lorentz, es una distribución de probabilidad continua. Es conocida como la distribución de Cauchy y en el ámbito de la física se conoce como la distribución de Lorentz, la función Lorentziana o la distribución de Breit-Wigner. Su importancia en la física es dada por ser la solución de la ecuación diferencial que describe la resonancia forzada. En espectroscopia describe la forma de las líneas espectrales que son ampliadas por diversos mecanismos, en particular, el mecanismo de ensanchamiento por colisión


- *Función de densidad*. En estadística la distribución de Cauchy (a veces también distribución de Lorentz) es una distribución de probabilidad continua cuya función de densidad es

$${\displaystyle {\begin{aligned}f(x;x_{0},\gamma )&={\frac {1}{\pi \gamma \left[1+\left({\frac {x-x_{0}}{\gamma }}\right)^{2}\right]}}\\[0.5em]&={1 \over \pi }\left[{\gamma \over (x-x_{0})^{2}+\gamma ^{2}}\right]\end{aligned}}}$$


- Donde $x_0$ es el parámetro de corrimiento que específica la ubicación del pico de la distribución, y $\gamma$ es el parámetro de escala que específica el ancho medio al máximo medio (half-width at half-maximum, HWHM).

- En el caso especial donde $x_0 = 0$ y $\gamma = 1$ es denominado la distribución estándar Cauchy con la función de densidad de probabilidad

$${\displaystyle f(x;0,1)={\frac {1}{\pi (1+x^{2})}}.\!}$$
- La distribución de Cauchy es un ejemplo de una distribución que no tiene valor esperado, varianza o momentos definidos. Su moda y su mediana están bien definidas y son ambas iguales a $x_0$.

# **97. La distribución Burr**
+ (a veces llamada distribución Burr Tipo XII o distribución Singh-Maddala) es una familia unimodal de distribuciones con una amplia variedad de formas.
+ Esta distribución se utiliza para modelar una amplia variedad de fenómenos, incluidos los precios de los cultivos, los ingresos familiares, las distribuciones de precios del mercado de opciones, el riesgo (seguro) y el tiempo de viaje.
+ Es particularmente útil para modelar histogramas . Aunque existen otras formas (el tipo III también es muy común), el término “distribución de rebabas” suele referirse al tipo XII.
+ La distribución de rebabas Tipo XII se define por los siguientes parámetros:
+ + c y k: parámetros de forma . Para Burr Tipo XII, ambos son positivos. El tipo III tiene un parámetro c negativo.
+ + $\alpha$: parámetro de escala .
+ + $\gamma$: parámetro de ubicación continua .
+ Cuando el cuarto parámetro, $\gamma$, es igual a cero, da una distribución de tres parámetros ($c$,$k$,$\alpha$).

# **100. Distribuciones continuas - La Distribución $F$**
# **a. Generalidades de la Distribución $F$**
+ **Wikipedia**. En teoría de probabilidad y estadística, la distribución F, también conocida como distribución de Fisher-Snedecor (nombrada por Ronald Fisher y George Snedecor), es una distribución de probabilidad continua, aparece frecuentemente como la distribución nula de una prueba estadística, especialmente en el análisis de varianza.
+ La distribución F es una distribución continua de muestreo de la relación de dos variables aleatorias independientes con distribuciones de chi-cuadrada, cada una dividida entre sus grados de libertad. La distribución F es asimétrica hacia la derecha y es descrita por los grados de libertad de su numerador (ν1) y denominador (ν2). Las siguientes gráficas muestran el efecto de los diferentes valores de grados de libertad en la forma de la distribución



# **b. La Distribución F - Definición**

+ Una variable aleatoria continua $X$ sigue una distribución *F* de parámetros $m$ y $n$, lo cual denotamos por $$X∼F_{(m,n)}$$ si tiene funcion FDP dada por la formula:
$$ f(x,m,n) = \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}}\frac{x^{\frac{m-2}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}, \ x>0$$


+ Tambien $$E(X)=\mu=\frac{n}{n-2}  \ \ \ \ V(X)=σ^2=\frac{2n^2(m+n-2)}{m(n-2)^2(n-4)}$$



# **c. Veamos que es una función de densidad**

+ **Densidad Distribución F**
$$ f(x,m,n) = \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}}\frac{x^{\frac{m-2}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}, \ x>0$$

+ **Integrando**.
$$ ∫_{-∞}^{∞}f(x)dx= ∫_{0}^{∞}\frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}}\frac{x^{\frac{m-2}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}dx$$


$$ ∫_{-∞}^{∞}f(x)dx= \ \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}} ∫_{0}^{∞}\frac{x^{\frac{m-2}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}dx$$

***
  
+ **hacemos** $$y = (1+\frac{mx}{n})^{-1}$$ esto es $$x =\frac{n}{m}y^{-1}(1-y) ⇒dx=-\frac{n}{m}y^{-2}dy$$
+
$$x^{\frac{m-2}{2}} =(\frac{n}{m}y^{-1}(1-y))^{\frac{m-2}{2}}=(\frac{n}{m})^{\frac{m-2}{2}}(y^{-{\frac{m-2}{2}}})(1-y)^{\frac{m-2}{2}}$$
***
+ **Sustituyendo en la Integral**.

$$ ∫_{-∞}^{∞}f(x)dx= - \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}} ∫_{1}^{0}\frac{(\frac{n}{m})^{\frac{m-2}{2}}(y^{-{\frac{m-2}{2}}})(1-y)^{\frac{m-2}{2}}}{y^{-\frac{m+n}{2}}}\frac{n}{m}y^{-2}dy$$
***
+ **Reorganizando la Integral**.

$$ ∫_{-∞}^{∞}f(x)dx=  \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}∫_{0}^{1}\frac{(y^{-{\frac{m-2}{2}}})(1-y)^{\frac{m-2}{2}}}{y^{-\frac{m+n}{2}}}y^{-2}dy$$

$$ ∫_{-∞}^{∞}f(x)dx=  \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}∫_{0}^{1}y^{{\frac{n}{2}}-1}(1-y)^{{\frac{m}{2}}-1}dy$$


+ **Recordando la función Beta**

 $$
{B(α,β)} = \int_0^{1} x^{α-1}(1-x)^{β-1}dx=\frac{\Gamma(α)\Gamma(β)}{\Gamma(α+β)}
$$

+ **Reemplazando en la  Integral**.
$$ ∫_{-∞}^{∞}f(x)dx=  \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}∫_{0}^{1}y^{{\frac{n}{2}}-1}(1-y)^{{\frac{m}{2}}-1}dy=\frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}\frac{Γ(\frac{m}{2})Γ(\frac{n}{2})}{Γ(\frac{m+n}{2})}=1$$




# **c. Valor Esperado**

+ **Densidad Distribución F**
$$ f(x,m,n) = \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}}\frac{x^{\frac{m-2}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}, \ x>0$$

+ **Integrando**.
$$ E(X)=∫_{-∞}^{∞}xf(x)dx= ∫_{0}^{∞}\frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}}\frac{x^{\frac{m-2}{2}+1}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}dx$$


$$ E(X)=∫_{-∞}^{∞}xf(x)dx= \ \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}} ∫_{0}^{∞}\frac{x^{\frac{m}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}dx$$

***
  
+ **hacemos** $$y = (1+\frac{mx}{n})^{-1}$$ esto es $$x =\frac{n}{m}y^{-1}(1-y) ⇒dx=-\frac{n}{m}y^{-2}dy$$
+
$$x^{\frac{m}{2}} =(\frac{n}{m}y^{-1}(1-y))^{\frac{m}{2}}=(\frac{n}{m})^{\frac{m}{2}}(y^{-{\frac{m}{2}}})(1-y)^{\frac{m}{2}}$$
***
+ **Sustituyendo en la Integral**.

$$ ∫_{-∞}^{∞}xf(x)dx= - \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}} ∫_{1}^{0}\frac{(\frac{n}{m})^{\frac{m}{2}}(y^{-{\frac{m}{2}}})(1-y)^{\frac{m}{2}}}{y^{-\frac{m+n}{2}}}\frac{n}{m}y^{-2}dy$$
***
+ **Reorganizando la Integral**.

$$ ∫_{-∞}^{∞}xf(x)dx=  \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}\frac{n}{m}∫_{0}^{1}\frac{(y^{-{\frac{m}{2}}})(1-y)^{\frac{m}{2}}}{y^{-\frac{m+n}{2}}}y^{-2}dy$$

$$ ∫_{-∞}^{∞}xf(x)dx=  \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}\frac{n}{m}∫_{0}^{1}y^{({\frac{n}{2}}-1)-1}(1-y)^{({\frac{m}{2}}+1)-1}dy$$


+ **Recordando la función Beta**

 $$
{B(α,β)} = \int_0^{1} x^{α-1}(1-x)^{β-1}dx=\frac{\Gamma(α)\Gamma(β)}{\Gamma(α+β)}
$$

+ **Reemplazando en la  Integral**.
$$ ∫_{-∞}^{∞}xf(x)dx=  \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}\frac{n}{m}\frac{Γ(\frac{n}{2}-1)Γ(\frac{m}{2}+1)}{Γ(\frac{m+n}{2})}$$

+ **Sabemos que**

$$Γ(\frac{m}{2}+1)=\frac{m}{2}Γ(\frac{m}{2})$$
$$Γ(\frac{n}{2})=Γ((\frac{n}{2}-1)+1)=(\frac{n}{2}-1)Γ(\frac{n}{2}-1)$$

+ **Reemplazando y simplificando la integral arriba**

$$ E(X)=\mu=∫_{-∞}^{∞}xf(x)dx=  \frac{n}{n-2}$$




# **c. Varianza**

+ **Densidad Distribución F**
$$ f(x,m,n) = \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}}\frac{x^{\frac{m-2}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}, \ x>0$$

+ **Integrando**.
$$ E(X^2)=∫_{-∞}^{∞}x^2f(x)dx= ∫_{0}^{∞}\frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}}\frac{x^{\frac{m-2}{2}+2}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}dx$$


$$ E(X^2)=∫_{-∞}^{∞}xf(x)dx= \ \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}} ∫_{0}^{∞}\frac{x^{\frac{m}{2}+1}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}dx$$

***
  
+ **hacemos** $$y = (1+\frac{mx}{n})^{-1}$$ esto es $$x =\frac{n}{m}y^{-1}(1-y) ⇒dx=-\frac{n}{m}y^{-2}dy$$
+
$$x^{\frac{m}{2}+1} =(\frac{n}{m}y^{-1}(1-y))^{\frac{m}{2}+1}=(\frac{n}{m})^{\frac{m}{2}+1}(y^{-{\frac{m}{2}}-1})(1-y)^{\frac{m}{2}+1}$$
***
+ **Sustituyendo en la Integral**.

$$ ∫_{-∞}^{∞}x^2f(x)dx= - \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}} ∫_{1}^{0}\frac{(\frac{n}{m})^{\frac{m}{2}+1}(y^{-{\frac{m}{2}}-1})(1-y)^{\frac{m}{2}+1}}{y^{-\frac{m+n}{2}}}\frac{n}{m}y^{-2}dy$$
***
+ **Reorganizando la Integral**.

$$ ∫_{-∞}^{∞}x^2f(x)dx=  \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{n}{m})^2∫_{0}^{1}y^{({\frac{n}{2}}-2)-1}(1-y)^{({\frac{m}{2}}+2)-1}$$


+ **Recordando la función Beta**

 $$
{B(α,β)} = \int_0^{1} x^{α-1}(1-x)^{β-1}dx=\frac{\Gamma(α)\Gamma(β)}{\Gamma(α+β)}
$$

+ **Reemplazando en la  Integral**.
$$ ∫_{-∞}^{∞}x^2f(x)dx=  \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{n}{m})^2\frac{Γ(\frac{n}{2}-2)Γ(\frac{m}{2}+2)}{Γ(\frac{m+n}{2})}$$

+ **Sabemos que**
$$Γ(α)=(α-1)Γ(α-1)=(α-1)(α-2)Γ(α-2)$$
+ **Cambiando $α=\frac{m}{2}+2$**
$$Γ(\frac{m}{2}+2)=(\frac{m}{2}+1)\frac{m}{2}Γ(\frac{m}{2})$$
+ **Cambiando $α=\frac{2}{2}$**
$$Γ(\frac{n}{2})=(\frac{n}{2}-1)(\frac{n}{2}-2)Γ(\frac{n}{2}-2)$$

+ **Reemplazando y simplificando la integral arriba**

$$ E(X^2)=∫_{-∞}^{∞}x^2f(x)dx= (\frac{n}{m})^2 \frac{(m+2)m}{(n-2)(n-4)}$$
+ **Por lo tanto**
$$Var(X)=σ^2=\frac{n^2}{m} \frac{2(m+n-2)}{(n-2)^2(n-4)}$$

#**a. Construya el gráfico de la Distribución dada**
+ **Densidad Distribución t - Student**


+ **Densidad Distribución F**
$$ f(x,m,n) = \frac{Γ(\frac{m+n}{2})}{Γ(\frac{m}{2})Γ(\frac{n}{2})}(\frac{m}{n})^{\frac{m}{2}}\frac{x^{\frac{m-2}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}, \ x>0$$


$$f_X(x)=q_0\frac{x^{\frac{m-2}{2}}}{(1+\frac{mx}{n})^{\frac{m+n}{2}}}$$

```python

#@title **a. Construya el gráfico de la Distribución dada**

# Definimos una funcion
m = 20 # parametro alpha
n = 25

a = ((m/n)**(m/2))*math.gamma((m+n)/2)
b = math.gamma(n/2)*math.gamma(m/2)
c = (m+n)/2
d = (m-2)/2

q_0 = a/b

def dis_F(x):
  return q_0*(x**d)/(((1+m*x/n)**c))
# los valores que va a tomar x
x = np.linspace(0,8,1000)
fig, ax = plt.subplots() #lugar donde vamos a estar graficando, ax objeto y subplot porque vamos hacer varios graficos
ax.plot(x,dis_F(x)) # variable independiente y la funcion

plt.title('Distribución F Snedecor')
plt.ylabel('Probabilidad')
plt.xlabel('X :Tasa de variabilidad entre Fabricas')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

```python
#@title **a. Construya el gráfico de la Distribución dada si  α=11.7 y β=7.24**
plt.figure(figsize=(8, 8))
m = 20 # parametro alpha
n = 25
df = 25 # parametro de forma.
F = stats.f(20,25)

x = np.linspace(F.ppf(0.0001),
                F.ppf(0.99999), 10000)
fp = F.pdf(x) # Función de Probabilidad
plt.plot(x, fp)


plt.title('Distribución F Snedecor')
plt.ylabel('Probabilidad')
plt.xlabel('X :Tasa de variabilidad entre Fabricas')
ax.grid() # le da mas elegancia al grafico
plt.show()
```

# **i. Suponga que la tasa de variabilidad sigue una distribució F con $(5,8)$ como parametros, calcule la probabilidad que la variabilidad en el contenido sea mayor a 1.5 o menor a 0.5**

#  **k. Suponga que la tasa de variabilidad sigue una distribució F con $(5,8)$ como parametros, calcule el valor de $k$  tal que $P (k < X < 2.55) = 0.42$**

```python
#@title  **k. Continuacion del Ejemplo**
fig, ax = plt.subplots(figsize=[6, 4])
x = np.linspace(0, 8, 401)
F = stats.f(5,8)

y = stats.f(5,8).pdf(x)
a = 2.55 # Valor de la variable a calcular a la izquierda
b = F.cdf(a)
c = F.cdf(a) - 0.42
d = round(F.ppf(c),2) # valor critico pedido
e = F.cdf(d) # Probabilidad acumulada
x1 = np.linspace(0, d, 401)
x2 = np.linspace(d, a, 401)
x3 = np.linspace(a, 8, 401)

ax.plot(x, y, c='g')

ax.fill_between(x1, 0,F.pdf(x1) , color="r", alpha = 0.3)
ax.fill_between(x2, 0,F.pdf(x2) , color="b", alpha =0.3)
ax.fill_between(x3, 0,F.pdf(x3) , color="y", alpha =0.3)

ax.text(d, 0.00001, d)
ax.text(a, 0.00001, a)

ax.text(a, 0.2, f'P(X>{a}) = {1-b:.4f}')
ax.text((a+d)/2, 0.1, f'P({d}=<X<={a})= {b-e:.4f}')
ax.text(d, 0.15, f'P(X<={d}) = {e:.4f}')

plt.title('Distribución F Snedecor')
plt.ylabel('Frecuencias')
plt.xlabel('X :Tasa de variabilidad entre Fabricas')
ax.grid() # le da mas elegancia al grafico
plt.show()

print(f'Area solicitada = {c:.4f}')

print(f'El valor critico solicitado es k ={d}')
print(f'P(X<={d})={e:.4f}')
print(f'P({d}=<X<={a})={b-e:.4f}')
print(f'P(X<={a})={b:.4f}')

```

# **102. CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN F DE SNEDECOR**

```python
# ==========================================================
# @title **CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN F DE SNEDECOR**
#
# La distribución F se utiliza principalmente para:
#   • Comparación de varianzas (prueba F)
#   • Análisis de varianza (ANOVA)
#   • Pruebas de significancia en regresión
#   • Comparación de dos muestras
#
# Parámetros:
#   d1 = grados de libertad del numerador
#   d2 = grados de libertad del denominador
#
# Fórmula: f(x) = Γ((d1+d2)/2) * (d1/d2)^(d1/2) * x^(d1/2-1) /
#           (Γ(d1/2) * Γ(d2/2) * (1 + d1*x/d2)^((d1+d2)/2))
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f, chi2, norm
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

#==========================================================
# Configuración de estilo
#==========================================================
try:
    plt.style.use('seaborn-v0_8')
except:
    try:
        plt.style.use('seaborn')
    except:
        plt.style.use('default')

plt.rcParams['figure.figsize'] = (12, 7)
plt.rcParams['font.size'] = 11

#==========================================================
# FUNCIÓN PRINCIPAL DE LA CALCULADORA
#==========================================================

def calculadora_f():
    """
    Calculadora interactiva para la distribución F de Snedecor
    """

    print("\n" + "="*70)
    print("📊 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN F DE SNEDECOR")
    print("="*70)
    print("\nLa distribución F se utiliza para:")
    print("  • Comparar dos varianzas poblacionales")
    print("  • Análisis de varianza (ANOVA)")
    print("  • Pruebas de significancia en modelos de regresión")
    print("  • Pruebas de igualdad de medias\n")

    #==========================================================
    # 1. ENTRADA DE PARÁMETROS
    #==========================================================

    while True:
        try:
            print("-"*70)
            print("INGRESA LOS PARÁMETROS:")
            print("-"*70)

            d1 = float(input("  Grados de libertad del numerador (d1 > 0): "))
            if d1 <= 0:
                print("  ❌ d1 debe ser mayor que 0. Intenta de nuevo.")
                continue

            d2 = float(input("  Grados de libertad del denominador (d2 > 0): "))
            if d2 <= 0:
                print("  ❌ d2 debe ser mayor que 0. Intenta de nuevo.")
                continue

            print("\nTipo de cálculo:")
            print("  1. P(X = a)  - Probabilidad puntual")
            print("  2. P(X ≤ a)  - Probabilidad acumulada a la izquierda")
            print("  3. P(X ≥ a)  - Probabilidad acumulada a la derecha")
            print("  4. P(a ≤ X ≤ b) - Probabilidad entre dos valores")
            print("  5. Cuantil - Encontrar X dado un percentil")
            print("  6. Comparar con diferentes grados de libertad")

            tipo = input("\nSelecciona el tipo de cálculo (1-6): ")

            if tipo == '1':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 0  # En distribuciones continuas, P(X=a)=0
                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X = {a}) = 0")
                print("En distribuciones continuas, la probabilidad puntual es siempre 0.")
                print(f"Densidad en X={a}: {f.pdf(a, d1, d2):.6f}")

                visualizar_f(d1, d2, a, None, prob, tipo)
                return {'d1': d1, 'd2': d2, 'a': a}

            elif tipo == '2':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = f.cdf(a, d1, d2)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≤ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con d1={d1:.1f}, d2={d2:.1f} grados de libertad")

                visualizar_f(d1, d2, a, None, prob, tipo)
                return {'d1': d1, 'd2': d2, 'a': a, 'prob': prob}

            elif tipo == '3':
                a = float(input("  Valor de a (≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                prob = 1 - f.cdf(a, d1, d2)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP(X ≥ {a}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con d1={d1:.1f}, d2={d2:.1f} grados de libertad")

                visualizar_f(d1, d2, a, None, prob, tipo)
                return {'d1': d1, 'd2': d2, 'a': a, 'prob': prob}

            elif tipo == '4':
                a = float(input("  Límite inferior (a ≥ 0): "))
                if a < 0:
                    print("  ❌ a debe ser ≥ 0. Intenta de nuevo.")
                    continue
                b = float(input("  Límite superior (b > a): "))
                if b <= a:
                    print("  ❌ b debe ser mayor que a. Intenta de nuevo.")
                    continue
                prob = f.cdf(b, d1, d2) - f.cdf(a, d1, d2)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nP({a} ≤ X ≤ {b}) = {prob:.6f} ({prob*100:.4f}%)")
                print(f"Con d1={d1:.1f}, d2={d2:.1f} grados de libertad")

                visualizar_f(d1, d2, a, b, prob, tipo)
                return {'d1': d1, 'd2': d2, 'a': a, 'b': b, 'prob': prob}

            elif tipo == '5':
                p = float(input("  Percentil (0-1): "))
                if p <= 0 or p >= 1:
                    print("  ❌ El percentil debe estar entre 0 y 1. Intenta de nuevo.")
                    continue
                x = f.ppf(p, d1, d2)

                print("\n" + "="*70)
                print("📈 RESULTADO")
                print("="*70)
                print(f"\nPara el percentil {p*100:.1f}%:")
                print(f"X = {x:.6f}")
                print(f"Con d1={d1:.1f}, d2={d2:.1f} grados de libertad")
                print(f"Interpretación: El {p*100:.1f}% de los datos están por debajo de {x:.4f}")

                visualizar_cuantil_f(d1, d2, x, p)
                return {'d1': d1, 'd2': d2, 'percentil': p, 'x': x}

            elif tipo == '6':
                comparar_df_f(d1, d2)
                return {'d1': d1, 'd2': d2}

            else:
                print("  ❌ Opción no válida. Intenta de nuevo.")
                continue

            break

        except ValueError:
            print("  ❌ Error: Ingresa valores numéricos válidos.")
            continue

#==========================================================
# FUNCIÓN PARA VISUALIZAR LA DISTRIBUCIÓN F
#==========================================================

def visualizar_f(d1, d2, a, b, prob, tipo):
    """Crea visualizaciones completas de la distribución F"""

    # Rango de valores para la gráfica
    x_max = max(f.ppf(0.999, d1, d2), a * 1.5 if a > 0 else f.ppf(0.999, d1, d2))
    x_max = max(x_max, 5)
    x = np.linspace(0.001, x_max, 1000)
    y = f.pdf(x, d1, d2)

    # Crear figura con subplots
    fig = plt.figure(figsize=(15, 10))

    #==========================================================
    # GRÁFICO 1: Distribución con área sombreada
    #==========================================================
    ax1 = plt.subplot(2, 2, 1)
    ax1.plot(x, y, 'b-', linewidth=2.5, label=f'F({d1:.1f}, {d2:.1f})')

    # Sombrear el área según el tipo de cálculo
    if tipo == '2':  # P(X ≤ a)
        mask = x <= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≤ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '3':  # P(X ≥ a)
        mask = x >= a
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P(X ≥ {a:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, label=f'X = {a:.2f}')

    elif tipo == '4':  # P(a ≤ X ≤ b)
        mask = (x >= a) & (x <= b)
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.5, label=f'P({a:.2f} ≤ X ≤ {b:.2f})')
        ax1.axvline(x=a, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)
        ax1.axvline(x=b, color='#E63946', linestyle='--', linewidth=2, alpha=0.7)

    # Estadísticas
    if d2 > 2:
        media = d2 / (d2 - 2)
        ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.2f}')

    if d2 > 4:
        varianza = (2 * d2**2 * (d1 + d2 - 2)) / (d1 * (d2 - 2)**2 * (d2 - 4))
        desv = np.sqrt(varianza)
        ax1.axvline(x=media + desv, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)
        ax1.axvline(x=media - desv, color='#2C3E50', linestyle=':', linewidth=1.5, alpha=0.5)

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Distribución F (d1={d1:.1f}, d2={d2:.1f})', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best', fontsize=10)
    ax1.set_xlim(0, x_max)

    # Agregar información de probabilidad
    if prob is not None:
        ax1.text(0.98, 0.95, f'Probabilidad = {prob:.6f}\n({prob*100:.2f}%)',
                transform=ax1.transAxes, fontsize=11, verticalalignment='top',
                horizontalalignment='right',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    #==========================================================
    # GRÁFICO 2: Comparación con diferentes grados de libertad
    #==========================================================
    ax2 = plt.subplot(2, 2, 2)

    # Fijar d2 y variar d1
    d1_vals = [1, 2, 5, 10, 20]
    colores = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    x2 = np.linspace(0.001, 5, 1000)
    for d1_val, color in zip(d1_vals, colores):
        y_temp = f.pdf(x2, d1_val, d2)
        ax2.plot(x2, y_temp, color=color, linewidth=2,
                label=f'd1 = {d1_val}')

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax2.set_title(f'Efecto de d1 (d2={d2:.1f} fijo)', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best', fontsize=9)
    ax2.set_xlim(0, 5)

    #==========================================================
    # GRÁFICO 3: Función de distribución acumulada
    #==========================================================
    ax3 = plt.subplot(2, 2, 3)
    cdf = f.cdf(x, d1, d2)
    ax3.plot(x, cdf, 'g-', linewidth=2.5, label='CDF')
    ax3.fill_between(x, 0, cdf, alpha=0.2, color='green')

    # Puntos de interés
    if tipo == '2':  # P(X ≤ a)
        ax3.scatter(a, prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {prob:.4f}')
        ax3.axhline(y=prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    elif tipo == '3':  # P(X ≥ a)
        ax3.scatter(a, 1-prob, color='#E63946', s=100, zorder=5, label=f'F({a:.2f}) = {1-prob:.4f}')
        ax3.axhline(y=1-prob, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
        ax3.axvline(x=a, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    # Cuantiles comunes
    for q in [0.25, 0.5, 0.75, 0.9, 0.95]:
        q_val = f.ppf(q, d1, d2)
        if q_val <= x_max:
            ax3.axvline(x=q_val, color='orange', linestyle=':', linewidth=1.5, alpha=0.5)
            ax3.text(q_val, 0.01, f'Q{int(q*100)}', ha='center', fontsize=8, color='orange')

    ax3.set_xlabel('X', fontsize=12)
    ax3.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax3.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax3.grid(True, alpha=0.3, linestyle='--')
    ax3.legend(loc='best', fontsize=10)
    ax3.set_ylim(-0.05, 1.05)
    ax3.set_xlim(0, x_max)

    #==========================================================
    # GRÁFICO 4: Tabla de valores críticos F
    #==========================================================
    ax4 = plt.subplot(2, 2, 4)
    ax4.axis('off')

    # Crear tabla de valores críticos comunes
    niveles = [0.90, 0.95, 0.975, 0.99, 0.995]
    valores_f = [f.ppf(nivel, d1, d2) for nivel in niveles]

    table_data = []
    table_data.append(['Nivel', 'F crítico', 'Interpretación'])
    for nivel, f_val in zip(niveles, valores_f):
        if f_val < x_max * 1.5 and f_val > 0:
            table_data.append([
                f'{nivel*100:.1f}%',
                f'{f_val:.4f}',
                f'P(X ≤ {f_val:.2f}) = {nivel:.3f}'
            ])

    # Si hay demasiados datos, mostrar solo los más relevantes
    if len(table_data) > 8:
        table_data = table_data[:1] + table_data[1::2]

    table = ax4.table(cellText=table_data, loc='center', cellLoc='center',
                      colWidths=[0.25, 0.3, 0.4])

    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.8)

    # Estilo de la tabla
    for i in range(len(table_data)):
        for j in range(len(table_data[0])):
            cell = table[(i, j)]
            if i == 0:
                cell.set_facecolor('#2C3E50')
                cell.set_text_props(weight='bold', color='white')
            else:
                cell.set_facecolor('#F5F5F5')

    ax4.set_title(f'Valores Críticos F\nd1={d1:.1f}, d2={d2:.1f}',
                  fontsize=13, fontweight='bold', pad=20)

    #==========================================================
    # Ajustar layout
    #==========================================================
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.3, wspace=0.3)

    #==========================================================
    # GRÁFICO ADICIONAL: Efecto de d2
    #==========================================================
    fig2, (ax5, ax6) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 5: Efecto de d2 (fijando d1)
    d2_vals = [1, 2, 5, 10, 30]
    colores2 = ['#E63946', '#F39C12', '#2ECC71', '#3498DB', '#2C3E50']

    x3 = np.linspace(0.001, 5, 1000)
    for d2_val, color in zip(d2_vals, colores2):
        y_temp = f.pdf(x3, d1, d2_val)
        ax5.plot(x3, y_temp, color=color, linewidth=2,
                label=f'd2 = {d2_val}')

    ax5.set_xlabel('X', fontsize=12)
    ax5.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax5.set_title(f'Efecto de d2 (d1={d1:.1f} fijo)', fontsize=13, fontweight='bold')
    ax5.grid(True, alpha=0.3, linestyle='--')
    ax5.legend(loc='best', fontsize=9)
    ax5.set_xlim(0, 5)

    # Gráfico 6: Relación con distribución Chi-Cuadrado
    ax6.plot(x, y, 'b-', linewidth=2.5, label=f'F({d1:.1f}, {d2:.1f})')

    # Si d1 o d2 son grandes, mostrar aproximación
    if d2 > 30 and d1 > 10:
        # Aproximación a Chi-Cuadrado
        chi2_df = d1
        y_chi2 = chi2.pdf(x * d1 / d2, chi2_df) * d1 / d2
        ax6.plot(x, y_chi2, 'r--', linewidth=2, label='Aprox. χ²')

    ax6.set_xlabel('X', fontsize=12)
    ax6.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax6.set_title('Comparación con otras distribuciones', fontsize=13, fontweight='bold')
    ax6.grid(True, alpha=0.3, linestyle='--')
    ax6.legend(loc='best')
    ax6.set_xlim(0, min(x_max, 10))

    plt.tight_layout()

    #==========================================================
    # MOSTRAR GRÁFICOS
    #==========================================================
    print("\n" + "="*70)
    print("📊 GENERANDO VISUALIZACIONES...")
    print("="*70)

    plt.show()

    #==========================================================
    # OPCIÓN DE GUARDAR GRÁFICOS
    #==========================================================

    print("\n" + "="*70)
    print("💾 OPCIÓN DE GUARDADO")
    print("="*70)
    guardar = input("¿Deseas guardar los gráficos? (s/n): ").lower()

    if guardar == 's':
        nombre_base = f"F_d1{d1:.1f}_d2{d2:.1f}"
        fig.savefig(f"{nombre_base}_distribucion.png", dpi=300, bbox_inches='tight')
        fig2.savefig(f"{nombre_base}_propiedades.png", dpi=300, bbox_inches='tight')
        print(f"✅ Gráficos guardados como:")
        print(f"   • {nombre_base}_distribucion.png")
        print(f"   • {nombre_base}_propiedades.png")

    print("\n" + "="*70)
    print("✅ ANÁLISIS COMPLETADO")
    print("="*70)

def visualizar_cuantil_f(d1, d2, x, p):
    """Visualiza el cuantil en la distribución F"""

    x_max = max(f.ppf(0.999, d1, d2), x * 1.5)
    x_vals = np.linspace(0.001, x_max, 1000)
    y = f.pdf(x_vals, d1, d2)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfico 1: Distribución con cuantil marcado
    ax1.plot(x_vals, y, 'b-', linewidth=2.5)
    mask = x_vals <= x
    ax1.fill_between(x_vals[mask], 0, y[mask], color='#E63946', alpha=0.5,
                     label=f'Área = {p*100:.1f}%')
    ax1.axvline(x=x, color='#E63946', linestyle='--', linewidth=2, label=f'X = {x:.2f}')

    # Media si existe
    if d2 > 2:
        media = d2 / (d2 - 2)
        ax1.axvline(x=media, color='#2C3E50', linestyle=':', linewidth=2, label=f'μ = {media:.2f}')

    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
    ax1.set_title(f'Percentil {p*100:.1f}% (d1={d1:.1f}, d2={d2:.1f})',
                  fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3, linestyle='--')
    ax1.legend(loc='best')
    ax1.set_xlim(0, x_max)

    # Gráfico 2: Función de distribución acumulada
    cdf = f.cdf(x_vals, d1, d2)
    ax2.plot(x_vals, cdf, 'g-', linewidth=2.5)
    ax2.scatter(x, p, color='#E63946', s=150, zorder=5, label=f'F({x:.2f}) = {p:.4f}')
    ax2.axhline(y=p, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)
    ax2.axvline(x=x, color='#E63946', linestyle='--', linewidth=1.5, alpha=0.5)

    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Probabilidad acumulada F(X)', fontsize=12)
    ax2.set_title('Función de Distribución Acumulada', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3, linestyle='--')
    ax2.legend(loc='best')
    ax2.set_ylim(-0.05, 1.05)
    ax2.set_xlim(0, x_max)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA COMPARAR DIFERENTES GRADOS DE LIBERTAD
#==========================================================

def comparar_df_f(d1_actual, d2_actual):
    """Compara la distribución actual con diferentes grados de libertad"""

    print("\n" + "="*70)
    print("📊 COMPARACIÓN DE GRADOS DE LIBERTAD")
    print("="*70)

    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # Combinaciones de d1 y d2 para comparar
    combinaciones = [
        (1, 1), (1, 5), (1, 10), (1, 30),
        (5, 1), (5, 5), (5, 10), (5, 30),
        (10, 1), (10, 5), (10, 10), (10, 30),
        (30, 1), (30, 5), (30, 10), (30, 30)
    ]

    # Seleccionar algunas combinaciones relevantes
    selected = [(1, 1), (1, 5), (5, 5), (5, 30), (10, 10), (30, 30)]

    x = np.linspace(0.001, 5, 1000)

    for idx, (d1, d2) in enumerate(selected):
        ax = axes[idx // 2, idx % 2]
        y = f.pdf(x, d1, d2)

        ax.plot(x, y, linewidth=2.5,
               label=f'F({d1}, {d2})',
               color=plt.cm.viridis(idx / len(selected)))

        # Resaltar la combinación actual si coincide
        if d1 == d1_actual and d2 == d2_actual:
            ax.fill_between(x, 0, y, alpha=0.2, color='red')
            ax.set_title(f'F({d1}, {d2}) ★ ACTUAL', fontsize=12, fontweight='bold')
        else:
            ax.set_title(f'F({d1}, {d2})', fontsize=12)

        ax.set_xlabel('X', fontsize=10)
        ax.set_ylabel('Densidad', fontsize=10)
        ax.grid(True, alpha=0.3)
        ax.legend(loc='best')
        ax.set_xlim(0, 5)
        ax.set_ylim(0, 1.5)

    plt.tight_layout()
    plt.show()

#==========================================================
# FUNCIÓN PARA PRUEBA F (COMPARACIÓN DE VARIANZAS)
#==========================================================

def prueba_f():
    """Realiza una prueba F para comparar dos varianzas"""

    print("\n" + "="*70)
    print("📊 PRUEBA F - COMPARACIÓN DE DOS VARIANZAS")
    print("="*70)
    print("\nEsta función realiza una prueba F para comparar")
    print("si dos varianzas poblacionales son iguales.\n")

    try:
        print("Ingresa los datos:")
        print("\nMuestra 1:")
        s1 = float(input("  Varianza muestral 1 (s₁²): "))
        n1 = int(input("  Tamaño de muestra 1 (n₁): "))

        print("\nMuestra 2:")
        s2 = float(input("  Varianza muestral 2 (s₂²): "))
        n2 = int(input("  Tamaño de muestra 2 (n₂): "))

        # Calcular estadístico F
        if s1 >= s2:
            f_stat = s1 / s2
            df1 = n1 - 1
            df2 = n2 - 1
            muestra_mayor = 1
        else:
            f_stat = s2 / s1
            df1 = n2 - 1
            df2 = n1 - 1
            muestra_mayor = 2

        # Calcular p-value (prueba de dos colas)
        p_value = 2 * min(f.cdf(f_stat, df1, df2), 1 - f.cdf(f_stat, df1, df2))

        print(f"\nResultados:")
        print(f"  • Estadístico F = {f_stat:.4f}")
        print(f"  • Grados de libertad: d1={df1}, d2={df2}")
        print(f"  • p-value = {p_value:.6f} ({p_value*100:.4f}%)")

        if p_value > 0.05:
            print("  ✅ No se rechaza H₀: Las varianzas son iguales (p > 0.05)")
        else:
            print("  ❌ Se rechaza H₀: Las varianzas son diferentes (p ≤ 0.05)")

        # Visualización
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

        # Gráfico 1: Distribución F con estadístico
        x = np.linspace(0.001, max(5, f_stat * 1.5), 1000)
        y = f.pdf(x, df1, df2)

        ax1.plot(x, y, 'b-', linewidth=2, label=f'F({df1}, {df2})')
        ax1.axvline(x=f_stat, color='#E63946', linestyle='--', linewidth=2,
                   label=f'F = {f_stat:.2f}')

        # Área de rechazo (p-value)
        mask = x >= f_stat
        ax1.fill_between(x[mask], 0, y[mask], color='#E63946', alpha=0.3,
                        label=f'p-value = {p_value:.4f}')

        ax1.set_xlabel('F', fontsize=12)
        ax1.set_ylabel('Densidad de probabilidad', fontsize=12)
        ax1.set_title('Prueba F - Comparación de Varianzas', fontsize=13, fontweight='bold')
        ax1.grid(True, alpha=0.3)
        ax1.legend(loc='best')
        ax1.set_xlim(0, max(5, f_stat * 1.5))

        # Gráfico 2: Comparación de varianzas
        varianzas = [s1, s2]
        muestras = ['Muestra 1', 'Muestra 2']
        colores = ['#2E86AB', '#E63946']

        bars = ax2.bar(muestras, varianzas, color=colores, alpha=0.8, edgecolor='black')

        # Agregar valores en las barras
        for bar, var in zip(bars, varianzas):
            height = bar.get_height()
            ax2.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                    f'{var:.4f}', ha='center', va='bottom', fontsize=10)

        ax2.set_xlabel('Muestra', fontsize=12)
        ax2.set_ylabel('Varianza', fontsize=12)
        ax2.set_title('Comparación de Varianzas Muestrales', fontsize=13, fontweight='bold')
        ax2.grid(True, alpha=0.3, axis='y')

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"❌ Error: {e}")

#==========================================================
# FUNCIÓN PARA EJEMPLOS CLÁSICOS
#==========================================================

def ejemplos_f():
    """Ejecuta ejemplos clásicos de la distribución F"""

    print("\n" + "="*70)
    print("📚 EJEMPLOS DE DISTRIBUCIÓN F DE SNEDECOR")
    print("="*70)

    ejemplos = [
        {
            'titulo': 'Ejemplo 1: Valor crítico F',
            'descripcion': 'Con d1=3, d2=15 grados de libertad, ¿cuál es el valor crítico F para α=0.05 (prueba de una cola)?',
            'tipo': 'critico',
            'd1': 3,
            'd2': 15,
            'alfa': 0.05
        },
        {
            'titulo': 'Ejemplo 2: Probabilidad en F',
            'descripcion': 'Con d1=4, d2=10, ¿cuál es la probabilidad de que X sea mayor que 3?',
            'tipo': 'prob',
            'd1': 4,
            'd2': 10,
            'valor': 3
        },
        {
            'titulo': 'Ejemplo 3: Prueba F de varianzas',
            'descripcion': 'Muestra 1: n=12, s²=25. Muestra 2: n=15, s²=16. ¿Son las varianzas iguales?',
            'tipo': 'prueba_f',
            's1': 25,
            'n1': 12,
            's2': 16,
            'n2': 15
        }
    ]

    for i, ej in enumerate(ejemplos, 1):
        print(f"\n{i}. {ej['titulo']}")
        print(f"   {ej['descripcion']}")

        if ej['tipo'] == 'critico':
            f_crit = f.ppf(1 - ej['alfa'], ej['d1'], ej['d2'])
            print(f"   F crítico (α={ej['alfa']}, d1={ej['d1']}, d2={ej['d2']}) = {f_crit:.4f}")

        elif ej['tipo'] == 'prob':
            prob = 1 - f.cdf(ej['valor'], ej['d1'], ej['d2'])
            print(f"   P(X > {ej['valor']}) = {prob:.6f} ({prob*100:.2f}%)")

        elif ej['tipo'] == 'prueba_f':
            # Calcular estadístico F
            f_stat = max(ej['s1'], ej['s2']) / min(ej['s1'], ej['s2'])
            df1 = (ej['n1'] - 1) if ej['s1'] >= ej['s2'] else (ej['n2'] - 1)
            df2 = (ej['n2'] - 1) if ej['s1'] >= ej['s2'] else (ej['n1'] - 1)
            p_value = 2 * min(f.cdf(f_stat, df1, df2), 1 - f.cdf(f_stat, df1, df2))

            print(f"   F = {f_stat:.4f}")
            print(f"   p-value = {p_value:.6f} ({p_value*100:.2f}%)")
            if p_value > 0.05:
                print("   ✅ Las varianzas son iguales (p > 0.05)")
            else:
                print("   ❌ Las varianzas son diferentes (p ≤ 0.05)")

    print("\n" + "="*70)

#==========================================================
# EJECUTAR LA CALCULADORA
#==========================================================

if __name__ == "__main__":

    print("\n" + "="*70)
    print("🎯 CALCULADORA ESTADÍSTICA - DISTRIBUCIÓN F DE SNEDECOR")
    print("="*70)

    while True:
        print("\n¿Qué deseas hacer?")
        print("  1. Análisis personalizado")
        print("  2. Ver ejemplos clásicos")
        print("  3. Prueba F - Comparación de varianzas")
        print("  4. Salir")

        opcion = input("\nSelecciona una opción (1-4): ")

        if opcion == '1':
            resultados = calculadora_f()
        elif opcion == '2':
            ejemplos_f()
        elif opcion == '3':
            prueba_f()
        elif opcion == '4':
            print("\n✅ ¡Gracias por usar la calculadora estadística!")
            print("="*70)
            break
        else:
            print("❌ Opción no válida. Por favor, selecciona 1, 2, 3 o 4.")
```
