````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

# Configuración de la animación
n_total = 2000
intervalo = 50  # milisegundos entre frames
frames_guardar = 40  # número de frames en el GIF

# Pre-generamos todos los lanzamientos
np.random.seed(42)  # para reproducibilidad
todos_resultados = np.random.randint(1, 7, size=n_total)

# Configuración de la figura
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_xlim(0.5, 6.5)
ax.set_ylim(0, 450)
ax.set_xticks(range(1, 7))
ax.set_xlabel("Cara del dado")
ax.set_ylabel("Frecuencia")
ax.set_title("Evolución del lanzamiento de un dado\nJulio Hurtado")
ax.grid(True, alpha=0.3)

# Inicializar el histograma
n, bins, patches = ax.hist([], bins=np.arange(0.5, 7.5, 1),
                           rwidth=0.8, color="skyblue", edgecolor="black")

# Texto para mostrar el número de lanzamientos
texto_lanzamientos = ax.text(0.02, 0.95, '', transform=ax.transAxes,
                             fontsize=12, verticalalignment='top',
                             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

def update(frame):
    """Actualiza el histograma para cada frame"""
    # Número de lanzamientos para este frame
    n_lanzamientos = int((frame + 1) * n_total / frames_guardar)

    # Obtener los resultados hasta este momento
    resultados_actuales = todos_resultados[:n_lanzamientos]

    # Actualizar el histograma
    ax.clear()
    ax.hist(resultados_actuales, bins=np.arange(0.5, 7.5, 1),
            rwidth=0.8, color="skyblue", edgecolor="black")
    ax.set_xlim(0.5, 6.5)
    ax.set_ylim(0, 450)
    ax.set_xticks(range(1, 7))
    ax.set_xlabel("Cara del dado")
    ax.set_ylabel("Frecuencia")
    ax.set_title(f"Lanzamiento de un dado - Julio Hurtado\n{frame+1}/{frames_guardar}")
    ax.grid(True, alpha=0.3)

    # Mostrar estadísticas
    if n_lanzamientos > 0:
        frecuencias = [np.sum(resultados_actuales == i) for i in range(1, 7)]
        max_freq = max(frecuencias)
        min_freq = min(frecuencias)
        texto = f"Lanzamientos: {n_lanzamientos}\n"
        texto += f"Máx: {max_freq} | Mín: {min_freq}\n"
        texto += f"Rango: {max_freq - min_freq}"
        ax.text(0.02, 0.95, texto, transform=ax.transAxes,
                fontsize=10, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    return patches

# Crear la animación
anim = FuncAnimation(fig, update, frames=frames_guardar,
                    interval=intervalo, repeat=True)

# Guardar como GIF
anim.save('dado_animacion.gif', writer=PillowWriter(fps=20))
print("✅ GIF guardado como 'dado_animacion.gif'")

# Mostrar la animación (si estás en Jupyter notebook)
from IPython.display import Image
Image(filename='dado_animacion.gif')
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

# Configuración de la animación
n_total = 2000
intervalo = 50  # milisegundos entre frames
frames_guardar = 40  # número de frames en el GIF

# Pre-generamos todos los lanzamientos
np.random.seed(42)  # para reproducibilidad
todos_resultados = np.random.choice(["Cara", "Sello"], size=n_total)

# Configuración de la figura
fig, ax = plt.subplots(figsize=(8, 6))
ax.set_ylim(0, 1200)
ax.set_ylabel("Frecuencia")
ax.set_title("Evolución del lanzamiento de una moneda\nHurtado Julio")
ax.grid(True, alpha=0.3, axis='y')

# Colores para Cara y Sello
colores = ["skyblue", "lightcoral"]

# Función para inicializar
def init():
    ax.bar(["Cara", "Sello"], [0, 0], color=colores, edgecolor="black")
    ax.set_xlabel("Resultado")
    ax.set_ylabel("Frecuencia")
    return []

# Función de actualización para cada frame
def update(frame):
    """Actualiza el gráfico de barras para cada frame"""
    # Número de lanzamientos para este frame
    n_lanzamientos = int((frame + 1) * n_total / frames_guardar)

    # Obtener los resultados hasta este momento
    resultados_actuales = todos_resultados[:n_lanzamientos]

    # Limpiar el eje
    ax.clear()

    # Calcular frecuencias
    conteos = {outcome: np.sum(resultados_actuales == outcome) for outcome in ["Cara", "Sello"]}

    # Crear gráfico de barras
    barras = ax.bar(conteos.keys(), conteos.values(),
                    color=colores, edgecolor="black", alpha=0.8)

    # Configurar el gráfico
    ax.set_ylim(0, max(1200, max(conteos.values()) * 1.1))
    ax.set_xlabel("Resultado", fontsize=12)
    ax.set_ylabel("Frecuencia", fontsize=12)
    ax.set_title(f"Lanzamiento de una moneda - Hurtado Julio\nFrame: {frame+1}/{frames_guardar}", fontsize=14)
    ax.grid(True, alpha=0.3, axis='y')

    # Agregar valores sobre las barras
    for i, (outcome, freq) in enumerate(conteos.items()):
        ax.text(i, freq + 10, str(freq), ha='center', va='bottom', fontsize=11, fontweight='bold')

    # Mostrar estadísticas en el gráfico
    if n_lanzamientos > 0:
        diferencia = abs(conteos["Cara"] - conteos["Sello"])
        porcentaje_cara = (conteos["Cara"] / n_lanzamientos) * 100
        porcentaje_sello = (conteos["Sello"] / n_lanzamientos) * 100

        texto = f"Lanzamientos: {n_lanzamientos}\n"
        texto += f"Diferencia: {diferencia}\n"
        texto += f"Cara: {porcentaje_cara:.1f}% | Sello: {porcentaje_sello:.1f}%"

        ax.text(0.02, 0.95, texto, transform=ax.transAxes,
                fontsize=10, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.9, edgecolor='gray'))

    # Mostrar línea del valor esperado (1000 para 2000 lanzamientos)
    if n_lanzamientos > 0:
        esperado = n_lanzamientos / 2
        ax.axhline(y=esperado, color='green', linestyle='--', linewidth=1.5, alpha=0.7, label=f'Esperado: {esperado:.0f}')
        ax.legend(loc='upper right', fontsize=9)

    return barras

# Crear la animación
print("Generando animación...")
anim = FuncAnimation(fig, update, frames=frames_guardar,
                    init_func=init, interval=intervalo, repeat=True)

# Guardar como GIF
anim.save('moneda_animacion.gif', writer=PillowWriter(fps=20))
print("✅ GIF guardado como 'moneda_animacion.gif'")

# Mostrar estadísticas finales
print("\n📊 Estadísticas finales:")
print(f"Total lanzamientos: {n_total}")
print(f"Cara: {np.sum(todos_resultados == 'Cara')} veces")
print(f"Sello: {np.sum(todos_resultados == 'Sello')} veces")
print(f"Diferencia: {abs(np.sum(todos_resultados == 'Cara') - np.sum(todos_resultados == 'Sello'))}")

# Si estás en Jupyter notebook, puedes mostrar el GIF con:
from IPython.display import Image
Image(filename='moneda_animacion.gif')
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from scipy import stats

# Configuración
n_total = 2000
frames = 50
np.random.seed(42)

# Generar resultados
resultados = np.random.choice(["Cara", "Sello"], size=n_total)
resultados_numericos = np.where(resultados == "Cara", 1, 0)

# Crear figura con dos subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10))
fig.suptitle("Evolución del lanzamiento de una moneda\nHurtado Julio", fontsize=16)

colores = ["skyblue", "lightcoral"]

def update(frame):
    # Limpiar ejes
    ax1.clear()
    ax2.clear()

    # Número de lanzamientos
    n = int((frame + 1) * n_total / frames)
    resultados_actuales = resultados[:n]
    numericos_actuales = resultados_numericos[:n]

    # Gráfico 1: Barras de frecuencia
    conteos = [np.sum(resultados_actuales == "Cara"), np.sum(resultados_actuales == "Sello")]
    barras = ax1.bar(["Cara", "Sello"], conteos, color=colores, edgecolor="black", alpha=0.8)

    # Valores sobre las barras
    for i, (outcome, freq) in enumerate(zip(["Cara", "Sello"], conteos)):
        ax1.text(i, freq + 10, str(freq), ha='center', va='bottom', fontsize=10, fontweight='bold')

    # Línea del valor esperado
    esperado = n / 2
    ax1.axhline(y=esperado, color='green', linestyle='--', alpha=0.7, label=f'Esperado: {esperado:.0f}')

    # Configurar gráfico 1
    ax1.set_ylim(0, max(1200, max(conteos) * 1.1))
    ax1.set_ylabel("Frecuencia", fontsize=11)
    ax1.set_title(f"Frecuencias (n={n})")
    ax1.grid(True, alpha=0.3, axis='y')
    ax1.legend(loc='upper right')

    # Calcular proporción acumulada
    proporcion_cara = np.cumsum(numericos_actuales) / np.arange(1, n+1)

    # Gráfico 2: Evolución de la proporción
    ax2.plot(range(1, n+1), proporcion_cara, 'b-', linewidth=1.5, alpha=0.7, label='Proporción de Caras')
    ax2.axhline(y=0.5, color='red', linestyle='--', alpha=0.7, label='Valor teórico (0.5)')

    # Intervalos de confianza (95%)
    if n > 30:
        z = 1.96
        ic_inf = 0.5 - z * np.sqrt(0.5 * 0.5 / np.arange(1, n+1))
        ic_sup = 0.5 + z * np.sqrt(0.5 * 0.5 / np.arange(1, n+1))
        ax2.fill_between(range(1, n+1), ic_inf, ic_sup, alpha=0.2, color='gray', label='IC 95%')

    # Configurar gráfico 2
    ax2.set_xlim(0, n_total)
    ax2.set_ylim(0, 1)
    ax2.set_xlabel("Número de lanzamientos", fontsize=11)
    ax2.set_ylabel("Proporción", fontsize=11)
    ax2.set_title("Convergencia a la probabilidad teórica (0.5)")
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc='best')

    # Mostrar estadísticas
    texto = f"Lanzamientos: {n}\n"
    texto += f"Caras: {conteos[0]} ({conteos[0]/n*100:.1f}%)\n"
    texto += f"Sellos: {conteos[1]} ({conteos[1]/n*100:.1f}%)\n"
    texto += f"Diferencia: {abs(conteos[0]-conteos[1])}"

    ax1.text(0.02, 0.95, texto, transform=ax1.transAxes,
            fontsize=9, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    return barras

# Crear animación
anim = FuncAnimation(fig, update, frames=frames, interval=50, repeat=True)

# Guardar GIF
anim.save('moneda_completa.gif', writer=PillowWriter(fps=20))
print("✅ GIF guardado como 'moneda_completa.gif'")

# Mostrar resultado final
print(f"\n📊 Resultados finales (n={n_total}):")
print(f"Cara: {np.sum(resultados == 'Cara')} veces ({np.sum(resultados == 'Cara')/n_total*100:.1f}%)")
print(f"Sello: {np.sum(resultados == 'Sello')} veces ({np.sum(resultados == 'Sello')/n_total*100:.1f}%)")

# Si estás en Jupyter notebook, puedes mostrar el GIF con:
from IPython.display import Image
Image(filename='moneda_completa.gif')

````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

# Configuración de la animación
n_total = 200000
intervalo = 30  # milisegundos entre frames
frames_guardar = 60  # número de frames en el GIF
probabilidades = [0.5, 0.3, 0.2]  # Azul: 50%, Roja: 30%, Verde: 20%
colores = ["blue", "red", "green"]
nombres = ["Azul", "Roja", "Verde"]

# Pre-generamos todos los lanzamientos
np.random.seed(42)  # para reproducibilidad
todos_resultados = np.random.choice(nombres, size=n_total, p=probabilidades)

# Configuración de la figura
fig, ax = plt.subplots(figsize=(10, 7))
ax.set_ylim(0, n_total * max(probabilidades) * 1.2)
ax.set_ylabel("Frecuencia", fontsize=12)
ax.set_title("Evolución de extracción de bolas de una urna\nJulio Hurtado", fontsize=14)
ax.grid(True, alpha=0.3, axis='y')

# Valores teóricos esperados
valores_esperados = [n_total * p for p in probabilidades]

def update(frame):
    """Actualiza el gráfico de barras para cada frame"""
    # Número de extracciones para este frame
    n_extracciones = int((frame + 1) * n_total / frames_guardar)

    # Obtener los resultados hasta este momento
    resultados_actuales = todos_resultados[:n_extracciones]

    # Limpiar el eje
    ax.clear()

    # Calcular frecuencias
    frecuencias = []
    for nombre in nombres:
        freq = np.sum(resultados_actuales == nombre)
        frecuencias.append(freq)

    # Crear gráfico de barras
    barras = ax.bar(nombres, frecuencias, color=colores, edgecolor="black", alpha=0.85)

    # Agregar valores sobre las barras
    for i, (nombre, freq) in enumerate(zip(nombres, frecuencias)):
        ax.text(i, freq + n_total*0.01, str(freq), ha='center', va='bottom',
                fontsize=10, fontweight='bold')

        # Mostrar porcentaje
        porcentaje = (freq / n_extracciones) * 100 if n_extracciones > 0 else 0
        ax.text(i, freq/2, f"{porcentaje:.1f}%", ha='center', va='center',
                fontsize=9, color='white', fontweight='bold', alpha=0.9)

    # Configurar el gráfico
    ax.set_ylim(0, max(max(frecuencias) * 1.1, max(valores_esperados) * 1.1))
    ax.set_xlabel("Color de la bola", fontsize=12)
    ax.set_ylabel("Frecuencia", fontsize=12)
    ax.set_title(f"Extracción de bola de urna - Julio Hurtado\nExtracciones: {n_extracciones:,} / {n_total:,}", fontsize=12)
    ax.grid(True, alpha=0.3, axis='y')

    # Mostrar líneas de valores esperados
    for i, (nombre, esperado, color, prob) in enumerate(zip(nombres, valores_esperados, colores, probabilidades)):
        # Línea del valor esperado para el total
        ax.axhline(y=esperado, color=color, linestyle='--', linewidth=1.5, alpha=0.5)
        # Mostrar valor esperado al lado derecho
        ax.text(2.95, esperado, f' Esperado: {esperado:.0f}', ha='left', va='center',
                fontsize=8, color=color, alpha=0.7)

    # Mostrar estadísticas detalladas
    if n_extracciones > 0:
        frecuencias_relativas = [f/n_extracciones*100 for f in frecuencias]

        texto = f"📊 Estadísticas (n={n_extracciones:,})\n"
        texto += f"{'─' * 25}\n"
        for nombre, freq_rel, prob, color in zip(nombres, frecuencias_relativas, probabilidades, colores):
            diff = freq_rel - (prob*100)
            signo = "+" if diff > 0 else ""
            texto += f"● {nombre}: {freq_rel:.1f}% (esperado: {prob*100:.0f}%) {signo}{diff:.1f}%\n"
        texto += f"{'─' * 25}\n"
        texto += f"📈 Diferencia máx-mín: {max(frecuencias_relativas)-min(frecuencias_relativas):.1f}%"

        # Agregar recuadro de estadísticas
        ax.text(0.02, 0.97, texto, transform=ax.transAxes,
                fontsize=9, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.9, edgecolor='gray'))

    return barras

# Crear la animación
print("🎲 Generando animación de extracción de bolas...")
anim = FuncAnimation(fig, update, frames=frames_guardar, interval=intervalo, repeat=True)

# Guardar como GIF
anim.save('urna_animacion.gif', writer=PillowWriter(fps=20))
print("✅ GIF guardado como 'urna_animacion.gif'")

# Mostrar estadísticas finales
print("\n📊 Estadísticas finales:")
print(f"Total extracciones: {n_total:,}")
print(f"Probabilidades teóricas: Azul=50%, Roja=30%, Verde=20%")
print("\nResultados obtenidos:")
for nombre, prob in zip(nombres, probabilidades):
    freq = np.sum(todos_resultados == nombre)
    freq_rel = (freq / n_total) * 100
    diff = freq_rel - (prob*100)
    print(f"{nombre}: {freq:,} veces ({freq_rel:.2f}%) - Diferencia: {diff:+.2f}%")

# Mostrar error absoluto medio
error_medio = np.mean([abs(np.sum(todos_resultados == nombre)/n_total - prob)
                       for nombre, prob in zip(nombres, probabilidades)])
print(f"\n📐 Error absoluto medio: {error_medio:.4f} ({error_medio*100:.2f}%)")

# Si estás en Jupyter notebook, puedes mostrar el GIF con:
from IPython.display import Image
Image(filename='urna_animacion.gif')
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

# Configuración
n_total = 200000
frames = 80
probabilidades = [0.5, 0.3, 0.2]
colores = ["blue", "red", "green"]
nombres = ["Azul", "Roja", "Verde"]

# Generar datos
np.random.seed(42)
resultados = np.random.choice(nombres, size=n_total, p=probabilidades)

# Crear figura con 3 subplots
fig = plt.figure(figsize=(14, 8))
gs = fig.add_gridspec(2, 2, height_ratios=[2, 1])
ax1 = fig.add_subplot(gs[0, :])  # Gráfico de barras
ax2 = fig.add_subplot(gs[1, 0])  # Convergencia de proporciones
ax3 = fig.add_subplot(gs[1, 1])  # Error vs tamaño muestra

fig.suptitle("Evolución de extracción de bolas de una urna\nJulio Hurtado", fontsize=14, fontweight='bold')

def update(frame):
    # Limpiar ejes
    ax1.clear()
    ax2.clear()
    ax3.clear()

    # Número de extracciones
    n = int((frame + 1) * n_total / frames)
    resultados_actuales = resultados[:n]

    # === GRÁFICO 1: Barras de frecuencia ===
    frecuencias = [np.sum(resultados_actuales == nombre) for nombre in nombres]
    barras = ax1.bar(nombres, frecuencias, color=colores, edgecolor="black", alpha=0.8)

    # Agregar valores y porcentajes
    for i, (nombre, freq) in enumerate(zip(nombres, frecuencias)):
        if n > 0:
            porcentaje = freq / n * 100
            ax1.text(i, freq + n_total*0.008, f"{freq:,}", ha='center', va='bottom', fontsize=9)
            ax1.text(i, freq/2, f"{porcentaje:.1f}%", ha='center', va='center',
                    fontsize=10, color='white', fontweight='bold')

    # Líneas de valores esperados teóricos
    for i, (nombre, prob, color) in enumerate(zip(nombres, probabilidades, colores)):
        esperado_total = n_total * prob
        ax1.axhline(y=esperado_total, color=color, linestyle='--', alpha=0.5, linewidth=1)

    ax1.set_ylim(0, max(max(frecuencias) * 1.1, n_total * max(probabilidades) * 1.1))
    ax1.set_ylabel("Frecuencia", fontsize=11)
    ax1.set_title(f"Frecuencias observadas (n = {n:,})", fontsize=11)
    ax1.grid(True, alpha=0.3, axis='y')

    # === GRÁFICO 2: Convergencia de proporciones ===
    for idx, (nombre, prob, color) in enumerate(zip(nombres, probabilidades, colores)):
        # Calcular proporción acumulada
        mascara = (resultados[:n] == nombre)
        proporciones = np.cumsum(mascara) / np.arange(1, n+1)

        # Mostrar solo cada cierto punto para rendimiento
        paso = max(1, n // 500)
        ax2.plot(np.arange(1, n+1, paso), proporciones[::paso],
                color=color, linewidth=1.5, alpha=0.7, label=f'{nombre} ({prob*100:.0f}%)')

    ax2.axhline(y=0.5, color='blue', linestyle=':', alpha=0.5, linewidth=1)
    ax2.axhline(y=0.3, color='red', linestyle=':', alpha=0.5, linewidth=1)
    ax2.axhline(y=0.2, color='green', linestyle=':', alpha=0.5, linewidth=1)

    ax2.set_xlim(1, n_total)
    ax2.set_ylim(0, 1)
    ax2.set_xlabel("Número de extracciones", fontsize=10)
    ax2.set_ylabel("Proporción", fontsize=10)
    ax2.set_title("Convergencia a probabilidades teóricas", fontsize=11)
    ax2.legend(loc='best', fontsize=8)
    ax2.grid(True, alpha=0.3)
    ax2.set_xscale('log')

    # === GRÁFICO 3: Error absoluto ===
    errores = []
    tamanos = []
    for step in range(1, n+1, max(1, n//200)):
        sub_resultados = resultados[:step]
        errores_actuales = []
        for nombre, prob in zip(nombres, probabilidades):
            freq_rel = np.sum(sub_resultados == nombre) / step
            errores_actuales.append(abs(freq_rel - prob))
        errores.append(np.mean(errores_actuales))
        tamanos.append(step)

    ax3.plot(tamanos, errores, 'purple', linewidth=2, alpha=0.7)
    ax3.set_xscale('log')
    ax3.set_yscale('log')
    ax3.set_xlabel("Número de extracciones", fontsize=10)
    ax3.set_ylabel("Error absoluto medio", fontsize=10)
    ax3.set_title("Evolución del error (escala log-log)", fontsize=11)
    ax3.grid(True, alpha=0.3, which='both')

    # Línea teórica 1/√n
    x_teorico = np.array([1, n_total])
    y_teorico = 0.3 / np.sqrt(x_teorico)
    ax3.plot(x_teorico, y_teorico, 'k--', alpha=0.5, linewidth=1, label='∝ 1/√n')
    ax3.legend(fontsize=8)

    # Estadísticas generales
    if n > 0:
        frecuencias_rel = [f/n for f in frecuencias]
        texto = f"n = {n:,}\n"
        texto += f"{'─' * 20}\n"
        for nombre, freq_rel, prob, color in zip(nombres, frecuencias_rel, probabilidades, colores):
            diff = (freq_rel - prob) * 100
            texto += f"{nombre}: {freq_rel*100:.2f}% ({diff:+.2f}%)\n"
        texto += f"{'─' * 20}\n"
        texto += f"Error medio: {np.mean([abs(freq_rel - prob) for freq_rel, prob in zip(frecuencias_rel, probabilidades)])*100:.3f}%"

        ax1.text(0.02, 0.97, texto, transform=ax1.transAxes,
                fontsize=8, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='white', alpha=0.9, edgecolor='gray'))

    return barras

# Crear animación
print("🎨 Generando animación completa...")
anim = FuncAnimation(fig, update, frames=frames, interval=40, repeat=True)

# Guardar GIF
anim.save('urna_completa.gif', writer=PillowWriter(fps=20))
print("✅ GIF guardado como 'urna_completa.gif'")

# Resultados finales
print("\n" + "="*50)
print("📊 RESULTADOS FINALES")
print("="*50)
for nombre, prob in zip(nombres, probabilidades):
    freq = np.sum(resultados == nombre)
    freq_rel = freq / n_total * 100
    diff = freq_rel - (prob*100)
    esperado = n_total * prob
    print(f"\n{nombre}:")
    print(f"  ● Observado: {freq:,} veces ({freq_rel:.2f}%)")
    print(f"  ● Esperado: {esperado:.0f} veces ({prob*100:.0f}%)")
    print(f"  ● Diferencia: {diff:+.2f}%")

print(f"\n📐 Error absoluto medio final: {np.mean([abs(np.sum(resultados == nombre)/n_total - prob) for nombre, prob in zip(nombres, probabilidades)])*100:.3f}%")
print(f"\n✨ Ley de los Grandes Números: A medida que n crece, las frecuencias relativas se acercan a las probabilidades teóricas.")

# Si estás en Jupyter notebook, puedes mostrar el GIF con:
from IPython.display import Image
Image(filename='urna_completa.gif')
````

````python
#@title **9. Distribución de Probabilidad para Lanzamos muchas veces un par de dados y contamos cuantas veces sale la suma de resultados - Ley de los grandes números**

from collections import defaultdict
import matplotlib.pyplot as plt

# Possible outcomes for each die
dado_outcomes = range(1, 7)

# Calculate the frequency of each sum
sum_frequencies = defaultdict(int)
for d1 in dado_outcomes:
    for d2 in dado_outcomes:
        sum_frequencies[d1 + d2] += 1

# Calculate the total number of possible outcomes
total_outcomes = len(dado_outcomes) * len(dado_outcomes)

# Calculate the probability for each sum
sum_probabilities = {sum_val: freq / total_outcomes for sum_val, freq in sum_frequencies.items()}

# Sort the outcomes for plotting
sorted_sums = sorted(sum_probabilities.keys())
probabilities = [sum_probabilities[s] for s in sorted_sums]

# Create a bar plot of the probability distribution
colors = plt.cm.viridis(np.linspace(0, 1, len(sorted_sums))) # Using a colormap for variety
plt.bar(sorted_sums, probabilities, color=colors, edgecolor="black")
plt.xticks(sorted_sums)
plt.title("Distribución de Probabilidad (Suma de dos dados) - Julio Hurtado")
plt.xlabel("Suma")
plt.ylabel("Probabilidad")

# Add probability labels on top of each bar
for i, prob in enumerate(probabilities):
    plt.text(sorted_sums[i], prob, f'{prob:.3f}', ha='center', va='bottom')

plt.show()
````
